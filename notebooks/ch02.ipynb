## Section 2.1 (Data Types)

### 2.1.1. Nominal attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های اسمی</span> 📚
</p>
<p style="font-size:16pt; text-align:right;">
<b>ویژگی‌های اسمی (Nominal Attributes)</b> متغیرهای کیفی هستند که نمایانگر دسته‌ها یا برچسب‌های متمایز با هیچ‌گونه ترتیب یا رتبه‌بندی ذاتی بین آن‌ها می‌باشند. این ویژگی‌ها داده‌ها را بر اساس تفاوت‌های کیفی به گروه‌های مختلف طبقه‌بندی می‌کنند. به عنوان مثال، <b>رنگ‌ها</b>، <b>جنسیت</b>، <b>انواع حیوانات</b> و موارد مشابه، همگی از ویژگی‌های اسمی به شمار می‌روند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای ویژگی‌های اسمی</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون ساده با استفاده از کتابخانه <b>pandas</b> برای کار با یک مجموعه داده حاوی ویژگی‌های اسمی ایجاد کنیم. ما یک مجموعه داده نمونه را بارگذاری کرده، ویژگی‌های اسمی را کاوش می‌کنیم و عملیات <b>One-Hot Encoding</b> را انجام می‌دهیم.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده، ما یه کتابخونه به اسم 'pandas' رو وارد می‌کنیم.
# 'pandas' یه ابزار خیلی پرکاربرده برای کار کردن با داده‌ها، مخصوصاً وقتی داده‌ها توی یه جدول مرتب باشن.
# ما از این کتابخونه استفاده می‌کنیم که داده‌ها رو مثل یه جدول اکسل بخونیم و دستکاریشون کنیم.

# این دیکشنری (dictionary) حاوی داده‌های نمونه ماست.
# ما این داده‌ها رو به صورت ستون به ستون تعریف کردیم، مثلاً ستون 'Animal' یا 'Color'.
# این داده‌ها از نوع اسمی (Nominal) هستن؛ یعنی فقط اسم یا دسته هستن و هیچ ترتیبی بینشون وجود نداره.
# مثلاً 'Dog' از 'Cat' بهتر یا بدتر نیست، فقط یه دسته متفاوته.
data = {
    'Animal': ['Dog', 'Cat', 'Fish', 'Bird', 'Snake'],
    'Color': ['Brown', 'Black', 'Gold', 'Blue', 'Green'],
    'Habitat': ['Forest', 'Home', 'Aquarium', 'Sky', 'Jungle']
}

# حالا با استفاده از 'pandas'، این داده‌ها رو به یه 'DataFrame' تبدیل می‌کنیم.
# 'DataFrame' همون جدولیه که 'pandas' باهاش کار می‌کنه و داده‌ها رو به شکل منظم نشون می‌ده.
df = pd.DataFrame(data)

# 'print' یه تابع ساده‌اس که هر چیزی رو بهمون نشون می‌ده. اینجا 'DataFrame' اصلی رو چاپ می‌کنیم
# تا ببینیم داده‌های خام ما قبل از هر پردازشی چه شکلی هستن.
print("دیتاست اصلی:")
print(df)
print("\n")

# این یه لیست از ستون‌های اسمیه که می‌خوایم بعداً روشون عملیات انجام بدیم.
nominal_attributes = ['Animal', 'Color', 'Habitat']

# این یه حلقه (for loop) ساده‌اس که روی تک‌تک ستون‌های اسمی که تو لیست بالا گذاشتیم، می‌چرخه.
# داخل حلقه، از تابع 'unique' استفاده می‌کنیم که مقادیر منحصر به فرد (Unique values) هر ستون رو پیدا کنه و بهمون نشون بده.
# این کار کمک می‌کنه بفهمیم هر ستون چند تا دسته یا 'Category' مختلف داره.
for attribute in nominal_attributes:
    print(f"مقادیر یکتا در {attribute}: {df[attribute].unique()}")

# این مهم‌ترین بخش کده و مفهوم 'One-Hot Encoding' رو اجرا می‌کنه.
# 'pd.get_dummies' یه تابع از 'pandas' هست که ستون‌های اسمی رو می‌گیره
# و هر کدوم از مقادیر اون ستون رو به یک ستون جدید تبدیل می‌کنه.
# تو این ستون‌های جدید، اگه مقدار اصلی وجود داشته باشه، عدد 1 و اگه نداشته باشه، عدد 0 می‌ذاره.
# این روش برای آماده‌سازی داده‌های اسمی برای الگوریتم‌های یادگیری ماشین خیلی مهمه،
# چون بیشتر الگوریتم‌ها با اعداد کار می‌کنن، نه با کلمات.
df_encoded = pd.get_dummies(df, columns=nominal_attributes)

# اینجا 'DataFrame' جدید رو بعد از این که 'One-Hot Encoding' انجام شده، چاپ می‌کنیم.
# می‌تونید ببینید که ستون‌های اسمی قدیمی حذف شدن و ستون‌های عددی جدیدی به جای اون‌ها اضافه شدن.
print("\nدیتاست بعد از One-Hot Encoding:")
print(df_encoded)
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیح مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال:
<ul>
<li>ما یک دیتافریم (DataFrame) پانداس با سه ویژگی اسمی: <b>'Animal'</b> (حیوان)، <b>'Color'</b> (رنگ) و <b>'Habitat'</b> (زیستگاه) ایجاد می‌کنیم.</li>
<li>ما مقادیر منحصربه‌فرد (Unique Values) را در هر ویژگی اسمی کاوش می‌کنیم.</li>
<li>ما از تابع <b>pd.get_dummies()</b> برای انجام <b>one-hot encoding</b> استفاده می‌کنیم و هر ویژگی اسمی را به ستون‌های دودویی (Binary) تبدیل می‌کنیم.</li>
<li>دیتافریم نهایی (<b>df_encoded</b>) پس از عملیات <b>one-hot encoding</b> نمایش داده می‌شود.</li>
</ul>
</p>
</div>

### 2.1.2. Binary attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های دودویی</span> 0️⃣1️⃣
</p>
<p style="font-size:16pt; text-align:right;">
<b>ویژگی‌های دودویی (Binary Attributes)</b> متغیرهای کیفی هستند که می‌توانند یکی از دو مقدار ممکن را بپذیرند و معمولاً نشان‌دهنده حضور یا عدم حضور یک ویژگی خاص می‌باشند. این ویژگی‌ها در بسیاری از مجموعه‌داده‌ها از اهمیت بالایی برخوردارند و نمونه‌هایی از آن‌ها شامل <b>Yes/No</b> (بله/خیر)، <b>True/False</b> (درست/نادرست)، <b>1/0</b> یا هر دو دسته متمایز دیگر می‌شوند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای ویژگی‌های دودویی</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای کار با یک مجموعه داده حاوی ویژگی‌های دودویی ایجاد کنیم. ما یک مجموعه داده نمونه را بارگذاری کرده، ویژگی‌های دودویی را کاوش می‌کنیم و عملیات‌های پایه را روی آن‌ها انجام می‌دهیم.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده. مثل قبل، کتابخونه 'pandas' رو وارد می‌کنیم.
# 'pandas' ابزار اصلی ما برای کار کردن با داده‌ها و جدول‌هاست.
# ما از این کتابخونه استفاده می‌کنیم که داده‌ها رو مثل یه جدول اکسل بخونیم و دستکاریشون کنیم.

# این دیکشنری (dictionary) داده‌های نمونه رو در خودش نگه داشته.
# این بار، داده‌های ما از نوع باینری (Binary) هستن. یعنی فقط دو مقدار ممکن دارن، مثل ۱ و ۰.
# اینجا 'PassedExam' نشون می‌ده که دانشجو امتحان رو قبول شده (۱) یا نه (۰).
# 'EnrolledInCourse' هم نشون می‌ده که دانشجو تو اون درس ثبت‌نام کرده (۱) یا نه (۰).
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'PassedExam': [1, 0, 1, 1, 0],
    'EnrolledInCourse': [1, 1, 0, 1, 0]
}

# حالا با استفاده از 'pandas'، این داده‌ها رو به یه 'DataFrame' تبدیل می‌کنیم.
# 'DataFrame' همون جدولیه که 'pandas' باهاش کار می‌کنه و داده‌ها رو به شکل منظم نشون می‌ده.
df = pd.DataFrame(data)

# 'print' یه تابع ساده‌اس که هر چیزی رو بهمون نشون می‌ده. اینجا 'DataFrame' اصلی رو چاپ می‌کنیم
# تا ببینیم داده‌های خام ما قبل از هر پردازشی چه شکلی هستن.
print("دیتاست اصلی:")
print(df)
print("\n")

# اینجا یه لیست از ستون‌های باینری رو تعریف می‌کنیم.
binary_attributes = ['PassedExam', 'EnrolledInCourse']

# این یه حلقه ساده‌اس که روی ستون‌های باینری می‌چرخه.
# تابع 'value_counts()' رو صدا می‌زنیم که تعداد تکرار هر مقدار رو تو اون ستون بشماره.
# مثلاً بهمون می‌گه چند تا دانشجو امتحان رو قبول شدن (۱) و چند تا رد (۰).
# این کار برای فهمیدن توزیع داده‌ها خیلی مفیده.
for attribute in binary_attributes:
    print(f"تعداد هر مقدار برای {attribute}:\n{df[attribute].value_counts()}\n")

# اینجا یه عملیات ساده ریاضی روی ستون‌های باینری انجام می‌دیم.
# 'df['TotalAttributes']' یه ستون جدید می‌سازه که جمع مقادیر دو ستون دیگه رو تو خودش نگه می‌داره.
# این کار بهمون نشون می‌ده که یه دانشجو چند تا از این دو ویژگی رو داره.
df['TotalAttributes'] = df['PassedExam'] + df['EnrolledInCourse']

# در نهایت، 'DataFrame' رو بعد از انجام عملیات نمایش می‌دیم.
# می‌تونید ستون جدید 'TotalAttributes' رو ببینید که اضافه شده.
print("دیتاست بعد از انجام عملیات روی ویژگی‌های باینری:")
print(df)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال:
<ul>
<li>ما یک دیتافریم (DataFrame) پانداس با دو ویژگی دودویی: <b>'PassedExam'</b> (آزمون گذرانده شده) و <b>'EnrolledInCourse'</b> (در دوره ثبت‌نام شده) ایجاد می‌کنیم.</li>
<li>ما تعداد رکوردهای هر مقدار منحصربه‌فرد را در ویژگی‌های دودویی با استفاده از تابع <b>value_counts()</b> کاوش می‌کنیم.</li>
<li>ما یک عملیات پایه (جمع) را روی ویژگی‌های دودویی انجام می‌دهیم تا یک ویژگی جدید به نام <b>'TotalAttributes'</b> (مجموع ویژگی‌ها) ایجاد کنیم.</li>
<li>دیتافریم نهایی (<b>df</b>) پس از این عملیات نمایش داده می‌شود.</li>
</ul>

### 2.1.3. Ordinal attributes

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های ترتیبی</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
<b>ویژگی‌های ترتیبی (Ordinal Attributes)</b> متغیرهای کیفی هستند که دارای یک ترتیب یا رتبه‌بندی معنادار بین دسته‌ها می‌باشند. برخلاف ویژگی‌های اسمی، ویژگی‌های ترتیبی یک توالی روشن و معنادار دارند، اما لزوماً فواصل بین آن‌ها یکنواخت یا به‌خوبی تعریف‌شده نیست. نمونه‌هایی از ویژگی‌های ترتیبی شامل <b>سطوح تحصیلی</b> (مانند ابتدایی، دبیرستان، دانشگاه)، <b>رتبه‌بندی رضایت مشتری</b> یا <b>طبقات اجتماعی-اقتصادی</b> می‌شوند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای ویژگی‌های ترتیبی</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای کار با یک مجموعه داده حاوی ویژگی‌های ترتیبی ایجاد کنیم. ما یک مجموعه داده نمونه را بارگذاری کرده، ویژگی‌های ترتیبی را کاوش می‌کنیم و نحوه رمزگذاری آن‌ها را برای حفظ رابطه ترتیبی نمایش می‌دهیم.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده و کتابخونه 'pandas' رو وارد می‌کنه.
# 'pandas' یه ابزار قدرتمند برای کار با داده‌هاست، مخصوصاً وقتی داده‌ها به شکل جدول باشن.
# ما از این کتابخونه برای ساختن و دستکاری 'DataFrame' استفاده می‌کنیم.

# این دیکشنری داده‌های نمونه ما رو داره. این بار، داده‌ها از نوع ترتیبی (Ordinal) هستن.
# ویژگی‌های ترتیبی دسته‌بندی‌هایی هستن که بینشون یه ترتیب یا رتبه مشخص وجود داره.
# مثلاً 'EducationLevel' یه ترتیب منطقی داره: 'Elementary' از 'High School' پایین‌تره و 'High School' از 'College' پایین‌تره.
# 'SatisfactionRating' هم یک ویژگی ترتیبیه، چون ۵ از ۴ بهتره و ۴ از ۳.
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'EducationLevel': ['High School', 'College', 'Elementary', 'College', 'High School'],
    'SatisfactionRating': [3, 5, 2, 4, 1]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
# 'DataFrame' همون جدول داده‌هاست که 'pandas' باهاش کار می‌کنه.
df = pd.DataFrame(data)

# 'DataFrame' اصلی رو چاپ می‌کنیم تا ببینیم داده‌های خام چه شکلی هستن.
print("دیتاست اصلی:")
print(df)
print("\n")

# این یه لیست از ستون‌های ترتیبیه که می‌خوایم بررسی‌شون کنیم.
ordinal_attributes = ['EducationLevel', 'SatisfactionRating']

# این یه حلقه (for loop) ساده‌اس که روی ستون‌های ترتیبی می‌چرخه.
# تابع 'unique' مقادیر منحصر به فرد هر ستون رو پیدا می‌کنه و بهمون نشون می‌ده.
# با دیدن مقادیر، می‌تونیم ترتیب موجود بین اون‌ها رو حدس بزنیم.
for attribute in ordinal_attributes:
    print(f"مقادیر یکتا در {attribute}: {df[attribute].unique()}")

# اینجا مهم‌ترین بخش کده که عملیات رمزگذاری (Encoding) رو انجام می‌ده.
# ما یه دیکشنری 'education_level_mapping' ساختیم که هر سطح تحصیلی رو به یه عدد مرتبط تبدیل می‌کنه.
# مثلاً 'Elementary' به عدد ۱، 'High School' به ۲ و 'College' به ۳ نگاشت (map) می‌شن.
# این کار داده‌های متنی رو به اعدادی تبدیل می‌کنه که ترتیب رو حفظ می‌کنن.
education_level_mapping = {'Elementary': 1, 'High School': 2, 'College': 3}
df['EducationLevelEncoded'] = df['EducationLevel'].map(education_level_mapping)

# 'print' نهایی برای نمایش 'DataFrame' بعد از عملیات رمزگذاری هست.
# می‌تونید ببینید که یک ستون جدید به اسم 'EducationLevelEncoded' اضافه شده که مقادیر عددی ترتیب‌دار رو داره.
print("\nدیتاست بعد از رمزگذاری ویژگی‌های ترتیبی:")
print(df)
```

<li>ما مقادیر منحصربه‌فرد (Unique Values) را در هر ویژگی ترتیبی کاوش می‌کنیم.</li>

<li>دیتافریم نهایی (<b>df</b>) پس از این عملیات نمایش داده می‌شود.</li>

### 2.1.4. Numeric attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های عددی</span> 🔢
</p>
<p style="font-size:16pt; text-align:right;">
<b>ویژگی‌های عددی (Numeric Attributes)</b> نمایانگر کمیت‌ها هستند و می‌توانند مقادیر عددی را بپذیرند. این ویژگی‌ها به دو دسته اصلی تقسیم می‌شوند: گسسته (Discrete) و پیوسته (Continuous).
<br>
<br>
<b>ویژگی‌های عددی گسسته</b> تنها می‌توانند مقادیر متمایز و جداگانه را بپذیرند (مثلاً تعداد اتاق خواب در یک خانه 🏠)، در حالی که <b>ویژگی‌های عددی پیوسته</b> می‌توانند هر مقداری را در یک بازه معین داشته باشند (مثلاً قد 📏 یا وزن).
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای ویژگی‌های عددی</span> 🔢
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای کار با یک مجموعه داده حاوی ویژگی‌های عددی ایجاد کنیم. ما یک مجموعه داده نمونه را بارگذاری کرده، ویژگی‌های عددی را کاوش می‌کنیم و عملیات‌های پایه را روی آن‌ها انجام می‌دهیم.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده و کتابخونه 'pandas' رو وارد می‌کنه.
# 'pandas' یه ابزار خیلی قدرتمنده برای کار کردن با داده‌ها و جدول‌ها.

# این دیکشنری حاوی داده‌های نمونه ماست.
# این بار، داده‌ها از نوع عددی (Numeric) هستن. یعنی مقادیرشون اعداد واقعی هستن که می‌تونیم روشون عملیات ریاضی انجام بدیم.
# مثلاً 'Age' (سن)، 'Height (cm)' (قد) و 'Score' (امتیاز) همگی ویژگی‌های عددی هستن.
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'Age': [21, 19, 22, 20, 23],
    'Height (cm)': [175, 160, 180, 165, 185],
    'Score': [85, 92, 78, 89, 95]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
# 'DataFrame' همون جدولیه که 'pandas' باهاش کار می‌کنه.
df = pd.DataFrame(data)

# 'DataFrame' اصلی رو چاپ می‌کنیم تا ببینیم داده‌های خام چه شکلی هستن.
print("دیتاست اصلی:")
print(df)
print("\n")

# این یه لیست از ستون‌های عددیه که می‌خوایم روی اون‌ها عملیات انجام بدیم.
numeric_attributes = ['Age', 'Height (cm)', 'Score']

# 'describe()' یه تابع خیلی مفیده که خلاصه‌ای از آمار توصیفی رو برای ستون‌های عددی بهمون می‌ده.
# مثلاً میانگین (mean)، انحراف معیار (std)، حداقل (min)، حداکثر (max) و چارک‌ها رو محاسبه می‌کنه.
# این آمار بهمون کمک می‌کنه که یه دید کلی از توزیع داده‌ها به دست بیاریم.
print("آمار توصیفی برای ویژگی‌های عددی:")
print(df[numeric_attributes].describe())

# این بخش، داده‌ها رو به صورت **نرمال‌سازی (Normalization)** درمیاره.
# نرمال‌سازی یه تکنیک پیش‌پردازشه که مقادیر عددی رو طوری تغییر می‌ده که میانگینشون صفر و انحراف معیارشون یک بشه.
# این کار برای بعضی از الگوریتم‌های یادگیری ماشین که به مقیاس داده‌ها حساس هستن، ضروریه.
# فرمولش اینه: مقدار نرمال‌شده = (مقدار اصلی - میانگین) / انحراف معیار
df['NormalizedScore'] = (df['Score'] - df['Score'].mean()) / df['Score'].std()

# در نهایت، 'DataFrame' رو بعد از انجام عملیات نمایش می‌دیم.
# می‌تونید ستون جدید 'NormalizedScore' رو ببینید که مقادیرش نرمال‌سازی شدن.
print("\nدیتاست بعد از انجام عملیات روی ویژگی‌های عددی:")
print(df)
```

در این مثال:
<ul>
<li>ما یک دیتافریم (DataFrame) پانداس با سه ویژگی عددی: <b>'Age'</b> (سن)، <b>'Height (cm)'</b> (قد به سانتی‌متر) و <b>'Score'</b> (امتیاز) ایجاد می‌کنیم.</li>

<li>ما یک عملیات پایه (نرمال‌سازی) را روی ویژگی <b>'Score'</b> انجام می‌دهیم تا یک ویژگی جدید به نام <b>'NormalizedScore'</b> (امتیاز نرمال‌شده) ایجاد کنیم.</li>
<li>دیتافریم نهایی (<b>df</b>) پس از این عملیات نمایش داده می‌شود.</li>

### 2.1.5. Discrete vs. continuous attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های گسسته و پیوسته</span> 🔢
</p>

<p style="font-size:22pt; text-align:right;">
<br>
<b>ویژگی‌های گسسته (Discrete Attributes):</b>
</p>
<p style="font-size:16pt; text-align:right;">
<b>تعریف:</b> ویژگی‌های گسسته تنها می‌توانند مقادیر مشخص و جداگانه را بپذیرند.
<br>
<b>مثال‌ها:</b> تعداد اتاق خواب در یک خانه 🏠، تعداد آیتم‌ها در یک سبد خرید 🛒، تعداد دانشجو در یک کلاس 🧑‍🎓.
<br>
<b>ماهیت:</b> این ویژگی‌ها اغلب با اعداد کامل (صحیح) شمارش می‌شوند و دارای مرزهای مشخصی بین مقادیر هستند.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>ویژگی‌های پیوسته (Continuous Attributes):</b>
</p>
<p style="font-size:16pt; text-align:right;">
<b>تعریف:</b> ویژگی‌های پیوسته می‌توانند هر مقداری را در یک بازه معین داشته باشند.
<br>
<b>مثال‌ها:</b> قد 📏، وزن ⚖️، دما 🌡️ و هر نوع اندازه‌گیری که می‌تواند مقادیر اعشاری داشته باشد.
<br>
<b>ماهیت:</b> این ویژگی‌ها دارای مجموعه‌ای پیوسته و نامحدود از مقادیر ممکن هستند، که آن‌ها را برای اندازه‌گیری مناسب می‌سازد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای ویژگی‌های گسسته و پیوسته</span> 🔢
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای کار با یک مجموعه داده حاوی هر دو ویژگی گسسته و پیوسته ایجاد کنیم. ما یک مجموعه داده نمونه را بارگذاری کرده، ماهیت هر نوع را کاوش می‌کنیم و عملیات‌های پایه را روی آن‌ها انجام می‌دهیم.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده و کتابخونه 'pandas' رو وارد می‌کنه.
# 'pandas' یه ابزار خیلی پرکاربرده برای کار کردن با داده‌ها و جدول‌ها.

# این دیکشنری (dictionary) حاوی داده‌های نمونه ماست.
# این بار، داده‌ها هم از نوع گسسته (Discrete) و هم از نوع پیوسته (Continuous) هستن.
# ویژگی‌های گسسته (Discrete) مقادیری رو می‌گیرن که قابل شمارش هستن، مثل 'NumCourses' (تعداد درس‌ها).
# ویژگی‌های پیوسته (Continuous) مقادیری رو می‌گیرن که می‌تونن هر عددی باشن، مثل 'GPA' (معدل) یا 'Income' (درآمد).
data = {
    'StudentID': [1, 2, 3, 4, 5],
    'NumCourses': [4, 5, 3, 6, 4],  # ویژگی گسسته (تعداد درس‌ها)
    'GPA': [3.5, 4.0, 3.2, 3.8, 3.9],  # ویژگی پیوسته (معدل)
    'Income': [25000, 30000, 20000, 35000, 32000]  # ویژگی پیوسته (درآمد به دلار)
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
# 'DataFrame' همون جدولیه که 'pandas' باهاش کار می‌کنه.
df = pd.DataFrame(data)

# 'DataFrame' اصلی رو چاپ می‌کنیم تا ببینیم داده‌های خام چه شکلی هستن.
print("دیتاست اصلی:")
print(df)
print("\n")

# این بخش نوع ویژگی‌ها رو بررسی و چاپ می‌کنه.
print("نوع ویژگی‌ها:")
# با یه حلقه 'for'، روی ستون‌های 'DataFrame' می‌چرخیم.
# با 'df[column].dtype' می‌تونیم نوع داده هر ستون رو بفهمیم.
# اگه نوع داده 'int' (عدد صحیح) باشه، می‌گیم گسسته‌اس.
# اگه نوع داده 'float' (عدد اعشاری) باشه، می‌گیم پیوسته‌اس.
for column in df.columns:
    if df[column].dtype == 'int':
        print(f"{column} یک ویژگی گسسته است.")
    elif df[column].dtype == 'float':
        print(f"{column} یک ویژگی پیوسته است.")

# اینجا یک عملیات ساده روی ویژگی پیوسته 'Income' انجام می‌دیم.
# ستون جدیدی به اسم 'ScaledIncome' می‌سازیم که مقدار 'Income' رو بر ۱۰۰۰ تقسیم می‌کنه.
# این کار داده‌ها رو برای خوندن راحت‌تر می‌کنه و بهش می‌گن 'مقیاس‌گذاری' (Scaling).
df['ScaledIncome'] = df['Income'] / 1000  # مقیاس‌گذاری درآمد برای خوانایی بیشتر

# در نهایت، 'DataFrame' رو بعد از انجام عملیات نمایش می‌دیم.
# می‌تونید ستون جدید 'ScaledIncome' رو ببینید که اضافه شده.
print("\nدیتاست بعد از انجام عملیات روی ویژگی‌های پیوسته:")
print(df)
```

<p style="font-size:24pt; text-align:right;">
در این مثال:
<ul>
<li>ما یک دیتافریم (DataFrame) پانداس با سه ویژگی: <b>'NumCourses'</b> (تعداد دوره‌ها) از نوع گسسته، <b>'GPA'</b> از نوع پیوسته، و <b>'Income'</b> (درآمد) از نوع پیوسته ایجاد می‌کنیم.</li>

### 2.2.1. Measuring the central tendency

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارهای گرایش به مرکز</span> 📏
</p>
<p style="font-size:16pt; text-align:right;">
اندازه‌گیری گرایش به مرکز، روشی است برای خلاصه کردن یک مجموعه داده با شناسایی مقدار مرکزی یا میانگین آن. سه معیار متداول برای گرایش به مرکز وجود دارد: <b>میانگین (mean)</b>، <b>میانه (median)</b> و <b>نما (mode)</b>.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیح معیارهای گرایش به مرکز</span> 📐
</p>

<p style="font-size:16pt; text-align:right;">
<br>
<b>میانگین (Average):</b>
<br>
این معیار با جمع کردن تمام مقادیر یک مجموعه داده و تقسیم حاصل جمع بر تعداد مقادیر محاسبه می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>میانه (Median):</b>
<br>
میانه مقدار میانی یک مجموعه داده است، زمانی که به ترتیب صعودی یا نزولی مرتب شده باشد. اگر مجموعه داده دارای تعداد زوجی از مقادیر باشد، میانه، میانگین دو مقدار میانی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>نما (Mode):</b>
<br>
نما مقداری است که بیشترین تکرار را در یک مجموعه داده دارد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای معیارهای گرایش به مرکز</span> 📐
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای محاسبه معیارهای گرایش به مرکز (میانگین، میانه و نما) در یک مجموعه داده نمونه ایجاد کنیم. ما یک دیتافریم را بارگذاری کرده، این معیارها را محاسبه و نتایج را نمایش می‌دهیم.
</p>
</div>

```python
import numpy as np

# این قسمت اول کده و کتابخونه 'numpy' رو وارد می‌کنه.
# 'numpy' یه کتابخونه خیلی مهم برای کار با آرایه‌ها و عملیات ریاضی روی اون‌هاست.
# بیشتر عملیات آماری و علمی تو پایتون با این کتابخونه انجام می‌شه.

# این یه آرایه (array) از داده‌های نمونه ماست.
# آرایه یه ساختار داده‌ایه که مقادیر عددی رو تو خودش نگه می‌داره.
data = np.array([15, 18, 2, 36, 12, 25, 18, 40, 28, 22])

# این بخش، میانگین (Mean) داده‌ها رو حساب می‌کنه.
# میانگین همون معدل یا متوسط حسابی داده‌هاست.
# 'np.mean()' تابعی از 'numpy' هست که میانگین رو برامون محاسبه می‌کنه.
mean_value = np.mean(data)
print(f"میانگین: {mean_value}")

# اینجا میانه (Median) داده‌ها رو محاسبه می‌کنیم.
# میانه، مقدار وسطی داده‌هاست، بعد از اینکه اون‌ها رو مرتب کنیم.
# این معیار، کمتر از میانگین تحت تأثیر داده‌های پرت یا 'Outliers' قرار می‌گیره.
# 'np.median()' تابعیه که میانه رو حساب می‌کنه.
median_value = np.median(data)
print(f"میانه: {median_value}")

# این قسمت مد (Mode) یا پرتکرارترین مقدار رو پیدا می‌کنه.
# 'np.bincount()' تعداد هر مقدار رو می‌شماره و 'np.argmax()' هم مقدار با بیشترین تکرار رو پیدا می‌کنه.
# در واقع، مد، مقداریه که تو مجموعه داده‌ها از همه بیشتر تکرار شده.
mode_value = np.argmax(np.bincount(data))
print(f"مد: {mode_value}")
```

<li>این کد از توابع NumPy برای محاسبه <b>میانگین</b>، <b>میانه</b> و <b>نما</b>ی مجموعه داده استفاده می‌کند. شما می‌توانید آرایه داده را با مجموعه داده خود جایگزین کنید.</li>

### 2.2.2. Measuring the dispersion of data

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">اندازه‌گیری پراکندگی داده</span> 📈
</p>
<p style="font-size:16pt; text-align:right;">
اندازه‌گیری پراکندگی داده برای درک میزان پخش‌شدگی یا خوشه‌بندی مقادیر در یک مجموعه داده مهم است. چندین معیار پراکندگی وجود دارد، از جمله <b>دامنه (range)</b>، <b>واریانس (variance)</b> و <b>انحراف معیار (standard deviation)</b>.
</p>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">اندازه‌گیری پراکندگی داده</span> 📈
</p>
<p style="font-size:16pt; text-align:right;">
اندازه‌گیری پراکندگی داده برای درک میزان پخش‌شدگی یا خوشه‌بندی مقادیر در یک مجموعه داده مهم است. چندین معیار پراکندگی وجود دارد، از جمله <b>دامنه (range)</b>، <b>واریانس (variance)</b> و <b>انحراف معیار (standard deviation)</b>.
</p>

<p style="font-size:16pt; text-align:right;">
<br>
<b>دامنه (Range):</b>
<br>
دامنه، اختلاف بین حداکثر و حداقل مقادیر در یک مجموعه داده است. این معیار یک اندازه‌گیری ساده از میزان پخش‌شدگی مقادیر ارائه می‌دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>واریانس (Variance):</b>
<br>
واریانس، میانگین اختلاف مجذور مقادیر از میانگین مجموعه داده است. واریانس بالاتر نشان‌دهنده پراکندگی بیشتر است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>انحراف معیار (Standard Deviation):</b>
<br>
انحراف معیار، ریشه دوم واریانس است. این معیار یک سنجش قابل تفسیرتر از پراکندگی داده‌ها ارائه می‌دهد، زیرا واحد آن با واحد داده اصلی یکسان است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کد پایتون کاربردی برای اندازه‌گیری پراکندگی داده</span> 📈
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>pandas</b> برای محاسبه معیارهای پراکندگی داده (دامنه، واریانس و انحراف معیار) در یک مجموعه داده نمونه ایجاد کنیم. ما یک دیتافریم را بارگذاری کرده، این معیارها را محاسبه و نتایج را نمایش می‌دهیم.
</p>
</div>

```python
import numpy as np

# این قسمت اول کده و کتابخونه 'numpy' رو وارد می‌کنه.
# 'numpy' یه کتابخونه خیلی مهم برای کار با آرایه‌ها و عملیات ریاضی روی اون‌هاست.
# بیشتر عملیات آماری و علمی تو پایتون با این کتابخونه انجام می‌شه.

# این یه آرایه (array) از داده‌های نمونه ماست.
# آرایه یه ساختار داده‌ایه که مقادیر عددی رو تو خودش نگه می‌داره.
data = np.array([15, 18, 2, 36, 12, 25, 18, 40, 28, 22])

# این بخش، میانگین (Mean) داده‌ها رو حساب می‌کنه.
# میانگین همون معدل یا متوسط حسابی داده‌هاست.
# 'np.mean()' تابعی از 'numpy' هست که میانگین رو برامون محاسبه می‌کنه.
mean_value = np.mean(data)
print(f"میانگین: {mean_value}")

# اینجا میانه (Median) داده‌ها رو محاسبه می‌کنیم.
# میانه، مقدار وسطی داده‌هاست، بعد از اینکه اون‌ها رو مرتب کنیم.
# این معیار، کمتر از میانگین تحت تأثیر داده‌های پرت یا 'Outliers' قرار می‌گیره.
# 'np.median()' تابعیه که میانه رو حساب می‌کنه.
median_value = np.median(data)
print(f"میانه: {median_value}")

# این قسمت مد (Mode) یا پرتکرارترین مقدار رو پیدا می‌کنه.
# 'np.bincount()' تعداد هر مقدار رو می‌شماره و 'np.argmax()' هم مقدار با بیشترین تکرار رو پیدا می‌کنه.
# در واقع، مد، مقداریه که تو مجموعه داده‌ها از همه بیشتر تکرار شده.
mode_value = np.argmax(np.bincount(data))
print(f"مد: {mode_value}")


# این بخش، دامنه (Range) داده‌ها رو حساب می‌کنه.
# دامنه، تفاوت بین بزرگترین و کوچکترین مقدار تو مجموعه داده‌هاست.
# 'np.ptp()' تابعیه که این مقدار رو حساب می‌کنه.
# دامنه بهمون یه دید کلی از پراکندگی داده‌ها می‌ده.
data_range = np.ptp(data)
print(f"دامنه: {data_range}")

# اینجا واریانس (Variance) داده‌ها رو محاسبه می‌کنیم.
# واریانس، میانگین مربع اختلاف هر داده با میانگین کل داده‌هاست.
# این معیار نشون می‌ده که داده‌ها چقدر حول میانگین پراکنده شدن.
# 'np.var()' تابعیه که واریانس رو حساب می‌کنه.
data_variance = np.var(data)
print(f"واریانس: {data_variance}")

# این قسمت انحراف معیار (Standard Deviation) داده‌ها رو پیدا می‌کنه.
# انحراف معیار، ریشه دوم واریانسه و همون واحدهای اصلی داده‌ها رو داره.
# این معیار هم میزان پراکندگی داده‌ها رو نشون می‌ده. هرچی انحراف معیار بیشتر باشه، داده‌ها پراکنده‌ترن.
# 'np.std()' تابعیه که انحراف معیار رو محاسبه می‌کنه.
data_stddev = np.std(data)
print(f"انحراف معیار: {data_stddev}")
```

</p>
<p style="font-size:16pt; text-align:right;">
در این کد، <b><code>np.ptp</code></b> دامنه (range)، <b><code>np.var</code></b> واریانس (variance) و <b><code>np.std</code></b> انحراف معیار (standard deviation) را محاسبه می‌کند. آرایه داده را با مجموعه داده خود جایگزین کنید.

### 2.2.3. Covariance and correlation analysis

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">

</p>
<p style="font-size:16pt; text-align:right;">
<b>کوواریانس (Covariance)</b> و <b>تحلیل همبستگی (Correlation Analysis)</b> معیارهای آماری هستند که میزان تغییر همزمان دو متغیر را توصیف می‌کنند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کوواریانس و همبستگی</span> 📈
</p>

<p style="font-size:16pt; text-align:right;">
<b>کوواریانس (Covariance):</b>
<br>
کوواریانس میزان تغییر مقادیر دو متغیر را نسبت به یکدیگر اندازه‌گیری می‌کند. یک کوواریانس مثبت نشان می‌دهد که با افزایش یک متغیر، متغیر دیگر نیز تمایل به افزایش دارد و برعکس برای کوواریانس منفی. با این حال، مقیاس کوواریانس استاندارد نیست، که تفسیر قدرت رابطه را دشوار می‌سازد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>همبستگی (Correlation):</b>
<br>
همبستگی یک معیار استاندارد برای سنجش قدرت و جهت رابطه خطی بین دو متغیر است. ضریب همبستگی در بازه -1 تا 1 قرار دارد. ضریب همبستگی 1 نشان‌دهنده یک همبستگی مثبت کامل، -1 نشان‌دهنده یک همبستگی منفی کامل، و 0 نشان‌دهنده عدم وجود همبستگی خطی است.
</p>

```python
import numpy as np
import pandas as pd

# این قسمت اول کده و کتابخونه‌های 'numpy' و 'pandas' رو وارد می‌کنه.
# 'numpy' برای کار با آرایه‌ها و محاسبات ریاضی پیشرفته و 'pandas' برای کار با داده‌ها تو قالب جدول استفاده می‌شه.

# این یه دیکشنری با داده‌های نمونه ماست.
# این داده‌ها شامل دو متغیر 'Variable1' و 'Variable2' هستن که می‌خوایم رابطه‌شون رو با هم بررسی کنیم.
data = {
    'Variable1': [15, 18, 2, 36, 12, 25, 18, 40, 28, 22],
    'Variable2': [10, 12, 5, 30, 8, 20, 15, 35, 25, 18]
}

# با استفاده از 'pandas'، داده‌ها رو به یه 'DataFrame' تبدیل می‌کنیم.
# 'DataFrame' همون جدولیه که داده‌ها توش منظم و مرتب هستن.
df = pd.DataFrame(data)

# این بخش، **ماتریس کوواریانس (Covariance Matrix)** رو محاسبه می‌کنه.
# کوواریانس، میزان تغییر هم‌جهت دو متغیر رو نشون می‌ده. اگه دو متغیر با هم زیاد بشن یا کم بشن، کوواریانسشون مثبت می‌شه و اگه یکی زیاد و دیگری کم بشه، کوواریانسشون منفی می‌شه.
# 'np.cov()' تابعی از 'numpy' هست که ماتریس کوواریانس رو حساب می‌کنه.
# 'rowvar=False' هم به تابع می‌گه که هر ستون یک متغیره، نه هر سطر.
cov_matrix = np.cov(df, rowvar=False)
print(f"ماتریس کوواریانس:\n{cov_matrix}")

# این قسمت **ماتریس همبستگی (Correlation Matrix)** رو محاسبه می‌کنه.
# همبستگی هم مثل کوواریانسه، با این تفاوت که مقادیرش بین -۱ و ۱ هستن.
# این مقدار نشون می‌ده که رابطه بین دو متغیر چقدر قوی و در چه جهتیه.
# اگه مقدار نزدیک به ۱ باشه، همبستگی مثبت و قویه. اگه نزدیک به -۱ باشه، همبستگی منفی و قویه. اگه نزدیک به ۰ باشه، همبستگی ضعیفه.
# 'df.corr()' تابعی از 'pandas' هست که ماتریس همبستگی رو حساب می‌کنه.
correlation_matrix = df.corr()
print(f"ماتریس همبستگی:\n{correlation_matrix}")
```

### 2.3.1. Data matrix vs. dissimilarity matrix

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ماتریس داده در مقابل ماتریس عدم تشابه</span> 📊
</p>

<p style="font-size:16pt; text-align:right;">
در داده‌کاوی، درک مفاهیم <b>ماتریس داده (Data Matrix)</b> و <b>ماتریس عدم تشابه (Dissimilarity Matrix)</b> بسیار حیاتی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>ماتریس داده:</b>
<br>
<b>توضیح:</b> ماتریس داده یک نمایش ساختاریافته از یک مجموعه داده است که در آن سطرها مربوط به مشاهدات یا نمونه‌های منفرد و ستون‌ها نمایانگر ویژگی‌ها یا صفات مختلف هستند. هر خانه در این ماتریس حاوی مقدار یک ویژگی خاص برای یک نمونه مشخص است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>ماتریس عدم تشابه:</b>
<br>
<b>توضیح:</b> ماتریس عدم تشابه، عدم تشابه یا تشابه بین جفت‌های نمونه در یک مجموعه داده را نشان می‌دهد. این ماتریس میزان تفاوت یا شباهت دو نمونه را بر اساس یک معیار عدم تشابه، مانند فاصله یا امتیاز عدم تشابه، کمی‌سازی می‌کند.
</p>
</div>

```python
import pandas as pd

# این قسمت اول کده و کتابخونه 'pandas' رو وارد می‌کنه.
# 'pandas' ابزار اصلی ما برای کار با داده‌هاست، به خصوص وقتی داده‌ها تو قالب جدول باشن.

# این یه دیکشنری (dictionary) حاوی داده‌های نمونه ماست.
# در این مثال، هر سطر (row) یک نمونه (Instance) یا شیء (Object) رو نشون می‌ده، مثل یک دانشجو.
# هر ستون (column) هم یک ویژگی (Feature) یا صفت (Attribute) رو نشون می‌ده، مثل سن یا نمره.
# به این ساختار جدولی، **ماتریس داده (Data Matrix)** گفته می‌شه.
data = {
    'ID': [1, 2, 3, 4],
    'Feature1': [10, 15, 20, 25],
    'Feature2': [0.5, 1.2, 0.8, 1.0]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
# 'DataFrame' همون ماتریس داده‌ست که 'pandas' باهاش کار می‌کنه و داده‌ها رو به شکل منظم نشون می‌ده.
df = pd.DataFrame(data)

# 'print' یه تابع ساده‌اس که هر چیزی رو بهمون نشون می‌ده.
# اینجا 'DataFrame' رو چاپ می‌کنیم تا ببینیم ماتریس داده چه شکلیه.
print("ماتریس داده:")
print(df)
```

```python
from scipy.spatial.distance import pdist, squareform
import pandas as pd

# این قسمت اول کده. دو تا کتابخونه اصلی رو وارد می‌کنیم:
# 'scipy.spatial.distance' که برای محاسبه فاصله بین نقاط در فضای چندبعدی استفاده می‌شه.
# 'pandas' که ابزار اصلی ما برای کار با داده‌هاست.

# این دیکشنری حاوی داده‌های نمونه ماست که یک ماتریس داده رو تشکیل می‌ده.
# 'ID' یک شناسه‌ست و 'Feature1' و 'Feature2' همون ویژگی‌های داده‌ها هستن که می‌خوایم فاصله‌شون رو حساب کنیم.
data = {
    'ID': [1, 2, 3, 4],
    'Feature1': [10, 15, 20, 25],
    'Feature2': [0.5, 1.2, 0.8, 1.0]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، **ماتریس عدم تشابه (Dissimilarity Matrix)** رو محاسبه می‌کنه.
# ماتریس عدم تشابه، فاصله بین هر جفت از نمونه‌ها رو نشون می‌ده. هرچی فاصله بیشتر باشه، عدم تشابه یا تفاوتشون هم بیشتره.
# 'pdist()' تابع اصلی برای محاسبه فاصله‌هاست.
# ما از 'euclidean' به عنوان 'metric' یا معیار فاصله استفاده می‌کنیم که همون فاصله اقلیدسی (فاصله مستقیم بین دو نقطه) هست.
# 'df[['Feature1', 'Feature2']]' هم به تابع می‌گه که فقط از این دو ستون برای محاسبه فاصله استفاده کنه.
# 'squareform()' هم فاصله‌های محاسبه شده رو به شکل یک ماتریس مربع درمیاره که خوندنش راحت‌تره.
dissimilarity_matrix = squareform(pdist(df[['Feature1', 'Feature2']], metric='euclidean'))

# در نهایت، ماتریس عدم تشابه رو چاپ می‌کنیم.
# 'pd.DataFrame()' بهمون کمک می‌کنه که ماتریس رو تو قالب یه جدول مرتب با برچسب‌های 'ID' ببینیم.
print("ماتریس عدم تشابه:")
print(pd.DataFrame(dissimilarity_matrix, index=df['ID'], columns=df['ID']))
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این کد، تابع <b>pdist</b> از کتابخانه <b>SciPy</b> فواصل اقلیدسی (Euclidean distances) بین جفت‌های نمونه را بر اساس ویژگی‌های انتخاب‌شده محاسبه می‌کند. تابع <b>squareform</b> نیز ماتریس فاصله فشرده‌شده را به یک ماتریس مربعی تبدیل می‌کند.
<br>
ماتریس‌های داده یک نمایش ساختاریافته از داده‌های خام ارائه می‌دهند، در حالی که ماتریس‌های عدم تشابه، عدم تشابه جفت‌ها را نشان می‌دهند که در کاربردهای مختلف داده‌کاوی مانند <b>خوشه‌بندی (clustering)</b>، <b>طبقه‌بندی (classification)</b> و <b>جستجوی شباهت (similarity search)</b> مفید هستند.
</p>

### 2.3.2. Proximity measures for nominal attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارهای نزدیکی برای ویژگی‌های اسمی</span> 🏷️
</p>
<p style="font-size:16pt; text-align:right;">
<b>معیارهای نزدیکی برای ویژگی‌های اسمی (Proximity measures for nominal attributes)</b> برای تعیین میزان تشابه یا عدم تشابه بین نمونه‌ها، هنگامی که با ویژگی‌های کیفی یا اسمی سروکار داریم، استفاده می‌شوند. برخلاف ویژگی‌های عددی، ویژگی‌های اسمی ترتیب طبیعی ندارند، که باعث می‌شود معیارهای فاصله سنتی برای آن‌ها نامناسب باشند. معیارهای نزدیکی این مشکل را با تمرکز بر تطابق یا عدم تطابق بین مقادیر کیفی حل می‌کنند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال معیار نزدیکی: شباهت جاکارد</span> 🏷️
</p>
<p style="font-size:16pt; text-align:right;">
<br>
<b>توضیح:</b>
<br>
شباهت <b>جاکارد (Jaccard Similarity)</b>، میزان شباهت بین دو مجموعه را با تقسیم اندازه اشتراک آن‌ها بر اندازه اجتماعشان اندازه‌گیری می‌کند. در زمینه ویژگی‌های اسمی، هر مقدار ویژگی به عنوان مجموعه‌ای از مقادیر دودویی (Binary) در نظر گرفته می‌شود (<b>۱</b> اگر ویژگی وجود داشته باشد، <b>۰</b> در غیر این صورت).
</p>
</div>

```python
from sklearn.metrics import jaccard_score
import pandas as pd

# این بخش اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'pandas' که یه ابزار خیلی پرکاربرده برای کار کردن با داده‌ها، به خصوص داده‌های جدولی.
# 'jaccard_score' هم یک تابع از کتابخونه 'sklearn.metrics' هست که برای محاسبه ضریب ژاکارد استفاده می‌شه.

# این یه دیکشنری (dictionary) حاوی داده‌های نمونه ماست.
# این داده‌ها از نوع اسمی (Nominal) هستن؛ یعنی فقط اسم یا دسته هستن و هیچ ترتیب مشخصی بینشون وجود نداره.
# مثلاً 'Red' و 'Blue' فقط دو دسته متفاوت از رنگ هستن.
data = {
    'Instance1': ['Red', 'Square', 'Large'],
    'Instance2': ['Blue', 'Circle', 'Small']
}

# برای اینکه بتونیم ضریب ژاکارد رو حساب کنیم، باید داده‌های اسمی رو به داده‌های باینری (صفر و یک) تبدیل کنیم.
# تابع 'pd.get_dummies()' از کتابخونه 'pandas' این کار رو برامون انجام می‌ده.
# این تابع برای هر مقدار منحصر به فرد (Unique) تو داده‌های اسمی، یه ستون جدید می‌سازه.
# اگه اون مقدار وجود داشته باشه، عدد ۱ و اگه نباشه، عدد ۰ رو قرار می‌ده.
# به این فرآیند **One-Hot Encoding** هم می‌گن.
binary_data = pd.get_dummies(pd.DataFrame(data))

# با اجرای این خط، 'DataFrame' نهایی که به صورت باینری تبدیل شده، نمایش داده می‌شه.
# این داده‌ها حالا آماده‌ان تا با استفاده از 'jaccard_score'، شباهتشون بررسی بشه.
binary_data
```

```python
# این بخش، محاسبه ضریب ژاکارد (Jaccard Similarity) رو انجام می‌ده.
# ضریب ژاکارد یه معیار برای سنجش شباهت بین دو مجموعه هست.
# 'binary_data.iloc[0]' و 'binary_data.iloc[1]' سطر اول و دوم 'DataFrame' رو انتخاب می‌کنن.
# این مقادیر باینری به تابع 'jaccard_score' داده می‌شن تا شباهتشون محاسبه بشه.
# اگه دو نمونه کاملاً مشابه باشن، مقدار ۱ و اگه هیچ شباهتی نداشته باشن، مقدار ۰ رو برمی‌گردونه.

jaccard_similarity = jaccard_score(binary_data.iloc[0], binary_data.iloc[1])

# اینجا هم ضریب ژاکارد یه نمونه با خودش حساب شده که همیشه باید برابر با ۱ باشه.

jaccard_similarity_1 = jaccard_score(binary_data.iloc[0], binary_data.iloc[0])


print(f"Jaccard Similarity: {jaccard_similarity}")
print(f"Jaccard Similarity: {jaccard_similarity_1}")
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، ویژگی‌های اسمی (<b>'Red'</b>, <b>'Square'</b>, <b>'Large'</b>) و (<b>'Blue'</b>, <b>'Circle'</b>, <b>'Small'</b>) با استفاده از one-hot encoding به بردارهای دودویی (Binary Vectors) تبدیل می‌شوند. سپس، <b>شباهت جاکارد (Jaccard Similarity)</b> با استفاده از تابع <b><code>jaccard_score</code></b> از کتابخانه <b><code>scikit-learn</code></b> محاسبه می‌شود.
<br>
معیارهای نزدیکی برای ویژگی‌های اسمی، برای خوشه‌بندی، طبقه‌بندی و سایر وظایف داده‌کاوی که شامل داده‌های کیفی هستند، ضروری‌اند. این مثال عملی کاربرد شباهت جاکارد را برای کمی‌سازی شباهت بین نمونه‌ها با ویژگی‌های اسمی نشان می‌دهد.

### 2.3.3. Proximity measures for binary attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارهای نزدیکی برای ویژگی‌های دودویی</span> ✅
</p>
<p style="font-size:16pt; text-align:right;">
<b>معیارهای نزدیکی برای ویژگی‌های دودویی (Proximity measures for binary attributes)</b> هنگام کار با مجموعه‌داده‌هایی که شامل ویژگی‌های دودویی (<b>0/1</b>) هستند، به کار گرفته می‌شوند. این معیارها تشابه یا عدم تشابه بین نمونه‌ها را بر اساس حضور یا عدم حضور ویژگی‌های دودویی خاص ارزیابی می‌کنند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال معیار نزدیکی: فاصله همینگ</span> ✅
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
<b>فاصله همینگ (Hamming Distance)</b> عدم تشابه بین دو رشته دودویی را با شمارش تعداد موقعیت‌هایی که بیت‌های متناظر در آن‌ها متفاوت هستند، اندازه‌گیری می‌کند.
</p>
</div>

```python
import numpy as np
from scipy.spatial.distance import hamming

# این قسمت اول کده. دو تا کتابخونه رو وارد می‌کنیم:
# 'numpy' که یه ابزار خیلی پرکاربرد برای کار با آرایه‌ها و محاسبات ریاضیه.
# 'hamming' که تابعی از 'scipy.spatial.distance' هست و برای محاسبه فاصله همینگ (Hamming Distance) استفاده می‌شه.

# این یه دیکشنری (dictionary) با داده‌های نمونه ماست.
# این داده‌ها از نوع باینری (Binary) هستن، یعنی فقط مقادیر ۰ و ۱ رو دارن.
# این مقادیر می‌تونن نشون‌دهنده حضور یا عدم حضور یک ویژگی خاص باشن.
data = {
    'Instance1': [1, 0, 1, 0],
    'Instance2': [0, 1, 1, 0]
}

# برای اینکه بتونیم فاصله همینگ رو با تابع 'hamming' محاسبه کنیم،
# باید داده‌ها رو به آرایه‌های 'numpy' تبدیل کنیم.
# این خط دو تا آرایه می‌سازه که هر کدوم شامل مقادیر یک نمونه هستن.
binary_data = np.array([data['Instance1'], data['Instance2']])

# این بخش، **فاصله همینگ (Hamming Distance)** رو محاسبه می‌کنه.
# فاصله همینگ، تعداد موقعیت‌هایی رو می‌شماره که مقادیر دو نمونه در اون‌ها با هم فرق دارن.
# مثلاً اگه دو رشته '1010' و '0110' رو مقایسه کنیم، می‌بینیم که فقط در دو موقعیت (اول و دوم) با هم فرق دارن.
# بنابراین، فاصله همینگشون برابر با ۲ هست.
# 'hamming(binary_data[0], binary_data[1])' این فاصله رو برامون حساب می‌کنه.
# نتیجه یه عدد اعشاریه، چون تابع مقدار رو به صورت میانگین اختلاف برمی‌گردونه.
hamming_distance = hamming(binary_data[0], binary_data[1])

print(f"Hamming Distance: {hamming_distance}")
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال فاصله همینگ</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، نمونه‌ها با بردارهای دودویی 
[۱,۰,۱,۰]
 و 
[۰,۱,۱,۰]
 نمایش داده شده‌اند. فاصله همینگ بین این بردارها با استفاده از تابع <b><code>hamming</code></b> از کتابخانه <b>SciPy</b> محاسبه می‌شود.
<br>
معیارهای نزدیکی برای ویژگی‌های دودویی، برای وظایفی مانند خوشه‌بندی، تشخیص الگو و طبقه‌بندی، هنگام کار با مجموعه‌داده‌های متشکل از ویژگی‌های دودویی بسیار حیاتی هستند. مثال فاصله همینگ نشان می‌دهد که چگونه می‌توان عدم تشابه بین نمونه‌ها را بر اساس ویژگی‌های دودویی کمی‌سازی کرد.

### 2.3.4. Dissimilarity of numeric data: Minkowski distance

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">فاصله مینکوفسکی</span> 📏
</p>
<p style="font-size:16pt; text-align:right;">
<b>فاصله مینکوفسکی (Minkowski distance)</b> یک معیار عدم تشابه است که برای ارزیابی میزان تشابه یا عدم تشابه بین دو نقطه در یک فضای چندبعدی استفاده می‌شود. این معیار یک تعمیم برای سایر معیارهای فاصله، از جمله <b>فاصله اقلیدسی (Euclidean distance)</b> و <b>فاصله منهتن (Manhattan distance)</b>، محسوب می‌شود و با فرمول زیر تعریف می‌شود:
</p>
<p style="font-size:20pt; text-align:center;">
    <img src="https://www.maartengrootendorst.com/images/posts/2021-01-02-distances/minkowski_distance.gif" alt="فرمول فاصله مینکوفسکی" style="max-width:100%; height:auto;">
</p>
<p style="font-size:16pt; text-align:center;">
که در آن <b>x</b> و <b>y</b> بردارهای نمایانگر نقاط داده عددی هستند، <b>n</b> تعداد ابعاد است، و <b>p</b> پارامتری است که مرتبه فاصله مینکوفسکی را تعیین می‌کند. هنگامی که <b>p=1</b> باشد، فاصله مینکوفسکی معادل فاصله منهتن است، و زمانی که <b>p=2</b> باشد، معادل فاصله اقلیدسی است.
</p>
</div>

```python
from scipy.spatial.distance import minkowski
import numpy as np

# این قسمت اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'scipy.spatial.distance' که برای محاسبه فاصله‌ها استفاده می‌شه.
# 'numpy' که برای کار با آرایه‌ها و محاسبات ریاضیه.

# این خطوط داده‌های نمونه ما هستن که به صورت لیست تعریف شدن.
# این‌ها دو نمونه با ویژگی‌های عددی هستن که می‌خوایم فاصله‌شون رو از هم محاسبه کنیم.
data1 = [2, 3, 5, 7]
data2 = [1, 4, 6, 8]

# این بخش **فاصله مینکوفسکی (Minkowski Distance)** رو با p=2 محاسبه می‌کنه.
# فاصله مینکوفسکی یه معیار کلی برای محاسبه فاصله بین دو نقطه در فضای چندبعدیه.
# وقتی p برابر با ۲ باشه، فاصله مینکوفسکی دقیقاً برابر با **فاصله اقلیدسی (Euclidean Distance)** می‌شه.
# فاصله اقلیدسی، کوتاه‌ترین فاصله بین دو نقطه در یک خط مستقیمه.
minkowski_distance = minkowski(data1, data2, p=2)
print(f"فاصله مینکوفسکی (p=2): {minkowski_distance}")

# این قسمت **فاصله مینکوفسکی** رو با p=1 محاسبه می‌کنه.
# وقتی p برابر با ۱ باشه، فاصله مینکوفسکی برابر با **فاصله منهتن (Manhattan Distance)** می‌شه.
# فاصله منهتن، جمع قدر مطلق تفاوت هر ویژگیه، مثل فاصله‌ای که در خیابان‌های یک شهر (مثل منهتن) طی می‌شه.
manhattan_distance = minkowski(data1, data2, p=1)
print(f"فاصله مینکوفسکی (p=1): {manhattan_distance}")
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، فاصله مینکوفسکی بین دو بردار عددی 
[۲,۳,۵,۷]
 و

 با استفاده از فاصله اقلیدسی (p=2) و فاصله منهتن (p=1) محاسبه می‌شود. برای این محاسبه از تابع minkowski از کتابخانه <b>SciPy</b> استفاده شده است.
<br>
فاصله مینکوفسکی یک معیار منعطف است که بر اساس مقدار انتخاب‌شده برای <b>p</b>، با سناریوهای مختلف سازگار می‌شود. این معیار در وظایف داده‌کاوی مختلفی مانند خوشه‌بندی، طبقه‌بندی و تشخیص نقاط پرت کاربرد دارد، به خصوص زمانی که با مجموعه‌داده‌های حاوی ویژگی‌های عددی سروکار داریم.
</p>

### 2.3.5. Proximity measures for ordinal attributes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">

معیارهای نزدیکی برای ویژگی‌های ترتیبی، هنگام کار با مجموعه‌داده‌هایی که شامل ویژگی‌هایی با ترتیب یا رتبه‌بندی ذاتی هستند، استفاده می‌شوند. برخلاف ویژگی‌های اسمی، ویژگی‌های ترتیبی دارای ترتیب معناداری هستند، اما ممکن است فواصل بین مقادیر آن‌ها یکنواخت نباشد.

<p style="font-size:16pt; text-align:right;">
<b>مثال معیار نزدیکی: ضریب همبستگی رتبه‌ای اسپیرمن</b>

<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
<b>ضریب همبستگی رتبه‌ای اسپیرمن (Spearman Rank Correlation Coefficient)</b> قدرت و جهت رابطه یکنواخت (monotonic) بین دو متغیر ترتیبی را اندازه‌گیری می‌کند. این معیار به جای مقادیر واقعی، بر اساس رتبه‌های (ranks) مقادیر محاسبه می‌شود.
</p>
</div>

```python
import numpy as np
from scipy.stats import spearmanr

# این قسمت اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'numpy' که برای کار با آرایه‌ها و محاسبات ریاضیه.
# 'spearmanr' که تابعی از 'scipy.stats' هست و برای محاسبه ضریب همبستگی رتبه‌ای اسپیرمن (Spearman Rank Correlation Coefficient) استفاده می‌شه.

# این یه دیکشنری (dictionary) با داده‌های نمونه ماست.
# این داده‌ها از نوع ترتیبی (Ordinal) هستن، یعنی فقط ترتیبشون مهمه.
# به جای خود مقادیر، رتبه‌شون (Rank) در نظر گرفته می‌شه. مثلاً تو 'Instance1'، مقدار ۲ از همه کوچیکتره، پس رتبه اول رو می‌گیره.
data = {
    'Instance1': [2, 3, 1, 4],
    'Instance2': [1, 4, 2, 3]
}

# این بخش، **ضریب همبستگی رتبه‌ای اسپیرمن** رو محاسبه می‌کنه.
# این ضریب، قدرت و جهت رابطه بین دو متغیر ترتیبی رو نشون می‌ده.
# اگه مقدارش نزدیک به ۱ باشه، یعنی هر دو متغیر با هم رشد می‌کنن.
# اگه نزدیک به -۱ باشه، یعنی یکی رشد می‌کنه و اون یکی کم می‌شه.
# 'spearmanr()' تابع اصلی برای این محاسبه‌اس.
# این تابع دو مقدار رو برمی‌گردونه: ضریب همبستگی و مقدار 'p-value' که ما فقط ضریب رو نگه می‌داریم.
spearman_corr, _ = spearmanr(data['Instance1'], data['Instance2'])

print(f"ضریب همبستگی رتبه‌ای اسپیرمن: {spearman_corr}")
```

 نمایش داده شده‌اند. ضریب همبستگی رتبه‌ای اسپیرمن با استفاده از تابع <b><code>spearmanr</code></b> از کتابخانه <b>SciPy</b> محاسبه می‌شود.

### 2.3.6. Dissimilarity for attributes of mixed types

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مدیریت عدم تشابه برای ویژگی‌های با نوع ترکیبی</span> 🧩
</p>
<p style="font-size:16pt; text-align:right;">
مدیریت عدم تشابه برای ویژگی‌های با نوع ترکیبی یک چالش رایج در داده‌کاوی است، زیرا مجموعه‌داده‌ها اغلب شامل ترکیبی از ویژگی‌های عددی (numeric)، کیفی (categorical) و ترتیبی (ordinal) هستند. استفاده از معیارهای عدم تشابه مناسب که بتوانند ماهیت متنوع این انواع ویژگی‌ها را در خود جای دهند، بسیار حیاتی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال معیار عدم تشابه: فاصله گاور (Gower's Distance)</b>
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
<b>فاصله گاور (Gower's Distance)</b> یک معیار عدم تشابه است که برای مجموعه‌داده‌های با انواع ویژگی‌های ترکیبی طراحی شده است. این معیار عدم تشابه بین دو نمونه را با در نظر گرفتن انواع ویژگی‌ها و اعمال معیارهای مناسب، مانند فاصله اقلیدسی برای ویژگی‌های عددی، شباهت جاکارد برای ویژگی‌های دودویی و ضریب تطابق ساده برای ویژگی‌های اسمی، محاسبه می‌کند.
</p>
</div>

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import pdist, squareform
from sklearn.preprocessing import StandardScaler

# این قسمت اول کده. ما چند تا کتابخونه رو وارد می‌کنیم:
# 'numpy' برای کار با آرایه‌ها، 'pandas' برای کار با داده‌های جدولی،
# 'pdist' و 'squareform' از 'scipy' برای محاسبه فاصله‌ها،
# و 'StandardScaler' از 'sklearn' برای استانداردسازی داده‌های عددی.

# این دیکشنری حاوی داده‌های نمونه ماست که انواع مختلفی از ویژگی‌ها رو داره:
# 'Numeric': ویژگی‌های عددی مثل سن یا قد.
# 'Binary': ویژگی‌های باینری مثل 'بله' یا 'خیر'.
# 'Nominal': ویژگی‌های اسمی مثل رنگ یا شهر.
# 'Ordinal': ویژگی‌های ترتیبی مثل سطح تحصیلات.
data = {
    'Numeric': [2.5, 1.8, 3.2, 4.0],
    'Binary': [1, 0, 1, 1],
    'Nominal': ['A', 'B', 'A', 'C'],
    'Ordinal': [2, 1, 3, 2]
}

# این بخش، داده‌های عددی رو استانداردسازی (Standardize) می‌کنه.
# استانداردسازی یعنی داده‌ها رو به صورتی تغییر می‌دیم که میانگینشون صفر و انحراف معیارشون یک بشه.
# این کار برای مقایسه داده‌ها با مقیاس‌های مختلف خیلی مهمه.
numeric_data = np.array([data['Numeric']]).T
numeric_data = StandardScaler().fit_transform(numeric_data)

# این قسمت ویژگی‌های باینری رو به یک 'DataFrame' تبدیل می‌کنه تا بتونیم باهاش کار کنیم.
# این داده‌ها به صورت طبیعی باینری هستن و نیازی به تبدیل ندارن.
binary_data = pd.get_dummies(pd.DataFrame(data['Binary'], columns=['Binary']))

# این بخش، ویژگی‌های اسمی رو به داده‌های باینری تبدیل می‌کنه.
# این کار همون **One-Hot Encoding** هست که قبلاً دیدیم و برای مقایسه داده‌های اسمی ضروریه.
nominal_data = pd.get_dummies(pd.DataFrame(data['Nominal'], columns=['Nominal']))

# حالا همه انواع داده‌ها (که همشون عددی یا باینری شدن) رو با هم ترکیب می‌کنیم.
# 'np.concatenate()' داده‌ها رو کنار هم می‌چینه و یک آرایه بزرگ می‌سازه.
# 'axis=1' بهش می‌گه که این ترکیب رو ستون به ستون انجام بده.
mixed_data = np.concatenate((numeric_data, binary_data, nominal_data, np.array([data['Ordinal']]).T), axis=1)

# این بخش **فاصله گُووِر (Gower's Distance)** رو محاسبه می‌کنه.
# فاصله گُووِر یک معیار فاصله مخصوصه برای داده‌هایی که انواع مختلفی دارن.
# این معیار، فاصله بین داده‌های عددی، اسمی و باینری رو به صورت همزمان حساب می‌کنه.
# در اینجا، از 'pdist' با 'metric='euclidean'' استفاده شده، چون قبلاً داده‌ها رو آماده و استاندارد کردیم.
gower_distance = squareform(pdist(mixed_data, metric='euclidean'))
print("ماتریس فاصله گُووِر:")
print(gower_distance)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، <b>فاصله گاور</b> برای یک مجموعه داده با ویژگی‌های عددی (numeric)، دودویی (binary)، اسمی (nominal) و ترتیبی (ordinal) محاسبه می‌شود. ویژگی‌های عددی استاندارد می‌شوند، ویژگی‌های دودویی و اسمی به مقادیر دودویی تبدیل می‌گردند، و همه انواع ویژگی‌ها در یک ماتریس واحد ترکیب می‌شوند. سپس، تابع <b><code>pdist</code></b> از <b>SciPy</b> برای محاسبه فواصل اقلیدسی جفت‌ها و تابع <b><code>squareform</code></b> برای تبدیل ماتریس فاصله فشرده‌شده به ماتریس مربعی استفاده می‌شود.
<br>
فاصله گاور یک معیار منعطف برای سنجش عدم تشابه در مجموعه‌داده‌های با انواع ویژگی‌های ترکیبی است و راه‌حلی جامع برای کار با ویژگی‌های متنوع داده ارائه می‌دهد.

### 2.3.7. Cosine similarity

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">شباهت کسینوسی</span> 📐
</p>
<p style="font-size:16pt; text-align:right;">
<b>شباهت کسینوسی (Cosine similarity)</b> معیاری است که برای اندازه‌گیری شباهت بین دو بردار غیرصفر در یک فضای ضرب داخلی (inner product space) استفاده می‌شود. در زمینه داده‌کاوی، این معیار اغلب برای ارزیابی شباهت بین اسناد در پردازش زبان طبیعی یا مقایسه بردارهای ویژگی در سیستم‌های توصیه‌گر به کار می‌رود. شباهت کسینوسی در بازه -1 تا 1 قرار دارد، که در آن 1 نشان‌دهنده شباهت کامل، 0 نشان‌دهنده عدم شباهت و -1 نشان‌دهنده عدم تشابه کامل است.
</p>
</div>

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# این قسمت اول کده. ما دو تا ابزار مهم رو از کتابخونه 'sklearn' وارد می‌کنیم:
# 'CountVectorizer' که برای تبدیل متن به بردار (vector) استفاده می‌شه.
# 'cosine_similarity' که برای محاسبه شباهت بین دو بردار از داده‌های متنی استفاده می‌شه.

# این لیست (list) حاوی سندهای (documents) متنی ماست.
# هر رشته (string) در این لیست، یک سند جداگانه در نظر گرفته می‌شه.
documents = [
    "Data mining is an exciting field of study.",
    "Machine learning techniques are used for data analysis.",
    "Python programming is widely used in data science applications."
]

# این بخش، داده‌های متنی رو به یک نمایش عددی تبدیل می‌کنه که بهش می‌گن 'Bag-of-Words'.
# 'CountVectorizer' هر کلمه رو می‌شماره و یک ماتریس می‌سازه که هر سطرش یک سند و هر ستونش یک کلمه‌اس.
# مقادیر داخل ماتریس، تعداد تکرار هر کلمه تو اون سند هستن.
vectorizer = CountVectorizer()
document_matrix = vectorizer.fit_transform(documents)

# حالا **تشابه کسینوسی (Cosine Similarity)** رو محاسبه می‌کنیم.
# تشابه کسینوسی، شباهت جهت‌گیری بین دو بردار رو می‌سنجه، نه اندازه‌شون رو.
# این معیار، میزان شباهت معنایی بین سندها رو نشون می‌ده.
# 'cosine_similarity()' تابع اصلی برای این محاسبه‌اس.
# نتیجه یک ماتریس مربعه که شباهت هر سند رو با بقیه سندها نشون می‌ده.
# مقادیر این ماتریس بین ۰ و ۱ هستن.
cosine_similarities = cosine_similarity(document_matrix, document_matrix)

# در نهایت، ماتریس تشابه کسینوسی رو چاپ می‌کنیم.
print("ماتریس تشابه کسینوسی:")
print(cosine_similarities)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از <code>CountVectorizer</code> از کتابخانه <code>scikit-learn</code> برای تبدیل مجموعه‌ای از اسناد متنی به یک ماتریس از تعداد توکن‌ها (نمایش bag-of-words) استفاده می‌شود. سپس، تابع <code>cosine_similarity</code> شباهت کسینوسی را بین بردارهای اسناد محاسبه می‌کند.
<br>
شباهت کسینوسی به خصوص در سناریوهایی مفید است که در آن‌ها بزرگی (magnitude) بردارها اهمیت چندانی ندارد و تمرکز بر جهت بردارها است. این معیار به طور گسترده در تحلیل متن، بازیابی اسناد و سیستم‌های فیلترینگ مشارکتی استفاده می‌شود. مثال ارائه‌شده نحوه محاسبه شباهت کسینوسی برای مجموعه‌ای از اسناد متنی را نشان می‌دهد.

### 2.3.8. Measuring similar distributions: the Kullback-Leibler divergence

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">واگرایی کولبک-لایبلر</span> 📉
</p>
<p style="font-size:16pt; text-align:right;">
<b>واگرایی کولبک-لایبلر (KL Divergence)</b> معیاری است که نشان می‌دهد یک توزیع احتمال چقدر از یک توزیع احتمال ثانویه و مورد انتظار منحرف می‌شود. در زمینه داده‌کاوی، اغلب برای کمی‌سازی تفاوت یا شباهت بین دو توزیع احتمال استفاده می‌شود. واگرایی KL یک معیار فاصله واقعی نیست، زیرا متقارن نمی‌باشد و از نامساوی مثلث پیروی نمی‌کند.
</p>
</div>

```python
import numpy as np
from scipy.stats import entropy

# این قسمت اول کده. ما دو تا ابزار رو وارد می‌کنیم:
# 'numpy' که برای کار با آرایه‌ها و داده‌های عددیه.
# 'entropy' که تابعی از 'scipy.stats' هست و برای محاسبه واگرایی کولبک-لایبلر (KL Divergence) استفاده می‌شه.

# این خطوط، دو تا توزیع احتمال (Probability Distribution) رو نشون می‌دن.
# هر توزیع نشون می‌ده که احتمال هر رویداد چقدره.
# مثلاً 'distribution1' می‌گه که احتمال رویداد اول ۰.۲، دوم ۰.۳ و سوم ۰.۵ هست.
distribution1 = np.array([0.2, 0.3, 0.5])
distribution2 = np.array([0.1, 0.6, 0.3])

# این بخش، **واگرایی کولبک-لایبلر** رو محاسبه می‌کنه.
# این واگرایی (Divergence) یک معیار برای سنجش تفاوت بین دو توزیع احتماله.
# هر چقدر مقدارش بیشتر باشه، یعنی دو توزیع با هم متفاوت‌تر هستن.
# واگرایی کولبک-لایبلر نامتقارنه، یعنی KL(p, q) با KL(q, p) برابر نیست.
# 'entropy()' تابعیه که این مقدار رو حساب می‌کنه. تو این تابع، توزیع اول رو به عنوان توزیع اصلی و توزیع دوم رو به عنوان توزیع مرجع در نظر می‌گیره.
kl_divergence = entropy(distribution1, distribution2)

# در نهایت، نتیجه محاسبه رو چاپ می‌کنیم.
print(f"واگرایی کولبک-لایبلر: {kl_divergence}")
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، دو توزیع احتمال با آرایه‌های <code>distribution1</code> و <code>distribution2</code> نمایش داده شده‌اند. تابع <b><code>entropy</code></b> از کتابخانه <b>SciPy</b>، واگرایی <b>KL Divergence</b> را بین این توزیع‌ها محاسبه می‌کند.
<br>
واگرایی <b>KL Divergence</b> اغلب در نظریه اطلاعات و آمار برای اندازه‌گیری تفاوت بین دو توزیع احتمال استفاده می‌شود. این معیار در زمینه‌های مختلفی از جمله پردازش زبان طبیعی، یادگیری ماشین و تشخیص الگو کاربرد دارد.

### 2.3.9. Capturing hidden semantics in similarity measures

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">درک معنای پنهان در معیارهای شباهت</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
<b>درک معنای پنهان (Capturing hidden semantics)</b> در معیارهای شباهت شامل یافتن الگوها، روابط یا شباهت‌های معنادار در داده‌ها است که ممکن است در نگاه اول آشکار نباشند. این فرآیند اغلب به تکنیک‌های پیشرفته‌ای نیاز دارد که فراتر از معیارهای فاصله یا شباهت ساده عمل می‌کنند. برای درک و بهره‌برداری از معنای پنهان در داده‌ها، از روش‌هایی مانند Word Embeddings (تعبیه‌های کلمه)، شباهت معنایی (Semantic Similarity) یا مدل‌های پیشرفته شبکه عصبی استفاده می‌شود.
</p>

```python
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# این قسمت اول کده. دو تا ابزار مهم رو از کتابخونه‌های مختلف وارد می‌کنیم:
# 'Word2Vec' از 'gensim' که برای ساختن بردارهای کلمات (word embeddings) استفاده می‌شه.
# 'cosine_similarity' از 'sklearn' که برای محاسبه شباهت بین دو بردار استفاده می‌شه.

# این لیست (list) حاوی سندهای (documents) متنی ماست.
# هر رشته (string) در این لیست، یک سند جداگانه در نظر گرفته می‌شه.
documents = [
    "Data mining is an exciting field of study.",
    "Machine learning techniques are used for data analysis.",
    "Python programming is widely used in data science applications."
]

# برای اینکه بتونیم از 'Word2Vec' استفاده کنیم، باید هر سند رو به کلماتش تقسیم کنیم.
# این فرآیند **توکنایز کردن (Tokenize)** نامیده می‌شه.
# این خط، هر سند رو به یه لیست از کلمات تبدیل می‌کنه.
tokenized_documents = [doc.split() for doc in documents]

# این بخش، یک مدل 'Word2Vec' رو روی داده‌های توکنایز شده، آموزش می‌ده.
# 'Word2Vec' کلمات رو به بردارهای عددی تبدیل می‌کنه، به طوری که کلماتی که معنی مشابهی دارن،
# بردارهای نزدیک به همی داشته باشن.
# 'vector_size=10': طول هر بردار رو ۱۰ می‌ذاره.
# 'window=3': به مدل می‌گه که ۳ کلمه قبل و بعد از کلمه فعلی رو نگاه کنه.
# 'min_count=1': کلماتی که حداقل یک بار تکرار شدن رو در نظر می‌گیره.
model = Word2Vec(tokenized_documents, vector_size=10, window=3, min_count=1, workers=4)

# حالا با استفاده از مدل آموزش‌دیده، **شباهت کسینوسی** رو محاسبه می‌کنیم.
# 'model.wv['data']' و 'model.wv['mining']' بردارهای مربوط به کلمات 'data' و 'mining' رو پیدا می‌کنن.
# تابع 'cosine_similarity()' هم شباهت جهت‌گیری این دو بردار رو حساب می‌کنه.
# چون 'data mining' یک اصطلاح مشترکه، انتظار داریم این دو کلمه شباهت زیادی به هم داشته باشن.
embedding_similarity = cosine_similarity([model.wv['data']], [model.wv['mining']])

# در نهایت، نتیجه محاسبه رو چاپ می‌کنیم.
print(f"Cosine Similarity using Word Embeddings: {embedding_similarity[0][0]}")
```

### 2.4.1. Data quality measures

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارهای کیفیت داده</span> ✅
</p>
<p style="font-size:16pt; text-align:right;">
<b>معیارهای کیفیت داده (Data quality measures)</b> برای ارزیابی اعتبار و دقت یک مجموعه داده به کار می‌روند. اطمینان از کیفیت بالای داده‌ها برای موفقیت در داده‌کاوی و تحلیل ضروری است. از معیارها و تکنیک‌های مختلفی برای ارزیابی جنبه‌های گوناگون کیفیت داده، از جمله کامل بودن (completeness)، دقت (accuracy)، سازگاری (consistency) و بهنگام بودن (timeliness) استفاده می‌شود.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال معیار کیفیت داده: مقادیر گمشده</span> ⚠️
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
<b>مقادیر گمشده (Missing Values)</b> یک مشکل رایج در مجموعه‌داده‌ها هستند و می‌توانند بر کیفیت تحلیل تأثیر بگذارند. یکی از معیارهای کیفیت داده، ارزیابی درصد مقادیر گمشده در هر ستون است.
</p>
</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'pandas' که ابزار اصلی ما برای کار با داده‌های جدولی هست.
# 'numpy' هم برای کار با داده‌های عددی و مقادیر خاص مثل 'None' استفاده می‌شه.

# این یه دیکشنری (dictionary) با داده‌های نمونه ماست.
# در این دیتاست، بعضی از مقادیر به صورت عمدی حذف شدن و با 'None' مشخص شدن.
# 'None' نشون‌دهنده مقادیر گمشده (Missing Values) هست که باید قبل از تحلیل داده‌ها، باهاشون برخورد کرد.
data = {
    'ID': [1, 2, 3, 4, 5],
    'Feature1': [10, 15, None, 25, 20],
    'Feature2': [5, None, 8, 12, 10],
    'Feature3': [2, 4, 6, None, 8]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، درصد مقادیر گمشده در هر ستون رو محاسبه می‌کنه.
# 'df.isnull()' یه 'DataFrame' جدید می‌سازه که هر مقداری که 'None' باشه رو با 'True' و بقیه رو با 'False' پر می‌کنه.
# '.sum()' مجموع 'True'ها (یعنی مقادیر گمشده) رو تو هر ستون حساب می‌کنه.
# 'len(df)' تعداد کل سطرهای 'DataFrame' رو می‌ده.
# در نهایت، با تقسیم تعداد مقادیر گمشده بر تعداد کل سطرها و ضرب در ۱۰۰، درصد رو به دست می‌آریم.
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("درصد مقادیر گمشده:")
print(missing_percentage)
```

در این مثال، مجموعه داده حاوی مقادیر گمشده‌ای است که با None نمایش داده شده‌اند. این کد درصد مقادیر گمشده در هر ستون را با استفاده از تابع <b><code>isnull()</code></b> و تابع <b><code>sum()</code></b> در پانداس محاسبه می‌کند.
<br>
ارزیابی کیفیت داده برای اتخاذ تصمیمات آگاهانه در طول فرآیند داده‌کاوی حیاتی است. این مثال بر مقادیر گمشده تمرکز دارد، اما سایر معیارهای کیفیت داده می‌توانند شامل بررسی نقاط پرت (outliers)، اطمینان از سازگاری در مقادیر کیفی و اعتبارسنجی داده‌ها در برابر قوانین تجاری از پیش تعریف‌شده باشند.

### 2.4.2. Data cleaning

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">پاکسازی داده (Data Cleaning)</span> 🧹

<p style="font-size:16pt; text-align:right;">
پاکسازی داده یک گام حیاتی در مرحله پیش‌پردازش داده‌ها (data preprocessing) است که با هدف شناسایی و رفع خطاها، ناسازگاری‌ها و نادرستی‌ها در مجموعه داده انجام می‌شود. این فرآیند شامل وظایفی مانند مدیریت مقادیر گمشده (missing values)، اصلاح انواع داده‌ها، حذف موارد تکراری و رسیدگی به نقاط پرت (outliers) است. اطمینان از داده‌های تمیز و قابل اعتماد برای دستیابی به نتایج معنادار و دقیق در داده‌کاوی ضروری است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال وظیفه پاکسازی داده: مدیریت مقادیر گمشده</b> 📝
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
یکی از وظایف رایج در پاکسازی داده، مدیریت مقادیر گمشده است. بسته به ماهیت داده‌های گمشده، می‌توان از راهکارهایی مانند تکمیل داده (imputation) یا حذف رکورد (removal) استفاده کرد.
</p>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'pandas' که ابزار اصلی ما برای کار با داده‌های جدولی هست.
# 'numpy' هم برای کار با داده‌های عددی و مقادیر خاص مثل 'None' استفاده می‌شه.

# این یه دیکشنری (dictionary) با داده‌های نمونه ماست.
# در این دیتاست، بعضی از مقادیر به صورت عمدی حذف شدن و با 'None' مشخص شدن.
# 'None' نشون‌دهنده مقادیر گمشده (Missing Values) هست که باید قبل از تحلیل داده‌ها، باهاشون برخورد کرد.
data = {
    'ID': [1, 2, 3, 4, 5],
    'Feature1': [10, 15, None, 25, 20],
    'Feature2': [5, None, 8, 12, 10],
    'Feature3': [2, 4, 6, None, 8]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، درصد مقادیر گمشده در هر ستون رو محاسبه می‌کنه.
# 'df.isnull()' یه 'DataFrame' جدید می‌سازه که هر مقداری که 'None' باشه رو با 'True' و بقیه رو با 'False' پر می‌کنه.
# '.sum()' مجموع 'True'ها (یعنی مقادیر گمشده) رو تو هر ستون حساب می‌کنه.
# 'len(df)' تعداد کل سطرهای 'DataFrame' رو می‌ده.
# در نهایت، با تقسیم تعداد مقادیر گمشده بر تعداد کل سطرها و ضرب در ۱۰۰، درصد رو به دست می‌آریم.
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("درصد مقادیر گمشده:")
print(missing_percentage)


# این بخش، مقادیر گمشده رو با استفاده از میانگین (mean) پر می‌کنه.
# این فرآیند به **imputation** یا **پر کردن داده‌ها** معروفه.
# 'df.fillna()' تابعیه که مقادیر گمشده رو پیدا می‌کنه و اون‌ها رو با یه مقدار مشخص جایگزین می‌کنه.
# 'df.mean()' میانگین هر ستون رو حساب می‌کنه.
# در نهایت، این خط مقادیر 'None' رو با میانگین ستون مربوطه پر می‌کنه.
df_filled = df.fillna(df.mean())

# این خط، دیتاست نهایی رو بعد از پر کردن داده‌ها چاپ می‌کنه.
print("\nدیتاست تمیز شده:")
print(df_filled)
```

### 2.4.3. Data integration

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">یکپارچه‌سازی داده</span> 🧩

<p style="font-size:16pt; text-align:right;">
یکپارچه‌سازی داده (Data Integration) فرآیند ترکیب داده‌ها از منابع مختلف در یک نمای یکپارچه و منسجم است. این فرآیند شامل مدیریت ناسازگاری‌ها، حل تعارضات و ایجاد یک مجموعه داده ادغام‌شده است که می‌تواند برای تحلیل و داده‌کاوی استفاده شود. یکپارچه‌سازی داده هنگام کار با مجموعه‌داده‌های متنوع که ممکن است از پایگاه‌های داده، فایل‌ها یا پلتفرم‌های مختلفی به دست آمده باشند، بسیار حیاتی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال وظیفه یکپارچه‌سازی داده: ادغام مجموعه‌داده‌ها</b> 🔄
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
یکی از وظایف رایج یکپارچه‌سازی داده، ادغام مجموعه‌داده‌ها با یک کلید یا شناسه مشترک است. این کار به ایجاد یک مجموعه داده یکپارچه کمک می‌کند که اطلاعات از چندین منبع را در بر می‌گیرد.
</p>
</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌های جدولی عالیه.

# این دو تا دیکشنری (dictionary) حاوی داده‌های نمونه ما هستن.
# این داده‌ها رو به صورت دو جدول مجزا در نظر بگیرید که یک ستون مشترک به نام 'ID' دارن.
data1 = {'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']}
data2 = {'ID': [2, 3, 4], 'Age': [25, 30, 22]}

# با استفاده از داده‌های بالا، دو 'DataFrame' جدید می‌سازیم.
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# این بخش، دو 'DataFrame' رو با هم ترکیب می‌کنه.
# این فرآیند به **ادغام (Merge)** معروفه.
# 'pd.merge()' تابعیه که این کار رو انجام می‌ده.
# 'on='ID'' به تابع می‌گه که بر اساس ستون مشترک 'ID' این دو جدول رو به هم وصل کنه.
# 'how='outer'' هم بهش می‌گه که تمام سطرها رو از هر دو جدول نگه داره.
# اگه یک 'ID' فقط تو یه جدول باشه، مقادیر مربوط به جدول دیگه رو با 'NaN' (مقدار گمشده) پر می‌کنه.
merged_df = pd.merge(df1, df2, on='ID', how='outer')

# در نهایت، 'DataFrame' ترکیب‌شده رو چاپ می‌کنیم.
print("دیتاست ترکیب شده:")
print(merged_df)
```

در این مثال، دو مجموعه داده (<b>df1</b> و <b>df2</b>) دارای یک کلید مشترک (<b>'ID'</b>) هستند. تابع <b><code>merge</code></b> در pandas برای ادغام این مجموعه‌داده‌ها بر اساس ستون <b>'ID'</b> استفاده می‌شود.

</p>
<p style="font-size:16pt; text-align:right;">
<b>نرمال‌سازی (Normalization)</b> یک تکنیک پیش‌پردازش داده است که برای مقیاس‌بندی و تبدیل ویژگی‌های عددی در یک مجموعه داده به یک مقیاس مشترک استفاده می‌شود. این فرآیند زمانی حیاتی است که ویژگی‌ها واحدهای یا مقیاس‌های متفاوتی داشته باشند، زیرا تضمین می‌کند که هر ویژگی به طور متناسب در تحلیل مشارکت کند. نرمال‌سازی معمولاً شامل تبدیل داده‌ها به یک مقیاس استاندارد، مانند بین 0 و 1 است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال نرمال‌سازی در پایتون: مقیاس‌بندی Min-Max</b>
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
<b>مقیاس‌بندی Min-Max</b> یک روش رایج نرمال‌سازی است که مقادیر یک ویژگی را به یک بازه مشخص، معمولاً بین 0 و 1، تبدیل می‌کند. فرمول مقیاس‌بندی Min-Max به صورت زیر است:
</p>
<p style="font-size:16pt; text-align:center;">
X<sub>normalized</sub>= (X − X<sub>min</sub>) / (X<sub>max</sub> − X<sub>min</sub>)
</p>
</div>

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import numpy as np

# این قسمت اول کده. ما سه تا کتابخونه اصلی رو وارد می‌کنیم:
# 'pandas' برای کار با داده‌های جدولی، 'MinMaxScaler' از 'sklearn' برای مقیاس‌گذاری،
# و 'matplotlib.pyplot' برای رسم نمودار.

# این دیکشنری حاوی داده‌های نمونه ماست که دو تا ویژگی عددی (Feature) دارن.
# این ویژگی‌ها مقیاس‌های متفاوتی دارن: 'Feature1' بین ۱۰ تا ۳۰ و 'Feature2' بین ۵۰۰ تا ۱۵۰۰.
# این تفاوت مقیاس می‌تونه بعضی از الگوریتم‌های یادگیری ماشین رو گمراه کنه.
data = {
    'Feature1': [10, 20, 15, 25, 30],
    'Feature2': [500, 1000, 750, 1250, 1500]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، داده‌ها رو با استفاده از **مقیاس‌گذاری Min-Max** نرمال‌سازی می‌کنه.
# مقیاس‌گذاری Min-Max داده‌ها رو به بازه مشخصی (معمولاً بین ۰ و ۱) منتقل می‌کنه.
# این کار باعث می‌شه که همه ویژگی‌ها به یک مقیاس یکسان برسن و هیچ کدوم بر دیگری غلبه نکنه.
# 'scaler.fit_transform(df)' داده‌ها رو به مدل آموزش می‌ده و هم‌زمان تبدیلشون می‌کنه.
scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(df)

# این بخش، داده‌های نرمال‌سازی شده رو به یه 'DataFrame' تبدیل می‌کنه تا راحت‌تر قابل نمایش باشن.
normalized_df = pd.DataFrame(normalized_data, columns=df.columns)
print("دیتاست نرمال‌سازی شده:")
print(normalized_df)

# این بخش، نمودارهای پراکندگی (Scatter Plots) رو برای مقایسه داده‌های اصلی و نرمال‌سازی شده رسم می‌کنه.
# 'plt.subplots()' یه فضای رسم با دو تا نمودار در کنار هم ایجاد می‌کنه.
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# نمودار اول: دیتاست اصلی
# 'ax1.scatter()' داده‌های اصلی رو رسم می‌کنه.
# 'set_title()' و 'set_xlabel()' و 'set_ylabel()' عنوان و برچسب محورها رو تنظیم می‌کنن.
ax1.scatter(df['Feature1'], df['Feature2'], c='blue', label='Original Data')
ax1.set_title('Original Dataset')
ax1.set_xlabel('Feature1')
ax1.set_ylabel('Feature2')
ax1.legend()

# نمودار دوم: دیتاست نرمال‌سازی شده
# 'ax2.scatter()' داده‌های نرمال‌سازی شده رو رسم می‌کنه.
# با نگاه به این نمودار می‌شه دید که داده‌ها فشرده شدن و به بازه کوچیکتری منتقل شدن.
ax2.scatter(normalized_df['Feature1'], normalized_df['Feature2'], c='red', label='Normalized Data')
ax2.set_title('Normalized Dataset')
ax2.set_xlabel('Normalized Feature1')
ax2.set_ylabel('Normalized Feature2')
ax2.legend()

# 'plt.show()' نمودارها رو نمایش می‌ده.
plt.show()
```

در این مثال، از <code>MinMaxScaler</code> از کتابخانه <code>scikit-learn</code> برای نرمال‌سازی ویژگی‌های عددی در مجموعه داده استفاده می‌شود.
<br>
نرمال‌سازی برای برخی از الگوریتم‌های یادگیری ماشین، به ویژه آن‌هایی که به مقیاس ویژگی‌های ورودی حساس هستند (مانند k-nearest neighbors یا support vector machines) بسیار حیاتی است. مثال مقیاس‌بندی Min-Max نحوه نرمال‌سازی ویژگی‌های عددی در یک مجموعه داده را نشان می‌دهد.

### 2.5.2. Discretization

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">گسسته‌سازی داده (Discretization)</span> ✂️
</p>
<p style="font-size:16pt; text-align:right;">
گسسته‌سازی (Discretization) یک تکنیک پیش‌پردازش داده است که شامل تبدیل ویژگی‌های عددی پیوسته به دسته‌ها یا سطل‌های (bins) گسسته می‌شود. این کار در شرایطی که ماهیت پیوسته داده‌ها ضروری نباشد یا هنگام کار با الگوریتم‌هایی که با داده‌های کیفی عملکرد بهتری دارند، مفید است. گسسته‌سازی، داده‌ها را ساده‌سازی کرده و می‌تواند آن‌ها را برای انواع خاصی از تحلیل یا مدل‌سازی مناسب‌تر سازد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال گسسته‌سازی در پایتون: سطل‌بندی با عرض یکسان (Equal Width Binning)</b> 📊
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
سطل‌بندی با عرض یکسان، یک روش ساده گسسته‌سازی است که بازه مقادیر پیوسته را به فواصل یا سطل‌هایی با عرض برابر تقسیم می‌کند. هر فاصله یک دسته گسسته را نشان می‌دهد و مقادیر در آن بازه به سطل مربوطه اختصاص می‌یابند.
</p>
</div>

```python
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer

# این قسمت اول کده. دو تا کتابخونه رو وارد می‌کنیم:
# 'pandas' که برای کار با داده‌ها و جدول‌ها عالیه.
# 'KBinsDiscretizer' از 'sklearn' که برای گسسته‌سازی (Discretization) داده‌ها استفاده می‌شه.

# این دیکشنری حاوی داده‌های نمونه ماست.
# این داده‌ها یک ویژگی عددی پیوسته (Continuous) به نام 'Age' (سن) دارن.
data = {
    'Age': [22, 35, 42, 28, 55, 48, 33, 40, 60, 70]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، داده‌ها رو با استفاده از **گسسته‌سازی (Binning) با عرض برابر (Equal-Width)** تبدیل می‌کنه.
# گسسته‌سازی، داده‌های پیوسته رو به دسته‌های محدود (گسسته) تبدیل می‌کنه.
# این روش، بازه کل داده‌ها رو به تعدادی سطل (bin) با عرض برابر تقسیم می‌کنه.
# 'n_bins=3' تعداد سطل‌ها رو ۳ تا می‌ذاره.
# 'strategy='uniform'' از روش عرض برابر استفاده می‌کنه.
# 'encode='ordinal'' هم به هر سطل یه عدد ترتیبی (مثلاً ۰، ۱، ۲) اختصاص می‌ده.
discretizer = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='uniform', subsample=None)
discretized_data = discretizer.fit_transform(df)

# در نهایت، داده‌های گسسته‌سازی شده رو به یه 'DataFrame' تبدیل می‌کنیم و چاپشون می‌کنیم.
print("دیتاست گسسته‌سازی شده:")
print(discretized_data)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از <code>KBinsDiscretizer</code> از کتابخانه <code>scikit-learn</code> برای اعمال سطل‌بندی با عرض یکسان بر روی ویژگی 'Age' استفاده می‌شود که آن را به سه سطل تقسیم می‌کند.
<br>
گسسته‌سازی زمانی ارزشمند است که با الگوریتم‌هایی کار می‌کنید که نیاز به ورودی‌های کیفی دارند یا زمانی که می‌خواهید تحلیل داده‌های پیوسته را ساده کنید. مثال سطل‌بندی با عرض یکسان نشان می‌دهد که چگونه می‌توان یک ویژگی عددی را به سطل‌های کیفی گسسته تبدیل کرد.
</p>
</div>

### 2.5.3. Data compression

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">فشرده‌سازی داده (Data Compression)</span> 💾
</p>
<p style="font-size:16pt; text-align:right;">
فشرده‌سازی داده فرآیندی است که برای کاهش حجم یا اندازه داده‌ها به منظور صرفه‌جویی در فضای ذخیره‌سازی و بهبود کارایی محاسباتی استفاده می‌شود. در زمینه داده‌کاوی، تکنیک‌های فشرده‌سازی می‌توانند برای مدیریت کارآمدتر مجموعه‌داده‌های بزرگ، افزایش سرعت پردازش و کاهش نیاز به منابع برای ذخیره‌سازی و انتقال به کار روند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال فشرده‌سازی داده در پایتون: کدگذاری طول اجرا (Run-Length Encoding)</b>
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
کدگذاری طول اجرا (Run-Length Encoding - RLE) یک تکنیک ساده فشرده‌سازی داده است که مقادیر تکراری متوالی را به عنوان یک مقدار واحد و یک شمارش (count) نمایش می‌دهد. این روش به ویژه برای مجموعه‌داده‌هایی با توالی‌های طولانی از مقادیر یکسان مؤثر است.
</p>
</div>

```python
def run_length_encoding(data):
    # این تابع، داده‌های تکراری پشت سر هم رو به صورت فشرده درمیاره.
    # مثلاً داده‌های '1, 1, 1' رو به '(1, 3)' تبدیل می‌کنه.
    encoded_data = []
    count = 1

    # این حلقه روی داده‌ها می‌چرخه تا مقادیر پشت سر هم رو پیدا کنه.
    for i in range(1, len(data)):
        # اگر مقدار فعلی با مقدار قبلی برابر بود، تعداد تکرار رو یکی زیاد می‌کنیم.
        if data[i] == data[i - 1]:
            count += 1
        # اگه مقادیر فرق داشتن، تکرار قبلی تموم شده.
        else:
            # اینجا، داده و تعداد تکرارش رو به لیست 'encoded_data' اضافه می‌کنیم.
            encoded_data.append((data[i - 1], count))
            # شمارنده رو برای داده جدید، روی ۱ تنظیم می‌کنیم.
            count = 1

    # این خط برای آخرین دسته از داده‌هاست که حلقه تموم شده، ولی هنوز به لیست اضافه نشده.
    encoded_data.append((data[-1], count))
    return encoded_data

# این یه لیست از داده‌های نمونه‌اس که مقادیر تکراری پشت سر هم دارن.
data = [1, 1, 1, 2, 3, 3, 3, 3, 4, 4]

# اینجا از تابع 'run_length_encoding' برای فشرده‌سازی داده‌ها استفاده می‌کنیم.
compressed_data = run_length_encoding(data)

# در نهایت، داده‌های فشرده‌شده رو چاپ می‌کنیم.
print("داده‌های فشرده‌شده:")
print(compressed_data)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از تابع <b><code>run_length_encoding</code></b> برای فشرده‌سازی یک مجموعه داده با مقادیر تکراری متوالی، با استفاده از کدگذاری طول اجرا استفاده می‌شود.
<br>
فشرده‌سازی داده برای کاهش نیازهای ذخیره‌سازی و افزایش سرعت پردازش داده‌ها مفید است. در حالی که کدگذاری طول اجرا یک مثال پایه است، تکنیک‌های فشرده‌سازی پیشرفته‌تری مانند کدگذاری هافمن (Huffman coding) یا فشرده‌سازی لمپل-زیو (Lempel-Ziv compression) نیز وجود دارند.
</p>
</div>

### 2.5.4. Sampling

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">نمونه‌برداری داده (Sampling)</span> 🎲
</p>
<p style="font-size:16pt; text-align:right;">
نمونه‌برداری (Sampling) یک تکنیک پیش‌پردازش داده است که شامل انتخاب یک زیرمجموعه از داده‌ها از یک مجموعه داده بزرگ‌تر می‌شود. این زیرمجموعه که به آن نمونه (sample) گفته می‌شود، نماینده کل جامعه آماری است و اغلب برای اهداف مختلفی مانند تحلیل اکتشافی (exploratory analysis)، آموزش مدل (model training) یا تست عملکرد (performance testing) استفاده می‌شود. نمونه‌برداری به ویژه هنگام کار با مجموعه‌داده‌های بزرگ ارزشمند است، زیرا امکان پردازش و تحلیل کارآمد یک زیرمجموعه قابل مدیریت را فراهم می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>مثال نمونه‌برداری در پایتون: نمونه‌برداری تصادفی (Random Sampling)</b> 🎲
</p>
<p style="font-size:16pt; text-align:right;">
<b>توضیح:</b>
<br>
نمونه‌برداری تصادفی یک تکنیک رایج نمونه‌برداری است که در آن نقاط داده به صورت تصادفی از مجموعه داده انتخاب می‌شوند و یک نمایش بی‌طرفانه از جامعه آماری ارائه می‌دهند.
</p>
</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما دو تا کتابخونه رو وارد می‌کنیم:
# 'pandas' که برای کار با داده‌های جدولی عالیه.
# 'numpy' که برای ایجاد آرایه‌ها و اعداد تصادفی ازش استفاده می‌شه.

# این دیکشنری حاوی داده‌های نمونه ماست.
# 'ID': یک لیست از ۱ تا ۱۰۰ رو ایجاد می‌کنه.
# 'Value': صد تا عدد تصادفی بین ۱ تا ۱۰۰ رو ایجاد می‌کنه.
# این داده‌ها رو به عنوان یک دیتاست بزرگ در نظر بگیرید که می‌خوایم ازش نمونه‌گیری کنیم.
data = {
    'ID': range(1, 101),
    'Value': np.random.randint(1, 100, size=100)
}

# با استفاده از داده‌های بالا، یک 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، **نمونه‌گیری تصادفی** رو انجام می‌ده.
# نمونه‌گیری تصادفی، انتخاب یک زیرمجموعه از داده‌ها به صورت کاملاً تصادفیه.
# این کار برای تحلیل داده‌های خیلی بزرگ که نمی‌تونیم همه‌شون رو بررسی کنیم، خیلی مفیده.
# 'df.sample()' تابعی از 'pandas' هست که این کار رو انجام می‌ده.
# 'n=20' به تابع می‌گه که ۲۰ سطر تصادفی رو انتخاب کنه.
# 'random_state=42' هم باعث می‌شه که هر بار که کد رو اجرا می‌کنیم، همون نمونه‌های تصادفی قبلی انتخاب بشن و نتیجه قابل تکرار باشه.
sample_size = 20
random_sample = df.sample(n=sample_size, random_state=42)

# در نهایت، نمونه تصادفی انتخاب‌شده رو چاپ می‌کنیم.
print("داده‌های نمونه‌گیری شده به صورت تصادفی:")
print(random_sample)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از تابع <b><code>sample</code></b> در <b>pandas</b> برای انجام نمونه‌برداری تصادفی از یک مجموعه داده استفاده می‌شود. پارامتر <b><code>random_state</code></b> تکرارپذیری نتایج را تضمین می‌کند.
<br>
نمونه‌برداری هنگام کار با مجموعه‌داده‌های بزرگ برای قابل مدیریت کردن محاسبات و تسریع فرآیند تحلیل ضروری است. نمونه‌برداری تصادفی تنها یکی از روش‌های متعدد نمونه‌برداری است.

### 2.6.1. Principal components analysis

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">تحلیل مؤلفه‌های اصلی (PCA)</span> 📉
</p>
<p style="font-size:16pt; text-align:right;">
<b>تحلیل مؤلفه‌های اصلی (Principal Components Analysis - PCA)</b> یک تکنیک کاهش ابعاد (dimensionality reduction) است که برای تبدیل یک مجموعه داده به فضایی با ابعاد کمتر، در عین حفظ حداکثر واریانس اصلی، استفاده می‌شود. PCA مؤلفه‌های اصلی را شناسایی می‌کند؛ این مؤلفه‌ها بردارهای عمودی هستند که جهت‌های حداکثر واریانس در داده‌ها را به دست می‌آورند. با تصویرسازی داده‌ها روی این مؤلفه‌ها، PCA امکان ارائه یک نمایش خلاصه‌تر از مجموعه داده را فراهم می‌کند که بصری‌سازی و تحلیل آن را آسان‌تر می‌سازد.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال تحلیل مؤلفه‌های اصلی در پایتون</span> 📉
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه‌های <b>Numpy</b>، <b>Matplotlib</b> ، <b>scikit-learn</b> و <b>pandas</b> برای انجام تحلیل مؤلفه‌های اصلی (PCA) روی یک مجموعه داده نمونه ایجاد کنیم. ما داده‌ها را تولید کرده، PCA را اعمال می‌کنیم و نتایج را بصری‌سازی می‌کنیم.
</p>
</div>

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np

# این قسمت اول کده. ما چند تا ابزار مهم رو وارد می‌کنیم:
# 'pandas' برای کار با داده‌ها تو قالب جدول.
# 'PCA' از 'sklearn' که برای کاهش ابعاد استفاده می‌شه.
# 'load_iris' از 'sklearn' برای وارد کردن یک دیتاست نمونه به اسم 'iris'.
# 'matplotlib.pyplot' برای رسم نمودار.

# این بخش، دیتاست معروف 'iris' رو بارگذاری می‌کنه.
# این دیتاست شامل ویژگی‌های مختلفی از سه گونه گل زنبق (iris) هست.
# ما ویژگی‌ها رو تو یه 'DataFrame' به اسم 'data' و گونه‌ها رو تو یه 'Series' به اسم 'target' می‌ذاریم.
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
target = pd.Series(data=iris.target, name='target')

# این قسمت، **کاهش ابعاد (Dimensionality Reduction)** رو با استفاده از PCA انجام می‌ده.
# PCA یا Principal Component Analysis، یه تکنیکه که تعداد ویژگی‌ها رو کم می‌کنه،
# در حالی که مهم‌ترین اطلاعات داده‌ها رو نگه می‌داره.
# 'n_components=2' بهش می‌گه که داده‌ها رو به ۲ تا ویژگی اصلی (Principal Components) تبدیل کنه.
# 'pca.fit_transform(data)' داده‌ها رو به مدل آموزش می‌ده و تبدیلشون می‌کنه.
pca = PCA(n_components=2)
principal_components = pca.fit_transform(data)

# حالا یه 'DataFrame' جدید می‌سازیم که شامل اون دو تا ویژگی اصلی و گونه‌های گل هست.
# این کار برای رسم نمودار بهمون کمک می‌کنه.
pc_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
pc_df['target'] = target

# این بخش، داده‌ها رو بعد از کاهش ابعاد، روی یه نمودار پراکندگی (scatter plot) رسم می‌کنه.
# 'plt.figure()' یه فضای رسم با اندازه مشخص ایجاد می‌کنه.
plt.figure(figsize=(8, 6))

# 'targets' و 'colors' برای رسم هر گونه گل با یک رنگ متفاوت هستن.
targets = [0, 1, 2]
colors = ['r', 'g', 'b']

# این حلقه (for loop) روی گونه‌های مختلف می‌چرخه.
for target, color in zip(targets, colors):
    # این خط، فقط داده‌های مربوط به یک گونه خاص رو انتخاب می‌کنه.
    indices_to_keep = pc_df['target'] == target
    # 'plt.scatter()' نقاط مربوط به اون گونه رو با رنگ مشخص رسم می‌کنه.
    # به این ترتیب، می‌تونیم ببینیم که گونه‌های مختلف گل چطور در فضای دو بعدی از هم جدا می‌شن.
    plt.scatter(pc_df.loc[indices_to_keep, 'PC1'],
                pc_df.loc[indices_to_keep, 'PC2'],
                c=color,
                label=target,
                alpha=0.7)

# این خطوط برای زیباتر شدن و قابل فهم‌تر شدن نمودار هستن.
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Principal Components Analysis')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، <b>PCA</b> بر روی مجموعه داده Iris اعمال می‌شود، که شامل اندازه‌گیری‌های طول کاسبرگ، عرض کاسبرگ، طول گلبرگ و عرض گلبرگ برای سه گونه از گل‌های زنبق است. کد ابعاد داده‌ها را به دو مؤلفه اصلی کاهش داده و داده‌ها را در فضای کاهش‌یافته بصری‌سازی می‌کند.
<br>
PCA به طور گسترده برای کاهش ویژگی‌ها، کاهش نویز و بصری‌سازی استفاده می‌شود. این مثال نشان می‌دهد که چگونه می‌توان PCA را برای ساده‌سازی و بصری‌سازی داده‌های با ابعاد بالا به کار برد.
</p>
</div>

### 2.6.2. Attribute subset selection

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">انتخاب زیرمجموعه ویژگی‌ها</span> 🎯
</p>
<p style="font-size:16pt; text-align:right;">
<b>انتخاب زیرمجموعه ویژگی‌ها (Attribute subset selection)</b> که به آن انتخاب ویژگی (feature selection) نیز گفته می‌شود، شامل شناسایی و انتخاب یک زیرمجموعه از ویژگی‌های مرتبط از مجموعه اصلی صفات در یک مجموعه داده است. هدف این فرآیند بهبود عملکرد مدل، کاهش ابعاد و افزایش قابلیت تفسیرپذیری با حفظ تنها ویژگی‌های آموزنده است. این فرآیند برای مقابله با "<b>نفرین ابعاد (curse of dimensionality)</b>" حیاتی است و می‌تواند به مدل‌های داده‌کاوی کارآمدتر و دقیق‌تر منجر شود.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال انتخاب زیرمجموعه ویژگی‌ها در پایتون</span> 🎯
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه <b>scikit-learn</b> برای انجام انتخاب ویژگی (Feature Selection) روی یک مجموعه داده نمونه ایجاد کنیم. ما داده‌ها را بارگذاری کرده، روش‌های مختلف انتخاب ویژگی را بررسی و نتایج را نمایش می‌دهیم.
</p>
</div>

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np
import matplotlib.pyplot as plt

# این قسمت اول کده. ما چند تا ابزار مهم رو از کتابخونه 'sklearn' وارد می‌کنیم:
# 'RandomForestClassifier' که یک مدل یادگیری ماشینه.
# 'SelectFromModel' که برای انتخاب ویژگی‌ها بر اساس اهمیتشون استفاده می‌شه.
# 'train_test_split' برای تقسیم داده‌ها به دو بخش آموزش و تست.
# 'accuracy_score' برای ارزیابی دقت مدل.

# این بخش، دیتاست معروف 'iris' رو بارگذاری می‌کنه.
# 'data' شامل ویژگی‌ها و 'target' شامل گونه‌های گل هست.
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
target = pd.Series(data=iris.target, name='target')

# این قسمت، دیتاست رو به دو بخش آموزش (training) و تست (testing) تقسیم می‌کنه.
# 'X_train' و 'y_train' برای آموزش مدل و 'X_test' و 'y_test' برای تست مدل استفاده می‌شن.
# 'test_size=0.2' به این معنیه که ۲۰ درصد داده‌ها برای تست استفاده می‌شن.
# 'random_state=42' هم باعث می‌شه که تقسیم‌بندی هر بار یکسان باشه.
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.2, random_state=42)

# این بخش، یک مدل 'RandomForestClassifier' رو آموزش می‌ده.
# 'RandomForest' یک الگوریتم خیلی قویه که می‌تونه اهمیت هر ویژگی رو برای پیش‌بینی، مشخص کنه.
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)

# حالا از 'SelectFromModel' استفاده می‌کنیم تا ویژگی‌های مهم رو انتخاب کنیم.
# 'threshold=0.2' بهش می‌گه که فقط ویژگی‌هایی رو انتخاب کنه که اهمیتشون بیشتر از ۰.۲ باشه.
# این کار تعداد ویژگی‌ها رو کم می‌کنه و به مدل کمک می‌کنه که بهتر عمل کنه.
sfm = SelectFromModel(clf, threshold=0.2)
sfm.fit(X_train, y_train)

# این بخش، دیتاست‌های آموزش و تست رو بر اساس ویژگی‌های انتخاب‌شده، تغییر می‌ده.
# به این ترتیب، فقط با ویژگی‌های مهم کار می‌کنیم.
X_train_selected = sfm.transform(X_train)
X_test_selected = sfm.transform(X_test)

# حالا یک مدل 'RandomForestClassifier' جدید رو با ویژگی‌های انتخاب‌شده آموزش می‌دیم.
# این مدل فقط از مهم‌ترین ویژگی‌ها برای پیش‌بینی استفاده می‌کنه.
clf_selected = RandomForestClassifier(n_estimators=100, random_state=42)
clf_selected.fit(X_train_selected, y_train)
y_pred_selected = clf_selected.predict(X_test_selected)

# در نهایت، دقت (accuracy) مدل رو با مقایسه پیش‌بینی‌ها با مقادیر واقعی محاسبه می‌کنیم.
# این مقدار نشون می‌ده که مدل ما چقدر خوب تونسته پیش‌بینی کنه.
accuracy = accuracy_score(y_test, y_pred_selected)
print(f'Accuracy with selected features: {accuracy * 100:.2f}%')
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از یک <b><code>RandomForestClassifier</code></b> برای ارزیابی اهمیت ویژگی‌ها استفاده می‌شود و تابع <b><code>SelectFromModel</code></b> برای انتخاب ویژگی‌ها بر اساس اهمیت آن‌ها به کار می‌رود. سپس، طبقه‌بندی‌کننده با استفاده از تنها ویژگی‌های انتخاب‌شده آموزش داده و ارزیابی می‌شود.
<br>
انتخاب زیرمجموعه ویژگی‌ها برای بهبود کارایی مدل، کاهش بیش‌برازش (overfitting) و افزایش قابلیت تفسیرپذیری بسیار حیاتی است. این مثال نحوه اعمال انتخاب زیرمجموعه ویژگی‌ها را با استفاده از اهمیت ویژگی‌ها در یک طبقه‌بندی‌کننده جنگل تصادفی نشان می‌دهد.
</p>

### 2.6.3. Nonlinear dimensionality reduction methods

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">روش‌های کاهش ابعاد غیرخطی</span> 🗜️
</p>
<p style="font-size:16pt; text-align:right;">
<b>روش‌های کاهش ابعاد غیرخطی (Nonlinear dimensionality reduction methods)</b> با هدف به تصویر کشیدن روابط و ساختارهای پیچیده در داده‌های با ابعاد بالا، که ممکن است به خوبی توسط تکنیک‌های خطی مانند <b>PCA</b> نمایش داده نشوند، به کار می‌روند. این روش‌ها داده‌ها را به یک فضای با ابعاد کمتر تبدیل می‌کنند، در حالی که روابط غیرخطی ذاتی درون داده‌ها را حفظ می‌کنند. کاهش ابعاد غیرخطی به ویژه هنگام کار با مجموعه‌داده‌هایی که حاوی الگوهای پیچیده، خوشه‌ها یا منیفلدها هستند، مفید است.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال کاهش ابعاد غیرخطی در پایتون: t-SNE</span> 🗜️
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک قطعه کد پایتون با استفاده از کتابخانه‌های <b>pandas</b> و <b>scikit-learn</b> برای انجام کاهش ابعاد غیرخطی با استفاده از روش t-Distributed Stochastic Neighbor Embedding (t-SNE) روی یک مجموعه داده نمونه ایجاد کنیم. ما داده‌ها را تولید کرده، t-SNE را اعمال می‌کنیم و نتایج را بصری‌سازی می‌کنیم.
</p>
</div>

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np

# این قسمت اول کده. ما چند تا ابزار مهم رو وارد می‌کنیم:
# 'pandas' برای کار با داده‌ها تو قالب جدول.
# 'TSNE' از 'sklearn' که برای کاهش ابعاد غیرخطی استفاده می‌شه.
# 'load_iris' برای وارد کردن یک دیتاست نمونه به اسم 'iris'.
# 'matplotlib.pyplot' برای رسم نمودار.

# این بخش، دیتاست معروف 'iris' رو بارگذاری می‌کنه.
# این دیتاست شامل ویژگی‌های مختلفی از سه گونه گل زنبق (iris) هست.
# ما ویژگی‌ها رو تو یه 'DataFrame' به اسم 'data' و گونه‌ها رو تو یه 'Series' به اسم 'target' می‌ذاریم.
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
target = pd.Series(data=iris.target, name='target')

# این قسمت، **کاهش ابعاد غیرخطی (Nonlinear Dimensionality Reduction)** رو با استفاده از t-SNE انجام می‌ده.
# t-SNE یا t-distributed Stochastic Neighbor Embedding، یه تکنیک قدرتمنده که داده‌ها رو از فضای با ابعاد بالا به فضای با ابعاد پایین (مثلاً دو بعدی) منتقل می‌کنه.
# t-SNE برای بصری‌سازی داده‌ها خیلی عالیه، چون الگوها و خوشه‌ها (clusters) رو که به صورت خطی قابل تشخیص نیستن، به خوبی نشون می‌ده.
# 'n_components=2' بهش می‌گه که داده‌ها رو به دو بعد تبدیل کنه.
# 'random_state=42' هم باعث می‌شه که نتیجه قابل تکرار باشه.
tsne = TSNE(n_components=2, random_state=42)
embedded_data = tsne.fit_transform(data)

# حالا یه 'DataFrame' جدید می‌سازیم که شامل داده‌های تبدیل‌شده و گونه‌های گل هست.
# این کار برای رسم نمودار بهمون کمک می‌کنه.
embedded_df = pd.DataFrame(data=embedded_data, columns=['Component 1', 'Component 2'])
embedded_df['target'] = target

# این بخش، داده‌ها رو بعد از کاهش ابعاد، روی یه نمودار پراکندگی (scatter plot) رسم می‌کنه.
# 'plt.figure()' یه فضای رسم با اندازه مشخص ایجاد می‌کنه.
plt.figure(figsize=(8, 6))

# 'targets' و 'colors' برای رسم هر گونه گل با یک رنگ متفاوت هستن.
targets = [0, 1, 2]
colors = ['r', 'g', 'b']

# این حلقه (for loop) روی گونه‌های مختلف می‌چرخه.
for target, color in zip(targets, colors):
    # این خط، فقط داده‌های مربوط به یک گونه خاص رو انتخاب می‌کنه.
    indices_to_keep = embedded_df['target'] == target
    # 'plt.scatter()' نقاط مربوط به اون گونه رو با رنگ مشخص رسم می‌کنه.
    # با نگاه به این نمودار، می‌شه دید که گونه‌های مختلف چطور در فضای دو بعدی از هم جدا شدن.
    plt.scatter(embedded_df.loc[indices_to_keep, 'Component 1'],
                embedded_df.loc[indices_to_keep, 'Component 2'],
                c=color,
                label=target,
                alpha=0.7)

# این خطوط برای زیباتر شدن و قابل فهم‌تر شدن نمودار هستن.
# عنوان و برچسب محورها به انگلیسی نوشته شدن.
plt.xlabel('Component 1')
plt.ylabel('Component 2')
plt.title('t-SNE: Nonlinear Dimensionality Reduction')
plt.legend()
plt.show()
```
