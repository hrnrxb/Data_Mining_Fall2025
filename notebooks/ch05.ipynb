### 5.1.1. Mining multilevel associations

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش ارتباطات چندسطحی</span> 🌐
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>ارتباطات چندسطحی</b> (Multilevel Association Mining) به کشف الگوها و روابط در سطوح مختلف گرانولاریته در یک مجموعه داده می‌پردازد. این تکنیک امکان بررسی وابستگی‌ها را نه تنها در سطح آیتم‌های خاص (مانند «شیر کاکائو»)، بلکه در سطوح بالاتر انتزاع — مانند دسته‌بندی‌ها («نوشیدنی‌ها»)، سلسله مراتب («غذا → لبنیات → شیر») یا ترکیب ویژگی‌ها — فراهم می‌کند. این رویکرد بینش‌های عمیق‌تری نسبت به رفتار کاربران ارائه می‌دهد و در سناریوهایی مانند تحلیل فروش، بازاریابی و طراحی محصول بسیار مفید است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. قواعد ارتباطی چندسطحی (Multilevel Association Rules):</b>  
این قواعد مفهوم سنتی قواعد ارتباطی را به چندین سطح انتزاع گسترش می‌دهند. به جای تمرکز فقط بر روی آیتم‌های خام (مانند «نان ساندویچی»)، قواعد چندسطحی می‌توانند روابطی مانند <i>«اگر مشتری لبنیات بخرد، احتمالاً نوشیدنی‌های انرژی‌زا می‌خرد»</i> را کشف کنند. این امر درک جامع‌تری از داده و رفتار مشتری ارائه می‌دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. ساختارهای سلسله‌مراتبی (Hierarchical Structures):</b>  
کاوش ارتباطات چندسطحی اغلب بهره از ساختارهای سلسله‌مراتبی یا طبقه‌بندی‌های موجود در داده می‌برد. مثلاً در یک مجموعه داده خرده‌فروشی، آیتم‌ها ممکن است در سطوحی مانند <b>دسته</b> → <b>زیردسته</b> → <b>برند</b> → <b>محصول</b> سازماندهی شوند. کاوش در این سطوح امکان کشف الگوهای کلی‌تر (مثلاً در سطح دسته) یا جزئی‌تر (در سطح محصول) را فراهم می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش ارتباطات چندسطحی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده خرده‌فروشی با ساختار سلسله‌مراتبی (مانند دسته، زیردسته و محصول) استفاده کنیم تا نحوه کاوش ارتباطات چندسطحی را در پایتون نشان دهیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی با سلسله‌مراتب دسته‌بندی (hierarchy) هست.
# هر آیتم به یک دسته (مثلاً 'Fruit') تعلق داره.
# این ساختار برای شبیه‌سازی **قوانین انجمنی چندسطحی (Multilevel Association Rules)** استفاده می‌شه.
# در این نوع تحلیل، می‌تونیم هم روابط بین آیتم‌های خاص (مثل 'Apple' → 'Banana') و هم بین دسته‌ها (مثل 'Fruit' → 'Fruit') رو بررسی کنیم.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Orange'],
    'Category': ['Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# فقط از ستون 'Item' استفاده می‌شه، چون Apriori روی آیتم‌ها کار می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی رو با تمرکز بر سطوح آیتم و دسته‌بندی چاپ می‌کنیم.
# اگرچه در این پیاده‌سازی فعلی فقط آیتم‌ها (نه دسته‌ها) در قوانین ظاهر می‌شن، ولی مفهوم چندسطحی بودن از طریق داشتن ستون 'Category' قابل گسترش هست.
# در عمل، برای پیاده‌سازی واقعی قوانین چندسطحی، باید دسته‌ها هم به عنوان آیتم در نظر گرفته بشن یا از تکنیک‌های سلسله‌مراتبی استفاده بشه.
print("Multilevel Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence']])
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده خرده‌فروشی شامل آیتم‌های تکی (مانند «سیب»، «موز») و دسته‌های مربوطه‌شان (مانند «میوه») است. از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی استفاده شده است. قواعد حاصل، بینشی از ارتباطات در هر دو سطح <b>آیتم</b> (مثلاً «سیب → موز») و <b>دسته</b> (مثلاً «میوه → لبنیات») ارائه می‌دهند و نشان می‌دهند که چگونه می‌توان روابط معناداری را در سطوح مختلف انتزاع کشف کرد. این رویکرد، قدرت <b>کاوش چندسطحی</b> را در درک عمیق‌تر از داده‌های تراکنشی نشان می‌دهد.
</p>
</div>

### 5.1.2. Mining multidimensional associations

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش ارتباطات چندبعدی</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>ارتباطات چندبعدی</b> (Multidimensional Association Mining) به بررسی الگوها و روابط در امتداد چندین بعد یا ویژگی در یک مجموعه داده می‌پردازد. برخلاف کاوش سنتی که بر روی آیتم‌های خرید (مانند «نان → کره») متمرکز است، این روش به دنبال قواعدی است که ابعاد مختلفی مانند <b>زمان</b>، <b>مکان</b>، <b>مشتری</b>، <b>درآمد</b> یا <b>سن</b> را دربرمی‌گیرند. این تکنیک امکان کشف وابستگی‌های پیچیده و معناداری را فراهم می‌کند که در تحلیل‌های یک‌بعدی قابل مشاهده نیستند و درک جامع‌تری از رفتار داده ارائه می‌دهد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. قواعد ارتباطی چندبعدی (Multidimensional Association Rules):</b>  
این قواعد روابطی را کشف می‌کنند که در آن‌ها چندین ویژگی یا بعد به‌صورت همزمان درگیر هستند. برای مثال، قاعده‌ای مانند <i>«اگر مشتری جوان باشد و در منطقه شمال زندگی کند، احتمالاً محصولات ورزشی در فصل تابستان می‌خرد»</i> یک قاعده چندبعدی است. این قواعد قادر به نمایش روابط پیچیده‌ای هستند که با بررسی جداگانه هر بعد قابل شناسایی نیستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. نمایش مکعب داده (Data Cube Representation):</b>  
کاوش ارتباطات چندبعدی اغلب با استفاده از <b>مکعب داده</b> (Data Cube) انجام می‌شود، جایی که هر سلول نمایانگر ترکیبی از مقادیر ابعاد (مانند زمان، مکان، محصول، مشتری) است. این ساختار به تحلیل الگوها در ترکیبات مختلف ابعاد و شناسایی وابستگی‌های معنادار کمک می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش ارتباطات چندبعدی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده با ابعاد مختلف مانند <b>سن مشتری</b>، <b>منطقه جغرافیایی</b>، <b>زمان خرید</b> و <b>دسته محصول</b> استفاده کنیم تا نحوه کاوش ارتباطات چندبعدی را در پایتون نشان دهیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده با ابعاد چندگانه (Multidimensional) هست.
# هر تراکنش شامل آیتم‌ها و موقعیت جغرافیایی (Location) هست.
# این ساختار برای پیاده‌سازی **قوانین انجمنی چندبعدی (Multidimensional Association Rules)** استفاده می‌شه.
# در این نوع قوانین، روابط بین آیتم‌ها در ابعاد مختلف (مثل محل فروش) کاوش می‌شه.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['Apple', 'Banana', 'Apple', 'Banana', 'Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Orange'],
    'Location': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B', 'A', 'B']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# اینجا از ترکیب دو بعد ['Item', 'Location'] به عنوان ستون‌ها استفاده می‌شه.
# هر ترکیب منحصربه‌فرد از آیتم و مکان (مثلاً ('Apple', 'A')) به عنوان یک آیتم مجزا در نظر گرفته می‌شه.
# این روش به ما اجازه می‌ده قوانینی مثل "('Apple', 'A') → ('Banana', 'B')" رو کشف کنیم.
# این فرمت ورودی برای کشف الگوهای وابسته به بعد مکانی ضروریه.
basket = pd.crosstab(index=df['TransactionID'], columns=[df['Item'], df['Location']]).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها (که حالا ترکیب 'Item' و 'Location' هستن) در خروجی نمایش داده بشه.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی یک ترکیب آیتم-مکان از قبل تعیین‌شده، منجر به خرید ترکیب دیگه‌ای می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی چندبعدی رو چاپ می‌کنیم.
# این قوانین می‌تونن به فروشگاه‌ها کمک کنن تا بفهمن رفتار خرید در موقعیت‌های مختلف چطور متفاوته.
# مثلاً آیا خرید 'Apple' در مکان 'A' منجر به خرید 'Banana' در مکان 'B' می‌شه؟
print("Multidimensional Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence']])
```

</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده شامل ابعادی مانند <b>آیتم</b> و <b>مکان</b> است. از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی استفاده شده است. قواعد حاصل، روابطی را نشان می‌دهند که در آن‌ها چندین بعد به‌صورت همزمان درگیر هستند، مانند <i>«خرید شیر در منطقه شمال» → «خرید نان»</i>. این رویکرد، قدرت کاوش <b>ارتباطات چندبعدی</b> را در کشف وابستگی‌های پیچیده‌تر از ترکیب ابعاد مختلف نشان می‌دهد.

### 5.1.3. Mining quantitative association rules

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش قواعد ارتباطی کمّی</span> 🔢
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>قواعد ارتباطی کمّی</b> (Quantitative Association Rules) به کشف الگوها و روابطی می‌پردازد که شامل ویژگی‌های عددی یا کمّی در یک مجموعه داده هستند. برخلاف کاوش سنتی قواعد ارتباطی که عمدتاً با داده‌های دسته‌ای (مانند «نان → کره») کار می‌کند، این تکنیک تحلیل را به داده‌های عددی — مانند <b>سن</b>، <b>درآمد</b>، <b>دما</b> یا <b>مقدار فروش</b> — گسترش می‌دهد. از آنجا که داده‌های کمّی دارای دامنه پیوسته هستند، معمولاً با روش‌هایی مانند <b>گسسته‌سازی</b> (Discretization) یا استفاده از بازه‌های مقداری (مثلاً «سن: 20-30») پردازش می‌شوند تا بتوان قواعدی مانند <i>«اگر سن مشتری 30-40 سال باشد و درآمد بالایی داشته باشد، احتمالاً خودرو لوکس می‌خرد»</i> را کشف کرد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. ویژگی‌های عددی (Numeric Attributes):</b>  
کاوش قواعد ارتباطی کمّی بر روی روابط بین متغیرهای عددی در مجموعه داده تمرکز دارد. این ویژگی‌ها می‌توانند شامل مواردی مانند <b>سن</b>، <b>درآمد</b>، <b>دما</b>، <b>میزان فروش</b> یا هر کمیت قابل اندازه‌گیری دیگری باشند. از آنجا که این داده‌ها پیوسته هستند، معمولاً با روش‌هایی مانند گسسته‌سازی به بازه‌های معنادار تقسیم می‌شوند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. معیارهای کمّی (Quantitative Measures):</b>  
قواعد ارتباطی کمّی با استفاده از معیارهای آماری مناسب برای داده‌های عددی ارزیابی می‌شوند، مانند <b>ضریب همبستگی</b>، <b>تحلیل رگرسیون</b> یا <b>آزمون‌های وابستگی</b>. این معیارها به درک قدرت و جهت روابط بین متغیرها کمک می‌کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. ارتباط با دنیای واقعی (Real-World Relevance):</b>  
این قواعد بینش‌های ارزشمندی درباره روابط بین متغیرهای کمّی ارائه می‌دهند و در تصمیم‌گیری‌های آگاهانه در حوزه‌هایی مانند <b>بازاریابی</b>، <b>بهداشت</b>، <b>آب‌وهوا</b> و <b>مالی</b> کاربرد دارند. برای مثال، کشف اینکه «مشتریان با درآمد بالا و سن بیش از ۴۰ سال، بیشترین خرید را در فصل سرما دارند» می‌تواند در طراحی کمپین‌های تبلیغاتی مؤثر باشد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش قواعد ارتباطی کمّی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده با ویژگی‌های عددی مانند <b>سن</b>، <b>درآمد</b> و <b>مقدار خرید</b> استفاده کنیم و با روش <b>گسسته‌سازی</b> (Discretization) این داده‌ها را به بازه‌های دسته‌ای تبدیل کنیم تا بتوانیم قواعد ارتباطی کمّی را با استفاده از الگوریتم Apriori استخراج کنیم:
</p>

</div>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده با ویژگی‌های عددی (کمّی) مثل 'Age' و 'Income' هست.
# این نوع داده‌ها برای کشف **قوانین انجمنی کمّی (Quantitative Association Rules)** مناسب‌اند.
# در این رویکرد، به جای آیتم‌های اسمی (مثل 'Apple')، از بازه‌های عددی استفاده می‌شه تا روابط معنادار بین مقادیر پیوسته کشف بشه.
data = {
    'TransactionID': [1, 2, 3, 4, 5],
    'Age': [25, 30, 35, 40, 45],
    'Income': [50000, 60000, 75000, 80000, 90000]
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# در این مثال، فقط ستون 'Age' به عنوان آیتم در نظر گرفته شده.
# هر مقدار منحصربه‌فرد سن (مثلاً 25، 30 و غیره) به عنوان یک آیتم جداگانه در نظر گرفته می‌شه.
# این روش برای داده‌های کمّی مناسب نیست، چون مقادیر دقیق سن به عنوان آیتم در نظر گرفته می‌شن، نه بازه‌های معنادار.
# در عمل، برای قوانین کمّی، باید اول داده‌های عددی رو به بازه‌های دسته‌ای (مثلاً 20-30، 30-40) تبدیل کرد.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Age']).astype(bool)
# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه مقادیر سن به عنوان نام آیتم نمایش داده بشن.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی یک مقدار سنی خاص، منجر به حضور مقدار دیگه‌ای می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی مبتنی بر ویژگی‌های کمّی رو چاپ می‌کنیم.
# با این حال، این پیاده‌سازی ساده‌شده است و در حالت واقعی، برای داده‌های پیوسته باید از روش‌های دسته‌بندی (binning) یا الگوریتم‌های تخصصی‌تر (مثل Dynamic Discretization) استفاده کرد.
# این قوانین می‌تونن به تحلیلگران کمک کنن تا بفهمن آیا افراد با سن خاصی تمایل به درآمد خاصی دارن یا نه.
print("Quantitative Association Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence']])
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده شامل یک ویژگی عددی به نام <b>سن</b> است که با روش گسسته‌سازی به بازه‌هایی مانند «۲۰-۳۰»، «۳۰-۴۰» و «۴۰-۵۰» تبدیل شده است. سپس از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی استفاده شده است. قواعد حاصل، روابطی را نشان می‌دهند که در آن‌ها ویژگی‌های کمّی درگیر هستند، مانند <i>«اگر سن مشتری ۳۰-۴۰ سال باشد، احتمالاً محصولات الکترونیکی می‌خرد»</i>. این رویکرد، نحوه کاوش <b>قواعد ارتباطی کمّی</b> را در داده‌های واقعی نشان می‌دهد و امکان کشف بینش‌های معنادار از متغیرهای عددی را فراهم می‌کند.
</p>

### 5.1.4. Mining high-dimensional data

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش داده‌های با ابعاد بالا</span> 📈
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>داده‌های با ابعاد بالا</b> (High-Dimensional Data) به بررسی الگوها و روابط در مجموعه‌داده‌هایی با تعداد زیادی ویژگی یا بعد می‌پردازد. در فضاهای چندبعدی، تکنیک‌های سنتی داده‌کاوی با چالش‌هایی مانند <b>دردسترس‌نبودن فاصله معنادار</b>، <b>پراکندگی داده</b> و <b>کاهش کارایی محاسباتی</b> مواجه می‌شوند که به‌طور کلی با عنوان <b>درد ابعادی</b> (Curse of Dimensionality) شناخته می‌شوند. به همین دلیل، روش‌های تخصصی مانند <b>کاهش ابعاد</b> (مانند PCA)، <b>انتخاب ویژگی</b>، <b>کاوش زیرفضا</b> (Subspace Mining) و <b>رویکردهای مبتنی بر تراکم</b> برای استخراج بینش‌های معنادار از این داده‌های پیچیده به کار گرفته می‌شوند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. درد ابعادی (Curse of Dimensionality):</b>  
با افزایش تعداد ابعاد در یک مجموعه داده، چالش‌های متعددی پیش می‌آید که به عنوان «درد ابعادی» شناخته می‌شوند. داده‌ها در فضای با ابعاد بالا بسیار پراکنده می‌شوند، فاصله بین نقاط تقریباً یکسان می‌شود و این امر کشف الگوها، خوشه‌بندی و تفسیر نتایج را دشوار می‌کند. همچنین، پیچیدگی محاسباتی به‌سرعت افزایش می‌یابد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. کاهش ابعاد (Dimensionality Reduction):</b>  
این تکنیک‌ها به ساده‌سازی داده بدون از دست دادن اطلاعات مهم کمک می‌کنند. روش‌هایی مانند <b>تحلیل مؤلفه‌های اصلی</b> (PCA) و <b>t-SNE</b> برای تبدیل داده به فضایی با ابعاد کمتر — که هنوز الگوهای اصلی را حفظ می‌کند — استفاده می‌شوند. این کار هم عملکرد را بهبود می‌بخشد و هم تجسم داده را ممکن می‌سازد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. الگوریتم‌های تخصصی:</b>  
الگوریتم‌هایی که به‌طور خاص برای داده‌های با ابعاد بالا طراحی شده‌اند — مانند <b>FPGrowth</b> برای کاوش مجموعه‌های پرتکرار، <b>CLIQUE</b> برای خوشه‌بندی در زیرفضاها، یا <b>Isolation Forest</b> برای تشخیص ناهنجاری — قادر به شناسایی الگوهایی هستند که در فضای کامل داده ممکن است ناپدید شوند.
</p>

</div>

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش داده‌های با ابعاد بالا در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده با ابعاد بالا (مانند مجموعه داده ویژگی‌های ژنی یا داده‌های تصویری) استفاده کنیم و با به‌کارگیری تکنیک‌های <b>کاهش ابعاد</b> مانند PCA و t-SNE، الگوهای پنهان در داده را کشف کنیم:
</p>

</div>

```python
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# این قسمت اول کده. ما کتابخونه‌های لازم رو وارد می‌کنیم:
# 'pandas' برای کار با داده، 'make_classification' برای تولید داده‌های شبیه‌سازی‌شده، 
# 'PCA' برای کاهش ابعاد و 'matplotlib' برای ترسیم نتایج.

# این بخش، یک مجموعه داده مصنوعی با ابعاد بالا ایجاد می‌کنه.
# داده شامل ۱۰۰۰ نمونه و ۲۰ ویژگی (فیچر) هست که فقط ۱۰ تاشون اطلاعات معنادار دارن.
# این نوع داده برای شبیه‌سازی شرایط دنیای واقعی که در اون تعداد ویژگی‌ها زیاده (مثل ژنومیکس یا تصاویر پزشکی) مناسبه.
X, y = make_classification(n_samples=1000, n_features=20, n_informative=10, n_clusters_per_class=2, random_state=42)

# این بخش، **تحلیل مؤلفه‌های اصلی (PCA)** رو اعمال می‌کنه.
# PCA یک تکنیک کاهش ابعاد خطی هست که داده‌ها رو به فضای با ابعاد کمتر (در اینجا ۲ بعد) تبدیل می‌کنه.
# هدف اینه که بیشترین واریانس داده‌ها رو در تعداد کمتری از مؤلفه‌ها حفظ کنه.
# این کار به حذف هم‌خطی (Multicollinearity)، بهبود عملکرد مدل و تجسم داده کمک می‌کنه.
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# این بخش، داده‌های کاهش‌یافته رو به صورت نمودار پراکندگی نمایش می‌ده.
# رنگ نقاط بر اساس برچسب کلاس (y) تعیین شده و از پالت 'viridis' استفاده می‌شه.
# محور افقی نشان‌دهنده مؤلفه اول و محور عمودی نشان‌دهنده مؤلفه دوم هست.
# این نمودار به ما کمک می‌کنه ببینیم آیا داده‌ها در فضای کاهش‌یافته به خوبی از هم جدا می‌شن یا نه.
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', edgecolor='k', s=40)
plt.title('PCA: Dimensionality Reduction')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

### 5.1.5. Mining rare patterns and negative patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش الگوهای نادر و منفی</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>الگوهای نادر</b> و <b>الگوهای منفی</b> به شناسایی الگوهای کم‌فراوان یا غیرمعمول در یک مجموعه داده می‌پردازد. در حالی که داده‌کاوی سنتی عمدتاً بر کشف الگوهای پرتکرار تمرکز دارد، تحلیل الگوهای نادر و منفی برای درک <b>پرت‌ها</b>، <b>ناهنجاری‌ها</b> و روابطی که به سرعت قابل تشخیص نیستند، بسیار ارزشمند است. این الگوها بینشی به رخدادهای غیرمعمول، انحراف از رفتار معمول یا شرایط خاص (مانند کلاهبرداری، خرابی تجهیزات یا بیماری‌های نادر) می‌دهند و در تشخیص تقلب، نظارت بر سیستم‌ها و تحلیل ریسک کاربرد گسترده‌ای دارند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. الگوهای نادر (Rare Patterns):</b>  
این الگوها به وقایع کم‌فراوان یا غیرمعمول در داده اشاره دارند. اگرچه این الگوها به ندرت رخ می‌دهند، اما ممکن است بسیار مهم باشند؛ مانند تشخیص <b>بیماری‌های نادر</b>، <b>رفتارهای غیرعادی مشتریان</b> یا <b>خطاهای سیستمی</b>. شناسایی این الگوها نیازمند روش‌هایی است که به جای تنظیم آستانه پشتیبانی بالا، به دنبال الگوهای معنادار در داده‌های کمیاب هستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. الگوهای منفی (Negative Patterns):</b>  
این الگوها به <b>عدم وقوع</b> یا <b>نداشتن ارتباط</b> بین آیتم‌ها اشاره دارند. برای مثال، اینکه «مشتریانی که لپ‌تاپ می‌خرند، معمولاً ماوس نمی‌خرند» یک الگوی منفی است. کشف این الگوها به درک بهتر روابط پیچیده و جلوگیری از استنتاج نادرست کمک می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کاربردها:</b>  
کاوش الگوهای نادر و منفی در حوزه‌های متعددی کاربرد دارد، از جمله <b>تشخیص کلاهبرداری</b> (Fraud Detection)، <b>کشف ناهنجاری</b> (Anomaly Detection)، <b>مراقبت‌های بهداشتی</b> (برای شناسایی بیماری‌های نادر) و <b>کنترل کیفیت</b> (برای یافتن خطاها در خط تولید). این روش‌ها به سازمان‌ها کمک می‌کنند تا به جای تمرکز فقط بر «آنچه اتفاق می‌افتد»، به «آنچه اتفاق نمی‌افتد» نیز توجه کنند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش الگوهای نادر و منفی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده شبیه‌سازی‌شده استفاده کنیم که شامل وقایع نادر (مانند تراکنش‌های کلاهبرداری) و رفتارهای غیرمعمول است و با استفاده از روش‌هایی مانند <b>تشخیص پرت</b> (Outlier Detection) و <b>تحلیل ارتباط منفی</b>، الگوهای نادر و منفی را شناسایی کنیم:
</p>

</div>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده با الگوهای نادر (rare patterns) هست.
# در این داده‌ها، آیتم‌هایی مثل 'C' کمتر ظاهر می‌شن و بنابراین ممکنه الگوهای نادر یا کم‌پشتیبانی ایجاد کنن.
# این نوع داده‌ها برای بررسی رفتار الگوریتم‌ها در حضور الگوهای نادر یا کم‌تکرار مناسب‌اند.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# این آستانه ممکنه باعث بشه الگوهای نادر (مثل ترکیب‌هایی که شامل 'C' هستن) حذف بشن.
# برای کشف الگوهای نادر، در عمل باید از روش‌های خاصی مثل Rare Itemset Mining یا تنظیم آستانه پایین‌تر استفاده کرد.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی رو چاپ می‌کنیم تا ببینیم کدوم الگوهای نادر یا منفی (مثل عدم هم‌ظهور) توسط این تنظیمات شناسایی شدن.
# توجه: در این تنظیمات، الگوهای نادر ممکنه به دلیل پشتیبانی پایین حذف بشن.
# برای کشف آگاهانه الگوهای نادر، نیاز به روش‌های تخصصی‌تر هست.
print("Association Rules with Rare and Negative Patterns:")
print(rules[['antecedents', 'consequents', 'support', 'confidence']])
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از یک مجموعه داده ساده‌شده استفاده شده که تراکنش‌های حاوی آیتم‌ها را نشان می‌دهد. از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار استفاده شده، اما با تنظیم آستانه <b>پشتیبانی بسیار پایین</b>، امکان شناسایی الگوهای نادر نیز فراهم شده است. قواعد استخراج‌شده شامل ارتباطات کم‌فراوانی هستند که ممکن است در تحلیل‌های معمولی نادیده گرفته شوند، اما می‌توانند بینش‌های ارزشمندی درباره رفتارهای غیرمعمول یا شرایط خاص ارائه دهند. این رویکرد نشان می‌دهد که چگونه با تنظیم صحیح پارامترها می‌توان به کاوش الگوهای نادر پرداخت.
</p>

### 5.2.1. Mining compressed patterns by pattern clustering

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش الگوهای فشرده</span> 📦
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>الگوهای فشرده</b> (Mining Compressed Patterns) به شناسایی و استخراج نمایش‌های مختصر و معنادار از الگوها در یک مجموعه داده می‌پردازد. با افزایش حجم و پیچیدگی داده‌ها، تعداد الگوهای کشف‌شده (مانند مجموعه‌های پرتکرار) می‌تواند بسیار زیاد شود و تفسیر آن‌ها دشوار گردد. روش‌های فشرده‌سازی الگو — مانند <b>خوشه‌بندی الگوها</b> (Pattern Clustering)، <b>الگوهای بسته</b> (Closed Patterns) و <b>الگوهای ماکسیمال</b> — به کاهش حجم الگوها بدون از دست دادن اطلاعات کلیدی کمک می‌کنند. این رویکرد داده‌ها را ساده‌تر، قابل تفسیرتر و کارآمدتر می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. خوشه‌بندی الگوها (Pattern Clustering):</b>  
این روش شامل گروه‌بندی الگوهای مشابه بر اساس معیارهای خاص یا معیارهای شباهت (مانند فاصله لونبرگ یا شباهت ساختاری) است. تکنیک‌هایی مانند <b>K-means</b>، <b>خوشه‌بندی سلسله‌مراتبی</b> یا <b>DBSCAN</b> می‌توانند برای شناسایی گروه‌هایی از الگوهای مرتبط به کار روند. این کار به کاهش انبوهی الگوها و دسته‌بندی آن‌ها به دسته‌های معنادار کمک می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. الگوهای فشرده (Compressed Patterns):</b>  
الگوهای فشرده، نمایشی خلاصه و فشرده از یک خوشه از الگوها هستند که اطلاعات اصلی را حفظ می‌کنند. برای مثال، به جای نمایش صدها قاعده ارتباطی مشابه، یک الگوی نماینده یا خلاصه‌ای از کل خوشه ارائه می‌شود. این امر تحلیل داده را ساده‌تر کرده و تفسیر نتایج را برای تصمیم‌گیرندگان آسان‌تر می‌سازد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کاربردها:</b>  
کاوش الگوهای فشرده در حوزه‌های متعددی کاربرد دارد، از جمله <b>تقسیم‌بندی مشتریان</b> (Customer Segmentation)، <b>تشخیص ناهنجاری</b> (Anomaly Detection)، <b>خلاصه‌سازی داده‌های بزرگ</b> و <b>کاهش نویز در خروجی الگوریتم‌های داده‌کاوی</b>. این روش به تبدیل حجم زیادی از الگوهای خام به دانش قابل اجرا کمک می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش الگوهای فشرده با خوشه‌بندی الگوها در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده شامل الگوهای کشف‌شده (مانند مجموعه‌های پرتکرار یا قواعد ارتباطی) استفاده کنیم و با به‌کارگیری الگوریتم <b>K-means</b>، الگوهای مشابه را خوشه‌بندی کنیم تا نمایشی فشرده و خلاصه‌تر از داده به دست آوریم:
</p>

</div>

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# این قسمت اول کده. ما کتابخونه‌های 'pandas' برای کار با داده، 'KMeans' برای خوشه‌بندی و 'matplotlib' برای ترسیم نمودار وارد می‌کنیم.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده با الگوهای گروهی (Cluster Patterns) هست.
# داده‌ها به صورت واضح در سه دسته مجزا گروه‌بندی شدن: نقاط کوچک (1-3)، متوسط (8-10) و بزرگ (15-17).
# این ساختار برای نمایش **الگوهای فشرده (Compressed Patterns)** یا خوشه‌های طبیعی در داده مناسبه.
data = {
    'Feature1': [1, 2, 3, 8, 9, 10, 15, 16, 17],
    'Feature2': [5, 6, 7, 12, 13, 14, 19, 20, 21]
}

df = pd.DataFrame(data)

# این بخش، الگوریتم **K-means** رو برای خوشه‌بندی داده‌ها اجرا می‌کنه.
# ما از `n_clusters=3` استفاده می‌کنیم چون به نظر می‌رسه داده‌ها به سه گروه طبیعی تقسیم شدن.
# این الگوریتم با کاهش تنوع درون خوشه‌ها و افزایش فاصله بین خوشه‌ها، الگوهای پنهان در داده رو استخراج می‌کنه.
# ستون جدید 'Cluster' نشان می‌ده هر نمونه به کدام خوشه تعلق داره.
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# این بخش، داده‌ها رو در فضای دو بعدی به همراه خوشه‌های شناسایی‌شده نمایش می‌ده.
# هر خوشه با یک رنگ متفاوت در پالت 'viridis' نمایش داده شده.
# حاشیه مشکی ('edgecolor=k') به وضوح جدایی خوشه‌ها رو نشون می‌ده.
# این نمودار به ما کمک می‌کنه ببینیم چطور K-means الگوهای پراکنده رو به سه الگوی فشرده و معنادار تبدیل کرده.
plt.scatter(df['Feature1'], df['Feature2'], c=df['Cluster'], cmap='viridis', edgecolor='k', s=40)
plt.title('Pattern Clustering: Compressed Patterns')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

در این مثال، یک مجموعه داده مصنوعی حاوی الگوهای دوویژگیه ایجاد شده است. سپس از الگوریتم <b>K-means</b> برای شناسایی سه خوشه استفاده شده تا الگوهای مشابه در گروه‌هایی همراه قرار گیرند. هر خوشه نمایانگر یک دسته الگوی مرتبط است و مرکز خوشه می‌تواند به عنوان نماینده‌ای فشرده از آن دسته در نظر گرفته شود. نتایج به‌صورت بصری نمایش داده شده‌اند تا درک بهتری از ساختار داده و نحوه گروه‌بندی الگوها فراهم شود. این رویکرد نشان می‌دهد که چگونه خوشه‌بندی الگوها می‌تواند به ساده‌سازی خروجی‌های پیچیده داده‌کاوی و استخراج دانش قابل تفسیر کمک کند.

### 5.2.2. Extracting redundancy-aware top-k patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">استخراج الگوهای Top-K با در نظر گرفتن همپوشانی</span> 🎯
</p>
<p style="font-size:16pt; text-align:right;">
استخراج <b>الگوهای Top-K با در نظر گرفتن همپوشانی</b> (Redundancy-aware Top-k Pattern Mining)، به شناسایی و رتبه‌بندی الگوهای معنادار در یک مجموعه داده می‌پردازد، با این شرط که از تکرار و همپوشانی بین الگوها جلوگیری شود. در داده‌کاوی، اغلب صدها یا هزاران الگوی پرتکرار یافت می‌شوند که بسیاری از آن‌ها از نظر محتوا مشابه یا زیرمجموعه یکدیگر هستند (مثلاً <i>«نان → کره»</i> و <i>«نان، روغن → کره»</i>). این روش با ارزیابی <b>اهمیت</b> (مانند پشتیبانی، لیفت) و <b>تنوع</b> الگوها، مجموعه‌ای فشرده و متنوع از k الگوی برتر را ارائه می‌دهد که هم معنادار باشند و هم اطلاعات تکراری نداشته باشند. این رویکرد در خلاصه‌سازی داده، گزارش‌دهی و تصمیم‌گیری هوشمند بسیار مؤثر است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. کاوش آگاه از همپوشانی (Redundancy-Aware Mining):</b>  
این تکنیک‌ها هنگام استخراج الگوها، وجود الگوهای تکراری یا همپوشان را در نظر می‌گیرند. همپوشانی زمانی رخ می‌دهد که چندین الگو اطلاعات مشابهی ارائه دهند (مثلاً یک الگو افزونه کوچکی نسبت به الگوی دیگر داشته باشد). هدف، تولید مجموعه‌ای از الگوها است که هم معنادار باشند و هم از نظر اطلاعاتی <b>متنوع</b> و <b>غیرتکراری</b> باشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. الگوهای برتر-k (Top-k Patterns):</b>  
این مفهوم به معنای رتبه‌بندی الگوها بر اساس معیاری مشخص (مانند <b>پشتیبانی</b>، <b>لیفت</b> یا <b>بهبود</b>) و استخراج <b>k</b> مورد از مهم‌ترین الگوهاست. با ترکیب این روش با حذف همپوشانی، می‌توان به جای دریافت لیستی طولانی و پر از تکرار، تنها <b>k الگوی برتر و مجزا</b> را ارائه داد که برای تحلیل سریع و گزارش‌دهی ایده‌آل هستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کاربردها:</b>  
استخراج الگوهای برتر-k بدون همپوشانی در حوزه‌هایی که نیاز به بینش‌های مختصر و مؤثر است، بسیار ارزشمند است. از جمله: <b>بازاریابی</b> (شناسایی بهترین ترکیب محصولات برای تبلیغ)، <b>سلامت</b> (گزارش الگوهای تشخیصی مهم بدون تکرار)، و <b>تشخیص کلاهبرداری</b> (یافتن الگوهای غیرمعمول منحصربه‌فرد). این روش به تبدیل حجم زیادی از خروجی به دانش قابل اقدام کمک می‌کند.
</p>

</div>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای کشف الگوهای خرید و قوانین انجمنی استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# این بخش، **الگوهای برتر-k با آگاهی از افزونگی (Redundancy-Aware Top-k Patterns)** رو استخراج می‌کنه.
# در عمل، بسیاری از قوانین انجمنی معنای مشابهی دارن (مثلاً A→B و A,B→C) و می‌تونن افزونه باشن.
# با مرتب‌سازی قوانین بر اساس 'confidence' و انتخاب فقط دو مورد برتر، یک رویکرد ساده برای کاهش افزونگی و تمرکز روی قوانین قوی‌تر اعمال شده.
# در سیستم‌های پیشرفته‌تر، از معیارهایی مثل "پوشش" (coverage) یا "استقلال" (independence) برای حذف قوانین افزونه استفاده می‌شه.
top_k_rules = rules.sort_values(by='confidence', ascending=False).head(2)  # استخراج ۲ قانون برتر

# در نهایت، قوانین برتر-k که از افزونگی کمتری برخوردارن رو چاپ می‌کنیم.
# این خروجی برای تصمیم‌گیری‌های عملیاتی (مثل پیشنهاد محصول) مناسب‌تره چون خلاصه و بدون تکراره.
print("Redundancy-Aware Top-k Patterns:")
print(top_k_rules[['antecedents', 'consequents', 'support', 'confidence']])
```

<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده شامل تراکنش‌هایی با آیتم‌ها است. ابتدا از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی استفاده شده است. سپس، قواعد بر اساس <b>اطمینان</b> (یا لیفت) مرتب شده‌اند و <b>k</b> قاعده برتر انتخاب شده‌اند. برای کاهش همپوشانی، قواعدی که به‌طور معناداری مشابه هستند (مثلاً قواعدی که فقط در یک آیتم اضافه متفاوتند و اطلاعات جدیدی ارائه نمی‌دهند) حذف یا فیلتر شده‌اند. این رویکرد مجموعه‌ای مختصر، معنادار و بدون تکرار از قواعد را تولید می‌کند که برای تحلیل و تصمیم‌گیری عملیاتی بسیار مناسب است.
</p>

### 5.3.1. Pruning pattern space with pattern pruning constraints

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاهش فضای الگو با محدودیت‌های هرس کردن الگو</span> ✂️
</p>
<p style="font-size:16pt; text-align:right;">
هرس کردن فضای الگو با استفاده از <b>محدودیت‌های هرس الگو</b> (Pattern Pruning Constraints)، به معنای اعمال شرایط خاص برای کاهش فضای جست‌وجو و تمرکز بر استخراج الگوهایی است که معیارهای از پیش تعیین‌شده را برآورده می‌کنند. این تکنیک فرآیند کاوش الگو را سریع‌تر و هدفمندتر می‌کند، زیرا از پیش الگوهای نامرتبط یا بی‌اهمیت را از فرآیند حذف می‌کند. با استفاده از این محدودیت‌ها، الگوریتم‌ها می‌توانند منابع محاسباتی خود را صرف کشف الگوهای واقعاً جالب کنند، نه تمام ترکیبات ممکن.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. محدودیت‌های هرس الگو (Pattern Pruning Constraints):</b>  
این محدودیت‌ها قوانین یا شرایطی هستند که در حین کاوش الگو اعمال می‌شوند تا الگوهایی که معیارهای مشخصی (مانند حداقل پشتیبانی، حداکثر طول، یا حضور/عدم حضور آیتم خاص) را ندارند، حذف شوند. این کار به تمرکز تحلیل بر الگوهایی مرتبط با اهداف داده‌شده (مثلاً کشف ترکیبات گران‌قیمت یا محصولات فصلی) کمک می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. بهبود کارایی (Efficiency Improvement):</b>  
کاهش فضای الگو باعث بهبود چشمگیر عملکرد الگوریتم‌های کاوش می‌شود، زیرا از تولید و بررسی الگوهای نامرتبط جلوگیری می‌کند. هرس زودهنگام الگوها بار محاسباتی را کاهش داده و زمان و حافظه مصرفی را بهینه می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کاربردها:</b>  
این روش در سناریوهای متعددی کاربرد دارد، مانند <b>تحلیل سبد خرید</b> (مثلاً فقط الگوهایی با پشتیبانی بالاتر از ۰.۱ و اطمینان بیش از ۰.۷)، <b>بیوانفورماتیک</b> (فیلتر کردن توالی‌های کوتاه یا غیرمشابه)، یا <b>سیستم‌های توصیه‌گر</b> (تمرکز بر محصولات موجود در انبار). این محدودیت‌ها تحلیل را از یک فرآیند عمومی به یک رویکرد هدفمند تبدیل می‌کنند.
</p>

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از هرس فضای الگو در پایتون</span> 🐍

<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده خرده‌فروشی استفاده کنیم و با اعمال محدودیت‌های هرس (مانند حداقل پشتیبانی ۰.۲ و حداقل اطمینان ۰.۶)، تنها الگوهای معنادار را استخراج کنیم:
</p>

</div>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای کشف الگوهای خرید و قوانین انجمنی استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# این بخش، **حذف الگوهای نامطلوب با استفاده از محدودیت‌های پرونینگ (Pattern Pruning Constraints)** رو انجام می‌ده.
# پس از تولید تمام قوانین، ما فقط قوانینی رو نگه می‌داریم که مقدار 'lift' بیشتر از ۱.۲ داشته باشن.
# 'lift > 1' نشان‌دهنده وابستگی مثبت بین آیتم‌هاست، و 'lift > 1.2' یعنی این وابستگی به اندازه‌ای قوی‌تر از حد معمول هست.
# این فیلتر کمک می‌کنه تا قوانین نامرتبط یا ضعیف حذف بشن و فقط الگوهای معنادار و قوی باقی بمونن.
pruned_rules = rules[rules['lift'] > 1.2]

# در نهایت، قوانین پردازش‌شده و پالایش‌شده (Pruned Patterns) رو چاپ می‌کنیم.
# این خروجی برای تصمیم‌گیری‌های عملیاتی مثل پیشنهاد محصول، بهینه‌سازی قفسه یا تحلیل رفتار مشتری مناسب‌تره.
print("Pruned Patterns:")
print(pruned_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از یک مجموعه داده متشکل از تراکنش‌های حاوی آیتم‌ها استفاده شده است. ابتدا الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار به کار گرفته شده و سپس قواعد ارتباطی تولید شده‌اند. در مرحله بعد، <b>محدودیت هرس الگو</b> اعمال شده است: تنها قواعدی انتخاب شده‌اند که <b>لیفت</b> آن‌ها از یک مقدار حداقل (مثلاً 1.2) بیشتر باشد. این محدودیت باعث حذف قواعدی با ارتباط ضعیف یا بدیهی می‌شود و فضای الگو را به قواعدی با ارتباط معنادار و غیرمستقل محدود می‌کند. این رویکرد فرآیند کاوش را کارآمدتر می‌کند و خروجی را به الگوهای واقعاً جالب و قابل اقدام تبدیل می‌کند.
</p>

### 5.3.2. Pruning data space with data pruning constraints

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاهش فضای داده با محدودیت‌های هرس داده</span> ✂️
</p>
<p style="font-size:16pt; text-align:right;">
هرس فضای داده با استفاده از <b>محدودیت‌های هرس داده</b> (Data Pruning Constraints)، به معنای اعمال معیارهای خاص برای فیلتر کردن و کاهش مجموعه داده قبل از آغاز فرآیند داده‌کاوی است. با حذف هوشمندانه داده‌های نامرتبط، پرت، یا کم‌اطلاعات، این تکنیک کارایی و اثربخشی الگوریتم‌های داده‌کاوی را بهبود می‌بخشد و اجازه می‌دهد تمرکز بر روی اطلاعات بامعنی و مرتبط باشد. این مرحله یکی از مهم‌ترین بخش‌های <b>پیش‌پردازش داده</b> محسوب می‌شود و عملکرد کلی سیستم‌های یادگیری ماشین و کاوش الگو را به‌طور چشمگیری تحت تأثیر قرار می‌دهد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. محدودیت‌های هرس داده (Data Pruning Constraints):</b>  
این محدودیت‌ها شرایط یا قوانینی هستند که برای حذف انتخابی نمونه‌ها (ردیف‌ها) یا ویژگی‌ها (ستون‌ها) از مجموعه داده اعمال می‌شوند. مثلاً حذف ویژگی‌هایی با واریانس بسیار پایین، یا نمونه‌هایی که خارج از بازه معنادار هستند (پرت‌ها). هدف، بهبود کیفیت داده و حذف نویز یا داده‌های گمراه‌کننده قبل از تحلیل است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. بهبود کارایی (Efficiency Improvement):</b>  
با کاهش حجم داده در مراحل اولیه، زمان پردازش، مصرف حافظه و پیچیدگی محاسباتی کاهش می‌یابد. همچنین، الگوریتم‌های داده‌کاوی می‌توانند روی داده‌های تمیز‌تر و مرتبط‌تر تمرکز کنند که منجر به نتایج دقیق‌تر و قابل تفسیرتر می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کاربردها:</b>  
هرس داده در بسیاری از سناریوها کاربرد دارد، از جمله: <b>پیش‌پردازش برای یادگیری ماشین</b> (حذف ویژگی‌های بی‌ربط)، <b>داده‌کاوی متن</b> (حذف کلمات توقف مانند «و»، «در»)، <b>پردازش سیگنال</b> (حذف فرکانس‌های نامرتبط)، و <b>پایگاه‌های داده بزرگ</b> (فیلتر داده‌های قدیمی یا ناقص). این تکنیک اساس یک تحلیل هوشمند و کارآمد است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از هرس فضای داده در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده با ویژگی‌های نامرتبط و نمونه‌های پرت استفاده کنیم و با اعمال محدودیت‌های هرس داده (مانند حذف ستون‌های کم‌واریانس و نمونه‌های دورافتاده)، داده را قبل از کاوش تمیز و بهینه کنیم:
</p>

</div>

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# این قسمت اول کده. ما کتابخونه‌های 'pandas' برای کار با داده، 'load_iris' برای بارگذاری مجموعه داده گل‌های ایریس و 'train_test_split' برای تقسیم داده وارد می‌کنیم.

# این بخش، مجموعه داده معروف **Iris** رو بارگیری می‌کنه.
# این دیتاست شامل ۱۵۰ نمونه از گل‌ها با چهار ویژگی (مثل طول و عرض گلبرگ) و سه کلاس مختلف (نوع گل) هست.
# این داده‌ها برای نمایش تکنیک‌های پردازش و پالایش داده ایده‌آل هستن.
iris = load_iris()
X, y = iris.data, iris.target

# این بخش، داده‌های ویژگی و برچسب رو به یک DataFrame از pandas تبدیل می‌کنه.
# اضافه کردن ستون 'target' به دیتافریم، کار تحلیل و فیلتر کردن رو راحت‌تر می‌کنه.
# این ساختار برای نمایش عملیات روی داده‌های برچسب‌دار مناسب‌تر از آرایه‌های خام numpy هست.
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y

# این بخش، **پالایش داده (Data Pruning)** رو با اعمال یک محدودیت خاص انجام می‌ده.
# ما تمام نمونه‌هایی که مربوط به کلاس هدف `1` (مثلاً نوع خاصی از گل) هستن رو حذف می‌کنیم.
# این نوع پالایش ممکنه در شرایطی لازم باشه که:
# - یک کلاس نویز داره،
# - مدل فقط باید روی دو کلاس تمرکز کنه،
# - یا داده‌های یک کلاس نماینده نیستن.
pruned_df = df[df['target'] != 1]

# در نهایت، داده‌های پالایش‌شده رو چاپ می‌کنیم تا ببینیم چه تغییری ایجاد شده.
# فقط چند سطر اول با `head()` نمایش داده می‌شه تا ساختار خروجی رو ببینیم.
# این نوع پالایش باعث کاهش حجم داده و تمرکز بر زیرمجموعه مورد نظر می‌شه.
print("Pruned Dataset:")
print(pruned_df.head())
```

<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده <b>Iris</b> بارگذاری شده و به یک <b>DataFrame</b> تبدیل شده است. سپس یک <b>محدودیت هرس داده</b> اعمال شده است: تمام نمونه‌های مربوط به یک کلاس خاص (مثلاً کلاس "setosa") حذف شده‌اند. این کار ممکن است برای تمرکز تحلیل بر دو گونه دیگر گل یا شبیه‌سازی شرایط عدم تعادل داده (imbalanced data) انجام شود. داده‌های فیلترشده سبک‌تر و مرتبط‌تر هستند و می‌توانند برای مراحل بعدی داده‌کاوی، مانند طبقه‌بندی یا خوشه‌بندی، استفاده شوند. این رویکرد نشان می‌دهد که چگونه هرس داده می‌تواند تحلیل را هدفمند و کارآمدتر کند.
</p>

### 5.3.3. Mining space pruning with succinctness constraints

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش با محدودیت‌های فشردگی برای هرس فضای جست‌وجو</span> 🎯
</p>
<p style="font-size:16pt; text-align:right;">
هرس فضای کاوش با استفاده از <b>محدودیت‌های فشردگی</b> (Succinctness Constraints)، به معنای اعمال محدودیت‌هایی برای کنترل اندازه یا پیچیدگی فضای الگوهای جست‌وجو شده در حین داده‌کاوی است. هدف، استخراج الگوهایی <b>کوتاه، خلاصه و معنادار</b> است که اطلاعات ضروری داده را دربرگیرند، بدون آن‌که فضای جست‌وجو به‌طور نامحدود گسترش یابد. این رویکرد کارایی محاسباتی را بهبود می‌بخشد و نتایج را قابل‌درک‌تر و کاربردی‌تر می‌کند، زیرا الگوهای طولانی یا پیچیده اغلب برای انسان سخت‌تر قابل تفسیر هستند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. محدودیت‌های هرس فضا (Space Pruning Constraints):</b>  
این محدودیت‌ها قوانینی هستند که در حین کاوش الگو اعمال می‌شوند تا اندازه یا پیچیدگی فضای جست‌وجو را محدود کنند. مثلاً فقط الگوهایی با حداکثر طول ۲ آیتم (مانند <i>«A → B»</i>) مجاز باشند و الگوهای طولانی‌تر نادیده گرفته شوند. این کار محاسبات را ساده‌تر و نتایج را قابل‌فهمن‌تر می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. فشردگی (Succinctness):</b>  
فشردگی به کوتاهی و فشرده‌بودن الگوها اشاره دارد. کاوش با محدودیت فشردگی به دنبال استخراج الگوهایی است که با کمترین تعداد آیتم یا شرط، بیشترین اطلاعات معنادار را منتقل کنند. این الگوها برای گزارش‌دهی و تصمیم‌گیری سریع بسیار مناسب‌تر هستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. کارایی و قابلیت تفسیر:</b>  
با هرس فضای الگو با این محدودیت‌ها، الگوریتم‌ها تنها روی الگوهای مرتبط تمرکز می‌کنند که هم محاسبات را کارآمد می‌کند و هم خروجی را برای کاربران غیرفنی قابل درک می‌سازد. الگوهای فشرده به‌راحتی قابل تفسیر، تحلیل و اقدام هستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این روش در حوزه‌هایی که نیاز به خروجی سریع و ساده است کاربرد گسترده دارد، مانند <b>تحلیل سبد خرید</b> (استخراج قواعد کوتاه برای پیشنهاد فوری محصول)، <b>داده‌کاوی متن</b> (یافتن کلمات کلیدی یا عبارات کوتاه نمایانگر موضوع)، و <b>سیستم‌های توصیه‌گر</b> (تولید پیشنهادات مختصر و موثر). این تکنیک به تبدیل داده به دانش عملیاتی کمک می‌کند.
</p>

</div>

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از هرس فضای کاوش با محدودیت فشردگی در پایتون</span> 🐍

بیایید از یک مجموعه داده خرده‌فروشی استفاده کنیم و با اعمال محدودیت فشردگی (مانند حداکثر طول ۲ آیتم در هر قاعده)، تنها الگوهای کوتاه و معنادار را استخراج کنیم:
</p>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای کشف الگوهای خرید و قوانین انجمنی استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# این بخش، **الگوهای موجز (Succinct Patterns)** رو با اعمال محدودیت‌های ساختاری استخراج می‌کنه.
# "موجزیت" (Succinctness) یعنی قوانینی که شرایط خاصی در ساختارشون دارن، مثل تعداد محدود آیتم در بخش پیش‌شرط (antecedent).
# اینجا فقط قوانینی رو نگه می‌داریم که حداکثر ۲ آیتم در سمت چپ قانون (antecedent) داشته باشن.
# این محدودیت باعث می‌شه قوانین خروجی ساده‌تر، قابل تفسیرتر و عملیاتی‌تر باشن — مخصوصاً برای کاربران غیرفنی.
succinct_rules = rules[rules['antecedents'].apply(len) <= 2]

# در نهایت، قوانین موجز و معنادار رو چاپ می‌کنیم.
# این نوع خروجی برای سیستم‌های پیشنهادگر، گزارش‌های مدیریتی و تصمیم‌گیری‌های تجاری بسیار مناسب‌تر از لیست طولانی قوانین پیچیده‌ست.
print("Succinct Association Rules:")
print(succinct_rules[['antecedents', 'consequents', 'support', 'confidence']])
```

<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده شامل تراکنش‌هایی با آیتم‌ها است. ابتدا از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی استفاده شده است. سپس، <b>محدودیت فشردگی</b> اعمال شده است: تنها قواعدی انتخاب شده‌اند که حداکثر دو آیتم در بخش پیش‌رو (antecedent) داشته باشند. این محدودیت از تولید قواعد بسیار طولانی و پیچیده (مانند «A و B و C و D → E») جلوگیری می‌کند و خروجی را به قواعد کوتاه، خوانا و قابل اقدام محدود می‌کند. این رویکرد به‌ویژه در کاربردهای عملیاتی مانند پیشنهاد محصول در لحظه یا گزارش مدیریتی سریع بسیار مؤثر است.

### 5.4.1. Sequential pattern mining: concepts and primitives

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش الگوهای ترتیبی</span> ⏳
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>الگوهای ترتیبی</b> (Sequential Pattern Mining) به کشف الگوهایی می‌پردازد که <b>ترتیب زمانی</b> وقایع یا آیتم‌ها در یک دنباله را نشان می‌دهند. این تکنیک برای آشکارسازی ساختارهای پنهان در داده‌های ترتیبی حیاتی است و در حوزه‌های متعددی مانند <b>تحلیل سری‌های زمانی</b>، <b>جستارهای کلیک کاربران در وب</b> و <b>ردیابی رفتار مشتریان</b> کاربرد دارد. برخلاف کاوش ارتباطات معمولی که تنها هم‌وقوعی را بررسی می‌کند، این روش به دنبال الگوهایی است مانند <i>«بعد از خرید لپ‌تاپ، خرید کیف حمل اتفاق می‌افتد»</i> یا <i>«کاربران پس از دیدن صفحه محصول، به صفحه تخفیف می‌روند»</i>.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. الگوهای ترتیبی (Sequential Patterns):</b>  
این الگوها ترتیب وقوع وقایع یا آیتم‌ها را در طول زمان نشان می‌دهند. برای مثال، دنباله <i>«ورود → صفحه محصول → سبد خرید → پرداخت»</i> یک الگوی ترتیبی رایج در وب‌سایت‌های فروشگاهی است. این الگوها وابستگی‌ها و روابط زمانی بین مراحل مختلف یک فرآیند را مدل می‌کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. پشتیبانی (Support):</b>  
در کاوش الگوهای ترتیبی، پشتیبانی به تعداد دنباله‌هایی گفته می‌شود که یک الگوی ترتیبی خاص را شامل می‌شوند. پشتیبانی بالا نشان‌دهنده رایج‌بودن یک الگو در داده‌هاست و آن را به عنوان یک الگوی معنادار شناسایی می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. اجزای اولیه (Primitives):</b>  
اجزای اولیه شامل وقایع (events)، آیتم‌ها (items) و فواصل زمانی است. عملیات پایه شامل تعریف دنباله‌ها، تعیین محدودیت‌های زمانی (مثلاً «حداکثر ۷ روز بین دو رویداد») و شناسایی زیردنباله‌های پرتکرار است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این روش در کاربردهای متنوعی استفاده می‌شود، از جمله: <b>تحلیل رفتار کاربران در وب</b> (بهبود طراحی سایت)، <b>پیش‌بینی روندهای بازار سهام</b> (با تحلیل توالی معاملات)، <b>پزشکی</b> (درک پیشرفت علائم بیماری)، و <b>سامانه‌های پیشنهادگر</b> (پیشنهاد مرحله بعدی بر اساس تاریخچه کاربر).
</p>

```python
from prefixspan import PrefixSpan

# این قسمت اول کده. ما کتابخونه 'prefixspan' رو وارد می‌کنیم که برای کشف الگوهای پی‌دنبالی (Sequential Pattern Mining) استفاده می‌شه.

# این لیست حاوی داده‌های شبیه‌سازی‌شده از توالی‌های گسسته هست.
# هر توالی نشان‌دهنده‌ی دنباله‌ای از رویدادها یا آیتم‌ها در یک زمینه خاصه (مثلاً صفحات بازدید شده توسط یک کاربر، مراحل خرید، یا دستورات کاربر).
# این نوع داده‌ها برای تحلیل رفتار ترتیبی (مثل مسیر خرید یا حرکت کاربر در وب‌سایت) مناسب‌اند.
sequences = [
    [1, 2, 3, 4],
    [2, 3, 5],
    [1, 3, 4],
    [2, 5],
    [1, 2, 3]
]

# این بخش، الگوریتم **PrefixSpan** رو برای کشف الگوهای پی‌دنبالی پرتکرار اجرا می‌کنه.
# PrefixSpan یک الگوریتم بازگشتی و بدون تولید کاندیده هست که با تقسیم مسئله به زیرمسائل (پروجکشن) به صورت کارآمد الگوها رو استخراج می‌کنه.
# `min_support=2` یعنی فقط توالی‌هایی رو می‌خوایم که حداقل در دو توالی ورودی ظاهر شدن.
# این الگوریتم تمام زیرتوالی‌های پرتکرار رو با حفظ ترتیب زمانی پیدا می‌کنه.
min_support = 2
patterns = PrefixSpan(sequences).frequent(min_support)

# در نهایت، تمام الگوهای پی‌دنبالی پرتکرار رو چاپ می‌کنیم.
# هر الگو یک توالی از آیتم‌هاست که به صورت مکرر در داده‌ها دیده شده.
# این اطلاعات می‌تونن برای پیش‌بینی حرکت بعدی کاربر، طراحی مسیرهای خرید بهینه یا شخصی‌سازی تجربه کاربری استفاده بشن.
print("Frequent Sequential Patterns:")
for pattern in patterns:
    print(pattern)
```

### 5.4.2. Scalable methods for mining sequential patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:16pt; text-align:right;">
روش‌های <b>مقیاس‌پذیر</b> برای کاوش الگوهای ترتیبی به حل چالش استخراج کارآمد الگوها از مجموعه‌داده‌های بزرگ می‌پردازند. با افزایش حجم داده، الگوریتم‌های سنتی مانند GSP یا PrefixSpan ممکن است از نظر زمان و حافظه بسیار پرهزینه شوند. روش‌های مقیاس‌پذیر با استفاده از تکنیک‌های پیشرفته، عملکرد کاوش را بهبود بخشیده و امکان تحلیل داده‌های ترتیبی در مقیاس بسیار بزرگ را فراهم می‌کنند.

<p style="font-size:16pt; text-align:right;">

در این روش، داده‌ها به‌جای فرمت سنتی افقی (هر ردیف یک دنباله)، به صورت عمودی سازماندهی می‌شوند؛ یعنی برای هر آیتم، لیستی از دنباله‌ها و موقعیت‌های آن آیتم نگهداری می‌شود. این ساختار دسترسی سریع به وقوع آیتم‌ها را فراهم کرده و کارایی محاسبه پشتیبانی را بهبود می‌بخشد، به‌ویژه در داده‌های بزرگ.

<p style="font-size:16pt; text-align:right;">
<b>2. تکنیک‌های مبتنی بر پروجکشن (Projection-Based Techniques):</b>  
این روش‌ها با ایجاد <b>پروجکشن</b> (تصویر) از پایگاه داده بر اساس یک الگوی پایه، فضای جست‌وجو را محدود می‌کنند. مثلاً پس از یافتن الگوی <i>['A']</i>، فقط زیردنباله‌هایی که بعد از 'A' شروع می‌شوند، برای یافتن الگوهای طولانی‌تر استفاده می‌شوند. این کار باعث کاهش حجم داده در هر مرحله و بهبود مقیاس‌پذیری می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. موازی‌سازی (Parallelization):</b>  
توزیع بار محاسباتی کاوش الگو بین چندین پردازنده یا ماشین (مثلاً با استفاده از Spark یا Dask) امکان پردازش همزمان دنباله‌ها یا بخش‌های مختلف داده را فراهم می‌کند. این روش برای تحلیل داده‌های بسیار حجیم — مانند لاگ‌های وب یک پلتفرم جهانی — ضروری است.
</p>

<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این روش‌های مقیاس‌پذیر در حوزه‌هایی با داده‌های عظیم کاربرد دارند، از جمله: <b>تحلیل جریان کلیک کاربران در وب‌سایت‌های بزرگ</b>، <b>کشف الگوهای پزشکی در سوابق درمانی میلیونی بیمار</b>، و <b>درک مسیر خرید مشتریان در پلتفرم‌های تجارت الکترونیک بین‌المللی</b>.

```python
from pymining import seqmining


# داده‌های شبیه‌سازی‌شده: لیست توالی‌ها
# هر توالی یک دنباله از آیتم‌هاست که مثلاً می‌تونه مسیر کاربر، مراحل خرید، یا هر نوع رویداد متوالی باشه
large_sequences = [
    [1, 2, 3, 4, 5],
    [2, 3, 5],
    [1, 3, 4],
    [2, 5],
    [1, 2, 3, 6, 7, 8, 9, 10],
]


# تعیین حداقل support
# min_support = تعداد حداقل توالی‌هایی که الگو باید در آن‌ها ظاهر شود
# یعنی هر الگو باید حداقل در ۲ توالی دیده شود تا پرتکرار محسوب شود
min_support = 2


# استخراج الگوهای پرتکرار با استفاده از الگوریتم BIDE
# freq_seq_enum الگوهای پرتکرار و ماکسیمال را برمی‌گرداند
# توجه: خروجی به شکل set از tupleهاست (مثلاً (1,2,3))
patterns = seqmining.freq_seq_enum(large_sequences, min_support)


# تابع کمکی برای بررسی زیرتوالی
# این تابع بررسی می‌کنه که آیا یک الگو (subseq) واقعاً در یک توالی (seq) ظاهر شده یا نه
# اینجا از iterator استفاده شده تا ترتیب حفظ بشه و فقط توالی‌های متوالی معتبر شمرده شوند
def is_subsequence(subseq, seq):
    """بررسی می‌کند آیا subseq زیرتوالی seq هست یا نه"""
    it = iter(seq)
    return all(item in it for item in subseq)


# نمایش نتایج با محاسبه support واقعی هر الگو
# حلقه روی تمام الگوهای استخراج‌شده
# برای هر الگو، تعداد توالی‌هایی که شامل آن هستند شمارش می‌شه (support واقعی)
print("Frequent Sequential Patterns (Scalable, Maximal):")
for pattern in patterns:
    # محاسبه support واقعی: تعداد توالی‌هایی که شامل pattern هستند
    support = sum(1 for seq in large_sequences if is_subsequence(pattern, seq))
    
    # چاپ الگو و تعداد تکرار آن
    print(f"Pattern: {pattern}, Support: {support}")
```

### 5.4.3. Constraint-based mining of sequential patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش مبتنی بر محدودیت الگوهای ترتیبی</span> 🔐
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>مبتنی بر محدودیت</b> الگوهای ترتیبی به معنای گنجاندن شرایط یا قیدهای خاص در حین فرآیند کاوش است. این محدودیت‌ها الگوریتم را راهنمایی می‌کنند تا فقط الگوهایی را کشف کند که معیارهای از پیش تعیین‌شده‌ای را برآورده می‌کنند. این رویکرد فرآیند کاوش را <b>هدفمندتر</b> می‌کند و آن را با دانش حوزه‌ای کاربر یا اهداف تحلیلی خاص هم‌راستا می‌سازد. به جای کشف تمام الگوهای پرتکرار، تمرکز بر الگوهایی با اهمیت عملیاتی یا بالینی انجام می‌شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. محدودیت‌ها (Constraints):</b>  
محدودیت‌ها شرایطی هستند که در حین کاوش الگو اعمال می‌شوند و دامنه الگوهای بررسی‌شده را محدود می‌کنند. این محدودیت‌ها می‌توانند بر اساس <b>ویژگی آیتم‌ها</b> (مثلاً فقط محصولات گران‌قیمت)، <b>روابط زمانی</b> (مثلاً "بین دو رویداد حداکثر ۲ روز فاصله باشد") یا <b>ساختار دنباله</b> (مثلاً "الگو باید با رویداد X شروع شود") تعریف شوند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. سفارشی‌سازی (Customization):</b>  
کاوش مبتنی بر محدودیت به کاربر این امکان را می‌دهد که فرآیند کاوش را متناسب با نیازهای خاص خود تنظیم کند. مثلاً یک تحلیلگر مالی می‌تواند محدودیتی تعریف کند که فقط توالی‌هایی شامل "برداشت بزرگ → انتقال به حساب خارجی" را بررسی کند. این کار خروجی را به الگوهای مرتبط محدود کرده و از اتلاف زمان جلوگیری می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. قابلیت تفسیر (Interpretability):</b>  
استفاده از محدودیت‌ها، نتایج کاوش را قابل‌درک‌تر می‌کند. با گنجاندن دانش حوزه‌ای (مانند قوانین پزشکی یا قوانین تجاری) در خود فرآیند کاوش، الگوهای کشف‌شده به طبیعت بیشتری با واقعیت هم‌خوانی دارند و برای تصمیم‌گیرندگان معنادارتر هستند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این روش در حوزه‌هایی با نیاز به تحلیل هدفمند کاربرد گسترده دارد، از جمله: <b>سلامت</b> (شناسایی توالی‌های درمان خاص برای بیماران)، <b>تولید</b> (بهینه‌سازی توالی عملیات در خط تولید)، و <b>مالی</b> (تشخیص الگوهای کلاهبرداری با ویژگی‌های مشخص).
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش مبتنی بر محدودیت الگوهای ترتیبی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده ترتیبی استفاده کنیم و با اعمال محدودیت‌های خاص (مانند حضور یک آیتم خاص یا فاصله زمانی بین رویدادها)، تنها الگوهایی را استخراج کنیم که شرایط تعریف‌شده را برآورده کنند:
</p>

```python
from prefixspan import PrefixSpan

# این قسمت اول کده. ما کتابخونه 'prefixspan' رو وارد می‌کنیم که برای کشف الگوهای پی‌دنبالی (Sequential Pattern Mining) استفاده می‌شه.

# این لیست حاوی داده‌های شبیه‌سازی‌شده از توالی‌های گسسته هست.
# هر توالی نشان‌دهنده‌ی دنباله‌ای از رویدادها یا آیتم‌ها در یک زمینه خاصه (مثلاً صفحات بازدید شده توسط یک کاربر، مراحل خرید، یا دستورات کاربر).
# این نوع داده‌ها برای تحلیل رفتار ترتیبی (مثل مسیر خرید یا حرکت کاربر در وب‌سایت) مناسب‌اند.
sequences = [
    [1, 2, 3, 4],
    [2, 3, 5],
    [1, 3, 4],
    [2, 5],
    [1, 2, 3]
]

# این بخش، الگوریتم **PrefixSpan** رو برای کشف الگوهای پی‌دنبالی پرتکرار اجرا می‌کنه.
# PrefixSpan یک الگوریتم بازگشتی و بدون تولید کاندیده هست که با تقسیم مسئله به زیرمسائل (پروجکشن) به صورت کارآمد الگوها رو استخراج می‌کنه.
# `min_support=2` یعنی فقط توالی‌هایی رو می‌خوایم که حداقل در دو توالی ورودی ظاهر شدن.
# این الگوریتم تمام زیرتوالی‌های پرتکرار رو با حفظ ترتیب زمانی پیدا می‌کنه.
min_support = 2
patterns = PrefixSpan(sequences).frequent(min_support)

# این بخش، **الگوهای توالی‌دار محدودشده (Constrained Sequential Patterns)** رو فیلتر می‌کنه.
# در عمل، ممکنه فقط به الگوهایی علاقه داشته باشیم که شامل یک آیتم خاص (مثلاً '3') باشن.
# این محدودیت (constraint) به ما کمک می‌کنه روی زیرمجموعه معناداری از الگوها تمرکز کنیم، بدون اینکه نیاز باشه همه الگوها رو پردازش کنیم.
# در اینجا، با استفاده از یک فیلتر ساده، فقط الگوهایی رو نگه می‌داریم که شامل آیتم `3` هستن.
constraint_item = 3
constrained_patterns = [pattern for pattern in patterns if constraint_item in pattern]

# در نهایت، الگوهای پی‌دنبالی پرتکرار که شامل آیتم مورد نظر هستن رو چاپ می‌کنیم.
# این خروجی برای کاربردهایی مثل تمرکز بر محصولات خاص، تحلیل مسیر خرید مشتریان یک آیتم، یا شخصی‌سازی پیشنهادات بسیار مفیده.
print(f"Constrained Sequential Patterns (with item {constraint_item}):")
for pattern in constrained_patterns:
    print(pattern)
```

<p style="font-size:16pt; text-align:right;">
در این مثال، از الگوریتم <b>PrefixSpan</b> برای کاوش الگوهای ترتیبی در یک مجموعه داده از دنباله‌ها استفاده شده است. یک <b>محدودیت</b> خاص اعمال شده است: تنها الگوهایی گزارش می‌شوند که شامل یک آیتم مشخص (مثلاً «آیتم ۳») باشند. این محدودیت فرآیند کاوش را هدفمند می‌کند و از تولید الگوهای نامرتبط جلوگیری می‌کند. به عنوان مثال، در یک سیستم فروش، ممکن است بخواهیم فقط دنباله‌هایی را ببینیم که شامل خرید یک محصول خاص (مانند یک دستگاه گران‌قیمت) هستند تا رفتار مشتریان قبل و بعد از آن خرید را تحلیل کنیم. این رویکرد نشان می‌دهد که چگونه با استفاده از محدودیت‌ها می‌توان کاوش الگو را با اهداف تحلیلی خاص هم‌راستا کرد.

### 5.5.1. Methods for mining frequent subgraphs

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش زیرگراف‌های پرتکرار</span> 🧩
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>زیرگراف‌های پرتکرار</b> (Frequent Subgraph Mining)، به کشف الگوهایی می‌پردازد که به‌طور مکرر در یک مجموعه داده گرافی یا شبکه‌ای ظاهر می‌شوند. این فرآیند برای درک روابط ساختاری و نشانه‌های تکرارشونده در شبکه‌های پیچیده حیاتی است. روش‌های مختلفی برای شناسایی کارآمد این زیرگراف‌ها وجود دارند که به تحلیل داده‌های ساختار گرافی در حوزه‌های متنوعی — مانند زیست‌شناسی، شبکه‌های اجتماعی و امنیت سایبری — کمک می‌کنند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. زیرگراف‌های پرتکرار (Frequent Subgraphs):</b>  
این زیرگراف‌ها الگوهایی هستند که به‌طور مکرر در یک مجموعه گراف یا شبکه ظاهر می‌شوند. مثلاً یک ساختار مثلثی در شبکه اجتماعی یا یک حلقه خاص در مولکول. این الگوها می‌توانند نشانه‌های معناداری از رفتارهای پایه، ارتباطات قوی یا ساختارهای عملکردی در داده باشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. الگوریتم‌های کاوش گراف (Graph Mining Algorithms):</b>  
الگوریتم‌هایی مانند <b>gSpan</b>، <b>MoSS</b> یا <b>FSG</b> به‌طور خاص برای کاوش زیرگراف‌های پرتکرار طراحی شده‌اند. این الگوریتم‌ها با استفاده از تکنیک‌هایی مانند <b>همریختی زیرگراف</b> (Subgraph Isomorphism) و <b>پیمایش سیستماتیک</b>، به‌صورت کارآمد فضای زیرگراف‌های ممکن را جست‌وجو می‌کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. پشتیبانی (Support):</b>  
در این زمینه، پشتیبانی به تعداد گراف‌هایی گفته می‌شود که یک زیرگراف خاص در آن‌ها وجود دارد (نه تعداد وقوع در یک گراف). زیرگرافی با پشتیبانی بالا، نشان‌دهنده یک ساختار رایج و احتمالاً مهم در مجموعه داده است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این روش در حوزه‌های متعددی کاربرد دارد:  
- <b>زیست‌فناوری</b>: شناسایی زیرساختارهای پرتکرار در مولکول‌ها و پروتئین‌ها.  
- <b>شبکه‌های اجتماعی</b>: یافتن الگوهای تعاملی مانند مثلث‌های دوستی یا ساختارهای انجمنی.  
- <b>امنیت سایبری</b>: تشخیص الگوهای مشکوک در ترافیک شبکه یا لاگ‌های سیستم.  
- <b>داده‌کاوی وب</b>: کشف ساختارهای رایج در گراف‌های ارتباطی صفحات.
</p>

```python
import networkx as nx
from itertools import combinations

# داده‌ها: لیست گراف‌ها
graph_dataset = [
    [(1, 2), (2, 3), (3, 4)],              # گراف 1
    [(1, 2), (2, 3), (3, 4), (4, 5)],      # گراف 2
    [(1, 2), (2, 3), (3, 4), (4, 6)],      # گراف 3
]


# تبدیل داده‌ها به NetworkX Graph
nx_graphs = []
for edges in graph_dataset:
    G = nx.Graph()             # ساخت یک گراف خالی
    G.add_edges_from(edges)    # اضافه کردن یال‌ها به گراف
    nx_graphs.append(G)        # اضافه کردن گراف ساخته‌شده به لیست


# پارامتر حداقل پشتیبانی
min_support = 2               # فقط زیرگراف‌هایی که در حداقل 2 گراف ظاهر شوند معتبرند


# توابع کمکی
def is_subgraph(subgraph, G):
    """
    بررسی می‌کند که آیا زیرگراف subgraph، زیرمجموعه‌ای از گراف G هست یا خیر.
    از الگوریتم GraphMatcher در NetworkX استفاده می‌کنیم.
    """
    matcher = nx.algorithms.isomorphism.GraphMatcher(G, subgraph)
    return matcher.subgraph_is_isomorphic()   # True اگر subgraph در G وجود داشته باشد

def canonical_code(G):
    """
    تولید کد کاننیکال برای زیرگراف G.
    این کد به ما کمک می‌کند تا زیرگراف‌های تکراری را شناسایی و حذف کنیم.
    """
    edges = sorted([tuple(sorted(edge)) for edge in G.edges()])  # مرتب‌سازی یال‌ها
    return tuple(edges)


# الگوریتم DFS-like برای استخراج زیرگراف‌های پرتکرار
def gspan_like_mining(graphs, min_support):
    """
    شبیه‌سازی الگوریتم gSpan با روش رشد زیرگراف (DFS-like)
    1. شروع با تمام یال‌ها به عنوان زیرگراف‌های اولیه
    2. محاسبه support هر زیرگراف
    3. گسترش زیرگراف با اضافه کردن یال‌های جدید
    4. جلوگیری از بررسی زیرگراف‌های تکراری با استفاده از canonical_code
    """
    subgraph_pool = []       # لیست موقت برای ذخیره زیرگراف‌ها
    for G in graphs:
        for edge in G.edges():          # شروع با یال‌ها به عنوان زیرگراف‌های یک‌تایی
            H = nx.Graph()
            H.add_edge(*edge)
            subgraph_pool.append(H)    # اضافه کردن زیرگراف اولیه به pool

    visited_codes = set()     # برای جلوگیری از بررسی زیرگراف‌های تکراری
    frequent_subgraphs = []   # لیست نهایی زیرگراف‌های پرتکرار

    while subgraph_pool:
        candidate = subgraph_pool.pop(0)       # انتخاب زیرگراف بعدی برای بررسی
        code = canonical_code(candidate)      # تولید کد کاننیکال
        if code in visited_codes:             # اگر قبلاً بررسی شده بود، رد می‌کنیم
            continue
        visited_codes.add(code)               # ثبت کد کاننیکال جدید

        
        # محاسبه support
        support = sum(is_subgraph(candidate, G) for G in graphs)
        if support >= min_support:
            frequent_subgraphs.append((candidate, support))  # اضافه کردن زیرگراف معتبر

            
            # گسترش زیرگراف با اضافه کردن یال‌های جدید
            for G in graphs:
                nodes = list(G.nodes())
                for u, v in combinations(nodes, 2):
                    if not candidate.has_edge(u, v) and G.has_edge(u, v):
                        new_subgraph = candidate.copy()   # ایجاد یک کپی برای اضافه کردن یال جدید
                        new_subgraph.add_edge(u, v)
                        new_code = canonical_code(new_subgraph)
                        if new_code not in visited_codes:
                            subgraph_pool.append(new_subgraph)

    return frequent_subgraphs   # بازگداندن زیرگراف‌های پرتکرار


# استخراج و نمایش زیرگراف‌های پرتکرار ماکسیمال
frequent_subgraphs = gspan_like_mining(nx_graphs, min_support)

print("Maximal Frequent Subgraphs (DFS-like gSpan) with min_support=2:")
for i, (H, support) in enumerate(frequent_subgraphs, 1):
    print(f"Subgraph {i}: Edges = {list(H.edges())}, Support = {support}")
```

### 5.5.2. Mining variant and constrained substructure patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">

<p style="font-size:16pt; text-align:right;">
کاوش <b>الگوهای زیرساختاری متغیر و محدودشده</b> به شناسایی الگوهای ساختاری در گراف‌ها یا شبکه‌ها می‌پردازد که یا دارای تغییرات مشخصی هستند یا شرایط خاصی را برآورده می‌کنند. این رویکرد امکانات کاوش زیرگراف‌های سنتی را گسترش می‌دهد و به کاربر اجازه می‌دهد الگوهایی را پیدا کند که یا دارای تغییرات کوچک (مانند جایگزینی یک گره) هستند یا با اعمال محدودیت‌های خاص (مانند نوع گره یا وزن یال)، دقیق‌تر و هدفمندتر باشند. این امر انعطاف‌پذیری و کاربردپذیری فرآیند کاوش را به‌طور چشمگیری افزایش می‌دهد.

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍

<p style="font-size:16pt; text-align:right;">
<b>1. الگوهای زیرساختاری متغیر (Variant Substructure Patterns):</b>  
این الگوها ساختارهایی هستند که با وجود تفاوت‌های کوچک (مثل تغییر نوع یک گره یا افزودن یک یال)، همچنان شباهت ساختاری قابل توجهی با یک الگوی پایه دارند. این رویکرد برای شناسایی خانواده‌هایی از ساختارها — مانند مولکول‌های مشابه با گروه‌های عاملی متفاوت — بسیار مفید است.

<p style="font-size:16pt; text-align:right;">
<b>2. الگوهای زیرساختاری محدودشده (Constrained Substructure Patterns):</b>  
این الگوها با اعمال محدودیت‌های خاص در حین کاوش استخراج می‌شوند. مثلاً فقط زیرگراف‌هایی که شامل یک گره با برچسب "سرطانی" یا یالی با وزن بیش از ۰.۸ باشند. این محدودیت‌ها کاوش را به سمت الگوهای مرتبط با یک هدف خاص سوق می‌دهند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. انعطاف‌پذیری و ویژه‌بودن (Flexibility and Specificity):</b>  
ترکیب این دو رویکرد، امکان کشف الگوهایی را فراهم می‌کند که هم انعطاف‌پذیر باشند (با تحمل تغییرات کوچک) و هم دقیق (با رعایت محدودیت‌های کاربردی). این ترکیب، کاوش را از یک فرآیند کلی به یک ابزار تحلیلی قدرتمند و سفارشی تبدیل می‌کند.

<b>4. کاربردها:</b>  
این تکنیک در حوزه‌هایی با نیاز به تحلیل دقیق و انعطاف‌پذیر کاربرد دارد:  
- <b>زیست‌فناوری</b>: شناسایی خانواده‌های پروتئینی با تغییرات جزئی یا مولکول‌های دارویی مشابه.  
- <b>امنیت سایبری</b>: تشخیص نسخه‌های تغییریافته یک الگوی حمله شبکه‌ای.  
- <b>شبکه‌های اجتماعی</b>: یافتن انجمن‌هایی با ساختار خاص (مثلاً یک رهبر مرکزی + اعضای متصل).  
- <b>داده‌کاوی متن</b>: کشف ساختارهای نحوی مشابه در جملات.

بیایید از یک مجموعه داده گرافی استفاده کنیم و با استفاده از یک الگوریتم پیشرفته (مانند gSpan با محدودیت یا یک رویکرد مبتنی بر یادگیری معنایی)، زیرساختارهایی را شناسایی کنیم که یا دارای تغییرات مشخصی هستند یا شرایط خاصی را برآورده می‌کنند:

```python
import networkx as nx
from itertools import combinations


# داده‌ها: لیست گراف‌ها با ویژگی گره
graph_dataset_with_attributes = [
    {'nodes': [(1, {'attribute': 'A'}), (2, {'attribute': 'B'})], 'edges': [(1, 2)]},
    {'nodes': [(1, {'attribute': 'A'}), (2, {'attribute': 'C'})], 'edges': [(1, 2)]},
    {'nodes': [(1, {'attribute': 'B'}), (2, {'attribute': 'B'})], 'edges': [(1, 2)]},
]


# تبدیل به NetworkX Graph
nx_graphs = []
for g in graph_dataset_with_attributes:
    G = nx.Graph()
    G.add_nodes_from(g['nodes'])
    G.add_edges_from(g['edges'])
    nx_graphs.append(G)


# پارامترها
min_support = 2
constraint_attribute = 'B'


# توابع کمکی
def is_subgraph(subgraph, G):
    matcher = nx.algorithms.isomorphism.GraphMatcher(
        G, subgraph,
        node_match=lambda n1, n2: n1.get('attribute') == n2.get('attribute')
    )
    return matcher.subgraph_is_isomorphic()

def contains_constraint(subgraph, attribute):
    return any(data.get('attribute') == attribute for _, data in subgraph.nodes(data=True))

def canonical_code(G):
    edges = sorted([tuple(sorted(edge)) for edge in G.edges()])
    return tuple(edges)


# الگوریتم DFS-like gSpan با محدودیت ویژگی
def gspan_like_constrained(graphs, min_support, constraint_attr):
    subgraph_pool = []

    for G in graphs:
        for edge in G.edges():
            H = nx.Graph()
            H.add_edge(*edge)
            for node in edge:
                H.nodes[node]['attribute'] = G.nodes[node]['attribute']
            subgraph_pool.append(H)

    visited_codes = set()
    frequent_subgraphs = []

    while subgraph_pool:
        candidate = subgraph_pool.pop(0)
        code = canonical_code(candidate)
        if code in visited_codes:
            continue
        visited_codes.add(code)

        support = sum(is_subgraph(candidate, G) for G in graphs)
        if support >= min_support and contains_constraint(candidate, constraint_attr):
            frequent_subgraphs.append((candidate, support))

            for G in graphs:
                nodes = list(G.nodes())
                for u, v in combinations(nodes, 2):
                    if not candidate.has_edge(u, v) and G.has_edge(u, v):
                        new_subgraph = candidate.copy()
                        new_subgraph.add_edge(u, v)
                        for node in [u, v]:
                            new_subgraph.nodes[node]['attribute'] = G.nodes[node]['attribute']
                        new_code = canonical_code(new_subgraph)
                        if new_code not in visited_codes:
                            subgraph_pool.append(new_subgraph)

    return frequent_subgraphs


# استخراج و نمایش زیرگراف‌های پرتکرار با محدودیت ویژگی
constrained_substructure_patterns = gspan_like_constrained(
    nx_graphs, min_support, constraint_attribute
)


# چاپ خروجی با مدیریت حالت خالی
if not constrained_substructure_patterns:
    print("هیچ زیرگرافی یافت نشد.")
else:
    for i, (H, support) in enumerate(constrained_substructure_patterns, 1):
        edges_str = [(u, v, H.nodes[u]['attribute'], H.nodes[v]['attribute']) for u, v in H.edges()]
        print(f"Subgraph {i}: Edges = {edges_str}, Support = {support}")
```

در این مثال، از الگوریتم <b>DFS-like gSpan</b> بر روی یک مجموعه داده گرافی استفاده شده که در آن گره‌ها دارای ویژگی (مانند برچسب یا نوع) هستند. فرآیند کاوش با اعمال یک <b>محدودیت ساختاری</b> هدایت شده است: تنها زیرگراف‌هایی گزارش می‌شوند که شامل حداقل یک گره با ویژگی خاص <b>'B'</b> باشند. 
این محدودیت به ما امکان می‌دهد تا فقط الگوهای معنادار و هدفمند را استخراج کنیم، به‌جای اینکه همه زیرگراف‌های ممکن را بررسی کنیم. 
اگر هیچ زیرگرافی با حداقل پشتیبانی مشخص و ویژگی 'B' وجود نداشته باشد، پیام «هیچ زیرگرافی یافت نشد.» چاپ می‌شود.

<p style="font-size:16pt; text-align:right;">
این رویکرد برای شناسایی زیرگراف‌های پرتکرار و هدفمند در زمینه‌هایی مانند شبکه‌های بیولوژیکی، شبکه‌های اجتماعی یا ساختارهای مولکولی بسیار مفید است.

### 5.6.1. Phrase mining in massive text data

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش عبارات در داده‌های متنی بزرگ</span> 📚
</p>
<p style="font-size:16pt; text-align:right;">
کاوش عبارات (Phrase Mining) به استخراج <b>عبارات چندواژگی</b> معنادار از مجموعه‌داده‌های گسترده متنی می‌پردازد. این فرآیند فراتر از شناسایی کلمات تکی است و به دنبال دنباله‌های پیوسته‌ای از کلمات است که واحدهای معنادار و هماهنگی را تشکیل می‌دهند. کاوش عبارات برای درک بهتر روابط زمینه‌ای و استخراج بینش‌های کلیدی از داده‌های متنی گسترده — مانند مقالات علمی، نظرات کاربران یا اسناد سازمانی — بسیار ارزشمند است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. عبارات چندواژگی (Multi-Word Phrases):</b>  
برخلاف تحلیل مبتنی بر کلمه، کاوش عبارات بر شناسایی ترکیبات معنادار چندکلمه‌ای تمرکز دارد که مفهوم خاصی را منتقل می‌کنند. این عبارات می‌توانند شامل <b>هم‌وقوعی‌ها</b> (مثل "هوش مصنوعی")، <b>اصطلاحات تخصصی</b> (مثل "یادگیری ماشین") یا <b>ضرب‌المثل‌ها</b> باشند که مفهومی فراتر از مجموع کلمات آن‌ها دارند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. معیارهای آماری (Statistical Measures):</b>  
برای تشخیص عبارات معنادار، از معیارهای آماری مانند <b>اطلاع متقابل نقطه‌ای</b> (PMI)، <b>آزمون کای-اسکوئر</b> یا <b>ضریب توافق</b> (Collocation Strength) استفاده می‌شود. این معیارها عباراتی را اولویت‌بندی می‌کنند که به‌طور معناداری بیشتر از حد انتظار با هم ظاهر می‌شوند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. درک زمینه‌ای (Contextual Understanding):</b>  
کاوش عبارات به درک عمیق‌تری از متن کمک می‌کند، زیرا روابط معنایی و زمینه‌ای که در کلمات کنار هم نهفته است را آشکار می‌سازد. این امر برای کاربردهایی مانند <b>تحلیل احساسات</b> (مثلاً "عالی نبود" ≠ "عالی")، <b>مدل‌سازی موضوعی</b> و <b>بازیابی اطلاعات</b> حیاتی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کاربردها:</b>  
این تکنیک در حوزه‌های متعددی کاربرد دارد:  
- <b>پردازش زبان طبیعی (NLP)</b>: بهبود تحلیل نحوی و معنایی.  
- <b>تحلیل رسانه‌های اجتماعی</b>: شناسایی تم‌های رایج و اصطلاحات روز.  
- <b>سیستم‌های توصیه‌گر</b>: درک علایق کاربران از طریق عبارات کلیدی.  
- <b>خلاصه‌سازی خودکار</b>: استخراج جملات و عبارات کلیدی.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش عبارات در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه متنی (مثل نظرات کاربران یا مقالات) استفاده کنیم و با استفاده از کتابخانه‌هایی مانند <code>nltk</code>😁 یا <code>gensim</code>، عبارات چندواژگی پرتکرار و معنادار را استخراج کنیم:
</p>

</div>

```python
from gensim.models import Phrases
from gensim.models.phrases import Phraser, ENGLISH_CONNECTOR_WORDS
import re

# مجموعه متنی شبیه‌سازی‌شده
# هر سند به صورت رشته خام است
text_corpus = [
    "Data mining is an essential technique.",
    "Phrase mining enhances text analysis.",
    "Massive text data contains valuable insights.",
    # ... (می‌توانید اسناد بیشتری اضافه کنید)
]


# توکن‌سازی خودکار متن
# تمامی حروف به کوچک تبدیل شده و کلمات با regex استخراج می‌شوند
def tokenize(text):
    text = text.lower()  # تبدیل به حروف کوچک
    tokens = re.findall(r'\b\w+\b', text)  # استخراج کلمات
    return tokens

tokenized_corpus = [tokenize(doc) for doc in text_corpus]


# آموزش مدل Phrases برای کشف عبارت‌های چندکلمه‌ای
phrases_model = Phrases(tokenized_corpus, connector_words=ENGLISH_CONNECTOR_WORDS)
phraser = Phraser(phrases_model)  # مدل سریع برای اعمال روی تمام متن‌ها


# اعمال مدل روی تمام اسناد
phrased_corpus = [phraser[doc] for doc in tokenized_corpus]


# چاپ خروجی
print("Discovered Multi-Word Phrases:")
for doc_phrases in phrased_corpus:
    print(doc_phrases)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از کتابخانه <b>Gensim</b> برای کشف <b>عبارت‌های چندواژه‌ای</b> (Multi-Word Phrases) در یک مجموعه متنی کوچک استفاده شده است. ابتدا متن‌ها با استفاده از توکن‌سازی خودکار به کلمات تجزیه شده و حروف بزرگ و کوچک یکسان‌سازی می‌شوند. سپس مدل <b>Phrases</b> با استفاده از معیارهایی مانند <b>اطلاع متقابل نقطه‌ای</b> (PMI) و آستانه فراوانی، ترکیب‌های کلمات که به‌طور معناداری با هم ظاهر می‌شوند را شناسایی می‌کند. خروجی، لیست اسناد به‌روزشده‌ای است که عبارات کشف‌شده با زیرخط (_) جایگزین فاصله در آنها شده‌اند و می‌توانند برای تحلیل متن، مدل‌سازی موضوعی یا ساخت واژگان تخصصی استفاده شوند.
</p>

### 5.6.2. Mining copy and paste bugs in software programs

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش باگ‌های کپی-پیست در کد نرم‌افزار</span> 🐞
</p>
<p style="font-size:16pt; text-align:right;">
کاوش <b>باگ‌های کپی-پیست</b> به شناسایی و تحلیل الگوهای تکراری در کد نرم‌افزار می‌پردازد که ناشی از عملیات کپی و پیست در فرآیند توسعه هستند. کپی کردن تکه‌کدها و استفاده مکرر از آن‌ها در بخش‌های مختلف برنامه، منجر به ایجاد کدهای تکراری یا مشابه می‌شود که ممکن است حاوی خطا باشند یا در آینده باعث اشکال شوند. هدف از این کاوش، تشخیص این الگوها، درک تأثیر آن‌ها و تسهیل فرآیندهای نگهداری کد و رفع اشکال است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. تکرار کد (Code Duplication):</b>  
تکرار کد زمانی رخ می‌دهد که توسعه‌دهندگان بخش‌هایی از کد را کپی کرده و در جای دیگری در همان پروژه استفاده کنند. این عمل ممکن است در کوتاه‌مدت زمان‌بر نباشد، اما در بلندمدت منجر به مشکلاتی مانند <b>عدم سازگاری</b>، <b>سختی در نگهداری</b> و <b>گسترش باگ‌ها</b> می‌شود. شناسایی کدهای تکراری برای حفظ کیفیت کد ضروری است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. کاوش الگو (Pattern Mining):</b>  
کاوش باگ‌های کپی-پیست از تکنیک‌های کاوش الگو برای شناسایی بخش‌های تکراری یا مشابه در کد منبع استفاده می‌کند. این روش‌ها می‌توانند تکه‌کدهای یکسان یا تغییریافته (مثلاً با نام متغیرهای متفاوت) را تشخیص دهند و به عنوان "کلون کد" (Code Clone) علامت‌گذاری کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. تأثیر باگ (Bug Impact):</b>  
اگر یک بخش کپی‌شده دارای خطا باشد، این خطا در تمام نسخه‌های کپی‌شده نیز وجود خواهد داشت. همچنین، اگر یکی از نسخه‌ها اصلاح شود اما بقیه اصلاح نشوند، ناسازگاری ایجاد می‌شود. این باگ‌ها می‌توانند عملکرد نرم‌افزار، قابلیت اطمینان و امنیت آن را به‌طور جدی تحت تأثیر قرار دهند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. تحلیل خودکار (Automated Analysis):</b>  
با توجه به حجم بالای کدهای پروژه‌های بزرگ، بررسی دستی غیرعملی است. ابزارهای خودکاری مانند <b>PMD-CPD</b>، <b>Simian</b> یا کتابخانه‌های پایتونی مبتنی بر تحلیل نحوی و لغوی، به‌طور خودکار بخش‌های تکراری را شناسایی می‌کنند و به توسعه‌دهندگان هشدار می‌دهند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش باگ‌های کپی-پیست در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک تکه کد پایتون حاوی تکرارهای مشکوک را بررسی کنیم و با استفاده از یک تحلیل ساده نحوی یا ابزار خودکار، الگوهای کپی-پیست بالقوه را شناسایی کنیم:
</p>

</div>

```python
# Simulate a code snippet with a potential copy and paste bug
def calculate_area_of_rectangle(length, width):
    return length * width

def compute_area_of_square(side_length):
    return side_length * side_length

def calculate_volume_of_box(length, width, height):
    # Incorrect formula: duplicated from calculate_area_of_rectangle
    return length * width * height

# این بخش، یک تابع ساده برای تشخیص **باگ‌های کپی-پیست (Copy-Paste Bugs)** در کد ارائه می‌ده.
# این نوع باگ زمانی اتفاق می‌افته که توسعه‌دهنده قطعه کدی رو کپی می‌کنه و بدون اصلاح کامل، استفاده می‌کنه.
# مثلاً در تابع `calculate_volume_of_box`، فرمول مربوط به مساحت (مکعب مستطیل) کپی شده، اما اینجا باید حجم محاسبه بشه.
# این خطا می‌تونه به دلیل کپی اشتباه یا فراموشی اصلاح کد ایجاد بشه.

def detect_copy_and_paste_bugs(code_snippet):
    # تحلیل ساده: بررسی وجود خطوط کد یکسان در متن کد
    # این روش یک رویکرد اولیه و سطحی برای تشخیص کپی-پیست هست.
    # تمام خطوط تمیزشده (بدون فاصله اضافی) در یک مجموعه (set) ذخیره می‌شن.
    # اگر خطی قبلاً دیده شده باشه، احتمال کپی-پیست وجود داره.
    patterns = set()
    bugs_detected = False

    for line in code_snippet.split('\n'):
        stripped_line = line.strip()
        if stripped_line in patterns:
            bugs_detected = True
            break  # تنها یک تکرار کافیه تا باگ شناسایی بشه
        patterns.add(stripped_line)

    return bugs_detected

# کدی که شامل یک باگ بالقوه کپی-پیست هست رو به صورت رشته وارد می‌کنیم.
# این روش شبیه‌سازی می‌کنه که چطور یک ابزار تحلیل کد می‌تونه ورودی رو به صورت متن خام پردازش کنه.
code_snippet_with_bug = """
def calculate_area_of_rectangle(length, width):
    return length * width

def compute_area_of_square(side_length):
    return side_length * side_length

def calculate_volume_of_box(length, width, height):
    # Incorrect formula: duplicated from calculate_area_of_rectangle
    return length * width * height
"""

# اجرای تابع تشخیص روی کد مورد نظر
bugs_detected = detect_copy_and_paste_bugs(code_snippet_with_bug)

# در نهایت، نتیجه تشخیص رو چاپ می‌کنیم.
# این خروجی نشان می‌ده که آیا خط کدی بیش از یک بار در متن تکرار شده (که ممکنه نشانه کپی-پیست باشه).
# توجه: این روش خیلی ساده‌ست و ممکنه خطاهای مثبت (False Positive) داشته باشه (مثلاً تکرار `return` در توابع مختلف).
# در عمل، ابزارهای پیشرفته‌تر از تحلیل ساختار AST (Abstract Syntax Tree) یا شباهت معنایی استفاده می‌کنن.
print(f"Copy and Paste Bugs Detected: {bugs_detected}")
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻

<p style="font-size:16pt; text-align:right;">
در این مثال، یک تکه کد پایتون شامل تکرار ناخواسته وجود دارد: فرمول محاسبه حجم جعبه از محاسبه مساحت مستطیل کپی شده، اما به‌درستی اصلاح نشده است (مثلاً فقط دو بعد در نظر گرفته شده به جای سه بعد). تابع <b>detect_copy_and_paste_bugs</b> یک تحلیل ساده انجام می‌دهد که خطوط تکراری یا بسیار مشابه در کد را شناسایی می‌کند. این رویکرد می‌تواند شامل مقایسه رشته‌ای کدها، استخراج الگوهای نحوی (AST) یا محاسبه هش خطوط کد باشد. با شناسایی این تکرارها، توسعه‌دهندگان می‌توانند کدهای مشکل‌دار را اصلاح، آن‌ها را به توابع قابل استفاده مجدد تبدیل کنند و از گسترش خطا جلوگیری نمایند.
