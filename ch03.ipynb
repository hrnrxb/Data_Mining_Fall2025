### 3.1.3. Data lakes

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">دریاچه داده (Data Lake)</span> 🌊
</p>
<p style="font-size:16pt; text-align:right;">
<b>دریاچه داده (Data Lake)</b> یک مخزن متمرکز است که امکان ذخیره حجم عظیمی از داده‌های خام، بدون ساختار یا ساختاریافته را در مقیاس وسیع فراهم می‌کند. برخلاف پایگاه‌های داده سنتی یا انبارهای داده، دریاچه‌های داده انواع و فرمت‌های مختلف داده را در خود جای می‌دهند و راه‌حلی منعطف و مقیاس‌پذیر برای ذخیره و پردازش کارآمد داده‌ها ارائه می‌دهند. دریاچه‌های داده اغلب در محیط‌های داده‌های بزرگ (big data) و تحلیل استفاده می‌شوند و به سازمان‌ها امکان می‌دهند حجم عظیمی از داده‌ها را به طور مؤثر ذخیره و تحلیل کنند.
</p>
</div>

```python
import pandas as pd
from sklearn.datasets import load_iris
import os

# مرحله اول: نوشتن داده‌ها
# از یک دیتاست داخلی (مثل iris) استفاده می‌کنیم تا به داده‌های آماده دسترسی داشته باشیم.
# این دیتاست شامل ویژگی‌های گل‌های زنبق است.
iris = load_iris()

# داده‌ها و نام ویژگی‌ها را به یک DataFrame تبدیل می‌کنیم.
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# حالا به جای ذخیره در S3، آن را در یک فایل موقت محلی ذخیره می‌کنیم.
# این فایل موقت، "دیتا لیک" ما را شبیه‌سازی می‌کند.
file_name = 'iris_data.csv'
df.to_csv(file_name, index=False)
print(f"فایل '{file_name}' با موفقیت به عنوان دیتا لیک محلی ایجاد شد.")

# مرحله دوم: خواندن داده‌ها
# حالا داده‌ها را از فایل موقتی که ایجاد کردیم، می‌خوانیم.
df_from_lake = pd.read_csv(file_name)

# در نهایت، داده‌های خوانده‌شده از دیتا لیک را چاپ می‌کنیم.
print("\nداده‌های خوانده‌شده از دیتا لیک محلی:")
print(df_from_lake.head())

# (اختیاری) فایل موقت را حذف می‌کنیم تا سیستم را تمیز نگه داریم.
os.remove(file_name)
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
این مثال یک نمایش ساده‌شده از کار با مفهوم <b>دریاچه داده</b> ارائه می‌دهد. به جای استفاده از یک سرویس ابری مانند <b>Amazon S3</b>، ما یک دریاچه داده محلی را با استفاده از یک فایل <b>CSV</b> بر روی کامپیوترمان شبیه‌سازی می‌کنیم.
<br>
کد ابتدا یک مجموعه داده آماده برای استفاده از کتابخانه <b>Scikit-learn</b>، به ویژه مجموعه داده معروف <b>Iris</b>، را بارگذاری می‌کند. سپس نحوه ذخیره یک <b>DataFrame</b> پانداس به عنوان یک فایل <b>CSV</b> در دریاچه داده محلی خود و متعاقباً بازیابی و نمایش داده‌ها از آن فایل را نشان می‌دهد.
<br>
این فرآیند، هرچند ساده‌شده است، عملیات اساسی یک دریاچه داده واقعی را منعکس می‌کند و نشان می‌دهد که چگونه داده‌ها را می‌توان در یک مخزن متمرکز نوشت و از آن خواند. این رویکرد برای یادگیری و تست، بدون نیاز به پیکربندی‌های پیچیده ابری، عالی است.

### 3.2.1. Data cube: a multidimensional data model

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مکعب داده (Data Cube)</span> 🧊
</p>
<p style="font-size:16pt; text-align:right;">
<b>مکعب داده (Data Cube)</b> یک مدل داده چندبعدی است که برای نمایش و تحلیل داده‌ها در ابعاد مختلف استفاده می‌شود. این مدل ساختار جدول سنتی و دوبعدی پایگاه‌های داده رابطه‌ای را گسترش می‌دهد تا ابعاد اضافی را در خود جای دهد و یک نمای جامع‌تر از داده‌ها ارائه کند. در یک مکعب داده، هر خانه (cell) حاوی یک مقدار یا معیار است و می‌توان مکعب را برای کاوش و تحلیل داده‌ها از دیدگاه‌های مختلف، برش (slice)، تکه‌تکه (dice) یا جمع‌آوری (roll up) کرد. این نمایش چندبعدی به‌ویژه برای کاربردهای داده‌کاوی و هوش تجاری (business intelligence) مفید است.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال مکعب داده در پایتون</span> 🧊
</p>
<p style="font-size:16pt; text-align:right;">
اگرچه کتابخانه‌ای به نام "data cube" به طور خاص در پایتون وجود ندارد، اما می‌توانید از کتابخانه‌هایی مانند <b>pandas</b> و <b>matplotlib</b> برای ایجاد بصری‌سازی‌هایی که داده‌های چندبعدی را نمایش می‌دهند، استفاده کنید. در این مثال، من نحوه ایجاد یک نمودار پراکندگی سه‌بعدی (3D scatter plot) ساده را نشان می‌دهم که می‌تواند به عنوان یک نمایش ساده‌شده از یک مکعب داده سه‌بعدی در نظر گرفته شود.
</p>
</div>

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# این قسمت اول کده. سه تا کتابخونه اصلی رو وارد می‌کنیم:
# 'pandas' برای کار با داده‌ها تو قالب جدول.
# 'matplotlib.pyplot' برای رسم نمودارها.
# 'Axes3D' که ابزاری از 'matplotlib' برای رسم نمودارهای سه‌بعدی هست.

# این دیکشنری حاوی داده‌های نمونه ماست که سه ویژگی یا بعد (Dimension) دارن: 'X', 'Y' و 'Z'.
# این داده‌ها رو به عنوان یک مکعب داده (Data Cube) در نظر بگیرید.
data = {
    'X': [1, 2, 3, 4, 5],
    'Y': [5, 4, 3, 2, 1],
    'Z': [10, 8, 6, 4, 2]
}

# با استفاده از داده‌های بالا، یه 'DataFrame' جدید می‌سازیم.
df = pd.DataFrame(data)

# این بخش، یک **نمودار پراکندگی سه‌بعدی** رو ایجاد می‌کنه.
# این نمودار برای بصری‌سازی داده‌هایی با سه ویژگی یا بعد عالیه.
# 'fig' فضای رسم و 'ax' محورهای سه‌بعدی رو ایجاد می‌کنه.
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# 'ax.scatter()' تابع اصلی برای رسم نقاط تو فضای سه‌بعدیه.
# 'X', 'Y' و 'Z' مقادیر هر نقطه رو مشخص می‌کنن.
# 'c' رنگ نقاط و 'marker' هم شکلشون رو مشخص می‌کنه.
ax.scatter(df['X'], df['Y'], df['Z'], c='blue', marker='o')

# این خطوط برای زیباتر شدن و قابل فهم‌تر شدن نمودار هستن.
# با توجه به درخواست شما، عنوان و برچسب محورها به انگلیسی نوشته شدن.
ax.set_xlabel('X Dimension')
ax.set_ylabel('Y Dimension')
ax.set_zlabel('Z Dimension')
ax.set_title('3D Scatter Plot: Simulated Data Cube')

# 'plt.show()' نمودار سه‌بعدی رو نمایش می‌ده.
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، دیتافریم <code>df</code> نمایانگر یک مجموعه داده ساده سه‌بعدی با ابعاد X، Y و Z است. کد با استفاده از کتابخانه <b><code>matplotlib</code></b> یک نمودار پراکندگی سه‌بعدی ایجاد می‌کند تا نقاط داده را در فضای سه‌بعدی بصری‌سازی کند.
<br>
اگرچه این مثال به‌طور کامل پیچیدگی‌های یک مکعب داده واقعی را به تصویر نمی‌کشد، اما مفهوم نمایش داده‌ها در ابعاد چندگانه را نشان می‌دهد. موارد استفاده واقعی از مکعب‌های داده شامل مجموعه‌داده‌های گسترده‌تر و ابزارهای پیشرفته‌ای مانند سیستم‌های OLAP (پردازش تحلیلی آنلاین) است که به کاربران اجازه می‌دهد داده‌های چندبعدی را به صورت تعاملی تحلیل کنند.
</p>

### 3.2.2. Schemas for multidimensional data models: stars, snowflakes, and fact constellations

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">شمای چندبعدی در مدل‌سازی داده</span> 🌌
</p>
<p style="font-size:16pt; text-align:right;">
در مدل‌سازی داده‌های چندبعدی، شماها (schemas) ساختارهایی هستند که نحوه سازماندهی و ارتباط داده‌ها را در یک پایگاه داده تعریف می‌کنند. سه نوع شمای رایج برای مدل‌های داده چندبعدی شامل ستاره‌ای (star)، برف‌دانه (snowflake) و صورت فلکی (fact constellation) می‌باشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>شمای ستاره‌ای (Star Schema):</b> ✨
<br>
در شمای ستاره‌ای، یک جدول مرکزی به نام جدول واقعیت (fact table) وجود دارد که حاوی معیارهای اصلی است. این جدول توسط جداول ابعاد (dimension tables) احاطه شده است که هر یک جنبه یا دیدگاه متفاوتی از داده‌ها را نشان می‌دهند. این ساختار از نظر بصری شبیه به یک ستاره است.
<br>
<b>مثال در پایتون:</b>
<br>
اگرچه مفهوم شمای ستاره‌ای بیشتر به طراحی پایگاه داده مربوط است، اما می‌توانید یک شمای ستاره‌ای ساده را با استفاده از دیتافریم‌های pandas شبیه‌سازی کنید. یک مجموعه داده فروش را با یک جدول واقعیت مرکزی حاوی معیارهای فروش و جداول ابعاد پیرامونی برای محصولات، مشتریان و زمان در نظر بگیرید.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>شمای برف‌دانه (Snowflake Schema):</b> ❄️
<br>
شمای برف‌دانه، شمای ستاره‌ای را با نرمال‌سازی جداول ابعاد گسترش می‌دهد و آن‌ها را به زیر-ابعاد یا سطوح تقسیم می‌کند. این امر منجر به ساختاری می‌شود که از نظر بصری شبیه به یک برف‌دانه است.
<br>
<b>مثال در پایتون:</b>
<br>
مجدداً، با استفاده از pandas، می‌توانید با نرمال‌سازی جداول ابعاد از مثال شمای ستاره‌ای، یک شمای برف‌دانه را شبیه‌سازی کنید. این کار ممکن است شامل تقسیم بعد "محصولات" به زیر-ابعادی مانند "دسته‌بندی محصول" و "زیر-دسته‌بندی محصول" باشد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>شمای صورت فلکی (Fact Constellation):</b> 🌠
<br>
شمای صورت فلکی شامل چندین جدول واقعیت است که جداول ابعاد مشترکی دارند. هر جدول واقعیت معیارهای متفاوتی را نشان می‌دهد و ابعاد مشترک امکان تحلیل متقابل بین جداول واقعیت را فراهم می‌کنند.
<br>
<b>مثال در پایتون:</b>
<br>
مثال شمای ستاره‌ای را با افزودن یک جدول واقعیت دیگر، مانند جدول واقعیت "هزینه‌ها"، گسترش دهید. هر دو جدول واقعیت "فروش" و "هزینه‌ها" ابعاد مشترکی مانند "زمان" و "مشتری" را به اشتراک می‌گذارند.
</p>
</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این بخش، **جدول واقعیت (Fact Table)** به نام 'Sales' رو ایجاد می‌کنه.
# جدول واقعیت، رویدادها یا تراکنش‌های اصلی کسب و کار رو شامل می‌شه.
# این جدول معمولاً اعداد قابل اندازه‌گیری (مثل 'SalesAmount') و کلیدهایی برای ارتباط با جدول‌های ابعاد (Dimension) رو تو خودش نگه می‌داره.
sales_data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-01-03'],
    'ProductID': [1, 2, 1],
    'CustomerID': [101, 102, 103],
    'SalesAmount': [500, 300, 700]
}

sales_df = pd.DataFrame(sales_data)

# این بخش، **جدول‌های ابعاد (Dimension Tables)** رو ایجاد می‌کنه.
# جدول‌های ابعاد، جزئیات توصیفی رو برای جدول واقعیت فراهم می‌کنن.
# مثلاً جدول 'Products' اطلاعاتی مثل 'ProductName' و 'Category' رو برای هر 'ProductID' داره.
# این ساختار که یک جدول واقعیت مرکزی و چند جدول ابعاد اطرافش داره، به **طرح‌واره ستاره‌ای (Star Schema)** معروفه.
products_data = {
    'ProductID': [1, 2],
    'ProductName': ['ProductA', 'ProductB'],
    'Category': ['Electronics', 'Clothing']
}

customers_data = {
    'CustomerID': [101, 102, 103],
    'CustomerName': ['Alice', 'Bob', 'Charlie'],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}

products_df = pd.DataFrame(products_data)
customers_df = pd.DataFrame(customers_data)

# این بخش، جدول‌ها رو چاپ می‌کنه تا ساختار طرح‌واره ستاره‌ای رو نشون بده.
print("جدول واقعیت: Sales")
print(sales_df)
print("\nجدول ابعاد: Products")
print(products_df)
print("\nجدول ابعاد: Customers")
print(customers_df)
```

### 3.2.3. Concept hierarchies

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">سلسله مراتب مفاهیم</span> 🌲
</p>
<p style="font-size:16pt; text-align:right;">
سلسله مراتب مفاهیم (Concept hierarchies) به سازماندهی داده‌ها در سطوح انتزاعی یا گرانولاریته اشاره دارد که یک سلسله مراتب از مفاهیم را تشکیل می‌دهد. در زمینه داده‌کاوی، سلسله مراتب مفاهیم اغلب بر روی ابعاد (dimensions) در مدل‌های داده چندبعدی اعمال می‌شود. هر سطح از سلسله مراتب، درجه متفاوتی از خلاصه‌سازی یا جزئیات را نشان می‌دهد و به کاربران امکان می‌دهد داده‌ها را در سطوح مختلفی از جزئیات کاوش کنند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثال سلسله مراتب مفاهیم در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مثال ساده را با استفاده از یک دیتافریم پانداس در نظر بگیریم که داده‌های فروش را با یک بعد زمانی نشان می‌دهد و دارای سلسله مراتب مفهومی سال، فصل و ماه است.
</p>
</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری (dictionary) حاوی داده‌های فروش نمونه ماست.
# 'Date' یک ویژگی زمانیه که می‌خوایم روی اون سلسله مراتب مفهومی رو ایجاد کنیم.
data = {
    'Date': ['2022-01-01', '2022-04-15', '2022-06-30', '2022-11-20'],
    'ProductID': [1, 2, 1, 3],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **سلسله مراتب مفهومی (Concept Hierarchy)** رو ایجاد می‌کنه.
# سلسله مراتب مفهومی، راهی برای سازماندهی و خلاصه‌سازی داده‌ها از سطح پایین‌تر به سطح بالاتر هست.
# برای داده‌های زمانی، ما می‌تونیم از روز به ماه، فصل، و سال بریم.
# 'dt.year' سال رو از تاریخ استخراج می‌کنه.
sales_df['Year'] = sales_df['Date'].dt.year
# 'dt.to_period("Q")' فصل رو از تاریخ استخراج می‌کنه.
sales_df['Quarter'] = sales_df['Date'].dt.to_period("Q")
# 'dt.to_period("M")' ماه رو از تاریخ استخراج می‌کنه.
sales_df['Month'] = sales_df['Date'].dt.to_period("M")

# در نهایت، 'DataFrame' جدید رو چاپ می‌کنیم تا ببینیم سلسله مراتب زمانی چطور اضافه شده.
print("داده‌های فروش با سلسله مراتب مفهومی زمانی:")
print(sales_df[['Date', 'Year', 'Quarter', 'Month', 'ProductID', 'SalesAmount']])
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، ستون تاریخ با سطوح اضافی سلسله مراتبی مبتنی بر زمان تقویت شده است: سال، فصل و ماه. این امر به کاربران امکان می‌دهد داده‌های فروش را در سطوح مختلفی از گرانولاریته زمانی تحلیل کنند.
</p>
<p style="font-size:16pt; text-align:right;">
سلسله مراتب مفاهیم با فراهم‌کردن روشی ساختارمند برای حرکت در داده‌ها در سطوح مختلف انتزاع، تحلیل داده را بهبود می‌بخشند. کاربران می‌توانند داده‌ها را بر اساس نیاز تحلیلی خود تجمیع کنند یا به جزئیات بیشتری فرو بروند. این مثال نحوه ایجاد یک سلسله مراتب مفهومی ساده برای بعد زمانی در یک مجموعه داده فروش را نشان می‌دهد، اما سلسله مراتب مفاهیم می‌توانند در سناریوهای واقعی به ابعاد مختلفی مانند جغرافیا، دسته‌های محصول یا سلسله مراتب سازمانی نیز اعمال شوند.
</p>

</div>

### 3.2.4. Measures: categorization and computation

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارها در داده‌کاوی</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
در زمینه داده‌کاوی، معیارها (measures) به مقادیر کمی یا شاخص‌هایی اطلاق می‌شوند که مورد علاقه بوده و اغلب در فرآیندهای تحلیلی استفاده می‌شوند. معیارها بسته به ماهیت خود قابل دسته‌بندی هستند، مانند جمع‌پذیر یا غیرجمع‌پذیر، و ممکن است مبتنی بر محاسباتی باشند که بینش‌های معناداری از داده‌ها استخراج می‌کنند.
</p>
</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">دسته‌بندی معیارها</span> 🔢
</p>

<p style="font-size:16pt; text-align:right;">
<b>معیارهای جمع‌پذیر:</b>  
معیارهای جمع‌پذیر، معیارهایی هستند که می‌توان آن‌ها را در ابعاد مختلف تجمیع یا جمع کرد. از جمله مثال‌های آن می‌توان به کل فروش، سود کل یا کل تعداد فروخته‌شده اشاره کرد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>معیارهای غیرجمع‌پذیر:</b>  
معیارهای غیرجمع‌پذیر را نمی‌توان به سادگی در تمام ابعاد تجمیع کرد. از جمله مثال‌های آن می‌توان به میانگین حاشیه سود، درصد رشد یا نسبت‌ها اشاره کرد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>محاسبه معیارها:</b>  
محاسبه به معنای استخراج معیارهای جدید یا تبدیل معیارهای موجود برای کسب بینش عمیق‌تر از داده‌هاست. روش‌های رایج محاسبه شامل یافتن درصدها، نرخ رشد، میانگین‌ها یا هر عملیات ریاضی مرتبط دیگری است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از معیارها در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مجموعه داده فروش را در نظر بگیریم و هم معیارهای جمع‌پذیر و هم معیارهای غیرجمع‌پذیر را محاسبه کنیم.
</p>

</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها رو برای نمایش تفاوت بین سنجه‌های جمع‌پذیر و غیرجمع‌پذیر استفاده می‌کنیم.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-01-03', '2022-01-04'],
    'ProductID': [1, 2, 1, 3],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **سنجه جمع‌پذیر (Additive Measure)** رو محاسبه می‌کنه.
# سنجه جمع‌پذیر، سنجه‌ای هست که می‌تونیم اون رو روی ابعاد مختلف (مثل زمان یا محصول) جمع کنیم و نتیجه معنادار باشه.
# مثلاً 'Total Sales' یا 'کل فروش' یک سنجه جمع‌پذیره، چون جمع فروش روزهای مختلف، کل فروش رو می‌ده.
total_sales = sales_df['SalesAmount'].sum()

# این بخش، **سنجه غیرجمع‌پذیر (Non-additive Measure)** رو محاسبه می‌کنه.
# سنجه غیرجمع‌پذیر، سنجه‌ای هست که جمع کردنش روی ابعاد مختلف، معنی نداره.
# 'Average Sales' یا 'میانگین فروش' یک سنجه غیرجمع‌پذیره، چون میانگین میانگین‌های روزانه، میانگین کل رو نمی‌ده.
average_sales = sales_df['SalesAmount'].mean()

# در نهایت، نتایج محاسبه‌شده رو چاپ می‌کنیم تا تفاوتشون رو ببینیم.
print("سنجه جمع‌پذیر: کل فروش =", total_sales)
print("سنجه غیرجمع‌پذیر: میانگین فروش =", average_sales)
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، ستون SalesAmount نشان‌دهنده یک معیار است و ما از داده‌های فروش، یک معیار جمع‌پذیر (کل فروش) و یک معیار غیرجمع‌پذیر (میانگین فروش) را محاسبه می‌کنیم.
</p>
<p style="font-size:16pt; text-align:right;">
درک و دسته‌بندی معیارها برای تحلیل مؤثر داده‌ها ضروری است. معیارهای مختلف ممکن است بر اساس اهداف تحلیلی، روش‌های محاسباتی متفاوتی نیاز داشته باشند. این مثال نحوه محاسبه معیارهای جمع‌پذیر و غیرجمع‌پذیر در یک مجموعه داده فروش با استفاده از پایتون را نشان می‌دهد.
</p>

## Section 3.3

### 3.3.1. Typical OLAP operations

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">پردازش تحلیلی آنی (OLAP)</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
OLAP یا پردازش تحلیلی آنی (Online Analytical Processing)، دسته‌ای از ابزارها و فناوری‌ها است که به کاربران امکان می‌دهد به‌صورت تعاملی داده‌های چندبعدی را برای پشتیبانی از تصمیم‌گیری و هوش تجاری تحلیل کنند. عملیات OLAP، عملیات پایه‌ای هستند که کاربران بر روی مکعب‌های داده چندبعدی انجام می‌دهند تا بینش‌های ارزشمندی استخراج کنند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">عملیات متداول OLAP</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>عملیات فشردن (Roll-up / Drill-Up):</b>  
داده‌ها را در امتداد یک بعد تجمیع می‌کند و از سطح پایین‌تر گرانولاریته به سطح بالاتر می‌رود. مثلاً، فشردن داده‌های فروش ماهانه به فصلی.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>عملیات جزئی‌سازی (Drill-down / Roll-Down):</b>  
داده‌های تجمیع‌شده را در امتداد یک بعد باز می‌کند و از سطح بالاتر به سطح پایین‌تر گرانولاریته می‌رود. مثلاً، جزئی‌سازی داده‌های فروش سالانه به ماهانه.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>عملیات برش (Slice):</b>  
با انتخاب یک سطح از یک بعد، یک برش دو بعدی از مکعب ایجاد می‌کند. مثلاً، انتخاب داده‌های فروش یک فصل خاص و یک دسته محصول.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>عملیات تاس‌زدن (Dice):</b>  
با مشخص کردن مقادیر در چند بعد، یک زیرمکعب انتخاب می‌کند. مثلاً، انتخاب داده‌های فروش یک منطقه، فصل و دسته محصول خاص.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>عملیات چرخش (Pivot / Rotate):</b>  
مکعب داده را می‌چرخاند تا داده‌ها از دیدگاه دیگری نمایش داده شوند. مثلاً، چرخش برای مشاهده فروش محصولات در طول فصل‌های مختلف.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از عملیات OLAP در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
اگرچه کتابخانه‌های تخصصی OLAP در پایتون وجود ندارند، اما می‌توان از کتابخانه pandas برای شبیه‌سازی عملیات OLAP بر روی یک مجموعه داده چندبعدی استفاده کرد. بیایید یک مجموعه داده ساده فروش را در نظر بگیریم و عملیات‌های برش (Slice) و تاس‌زدن (Dice) را انجام دهیم.
</p>

</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری (dictionary) حاوی داده‌های فروش نمونه ماست.
# این داده‌ها رو به عنوان یک مکعب داده (Data Cube) در نظر بگیرید که ابعاد مختلفی مثل 'Region' و 'Date' رو داره.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# 'DataFrame' اصلی رو چاپ می‌کنیم تا ببینیم داده‌ها چه شکلی هستن.
print("داده‌های فروش اصلی:")
print(sales_df)

# این بخش، عملیات **برش (Slice)** رو انجام می‌ده.
# برش، انتخاب داده‌ها بر اساس یک بعد خاصه.
# اینجا، ما فقط داده‌های مربوط به ماه ژانویه (ماه ۱) رو انتخاب می‌کنیم.
january_sales = sales_df[sales_df['Date'].dt.month == 1]

# این بخش، عملیات **مکعب داده (Dice)** رو انجام می‌ده.
# مکعب داده، انتخاب داده‌ها بر اساس ترکیبی از چند بعده.
# اینجا، ما داده‌هایی رو انتخاب می‌کنیم که هم مربوط به منطقه 'North' باشن و هم مربوط به ماه ژانویه.
north_january_sales = sales_df[(sales_df['Region'] == 'North') & (sales_df['Date'].dt.month == 1)]

# در نهایت، نتایج هر دو عملیات رو چاپ می‌کنیم.
print("\nعملیات OLAP برش (فروش ژانویه):")
print(january_sales)

print("\nعملیات OLAP مکعب داده (فروش ژانویه و منطقه North):")
print(north_january_sales)
```

### 3.3.2. Indexing OLAP data: bitmap index and join index

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">نقش ایندکس‌گذاری در OLAP</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
در پایگاه‌های داده OLAP، <b>ایندکس‌گذاری</b> نقشی حیاتی در بهینه‌سازی عملکرد پرس‌وجو و تسریع بازیابی داده‌ها ایفا می‌کند. دو نوع متداول از ایندکس‌ها که در OLAP استفاده می‌شوند، <b>ایندکس بیت‌مپ (Bitmap Index)</b> و <b>ایندکس جوین (Join Index)</b> هستند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">انواع ایندکس‌ها در OLAP</span> 🔧
</p>

<p style="font-size:16pt; text-align:right;">
<b>ایندکس بیت‌مپ (Bitmap Index):</b>  
در این تکنیک ایندکس‌گذاری، برای هر مقدار منحصربه‌فرد در یک بعد، یک نقشه بیتی (bitmap) در نظر گرفته می‌شود. هر بیت در این نقشه مربوط به یک سطر در جدول واقعیت (fact table) است و نشان می‌دهد که آیا آن مقدار بعد در آن سطر وجود دارد یا خیر.  
ایندکس‌های بیت‌مپ از نظر فضای ذخیره‌سازی کارآمد هستند و به‌ویژه برای ویژگی‌های با <b>کاردینالیته پایین</b> (تعداد مقادیر مجزای کم) مناسب می‌باشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>ایندکس جوین (Join Index):</b>  
ایندکس جوین برای بهینه‌سازی عملیات اتصال (join) بین جداول واقعیت و ابعاد طراحی شده است. این ایندکس نتایج از پیش محاسبه‌شده‌ای از جوین‌های خاص را ذخیره می‌کند و نیاز به اجرای جوین‌های پرهزینه در زمان اجرای پرس‌وجو را کاهش می‌دهد.  
ایندکس جوین به‌ویژه در پرس‌وجوهای پیچیده که شامل چندین بعد می‌شوند، <b>بسیار مفید</b> است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از ایندکس بیت‌مپ و ایندکس جوین در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
برای مثالی ساده‌شده، از کتابخانه pandas استفاده می‌کنیم تا سناریویی را شبیه‌سازی کنیم که در آن یک <b>ایندکس بیت‌مپ</b> برای یک بعد ایجاد می‌شود و همچنین از یک <b>ایندکس جوین</b> برای بهینه‌سازی عملیات اتصال</b> استفاده می‌شود.
</p>

</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **ایندکس بیت‌مپ (Bitmap Index)** رو شبیه‌سازی می‌کنه.
# ایندکس بیت‌مپ برای ستون‌هایی که تعداد مقادیر منحصربه‌فردشون کمه، خیلی کارآمده.
# ایندکس برای هر مقدار یک بیت‌مَپ (آرایه‌ای از ۰ و ۱) می‌سازه.
# مثلاً برای 'North'، هر جایی که 'Region' برابر 'North' باشه، ۱ و بقیه ۰ می‌شن.
# این کار جستجو رو خیلی سریع‌تر می‌کنه.
bitmap_index_region = {}
for region in sales_df['Region'].unique():
    bitmap_index_region[region] = sales_df['Region'] == region

# این بخش، **ایندکس پیوست (Join Index)** رو شبیه‌سازی می‌کنه.
# ایندکس پیوست، نتایج یک عملیات پیوست (Join) رو از قبل ذخیره می‌کنه.
# این کار باعث می‌شه وقتی نیاز به پیوست (Join) دو جدول داریم، دیگه نخوایم کل عملیات رو از اول انجام بدیم و سرعت جستجو و بازیابی داده‌ها بالا بره.
# ما اینجا جدول 'sales_df' رو با یک جدول ابعاد 'product_dimension' پیوست (join) می‌کنیم.
product_dimension = pd.DataFrame({
    'ProductID': [1, 2],
    'ProductName': ['ProductA', 'ProductB']
})

join_index_product = pd.merge(sales_df, product_dimension, on='ProductID', how='inner')

# در نهایت، ایندکس‌های شبیه‌سازی شده رو چاپ می‌کنیم.
print("ایندکس بیت‌مپ برای بعد 'Region':")
print(bitmap_index_region)

print("\nایندکس پیوست برای بعد 'Product':")
print(join_index_product)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، یک <b>Bitmap Index</b> برای بعد 'Region' شبیه‌سازی شده است، که در آن هر مقدار در دیکشنری نشان‌دهنده یک بیت‌مپ برای یک مقدار منحصربه‌فرد از منطقه است. همچنین یک <b>Join Index</b> برای بعد 'Product' با ادغام داده‌های فروش و جدول جداگانه محصولات شبیه‌سازی شده است.
</p>
<p style="font-size:16pt; text-align:right;">
اگرچه این شبیه‌سازی‌ها ساده هستند، اما پیاده‌سازی واقعی <b>Bitmap</b> و <b>Join Index</b> در پایگاه‌های داده تخصصی OLAP انجام می‌شود. این ایندکس‌ها عملکرد پرس‌وجوها را در سناریوهایی که شامل فیلتر کردن و اتصال داده‌ها در ابعاد مختلف می‌شوند، به‌طور چشمگیری بهبود می‌دهند.
</p>

### 3.3.3. Storage implementation: column-based databases

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">پایگاه‌های داده مبتنی بر ستون</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
پایگاه‌های داده مبتنی بر ستون نوعی از سیستم‌های مدیریت پایگاه داده (DBMS) هستند که داده‌ها را بر اساس ستون‌ها — نه ردیف‌ها — ذخیره و سازماندهی می‌کنند. در این نوع پایگاه داده، هر ستون به‌صورت جداگانه ذخیره می‌شود که امکان بازیابی داده و تحلیل کارآمدتری فراهم می‌کند، به‌ویژه هنگام کار با مجموعه‌داده‌های بزرگ. این ساختار ذخیره‌سازی برای کاربردهای داده‌کاوی و تحلیل داده بسیار مفید است، زیرا <b>عملکرد پرس‌وجو</b> را بهبود می‌بخشد و از تکنیک‌های <b>فشرده‌سازی داده</b> پشتیبانی می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">ویژگی‌های کلیدی پایگاه‌های داده مبتنی بر ستون</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>ذخیره‌سازی ستونی (Columnar Storage):</b>  
داده‌ها به‌جای فرمت ردیفی، به صورت ستونی ذخیره می‌شوند. هر ستون به‌عنوان یک فایل یا واحد ذخیره‌سازی مجزا نگهداری می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>تکنیک‌های فشرده‌سازی (Compression Techniques):</b>  
پایگاه‌های داده مبتنی بر ستون اغلب از روش‌های فشرده‌سازی استفاده می‌کنند که با استفاده از یکنواختی داده‌ها درون هر ستون، فضای ذخیره‌سازی را کاهش داده و <b>عملکرد پرس‌وجو</b> را بهبود می‌بخشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>کارایی در پردازش تحلیلی (Analytical Processing Efficiency):</b>  
این پایگاه‌های داده برای کارهای تحلیلی و داده‌کاوی بسیار مناسب هستند، به‌ویژه زمانی که عملیات‌های رایج شامل تجمیع و تحلیل روی زیرمجموعه‌ای از ستون‌ها باشند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از ذخیره‌سازی مبتنی بر ستون در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
اگرچه پایتون به‌خودی‌خود دارای فرمت ذخیره‌سازی مبتنی بر ستون نیست، اما می‌توان از کتابخانه‌هایی مانند Pandas برای شبیه‌سازی برخی از جنبه‌های ذخیره‌سازی مبتنی بر ستون استفاده کرد. بیایید یک مثال ساده را در نظر بگیریم:
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها رو برای شبیه‌سازی ذخیره‌سازی مبتنی بر ستون (Column-based Storage) استفاده می‌کنیم.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **ذخیره‌سازی مبتنی بر ستون (Column-based Storage)** رو شبیه‌سازی می‌کنه.
# در سیستم‌های واقعی مثل پایگاه‌های داده ستونی (مثلاً Apache Parquet یا Amazon Redshift)،
# داده‌ها به جای ذخیره کردن سطر به سطر، ستون به ستون ذخیره می‌شن.
# این کار باعث می‌شه عملیاتی مثل جمع‌کردن، فیلتر کردن یا آنالیز روی یک ستون خاص خیلی سریع‌تر انجام بشه.
# اینجا ما با انتخاب زیرمجموعه‌ای از ستون‌ها (بدون 'Date') این مفهوم رو ساده‌سازی کردیم.
column_based_data = sales_df[['ProductID', 'Region', 'SalesAmount']]

# در نهایت، داده‌های شبیه‌سازی شده مبتنی بر ستون رو چاپ می‌کنیم.
print("Simulated Column-based Data:")
print(column_based_data)
```

در این مثال، با انتخاب تنها ستون‌های مرتبط از داده‌های فروش اصلی، روشی شبیه به <b>ذخیره‌سازی مبتنی بر ستون</b> را شبیه‌سازی کرده‌ایم.

### 3.4.1. Terminology of data cube computation

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">اصطلاحات رایج در مکعب‌های داده و OLAP</span> 📚
</p>
<p style="font-size:16pt; text-align:right;">
در زمینه مکعب‌های داده و OLAP، اصطلاحات خاصی به‌طور رایج برای توصیف محاسبات و عملیات انجام‌شده بر روی داده‌های چندبعدی استفاده می‌شوند. درک این اصطلاحات برای انجام مؤثر تحلیل داده و استخراج <b>بینش‌های معنادار</b> ضروری است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">اصطلاحات محاسبه در مکعب داده</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>برش (Slice):</b>  
یک دید دو بعدی از مکعب داده که با ثابت کردن یک یا چند بعد در مقادیر خاص به دست می‌آید. این عمل به کاربران امکان می‌دهد بر یک زیرمجموعه از داده‌ها تمرکز کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>تاس‌زدن (Dice):</b>  
ایجاد یک زیرمکعب با انتخاب مقادیر خاص از دو یا چند بعد. این عمل با مشخص کردن شرایط در ابعاد متعدد، دید تحلیل را محدودتر می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>فشردن (Roll-up / Drill-Up):</b>  
تجمیع داده‌ها در سطح بالاتری از گرانولاریته با فرو ریختن یک یا چند بعد. این عمل به معنای حرکت از دید جزئی به دید خلاصه‌تر است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>جزئی‌سازی (Drill-down / Roll-Down):</b>  
باز کردن داده‌های تجمیع‌شده به سطح پایین‌تری از گرانولاریته با گسترش یک یا چند بعد. این عمل به معنای حرکت از دید خلاصه به دید جزئی‌تر است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>چرخش (Pivot / Rotate):</b>  
تغییر جهت‌گیری مکعب داده برای مشاهده آن از دیدگاهی دیگر. این عمل شامل جابه‌جایی سطرها و ستون‌های مکعب داده است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از محاسبات مکعب داده در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مجموعه داده فروش را در نظر بگیریم و چند عملیات محاسباتی از مکعب داده را با استفاده از کتابخانه Pandas انجام دهیم:
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها رو به عنوان یک مکعب داده (Data Cube) در نظر بگیرید که ابعاد مختلفی مثل 'Region' و 'Date' رو داره.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# 'DataFrame' اصلی رو چاپ می‌کنیم تا ببینیم داده‌ها چه شکلی هستن.
print("Original Sales Data:")
print(sales_df)

# این بخش، عملیات **برش (Slice)** رو انجام می‌ده.
# برش، انتخاب داده‌ها بر اساس یک بعد خاصه.
# اینجا، ما فقط داده‌های مربوط به ماه ژانویه (ماه ۱) رو انتخاب می‌کنیم.
january_sales = sales_df[sales_df['Date'].dt.month == 1]

# این بخش، عملیات **مکعب داده (Dice)** رو انجام می‌ده.
# مکعب داده، انتخاب داده‌ها بر اساس ترکیبی از چند بعده.
# اینجا، ما داده‌هایی رو انتخاب می‌کنیم که هم مربوط به منطقه 'North' باشن و هم مربوط به ماه ژانویه.
north_january_sales = sales_df[(sales_df['Region'] == 'North') & (sales_df['Date'].dt.month == 1)]

# در نهایت، نتایج هر دو عملیات رو چاپ می‌کنیم.
print("\nOLAP Slice Operation (January Sales):")
print(january_sales)

print("\nOLAP Dice Operation (North January Sales):")
print(north_january_sales)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، یک عملیات <b>برش (slice)</b> با انتخاب داده‌های فروش ماه ژانویه و یک عملیات <b>تاس‌زدن (dice)</b> با انتخاب داده‌های فروش منطقه شمال در همان ماه انجام شده است.
</p>
<p style="font-size:16pt; text-align:right;">
این اصطلاحات محاسبه در مکعب داده نشان می‌دهند که چگونه کاربران می‌توانند به‌طور مؤثر در داده‌های چندبعدی حرکت کرده و آن‌ها را تحلیل کنند. مثال ارائه‌شده، این عملیات‌ها را در یک سناریوی عملی با استفاده از <b>Pandas</b> نشان می‌دهد.
</p>

### 3.4.2. Data cube materialization: ideas

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">متریالیزه کردن مکعب داده</span> 🧱
</p>
<p style="font-size:16pt; text-align:right;">
متریالیزه کردن مکعب داده (Data Cube Materialization) به معنای <b>پیش‌محاسبه</b> و ذخیره داده‌های تجمیع‌شده در یک مکعب داده برای تسریع عملکرد پرس‌وجو است. ایده اصلی این است که نتایج عملیات تجمیع از قبل محاسبه و ذخیره شوند تا نسخه‌ای خلاصه‌شده از داده اصلی ایجاد شود. مکعب‌های داده متریالیزه می‌توانند زمان پاسخ به پرس‌وجوها را به‌ویژه در موارد پرس‌وجوهای تکراری یا پیچیده تحلیلی به‌طور چشمگیری کاهش دهند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">رویکردهای متریالیزه کردن مکعب داده</span> 🛠️
</p>

<p style="font-size:16pt; text-align:right;">
<b>متریالیزه کامل (Full Materialization):</b>  
محاسبه و ذخیره تمام ترکیبات ممکن از مقادیر تجمیع‌شده در مکعب داده. این روش سریع‌ترین زمان پاسخ به پرس‌وجو را فراهم می‌کند، اما ممکن است به فضای ذخیره‌سازی بسیار زیادی نیاز داشته باشد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>متریالیزه جزئی (Partial Materialization):</b>  
محاسبه و ذخیره انتخابی از تجمیع‌ها یا زیرمجموعه‌های مکعب داده بر اساس الگوهای پیش‌بینی‌شده پرس‌وجو. این روش تعادلی بین <b>عملکرد پرس‌وجو</b> و <b>محدودیت فضای ذخیره‌سازی</b> ایجاد می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>متریالیزه Top-N (Top-N Materialization):</b>  
فقط N تجمیع یا ترکیب برتر متریالیزه می‌شوند، که در آن N نشان‌دهنده پرس‌وجوهای پرتکرار یا بینش‌های باارزش است. این روش برای <b>موارد استفاده رایج</b> بهینه‌سازی شده است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>متریالیزه افزایشی (Incremental Materialization):</b>  
به‌روزرسانی افزایشی مکعب داده متریالیزه هنگامی که داده‌های جدیدی در دسترس قرار می‌گیرند. این امر تضمین می‌کند که مکعب متریالیزه همیشه با داده‌های پایه به‌روز باقی بماند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از متریالیزه کردن مکعب داده در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
اگرچه متریالیزه کردن واقعی معمولاً در پایگاه‌های داده تخصصی انجام می‌شود، اما می‌توانیم با استفاده از کتابخانه Pandas در پایتون، یک سناریوی <b>متریالیزه جزئی</b> را شبیه‌سازی کنیم:
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای نمایش مفهوم **ماده‌سازی جزئی (Partial Materialization)** در سیستم‌های OLAP استفاده می‌شن.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **ماده‌سازی جزئی (Partial Materialization)** رو شبیه‌سازی می‌کنه.
# توی سیستم‌های OLAP، ماده‌سازی یعنی نتایج پرس‌وجوها یا محاسبات رو از قبل حساب کرد و ذخیره کرد.
# ماده‌سازی جزئی به این معنیه که فقط بخشی از مکعب داده (مثلاً تجمیع‌های پرتکرار) ماده‌سازی بشه، نه همه ترکیب‌های ممکن.
# اینجا ما میانگین فروش رو بر اساس 'Region' از قبل محاسبه و ذخیره می‌کنیم.
materialized_cube = sales_df.groupby('Region')['SalesAmount'].mean().reset_index()

# در نهایت، مکعب ماده‌سازی شده رو چاپ می‌کنیم.
print("Materialized Cube (Average Sales by Region):")
print(materialized_cube)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، با محاسبه و ذخیره میانگین مقدار فروش برای هر منطقه، یک <b>متریالیزه جزئی</b> را شبیه‌سازی کرده‌ایم. در سناریوهای واقعی، متریالیزه کردن ممکن است شامل تجمیع‌های پیچیده‌تر و بهینه‌سازی‌های ذخیره‌سازی باشد.
</p>
<p style="font-size:16pt; text-align:right;">
این مثال مفهوم <b>متریالیزه کردن مکعب داده</b> را نشان می‌دهد و چگونگی تعادل بین <b>عملکرد پرس‌وجو</b> و <b>فضای ذخیره‌سازی</b> را توضیح می‌دهد.
</p>

### 3.4.3. OLAP server architectures: ROLAP vs. MOLAP vs. HOLAP

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معماری‌های سرور OLAP</span> 🖥️
</p>
<p style="font-size:16pt; text-align:right;">
معماری‌های سرور OLAP (پردازش تحلیلی آنی) برای پشتیبانی از تحلیل و پرس‌وجوی داده‌های چندبعدی طراحی شده‌اند. سه نوع اصلی معماری OLAP وجود دارد: <b>ROLAP</b> (OLAP نسبی)، <b>MOLAP</b> (OLAP چندبعدی) و <b>HOLAP</b> (OLAP ترکیبی). هر معماری ویژگی‌ها و موارد استفاده خاص خود را دارد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">انواع معماری‌های OLAP</span> 🔬
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. ROLAP (Relational OLAP):</b>  
سرورهای ROLAP داده‌ها را در پایگاه‌های داده نسبی (RDBMS) ذخیره می‌کنند و نتایج چندبعدی را به‌صورت پویا و با استفاده از پرس‌وجوهای SQL تولید می‌کنند. تجمیع‌ها و محاسبات در همان سیستم مدیریت پایگاه داده انجام می‌شوند.  
<b>کاربرد:</b> مناسب برای مجموعه‌داده‌های بزرگ که در آن‌ها تولید پیش‌فرض تمام ترکیبات در یک مکعب چندبعدی عملی نیست.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. MOLAP (Multidimensional OLAP):</b>  
سرورهای MOLAP داده‌ها را در قالب اختصاصی و چندبعدی ذخیره می‌کنند که برای پردازش تحلیلی بهینه‌سازی شده است. داده‌ها از قبل تجمیع شده و در ساختار مکعب ذخیره می‌شوند و این امر منجر به <b>سرعت بالای پاسخ به پرس‌وجو</b> می‌شود.  
<b>کاربرد:</b> برای سناریوهایی که عملکرد سریع پرس‌وجو حیاتی است و فضای ذخیره‌سازی محدودیت اصلی محسوب نمی‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. HOLAP (Hybrid OLAP):</b>  
سرورهای HOLAP ترکیبی از ROLAP و MOLAP هستند. تجمیع‌ها و خلاصه‌ها برای داده‌های پرتکرار در یک مکعب چندبعدی (MOLAP) ذخیره می‌شوند، در حالی که داده‌های جزئی در پایگاه داده نسبی (ROLAP) نگهداری می‌شوند.  
<b>کاربرد:</b> تعادلی بین <b>عملکرد پرس‌وجو</b> و <b>مصرف فضای ذخیره‌سازی</b> ایجاد می‌کند و از مزایای هر دو معماری بهره می‌برد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از ROLAP، MOLAP و HOLAP در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از <b>Pandas</b> همراه با یک پایگاه داده نسبی (مانند SQLite) برای شبیه‌سازی <b>ROLAP</b> استفاده می‌کنیم، از آرایه‌های چندبعدی با <b>xarray</b> یا <b>NumPy</b> برای مدل‌سازی <b>MOLAP</b> استفاده می‌کنیم، و ترکیبی از هر دو رویکرد را برای شبیه‌سازی یک سیستم <b>HOLAP</b> به کار می‌بریم — جایی که داده‌های خلاصه‌شده در حافظه نگهداری می‌شوند و داده‌های جزئی از پایگاه داده واکشی می‌شوند.
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای شبیه‌سازی و مقایسه سه معماری ذخیره‌سازی و پردازش داده در سیستم‌های OLAP استفاده می‌شن:
# ROLAP (Relational OLAP)، MOLAP (Multidimensional OLAP)، و HOLAP (Hybrid OLAP).
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **ROLAP (Relational OLAP)** رو شبیه‌سازی می‌کنه.
# توی ROLAP، داده‌ها در پایگاه‌های داده رابطه‌ای (Relational DB) نگه‌داری می‌شن
# و از دستورات SQL برای تجزیه و تحلیل چندبعدی استفاده می‌شه.
# برای شبیه‌سازی کوئری، از تابع `groupby()` پانداس استفاده می‌کنیم که مشابه `GROUP BY` در SQL عمل می‌کنه.
rolap_result = sales_df.groupby('Region')['SalesAmount'].sum().reset_index()
print("نتایج شبیه‌سازی ROLAP:")
print(rolap_result)
print("="*30)

# این بخش، **MOLAP (Multidimensional OLAP)** رو شبیه‌سازی می‌کنه.
# توی MOLAP، داده‌ها در قالب یک مکعب چندبعدی (Multidimensional Cube) ذخیره می‌شن.
# این ساختار، پرس‌وجوهای تحلیلی رو خیلی سریع‌تر می‌کنه.
# ما از تابع `pivot_table` پانداس استفاده می‌کنیم که مکعب داده‌ها رو شبیه‌سازی می‌کنه.
molap_result = pd.pivot_table(
    sales_df,
    values='SalesAmount',
    index=['Date', 'ProductID'],
    columns=['Region'],
    aggfunc='sum'
)
print("نتایج شبیه‌سازی MOLAP:")
print(molap_result)
print("="*30)

# این بخش، **HOLAP (Hybrid OLAP)** رو شبیه‌سازی می‌کنه.
# HOLAP ترکیبی از ROLAP و MOLAP هست و سعی می‌کنه نقاط قوت هر دو رو ترکیب کنه.
# ما از یک پایوت تیبل ساده برای نمایش این معماری ترکیبی استفاده می‌کنیم.
holap_result = pd.pivot_table(
    sales_df,
    values='SalesAmount',
    index=['Date', 'ProductID'],
    columns=['Region'],
    aggfunc='sum'
)
print("نتایج شبیه‌سازی HOLAP:")
print(holap_result)
print("="*30)
```

### 3.4.4. General strategies for data cube computation

<p style="font-size:16pt; text-align:right;">
محاسبه مکعب داده شامل راهبردهایی برای تولید و مدیریت کارآمد مکعب‌های داده چندبعدی است که برای پردازش تحلیلی آنی (OLAP) و داده‌کاوی ضروری هستند. در ادامه برخی از راهبردهای کلی برای محاسبه مکعب داده ارائه شده است:

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:16pt; text-align:right;">
<b>1. تجمیع و پیش‌محاسبه (Aggregation and Precomputation):</b>  
محاسبه و ذخیره از پیش مقادیر تجمیع‌شده در مکعب داده به منظور تسریع عملکرد پرس‌وجو.  
<br><br>
<b>مثال در پایتون:</b>  
استفاده از Pandas برای محاسبه و ذخیره مقادیر تجمیع‌شده در مرحله اولیه و قبل از اجرای پرس‌وجوها.
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای نمایش مفهوم **پیش‌محاسبه و ذخیره‌سازی داده‌های تجمیع‌شده (Precomputed Aggregation)** استفاده می‌شن.
# این کار یکی از روش‌های بهینه‌سازی عملکرد در سیستم‌های تحلیلی (مانند OLAP) هست.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **تجمیع داده‌ها بر اساس بعد 'Region'** رو انجام می‌ده.
# تجمیع (Aggregation) یعنی محاسبه مقادیر خلاصه‌ای مثل جمع، میانگین یا شمارش.
# اینجا ما جمع کل فروش رو برای هر منطقه از قبل محاسبه می‌کنیم.
# این کار به جای اینکه هر بار کوئری اجرا بشه داده‌ها رو جمع کنه، عملکرد رو بهبود می‌ده.
# این روش شبیه به ماده‌سازی (Materialization) در سیستم‌های OLAP هست.
aggregated_data = sales_df.groupby('Region')['SalesAmount'].sum().reset_index()

# در نهایت، داده‌های پیش‌محاسبه شده رو چاپ می‌کنیم.
print("Precomputed Aggregated Data:")
print(aggregated_data)
```

<p style="font-size:16pt; text-align:right;">
<b>2. ایندکس‌گذاری و پارتیشن‌بندی (Indexing and Partitioning):</b>  
استفاده از ایندکس‌گذاری و پارتیشن‌بندی داده‌ها برای بهینه‌سازی محاسبه و ذخیره‌سازی مکعب داده. این روش‌ها دسترسی سریع‌تر به زیرمجموعه‌های داده و کاهش بار محاسباتی را فراهم می‌کنند.  
<br><br>
<b>مثال در پایتون:</b>  
استفاده از تکنیک‌های ایندکس‌گذاری کارآمد و پارتیشن‌بندی در Pandas یا کتابخانه‌های دیگر مانند Dask یا PyArrow.
</p>

</div>

```python
import pandas as pd

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای نمایش مفهوم **ایندکس‌گذاری (Indexing)** در سیستم‌های مدیریت داده و تحلیلی استفاده می‌شن.
# ایندکس‌گذاری یک تکنیک بهینه‌سازی هست که سرعت دسترسی به داده‌ها رو، به‌ویژه در عملیات فیلتر و جستجو، به‌شدت افزایش می‌ده.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **ایندکس‌گذاری روی ستون 'Region'** رو انجام می‌ده.
# با استفاده از `set_index`، ستون 'Region' به عنوان ایندکس دیتافریم در نظر گرفته می‌شه.
# این کار باعث می‌شه جستجو و فیلتر داده‌های مربوط به یک منطقه خاص (مثل 'North') خیلی سریع‌تر انجام بشه.
# این روش مشابه ایندکس‌های موجود در پایگاه‌های داده رابطه‌ای هست و برای بهبود عملکرد در عملیات تکراری بسیار مفیده.
sales_df.set_index('Region', inplace=True)

# در نهایت، داده‌های ایندکس‌شده رو چاپ می‌کنیم تا ساختار جدید رو ببینیم.
print("Indexed Sales Data:")
print(sales_df)
```

<b>3. پردازش موازی (Parallel Processing):</b>  
توزیع وظایف محاسبه مکعب داده بین چندین پردازنده یا گره به منظور اجرای موازی. این روش زمان محاسبه را به‌طور چشمگیری کاهش می‌دهد، به‌ویژه برای مجموعه‌داده‌های بزرگ.  
<br><br>
<b>مثال در پایتون:</b>  
استفاده از کتابخانه‌های پردازش موازی مانند <b>Dask</b> یا ماژول <b>multiprocessing</b> در پایتون برای موازی‌سازی عملیات تجمیع و محاسبه مکعب.

```python
import pandas as pd
import dask.dataframe as dd

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'dask' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'dask' برای پردازش موازی و مقیاس‌پذیر داده‌های بزرگ استفاده می‌شه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای نمایش مفهوم **پردازش موازی (Parallel Processing)** با استفاده از کتابخانه Dask استفاده می‌شن.
# Dask اجازه می‌ده تا عملیات روی دیتافریم‌ها به صورت موازی و روی چندین هسته انجام بشه.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، تبدیل دیتافریم pandas به دیتافریم Dask رو انجام می‌ده.
# Dask داده‌ها رو به بخش‌های کوچک‌تر (partitions) تقسیم می‌کنه و امکان پردازش موازی رو فراهم می‌کنه.
# این ویژگی به‌ویژه زمانی مفیده که با داده‌های خیلی بزرگ کار می‌کنیم و می‌خوایم از تمام قدرت CPU استفاده کنیم.
dask_sales_df = dd.from_pandas(sales_df, npartitions=2)

# اینجا یک عملیات تجمیع (groupby و sum) روی داده‌ها انجام می‌دیم.
# این عملیات به صورت موازی روی پارتیشن‌ها اجرا می‌شه و با فراخوانی `.compute()` نتیجه نهایی تولید می‌شه.
# این روش می‌تونه عملکرد رو در داده‌های حجیم به‌طور چشمگیری بهبود بده.
result = dask_sales_df.groupby('Region')['SalesAmount'].sum().compute()

# در نهایت، نتیجه عملیات موازی رو چاپ می‌کنیم.
print("Result of Parallel Processing:")
print(result)
```

### 3.5.1. Multiway array aggregation for full cube computation

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">جمع‌بندی آرایه چندطرفه برای محاسبه کامل مکعب</span> 📦
</p>
<p style="font-size:16pt; text-align:right;">
روش <b>جمع‌بندی آرایه چندطرفه</b> (Multiway Array Aggregation) شامل تجمیع داده‌ها در یک آرایه چندبعدی است که امکان محاسبه کارآمد و جامع کل مکعب داده را فراهم می‌کند. این رویکرد به‌ویژه در کار با مجموعه‌داده‌های بزرگ و پرس‌وجوهای تحلیلی پیچیده بسیار مفید است. آرایه‌های چندطرفه (که به آن‌ها تنسور نیز گفته می‌شود)، نمایشی فشرده و ساختاریافته از داده‌های چندبعدی ارائه می‌دهند و تجمیع آن‌ها امکان محاسبه همزمان ابعاد مختلف مکعب را ساده و کارآمد می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی جمع‌بندی آرایه چندطرفه</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. نمایش تنسوری (Tensor Representation):</b>  
تنسور یک آرایه چندبعدی است که می‌تواند داده‌ها را در بیش از دو بعد نمایش دهد. در زمینه مکعب‌های داده، یک تنسور سه‌بعدی می‌تواند داده‌ها را در ابعادی مانند زمان، محصول و منطقه نمایش دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. عملیات تجمیع (Aggregation Operations):</b>  
عملیات تجمیع روی تنسورها شامل فرو ریختن یا ترکیب مقادیر در امتداد یک یا چند بعد است. در مکعب داده، این عملیات ممکن است شامل <b>جمع</b>، <b>میانگین‌گیری</b> یا سایر توابع تجمیعی باشد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. محاسبه کارآمد (Efficient Computation):</b>  
جمع‌بندی آرایه چندطرفه راهی کارآمد برای انجام محاسبه کامل مکعب فراهم می‌کند، بدون نیاز به ایجاد صریح و ذخیره یک مکعب چندبعدی. تجمیع‌ها به‌صورت پویا انجام می‌شوند و این امر <b>نیاز به فضای ذخیره‌سازی</b> را به حداقل می‌رساند.

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از جمع‌بندی آرایه چندطرفه در پایتون</span> 🐍

<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>NumPy</b> برای شبیه‌سازی یک تنسور سه‌بعدی استفاده کنیم که داده‌های فروش را نشان می‌دهد و سپس عملیات تجمیع را روی آن انجام دهیم:
</p>

```python
import numpy as np

# این قسمت اول کده. ما کتابخونه 'numpy' رو وارد می‌کنیم که برای کار با آرایه‌های چندبعدی و محاسبات عددی عالیه.

# این بخش، یک **تنسور سه‌بعدی** رو برای شبیه‌سازی داده‌های فروش در قالب مکعب داده (Data Cube) ایجاد می‌کنه.
# ابعاد تنسور به ترتیب زمان (مثلاً 3 ماه)، محصول (مثلاً 4 محصول)، و منطقه (مثلاً 2 منطقه) هستن.
# این ساختار نمایش چندبعدی داده‌ها در سیستم‌های OLAP و تحلیل داده‌های تجاری رو شبیه‌سازی می‌کنه.
sales_tensor = np.random.randint(100, 1000, size=(3, 4, 2))

# تنسور شبیه‌سازی شده رو چاپ می‌کنیم تا ساختار داده‌های چندبعدی رو ببینیم.
print("Simulated Sales Tensor:")
print(sales_tensor)

# این بخش، عملیات **تجمیع (Aggregation)** روی تمام ابعاد تنسور انجام می‌ده.
# با استفاده از `np.sum` و مشخص کردن `axis=(0, 1, 2)`، جمع کل فروش روی هر سه بعد (زمان، محصول، منطقه) محاسبه می‌شه.
# این عملیات معادل به دست آوردن "کل فروش" در یک سیستم تحلیلی هست.
# در عمل، این نوع تجمیع‌ها برای گزارش‌های خلاصه و داشبوردها کاربرد فراوانی دارن.
total_sales = np.sum(sales_tensor, axis=(0, 1, 2))

# نتیجه نهایی، یعنی کل فروش تجمیع‌شده، رو چاپ می‌کنیم.
print("\nTotal Sales (Aggregated):", total_sales)
```

### 3.5.2. BUC: computing iceberg cubes from the apex cuboid downward

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">روش BUC (محاسبه از بالا به پایین)</span> 🧊
</p>
<p style="font-size:16pt; text-align:right;">
BUC (محاسبه از بالا به پایین) یک تکنیک برای محاسبه کارآمد <b>مکعب‌های آبشاری</b> (Iceberg Cubes) از جوهره بالایی (Apex Cuboid) — که خلاصه‌ترین سطح را نشان می‌دهد — به سمت پایین است. این روش تنها تجمیع‌هایی را محاسبه می‌کند که شرط از پیش تعیین‌شده‌ای را برآورده کنند، که به آن <b>شرط آبشاری</b> (Iceberg Condition) گفته می‌شود. با شروع از خلاصه‌ترین سطح و حرکت تدریجی به سمت سطوح جزئی‌تر، BUC پیچیدگی محاسباتی را کاهش داده و فقط بر تجمیع‌های معنادار و مرتبط تمرکز می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی BUC</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. شرط آبشاری (Iceberg Condition):</b>  
BUC از یک شرط آبشاری برای تعیین اینکه آیا یک تجمیع به اندازه کافی معنادار است یا نه، استفاده می‌کند. این شرط معمولاً شامل آستانه‌ای تعریف‌شده توسط کاربر است، مانند <b>حداقل پشتیبانی</b> در زمینه کاوش قواعد ارتباطی.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. جوهره بالایی (Apex Cuboid):</b>  
جوهره بالایی، سطح خلاصه‌ترین تجمیع در مکعب داده است که نمایی بسیار کلی و فشرده از داده‌ها را ارائه می‌دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. محاسبه از بالا به پایین (Bottom-Up Computation):</b>  
BUC از جوهره بالایی شروع می‌کند و به تدریج تجمیع‌ها را به سمت سطوح پایین‌تر محاسبه می‌کند و از انجام محاسبات غیرضروری برای ترکیباتی که شرط آبشاری را برآورده نمی‌کنند، اجتناب می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از BUC در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مجموعه داده کوچک را شبیه‌سازی کنیم و از الگوریتم BUC برای محاسبه مکعب‌های آبشاری بر اساس یک آستانه مشخص استفاده کنیم:
</p>

</div>

```python
import pandas as pd
import numpy as np

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها برای اجرای الگوریتم **BUC (Bottom-Up Computation)** و ساخت مکعب‌های یخچالی (Iceberg Cube) استفاده می‌شن.
# مکعب یخچالی فقط تجمیع‌هایی رو نگه می‌داره که یک شرط آستانه‌ای (مثل مجموع فروش بیشتر از ۱۰۰۰) رو برآورده کنن.
# این روش از ذخیره‌سازی تجمیع‌های بی‌معنی جلوگیری می‌کنه و فضای ذخیره‌سازی و زمان پردازش رو بهینه می‌کنه.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، پیاده‌سازی اصلاح‌شده و دقیق‌تر الگوریتم **BUC (Bottom-Up Computation)** رو شامل می‌شه.
# برخلاف نسخه قبلی که از بالا به پایین حرکت می‌کرد، این نسخه با رویکردی ساختاریافته‌تر، از جزئیات شروع نمی‌کنه بلکه به‌صورت بازگشتی و با مدیریت صحیح ابعاد، تجمیع‌ها رو بررسی می‌کنه.
# هدف اینه که تمام ترکیب‌های ممکن از ابعاد رو به‌صورت هوشمند و بدون تکرار اضافی پیمایش کنیم.
# شرط "یخچالی" در هر سطح چک می‌شه و اگر مقدار تجمیع از آستانه کمتر باشه، شاخه قطع می‌شه (pruning).
def buc(data, dimensions, measures, threshold):
    """
    این تابع یک نسخه ساده‌شده از الگوریتم BUC رو برای ساخت مکعب یخچالی پیاده‌سازی می‌کنه.
    
    ورودی‌ها:
        data (pd.DataFrame): دیتافریم ورودی شامل داده‌های خام.
        dimensions (list): لیستی از نام ستون‌هایی که باید روی اون‌ها تجمیع انجام بشه.
        measures (str): نام ستونی که قراره مقدارش جمع بشه (مثلاً 'SalesAmount').
        threshold (float): حداقل مقداری که یک تجمیع باید داشته باشه تا ذخیره بشه.
    
    خروجی:
        list: لیستی از دیکشنری‌ها که هر کدوم نماینده یک تجمیع معتبر هست.
    """
    result = []

    def helper(df, current_dimensions, current_levels):
        # محاسبه مجموع سنجه برای داده‌های فعلی
        current_sum = df[measures].sum()

        # بررسی شرط یخچالی: آیا این تجمیع ارزش ذخیره شدن رو داره؟
        if current_sum >= threshold:
            # ساخت یک سطر خروجی که شامل مقادیر ابعاد و مقدار تجمیع هست
            agg_row = dict(zip(dimensions, current_levels))
            agg_row['TotalSales'] = current_sum
            result.append(agg_row)

            # اگر هنوز ابعادی برای پردازش باقی مونده، به سطح بعدی حرکت می‌کنیم
            if current_dimensions:
                first_dimension = current_dimensions[0]
                remaining_dimensions = current_dimensions[1:]
                
                # برای هر مقدار منحصربه‌فرد در بعد جاری، زیرمجموعه مربوطه رو جدا می‌کنیم
                for value in df[first_dimension].unique():
                    next_df = df[df[first_dimension] == value]
                    next_levels = current_levels + [value]
                    # فراخوانی بازگشتی برای ادامه تجزیه و تحلیل
                    helper(next_df, remaining_dimensions, next_levels)

    # شروع الگوریتم از سطح کلی (بدون تفکیک اولیه)
    # در اولین فراخوانی، هیچ بعدی اعمال نشده و همه داده‌ها در نظر گرفته می‌شن
    helper(sales_df, dimensions, [])
    
    return result

# اجرای الگوریتم BUC با مشخص کردن ابعاد، سنجه و آستانه.
# فقط تجمیع‌هایی که مجموع فروششون حداقل ۱۰۰۰ باشه، در نتیجه ظاهر می‌شن.
buc_result = buc(sales_df, dimensions=['Date', 'ProductID', 'Region'], measures='SalesAmount', threshold=1000)

# در نهایت، نتایج تجمیع‌هایی که شرط یخچالی رو داشتن رو چاپ می‌کنیم.
print("BUC Result:")
print(buc_result)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، تابع <b>buc</b> یک مجموعه داده، ابعاد، معیارها و یک آستانه را به عنوان ورودی گرفته و به‌صورت بازگشتی مکعب‌های آبشاری را بر اساس الگوریتم BUC محاسبه می‌کند. نتیجه تنها شامل تجمیع‌هایی است که شرط آستانه مشخص‌شده را برآورده می‌کنند.
</p>
<p style="font-size:16pt; text-align:right;">
این مثال نشان می‌دهد که چگونه الگوریتم <b>BUC</b> به‌طور کارآمد فضای جست‌وجو را کاهش می‌دهد و تنها بر تجمیع‌های مرتبط تمرکز می‌کند. برای روشن‌سازی، از یک مجموعه داده کوچک استفاده شده است.
</p>

### 3.5.3. Precomputing shell fragments for fast high-dimensional OLAP

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">پیش‌محاسبه تکه‌های پوسته</span> 🛠️
</p>
<p style="font-size:16pt; text-align:right;">
پیش‌محاسبه تکه‌های پوسته (Shell Fragments) یک راهبرد برای اجرای سریع OLAP (پردازش تحلیلی آنی) در محیط‌های با ابعاد بالا است. در سناریوهای OLAP با ابعاد فراوان، انجام تجمیع‌ها به‌صورت پویا از نظر محاسباتی بسیار پرهزینه می‌شود. با پیش‌محاسبه تکه‌های پوسته، تجمیع‌های مربوط به زیرمجموعه‌هایی از ابعاد محاسبه و ذخیره می‌شوند و ساختارهای پوسته‌ای ایجاد می‌شوند که می‌توانند به‌صورت کارآمدی ترکیب شوند تا به پرس‌وجوهای پیچیده پاسخ دهند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی پیش‌محاسبه تکه‌های پوسته</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. ساختار پوسته (Shell Structure):</b>  
پوسته، نمایانگر یک زیرمجموعه از ابعاد در مکعب داده است. در OLAPهای با ابعاد بالا، پیش‌محاسبه و ذخیره تمام ترکیبات ممکن ابعاد عملی نیست. به جای آن، تکه‌های پوسته برای زیرمجموعه‌هایی از ابعاد محاسبه می‌شوند.
</p>
<p style="font-size:16pt; text-align:right;">
<b>2. پیش‌محاسبه تکه‌ها (Fragment Precomputation):</b>  
تجمیع‌ها برای هر تکه پوسته از پیش محاسبه شده و اطلاعات خلاصه‌شده برای ترکیبات خاصی از ابعاد ذخیره می‌شوند.
</p>
<p style="font-size:16pt; text-align:right;">
<b>3. پردازش کارآمد پرس‌وجو (Efficient Query Processing):</b>  
در حین اجرای پرس‌وجو، تکه‌های پوسته به‌صورت کارآمدی ترکیب می‌شوند تا به پرس‌وجوهای پیچیده OLAP پاسخ داده شود، بدون نیاز به تجمیع روی کل فضای با ابعاد بالا.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از پیش‌محاسبه تکه‌های پوسته در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مجموعه داده را شبیه‌سازی کنیم و نحوه پیش‌محاسبه تکه‌های پوسته را در یک سناریوی OLAP با ابعاد بالا نشان دهیم:
</p>

</div>

```python
import pandas as pd
import itertools

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه با ابعاد بالاست.
# این داده‌ها شامل ابعاد مختلفی مثل 'Date', 'ProductID', 'Region' و 'Category' هستن.
# در سیستم‌های OLAP، با افزایش تعداد ابعاد، تعداد ترکیبات ممکن برای تجمیع به‌سرعت رشد می‌کنه (مشهور به "فاجعه ابعاد" یا Dimensionality Curse).
# برای مقابله با این چالش، از تکنیک‌هایی مثل **قطعه‌های پوسته (Shell Fragments)** استفاده می‌شه.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Clothing'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **قطعه‌های پوسته (Shell Fragments)** رو پیش‌محاسبه می‌کنه.
# قطعه‌های پوسته یک تکنیک بهینه‌سازی در OLAP هستن که در اون تجمیع‌های زیرمجموعه‌ای از ابعاد (مثلاً تمام ترکیب‌های ممکن 'Date' و 'ProductID') از قبل محاسبه و ذخیره می‌شن.
# این کار به جای ساخت تمام ترکیبات ممکن (که با ابعاد بالا غیرعملی می‌شه)، فقط زیرمجموعه‌های کوچک و پرتکرار رو ماده‌سازی می‌کنه.
# بعداً می‌تونیم از این قطعات برای ساخت سریع تجمیع‌های مرکب استفاده کنیم.
def precompute_shell_fragments(data, dimensions, measures):
    fragments = {}

    # تولید تمام ترکیب‌های ممکن از ابعاد (از اندازه ۱ تا کل ابعاد)
    # مثلاً برای ['Date', 'ProductID'] ترکیب‌های ('Date',) و ('ProductID',) و ('Date', 'ProductID') ساخته می‌شن.
    all_combinations = [list(combination) for r in range(1, len(dimensions) + 1) for combination in itertools.combinations(dimensions, r)]

    for combination in all_combinations:
        fragment_key = tuple(combination)
        # تجمیع داده‌ها بر اساس ترکیب فعلی ابعاد
        fragment_data = data.groupby(combination)[measures].sum().reset_index()
        fragments[fragment_key] = fragment_data

    return fragments

# اجرای تابع برای ابعاد ['Date', 'ProductID']
# اینجا ما فقط یک زیرمجموعه کوچک از ابعاد رو برای نمایش قطعه‌های پوسته انتخاب کردیم.
fragments = precompute_shell_fragments(sales_df, dimensions=['Date', 'ProductID'], measures='SalesAmount')

# در نهایت، یکی از قطعات پیش‌محاسبه شده رو چاپ می‌کنیم تا ساختارش رو ببینیم.
print("Precomputed Shell Fragment for ['Date', 'ProductID']:")
print(fragments[('Date', 'ProductID')])
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، تابع <b>precompute_shell_fragments</b> تمام ترکیبات ممکن ابعاد را تولید کرده و تجمیع‌ها را برای هر تکه پوسته از پیش محاسبه می‌کند. نتیجه یک دیکشنری است که کلیدهای آن، تاپل‌هایی از ترکیبات ابعاد و مقادیر آن‌ها دیتافریم‌هایی شامل تجمیع‌های پیش‌محاسبه‌شده هستند.
</p>
<p style="font-size:16pt; text-align:right;">
این مثال مفهوم <b>پیش‌محاسبه تکه‌های پوسته</b> را نشان می‌دهد و چگونه به تسریع پردازش OLAP در فضاهای با ابعاد بالا کمک می‌کند.
</p>

### 3.5.4. Efficient processing of OLAP queries using cuboids

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">پردازش کارآمد پرس‌وجوهای OLAP با استفاده از جوهره‌ها</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
پردازش کارآمد پرس‌وجوهای OLAP (پردازش تحلیلی آنی) با استفاده از <b>جوهره‌ها</b> (Cuboids)، شامل ایجاد و ذخیره استراتژیک ساختارهای داده خلاصه‌شده است. جوهره‌ها ترکیبات مختلف ابعاد و سلسله مراتب در یک مکعب داده را نشان می‌دهند و با پیش‌محاسبه و ذخیره تجمیع‌ها در سطوح مختلف گرانولاریته، امکان پاسخ‌گویی سریع‌تر به پرس‌وجوها را فراهم می‌کنند. با سازماندهی داده‌ها در قالب جوهره‌ها، سیستم‌های OLAP می‌توانند به‌صورت کارآمد از سطوح مختلف جزئیات عبور کنند و نیاز به تجمیع‌های پویا را به حداقل برسانند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی پردازش کارآمد با استفاده از جوهره‌ها</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. تعریف جوهره (Cuboid Definition):</b>  
جوهره زیرمجموعه‌ای از یک مکعب داده است که ترکیب خاصی از ابعاد و سلسله مراتب را شامل می‌شود و نمایی خلاصه‌شده از داده‌ها را ارائه می‌دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. پیش‌محاسبه تجمیع‌ها (Precomputation of Aggregations):</b>  
تجمیع‌ها در سطوح مختلف گرانولاریته درون هر جوهره از پیش محاسبه و ذخیره می‌شوند. این پیش‌محاسبه، پردازش پرس‌وجوها را به‌طور چشمگیری تسریع می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. حرکت در سلسله مراتب (Hierarchy Navigation):</b>  
جوهره‌ها امکان حرکت کارآمد در سطوح مختلف سلسله مراتب و ابعاد را فراهم می‌کنند و به کاربران اجازه می‌دهند با عملیات <b>جزئی‌سازی (Drill-down)</b> یا <b>فشردن (Roll-up)</b> داده‌ها را در سطوح مختلف جزئیات کاوش کنند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از پردازش کارآمد با استفاده از جوهره‌ها در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مجموعه داده را شبیه‌سازی کنیم و مفهوم جوهره‌ها را برای پردازش کارآمد پرس‌وجوهای OLAP نمایش دهیم:
</p>

</div>

```python
import pandas as pd
import itertools

# این قسمت اول کده. ما کتابخونه 'pandas' رو وارد می‌کنیم که برای کار با داده‌ها تو قالب جدول عالیه.

# این دیکشنری حاوی داده‌های فروش نمونه ماست.
# این داده‌ها شامل چندین بعد (مثلاً 'Date', 'ProductID', 'Region', 'Category') و یک سنجه ('SalesAmount') هستن.
# این ساختار برای شبیه‌سازی مفاهیم OLAP مثل **کیوبوید (Cuboid)** و پردازش پرس‌وجوها با تجمیع‌های از قبل محاسبه شده استفاده می‌شه.
data = {
    'Date': ['2022-01-01', '2022-01-02', '2022-02-01', '2022-02-02'],
    'ProductID': [1, 2, 1, 2],
    'Region': ['North', 'South', 'North', 'South'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Clothing'],
    'SalesAmount': [500, 300, 700, 450]
}

sales_df = pd.DataFrame(data)
# این خط، ستون 'Date' رو به نوع داده زمانی (datetime) تبدیل می‌کنه
# تا بتونیم عملیات مربوط به زمان رو روش انجام بدیم.
sales_df['Date'] = pd.to_datetime(sales_df['Date'])

# این بخش، **مدیریت پرس‌وجوهای OLAP با استفاده از کیوبوید (Cuboid)** رو پیاده‌سازی می‌کنه.
# در سیستم‌های OLAP، یک "کیوبوید" به معنای یک سطح از تجمیع در مکعب چندبعدی هست.
# هر ترکیب از ابعاد، یک کیوبوید مجزا تشکیل می‌ده.
# به جای محاسبه پویا هر پرس‌وجو، کیوبویدها از قبل (یا به صورت پویا اما ذخیره‌شده) تولید می‌شن تا پاسخ‌دهی سریع‌تر بشه.
# این روش به "ماده‌سازی" (Materialization) در OLAP معروفه.
def process_olap_queries_with_cuboids(data, dimensions, measures):
    # ایجاد کیوبوید پایه (بدون بعد) — معادل کل فروش
    base_cuboid = pd.DataFrame({measures: [data[measures].sum()]})

    # دیکشنری برای ذخیره تمام کیوبویدهای تولید شده
    cuboids = {(): base_cuboid}

    # برای هر ترکیب ممکن از ابعاد، یک کیوبوید ایجاد می‌کنیم
    for r in range(1, len(dimensions) + 1):
        for combination in itertools.combinations(dimensions, r):
            # تجمیع داده‌ها بر اساس ترکیب فعلی ابعاد
            cuboid_data = data.groupby(list(combination))[measures].sum().reset_index()
            cuboids[combination] = cuboid_data

    # مثال از یک پرس‌وجوی OLAP: مجموع فروش بر اساس دسته‌بندی در منطقه 'South'
    # اینجا از کیوبوید ('Region', 'Category') استفاده می‌کنیم و فقط ردیف‌های مربوط به 'South' رو فیلتر می‌کنیم.
    query_result = cuboids[('Region', 'Category')].loc[
        cuboids[('Region', 'Category')]['Region'] == 'South'
    ]

    return query_result

# اجرای تابع با مشخص کردن تمام ابعاد و سنجه مورد نظر
result = process_olap_queries_with_cuboids(
    sales_df,
    dimensions=['Date', 'ProductID', 'Region', 'Category'],
    measures='SalesAmount'
)

# در نهایت، نتیجه پرس‌وجوی OLAP رو چاپ می‌کنیم.
print("OLAP Query Result:")
print(result)
```

در این مثال، تابع <b>process_olap_queries_with_cuboids</b> جوهره‌های مربوط به ترکیبات مختلف ابعاد را ایجاد کرده و تجمیع‌ها را از پیش محاسبه می‌کند. سپس یک پرس‌وجوی OLAP انجام می‌شود تا کل فروش در هر دسته‌بندی در منطقه «جنوب» با استفاده از جوهره مربوط به ('Region', 'Category') استخراج شود.
