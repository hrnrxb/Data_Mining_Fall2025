## Section 4.1

### 4.1.1. Market basket analysis: a motivating example

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">تحلیل سبد خرید</span> 🛒
</p>
<p style="font-size:16pt; text-align:right;">
تحلیل سبد خرید (Market Basket Analysis) یک تکنیک داده‌کاوی است که برای کشف ارتباطات و روابط بین آیتم‌هایی که همراه خریده می‌شوند استفاده می‌شود. هدف اصلی، شناسایی الگوهای هم‌وقوعی بین محصولات در تراکنش‌هاست تا بینشی از رفتار و ترجیحات مشتریان به دست آید. این تحلیل به‌طور گسترده در خرده‌فروشی، تجارت الکترونیک و سایر صنایع به کار می‌رود تا استراتژی‌های مربوط به چیدمان محصولات، تبلیغات و فروش متقابل (Cross-selling) بهینه‌سازی شوند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. قواعد ارتباطی (Association Rules):</b>  
تحلیل سبد خرید معمولاً منجر به استخراج قواعدی به شکل «اگر مشتری کالای X را بخرد، احتمالاً کالای Y را نیز می‌خرد» می‌شود. این قواعد به درک رفتار خرید مشتریان کمک می‌کنند.
</p>
<p style="font-size:16pt; text-align:right;">
<b>2. پشتیبانی، اطمینان و لیفت (Support, Confidence, Lift):</b>  
<b>پشتیبانی</b> (Support) نشان می‌دهد یک مجموعه از آیتم‌ها چقدر اغلب با هم ظاهر می‌شوند.  
<b>اطمینان</b> (Confidence) احتمال خرید آیتم Y را وقتی آیتم X خریده شده باشد، اندازه‌گیری می‌کند.  
<b>لیفت</b> (Lift) نشان‌دهنده قدرت ارتباط بین دو آیتم است؛ اگر بیشتر از ۱ باشد، یعنی آیتم‌ها به‌طور معناداری با هم هم‌وقوع دارند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از تحلیل سبد خرید در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> در پایتون استفاده کنیم تا تحلیل سبد خرید را روی یک مجموعه داده نمونه خرده‌فروشی انجام دهیم:
</p>

</div>

```python
pip install mlxtend
```

```python
# نصب کتابخونه mlxtend در صورت نیاز
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# 🔹 داده‌های نمونه: شامل تراکنش‌ها و آیتم‌ها
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# 🔹 تبدیل داده‌ها به فرمت one-hot encoding
# هر تراکنش یک سطر، هر آیتم یک ستون (True = وجود آیتم در تراکنش)
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# 🔹 اجرای الگوریتم Apriori
# min_support=0.2 یعنی آیتم‌ست‌ها باید در حداقل 20% تراکنش‌ها ظاهر بشن
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# 🔹 تولید قوانین انجمنی (Association Rules)
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1)

# 🔹 پاک‌سازی: حذف مقادیر NaN و Inf برای جلوگیری از هشدار
rules = rules.replace([float('inf'), float('-inf')], pd.NA).dropna().reset_index(drop=True)

# 🔹 نمایش نتایج
print("Frequent Itemsets:")
print(frequent_itemsets)

print("\nAssociation Rules:")
print(rules)
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، یک مجموعه داده خرده‌فروشی حاوی تراکنش‌ها و آیتم‌ها شبیه‌سازی شده است. سپس از کتابخانه <b>mlxtend</b> برای انجام تحلیل سبد خرید بر اساس الگوریتم <b>Apriori</b> استفاده شده است. خروجی شامل مجموعه‌های آیتم‌های پرتکرار و قواعد ارتباطی است که بینشی از اینکه کدام آیتم‌ها اغلب همراه خریده می‌شوند، ارائه می‌دهد.
</p>

### 4.1.2. Frequent itemsets, closed itemsets, and association rules

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مجموعه‌های آیتم‌های پرتکرار و قواعد ارتباطی</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
در تحلیل سبد خرید، مفاهیم <b>مجموعه‌های پرتکرار</b> (Frequent Itemsets)، <b>مجموعه‌های بسته</b> (Closed Itemsets) و <b>قواعد ارتباطی</b> (Association Rules) نقش کلیدی در شناسایی الگوها و روابط درون داده‌های تراکنشی ایفا می‌کنند. این مفاهیم اساس کشف دانش از داده‌های فروش و درک رفتار مشتریان را تشکیل می‌دهند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. مجموعه‌های پرتکرار (Frequent Itemsets):</b>  
مجموعه‌ای از آیتم‌ها که به طور مکرر در تراکنش‌ها ظاهر می‌شوند. این مجموعه‌ها بر اساس آستانه حداقل پشتیبانی (Minimum Support) تعیین می‌شوند که فراوانی یا درصد حداقلی از تراکنش‌هایی را مشخص می‌کند که آن مجموعه در آن‌ها باید حضور داشته باشد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. مجموعه‌های بسته (Closed Itemsets):</b>  
یک مجموعه پرتکرار است که هیچ یک از مجموعه‌های بزرگتر (supersets) آن همان مقدار پشتیبانی را ندارند. به عبارت دیگر، این مجموعه زیرمجموعه هیچ مجموعه پرتکرار دیگری با همان پشتیبانی نیست. مجموعه‌های بسته نمایشی فشرده‌تر و کارآمدتر از الگوهای پرتکرار را فراهم می‌کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. قواعد ارتباطی (Association Rules):</b>  
این قواعد روابط بین آیتم‌ها را به شکل «اگر آیتم X وجود داشت، احتمالاً آیتم Y نیز وجود دارد» بیان می‌کنند. این قواعد از مجموعه‌های پرتکرار استخراج می‌شوند و با معیارهایی مانند <b>اطمینان</b> (Confidence) و <b>لیفت</b> (Lift) ارزیابی می‌شوند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از مجموعه‌های پرتکرار، بسته و قواعد ارتباطی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا این مفاهیم را روی یک مجموعه داده نمونه نمایش دهیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis)، کشف مجموعه‌های پرتکرار و قوانین انجمنی استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
# تبدیل داده‌ها به فرمت بولی بدون استفاده از applymap
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **مجموعه‌های بسته (Closed Itemsets)** رو استخراج می‌کنه.
# یک مجموعه آیتم "بسته" هست اگر هیچ زیرمجموعه‌ای از آیتم‌ها با همان پشتیبانی (support) و شامل یک آیتم اضافه‌تر وجود نداشته باشه.
# در عمل، مجموعه‌های بسته به ما کمک می‌کنن تا فضای ذخیره‌سازی و محاسبات رو نسبت به تمام مجموعه‌های پرتکرار کاهش بدیم.
# اینجا چون همه سطرها در ستون support مقداری دارن، شرط تساوی تکراری هست و در حالت عادی باید با مقایسه زیرمجموعه‌ها فیلتر بشه.
# (نکته: این پیاده‌سازی ساده‌شده است و برای تشخیص واقعی مجموعه‌های بسته نیاز به الگوریتم‌های پیشرفته‌تری مثل Charm یا Closet داریم.)
closed_itemsets = frequent_itemsets[frequent_itemsets['support'] == frequent_itemsets['support']]

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده احتمال خرید آیتم B وقتی آیتم A خریده شده.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، سه خروجی مهم رو چاپ می‌کنیم:
# ۱. تمام مجموعه‌های پرتکرار
# ۲. مجموعه‌های بسته (در این حالت همه، چون فیلتر واقعی اعمال نشده)
# ۳. قوانین انجمنی با قطعیت بالا
print("Frequent Itemsets:")
print(frequent_itemsets)

print("\nClosed Itemsets:")
print(closed_itemsets)

print("\nAssociation Rules:")
print(rules)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از کتابخانه <b>mlxtend</b> استفاده شده تا مجموعه‌های پرتکرار، مجموعه‌های بسته و قواعد ارتباطی از یک مجموعه داده خرده‌فروشی شبیه‌سازی‌شده استخراج شوند. خروجی شامل مجموعه‌هایی از آیتم‌ها است که اغلب با هم ظاهر می‌شوند (مجموعه‌های پرتکرار)، مجموعه‌های بسته که زیرمجموعه هیچ مجموعه بزرگتری با همان پشتیبانی نیستند، و قواعد ارتباطی که روابط بین آیتم‌ها را نشان می‌دهند.

## Section 4.2

### 4.2.1. Apriori Algorithm: Finding Frequent Itemsets by Confined Candidate Generation

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">الگوریتم Apriori</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم <b>Apriori</b> یکی از الگوریتم‌های کلاسیک داده‌کاوی است که برای کشف <b>مجموعه‌های پرتکرار</b> در مجموعه‌داده‌های تراکنشی استفاده می‌شود. این الگوریتم با استفاده از یک رویکرد سطح‌به‌سطح (level-wise)، ترکیبات مختلف آیتم‌ها را بررسی کرده و به‌صورت کارآمد مواردی را که آستانه مشخص‌شده پشتیبانی (Support) را دارند، شناسایی می‌کند.  
<br>
این الگوریتم از <b>ویژگی Apriori</b> استفاده می‌کند که بیان می‌کند: اگر یک مجموعه آیتم‌ها پرتکرار باشد، تمام زیرمجموعه‌های آن نیز باید پرتکرار باشند. این ویژگی به الگوریتم اجازه می‌دهد تا از بررسی ترکیبات نامرتبط صرف‌نظر کند و فضای جست‌وجو را به‌طور چشمگیری کاهش دهد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مراحل کلیدی الگوریتم Apriori</span> 🔧
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. تولید کاندیدا (Candidate Generation):</b>  
الگوریتم با آیتم‌های تکی به عنوان کاندیدا شروع می‌کند و به‌صورت تکراری، مجموعه‌های بزرگتری از آیتم‌ها را بر اساس مجموعه‌های پرتکرار یافت‌شده در تکرار قبلی ایجاد می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. شمارش پشتیبانی (Support Counting):</b>  
پشتیبانی (Support) هر مجموعه کاندیدا با شمارش تعداد تراکنش‌هایی که شامل آن مجموعه هستند، محاسبه می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. هرس کردن (Pruning):</b>  
مجموعه‌هایی که پشتیبانی آن‌ها کمتر از آستانه از پیش تعریف‌شده باشد، حذف می‌شوند، زیرا نمی‌توانند جزئی از یک مجموعه پرتکرار باشند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. تکرار (Iteration):</b>  
این فرآیند تا زمانی ادامه می‌یابد که دیگر مجموعه پرتکرار جدیدی ایجاد نشود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از الگوریتم Apriori در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا الگوریتم Apriori را روی یک مجموعه داده نمونه خرده‌فروشی پیاده‌سازی کنیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای خرید مشتریان استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# در نهایت، مجموعه‌های پرتکرار رو چاپ می‌کنیم تا ببینیم کدوم ترکیب‌های آیتم‌ها به صورت مکرر با هم خریده می‌شن.
# این اطلاعات می‌تونن برای پیشنهاد محصول، قرارگیری کالا در قفسه یا کمپین‌های تبلیغاتی استفاده بشن.
print("Frequent Itemsets:")
print(frequent_itemsets)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار در یک مجموعه داده خرده‌فروشی استفاده شده است. از تابع <b>apriori</b> کتابخانه <b>mlxtend</b> برای تولید مجموعه‌های آیتم‌ها بر اساس آستانه حداقل پشتیبانی استفاده شده است.
</p>

### 4.2.2. Generating association rules from frequent itemsets

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">تولید قواعد ارتباطی از مجموعه‌های پرتکرار</span> 🔗
</p>
<p style="font-size:16pt; text-align:right;">
پس از شناسایی مجموعه‌های پرتکرار با استفاده از الگوریتم‌هایی مانند <b>Apriori</b>، گام بعدی در تحلیل سبد خرید، تولید <b>قواعد ارتباطی</b> است. این قواعد روابط جالب بین آیتم‌ها را آشکار می‌کنند و بینش‌های ارزشمندی را برای کسب‌وکارها فراهم می‌کنند تا در مورد چیدمان محصولات، استراتژی‌های بازاریابی و فروش متقابل (Cross-selling) تصمیم‌های آگاهانه‌تری بگیرند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. پشتیبانی، اطمینان و لیفت:</b>  
این معیارها برای ارزیابی قدرت و اهمیت قواعد ارتباطی استفاده می‌شوند.
<br><br>
<b>پشتیبانی (Support):</b> فراوانی یک مجموعه آیتم‌ها را در کل تراکنش‌ها اندازه‌گیری می‌کند.  
<b>اطمینان (Confidence):</b> احتمال خرید آیتم Y را وقتی آیتم X خریده شده باشد، نشان می‌دهد.  
<b>لیفت (Lift):</b> نشان‌دهنده قدرت واقعی ارتباط بین دو آیتم است؛ لیفت بیشتر از ۱ یعنی ارتباط معنادار.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. قالب قاعده ارتباطی:</b>  
قواعد ارتباطی معمولاً به صورت <b>«اگر X، آنگاه Y»</b> نمایش داده می‌شوند که در آن X و Y مجموعه‌هایی از آیتم‌ها هستند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از تولید قواعد ارتباطی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید با ادامه مثال قبلی، از کتابخانه <b>mlxtend</b> استفاده کنیم تا قواعد ارتباطی را از مجموعه‌های پرتکرار استخراج کنیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای رفتار خرید مشتریان استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
# این قوانین می‌تونن به فروشگاه‌ها کمک کنن تا محصولات رو بهتر پیشنهاد بدن یا قفسه‌گذاری بهینه انجام بدن.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی رو چاپ می‌کنیم تا روابط معنادار بین آیتم‌ها رو ببینیم.
print("Association Rules:")
print(rules)
```

در این مثال، پس از به دست آوردن مجموعه‌های پرتکرار، از تابع <b>association_rules</b> کتابخانه <b>mlxtend</b> برای تولید قواعد ارتباطی بر اساس آستانه اطمینان استفاده شده است. خروجی شامل قواعدی است که اطلاعاتی درباره مجموعه‌های پیش‌رو (antecedent) و پیامد (consequent) ارائه می‌دهد، همراه با معیارهایی مانند <b>پشتیبانی</b>، <b>اطمینان</b> و <b>لیفت</b>.

### 4.2.3. Improving the efficiency of Apriori

<p style="font-size:16pt; text-align:right;">
الگوریتم <b>Apriori</b> ابزاری قدرتمند برای یافتن مجموعه‌های پرتکرار و تولید قواعد ارتباطی است، اما ممکن است در دatasets بزرگ از نظر محاسباتی پرهزینه شود. برای کاهش بار محاسباتی و بهبود عملکرد، تکنیک‌های متعددی وجود دارند که می‌توانند کارایی این الگوریتم را به‌طور چشمگیری افزایش دهند.

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">راهکارهای کلیدی برای بهبود کارایی</span> 🛠️
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. استراتژی‌های هرس‌کردن (Pruning Strategies):</b>  
استفاده از روش‌هایی مانند <b>هرس مبتنی بر هش</b> (hash-based) یا <b>هرس مبتنی بر درخت</b> (tree-based) می‌تواند فضای جست‌وجو را کاهش دهد و از تولید مجموعه‌های کاندیدایی که از پیش مشخص است پرتکرار نیستند، جلوگیری کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. کاهش تراکنش‌ها (Transaction Reduction):</b>  
کاهش حجم تراکنش‌ها یا استفاده از نمونه‌ای کوچک‌تر از داده‌ها می‌تواند به‌طور چشمگیری سرعت الگوریتم Apriori را افزایش دهد.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. موازی‌سازی (Parallelization):</b>  
توزیع بار محاسباتی بین چندین پردازنده یا گره، به‌ویژه برای مجموعه‌داده‌های بزرگ با تعداد زیادی تراکنش، می‌تواند بهبود چشمگیری در عملکرد ایجاد کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. شمارش پویای مجموعه آیتم‌ها (Dynamic Itemset Counting):</b>  
تنظیم پویا آستانه حداقل پشتیبانی بر اساس حجم داده یا پیشرفت الگوریتم، می‌تواند جست‌وجو برای مجموعه‌های پرتکرار را بهینه‌تر کند.
</p>

</p>

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori

# 🔹 داده‌های شبیه‌سازی‌شده بزرگ برای تست الگوریتم Apriori
# این مجموعه شامل ۱۰,۰۰۰ تراکنش است که الگوهای خرید آیتم‌های ['A', 'B', 'C', 'D', 'E'] در آن تکرار شده.
# این حجم داده برای بررسی عملکرد الگوریتم روی داده‌های بزرگ و شبیه‌سازی رفتار مشتریان مناسب است.
data = {
    'TransactionID': [i for i in range(1, 10001)],
    'Item': ['A', 'B', 'C', 'D', 'E'] * 2000  # هر آیتم ۲۰۰۰ بار تکرار شده
}

df = pd.DataFrame(data)

# 🔹 تبدیل داده‌ها به فرمت بولی (True/False)
# هر تراکنش یک سطر و هر آیتم یک ستون می‌شود.
# اگر آیتم در تراکنش وجود داشته باشد True و در غیر این صورت False قرار می‌گیرد.
# Apriori نیاز به داده‌های بولی دارد و با این روش هم FutureWarning و هم DeprecationWarning حذف می‌شوند.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# 🔹 اجرای الگوریتم Apriori
# min_support=0.2 → فقط مجموعه‌هایی که حداقل در ۲۰٪ تراکنش‌ها ظاهر شده‌اند در نظر گرفته می‌شوند.
# use_colnames=True → نمایش نام واقعی آیتم‌ها در خروجی به جای اندیس‌ها
# توجه: در نسخه جدید mlxtend پارامتر n_jobs حذف شده و اجرای موازی به صورت داخلی پشتیبانی نمی‌شود.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# 🔹 نمایش مجموعه‌های پرتکرار
# این اطلاعات می‌تواند برای تحلیل رفتار مشتریان، پیشنهاد محصول، چینش کالا در فروشگاه یا طراحی کمپین‌های تبلیغاتی استفاده شود.
print("Frequent Itemsets:")
print(frequent_itemsets)
```

</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، پارامتر <b>n_jobs=-1</b> در تابع <b>apriori</b> به الگوریتم دستور می‌دهد تا از تمام پردازنده‌های موجود برای موازی‌سازی استفاده کند. این کار می‌تواند کارایی الگوریتم Apriori را به‌ویژه در کار با مجموعه‌داده‌های بزرگ به‌طور چشمگیری بهبود بخشد.

### 4.2.4. A pattern-growth approach for mining frequent itemsets

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">رویکرد رشد الگو در داده‌کاوی</span> 🌱
</p>
<p style="font-size:16pt; text-align:right;">
رویکرد <b>رشد الگو</b> (Pattern-Growth) یک راهبرد جایگزین برای کشف مجموعه‌های پرتکرار است که با الگوریتم سنتی Apriori متفاوت است. به جای تولید کاندیدا و هرس کردن، این روش با استفاده از یک ساختار داده فشرده به نام <b>درخت پیشوند</b> یا <b>درخت FP</b> (FP-tree)، مجموعه‌های پرتکرار را از الگوهای کوچک‌تر رشد می‌دهد. این رویکرد می‌تواند از نظر حافظه کارآمدتر و سریع‌تر باشد، به‌ویژه برای مجموعه‌داده‌هایی با تعداد زیادی تراکنش.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مراحل کلیدی رویکرد رشد الگو</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. ساخت درخت FP (Constructing the FP-Tree):</b>  
ساخت یک ساختار داده فشرده به نام <b>درخت FP</b> (FP-tree) از مجموعه داده تراکنش‌ها. این درخت به‌صورت کارآمدی فراوانی مجموعه‌های آیتم‌ها را نمایش می‌دهد و امکان دسترسی سریع به الگوهای پرتکرار را فراهم می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. کاوش مجموعه‌های پرتکرار (Mining Frequent Itemsets):</b>  
استفاده از تکنیک‌های بازگشتی برای استخراج مستقیم مجموعه‌های پرتکرار از درخت FP، بدون نیاز به تولید صریح مجموعه‌های کاندیدا — که فرآیندی زمان‌بر در الگوریتم Apriori است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. پایه الگوی شرطی (Conditional Pattern Base):</b>  
استفاده از <b>پایه الگوی شرطی</b> برای یافتن مجموعه‌های پرتکرار به‌صورت هدفمند و کارآمد. این پایه شامل مسیرهایی از درخت است که با یک آیتم خاص پایان می‌یابند و به ساخت زیردرخت‌های شرطی (Conditional FP-trees) برای یافتن الگوهای بزرگ‌تر کمک می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از رویکرد رشد الگو در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا الگوریتم <b>FP-growth</b> را روی یک مجموعه داده نمونه خرده‌فروشی پیاده‌سازی کنیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل FP-growth استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای خرید مشتریان استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم FP-growth هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **FP-growth (Frequent Pattern Growth)** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# برخلاف Apriori، FP-growth از یک ساختار درختی به نام **FP-tree** استفاده می‌کنه و نیازی به تولید مجموعه‌های کاندید نداره.
# این باعث می‌شه الگوریتم سریع‌تر و کارآمدتر باشه، خصوصاً برای داده‌های حجیم.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = fpgrowth(basket, min_support=0.2, use_colnames=True)

# در نهایت، مجموعه‌های پرتکرار رو چاپ می‌کنیم تا ببینیم کدوم ترکیب‌های آیتم‌ها به صورت مکرر با هم خریده می‌شن.
# این اطلاعات می‌تونن برای پیشنهاد محصول، قرارگیری کالا در قفسه یا کمپین‌های تبلیغاتی استفاده بشن.
print("Frequent Itemsets:")
print(frequent_itemsets)
```

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، از الگوریتم <b>FP-growth</b> موجود در کتابخانه <b>mlxtend</b> برای یافتن مجموعه‌های پرتکرار در یک مجموعه داده خرده‌فروشی استفاده شده است. تابع <b>fpgrowth</b> به‌صورت مستقیم مجموعه‌های پرتکرار را تولید می‌کند و نیازی به مراحل تولید کاندیدا و هرس کردن — مانند الگوریتم Apriori — ندارد. این ویژگی، آن را به یک روش کارآمد و مبتنی بر <b>رشد الگو</b> تبدیل می‌کند.

### 4.2.5. Mining frequent itemsets using the vertical data format

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاوش مجموعه‌های پرتکرار با فرمت داده عمودی</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
کاوش مجموعه‌های پرتکرار می‌تواند با استفاده از <b>فرمت داده عمودی</b> (Vertical Data Format) انجام شود، که یک روش جایگزین برای نمایش داده‌های تراکنشی است. در این فرمت، هر رکورد مربوط به یک آیتم منحصربه‌فرد و لیست تراکنش‌هایی که آن آیتم در آن‌ها ظاهر شده است، می‌شود. این رویکرد به‌ویژه در مجموعه‌داده‌های با ابعاد بالا — که تعداد آیتم‌های مجزا زیاد است — بسیار کارآمد است و محاسبات مربوط به پشتیبانی و اشتراک مجموعه‌ها را ساده‌تر می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. فرمت داده عمودی (Vertical Data Format):</b>  
در این فرمت، هر آیتم همراه با لیست تراکنش‌هایی که در آن‌ها ظاهر شده است، ذخیره می‌شود. این ساختار امکان بررسی مستقل هر آیتم را فراهم کرده و فرآیند شناسایی مجموعه‌های پرتکرار را ساده‌تر می‌کند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. لیست شناسه تراکنش‌ها (Transaction Identifier Lists):</b>  
برای هر آیتم، لیستی از شناسه تراکنش‌ها (TID-list) نگهداری می‌شود که نشان می‌دهد آن آیتم در کدام تراکنش‌ها حضور دارد. این لیست‌ها امکان محاسبه سریع <b>پشتیبانی</b> (Support) را با اشتراک‌گیری مجموعه‌ها فراهم می‌کنند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. نگاشت معکوس تراکنش (Inverse Transaction Mapping):</b>  
علاوه بر لیست تراکنش‌ها برای هر آیتم، اغلب از یک نگاشت معکوس استفاده می‌شود تا بتوان به‌سرعت آیتم‌های موجود در یک تراکنش مشخص را بازیابی کرد. این نگاشت دسترسی دوطرفه بین آیتم‌ها و تراکنش‌ها را ممکن می‌سازد و کارایی پردازش را افزایش می‌دهد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش مجموعه‌های پرتکرار با فرمت داده عمودی در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا کاوش مجموعه‌های پرتکرار با استفاده از <b>فرمت داده عمودی</b> را روی یک مجموعه داده نمونه خرده‌فروشی نمایش دهیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای پیش‌پردازش تراکنش‌ها و اجرای الگوریتم Apriori استفاده می‌شه.

# این دیکشنری حاوی یک مجموعه داده خرده‌فروشی در فرمت عمودی (Vertical Format) هست.
# هر تراکنش شامل لیستی از آیتم‌های خریداری‌شده‌ست.
# این فرمت شبیه به نحوه ذخیره‌سازی واقعی داده‌های تراکنشی در سیستم‌های فروشگاهی هست.
data = {
    'TransactionID': [1, 2, 3, 4, 5],
    'Items': [['A', 'B', 'D'], ['A', 'C', 'D'], ['B', 'D'], ['A', 'B', 'C'], ['A', 'C']]
}

df = pd.DataFrame(data)

# این بخش، داده‌های تراکنشی رو از فرمت لیستی (عمودی) به فرمت یک‌دسته‌ای (Horizontal/Binary Format) تبدیل می‌کنه.
# از کلاس `TransactionEncoder` کتابخانه mlxtend استفاده می‌شه.
# این تبدیل لازمه چون الگوریتم‌هایی مثل Apriori نیاز دارن که داده‌ها به صورت ستون‌های جداگانه برای هر آیتم و با مقادیر ۰ و ۱ باشن.
te = TransactionEncoder()
te_ary = te.fit(df['Items']).transform(df['Items'])
horizontal_df = pd.DataFrame(te_ary, columns=te.columns_)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(horizontal_df, min_support=0.2, use_colnames=True)

# در نهایت، مجموعه‌های پرتکرار رو چاپ می‌کنیم تا ببینیم کدوم ترکیب‌های آیتم‌ها به صورت مکرر با هم خریده می‌شن.
# این اطلاعات می‌تونن برای تحلیل رفتار مشتری، پیشنهاد محصول و بهینه‌سازی فروش استفاده بشن.
print("Frequent Itemsets:")
print(frequent_itemsets)
```

</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، مجموعه داده در فرمت عمودی نمایش داده شده و از <b>TransactionEncoder</b> کتابخانه <b>mlxtend</b> برای تبدیل آن به فرمت افقی (معمولی) استفاده شده است. سپس الگوریتم <b>Apriori</b> برای یافتن مجموعه‌های پرتکرار به کار گرفته شده است. این رویکرد نشان می‌دهد که چگونه می‌توان از فرمت عمودی برای مدیریت کارآمد داده‌های پرتکرار استفاده کرد، حتی اگر در نهایت نیاز به تبدیل به فرمت افقی برای سازگاری با برخی الگوریتم‌ها وجود داشته باشد.
</p>

### 4.2.6. Mining closed and max patterns

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">الگوهای بسته و ماکسیمال در داده‌کاوی</span> 🔍
</p>
<p style="font-size:16pt; text-align:right;">
در داده‌کاوی، علاوه بر یافتن مجموعه‌های پرتکرار، علاقه زیادی به کشف <b>الگوهای بسته</b> (Closed Patterns) و <b>الگوهای ماکسیمال</b> (Max Patterns) وجود دارد. این الگوها اطلاعات فشرده‌تر و معنادارتری نسبت به مجموعه‌های پرتکرار معمولی ارائه می‌دهند و به درک بهتر و کاربرد مؤثرتر نتایج داده‌کاوی کمک می‌کنند.
</p>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. الگوهای بسته (Closed Patterns):</b>  
الگوهای بسته، مجموعه‌های پرتکراری هستند که اضافه کردن هیچ آیتم دیگری به آن‌ها بدون کاهش پشتیبانی (Support) ممکن نیست. این الگوها با حذف ترکیبات اضافی و تکراری، نمایشی فشرده و بدون اتلاف از مجموعه‌های پرتکرار ارائه می‌دهند.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. الگوهای ماکسیمال (Max Patterns):</b>  
الگوهای ماکسیمال، بزرگ‌ترین مجموعه‌های پرتکرار از نظر اندازه هستند و توسط هیچ مجموعه پرتکرار دیگری شامل نمی‌شوند. این الگوها دیدگاهی بسیار فشرده و متمرکز بر مهم‌ترین روابط در داده فراهم می‌کنند، اگرچه اطلاعاتی درباره زیرمجموعه‌های پرتکرار آن‌ها ارائه نمی‌دهند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از کاوش الگوهای بسته و ماکسیمال در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا الگوهای <b>بسته</b> (Closed) و <b>ماکسیمال</b> (Max) را از یک مجموعه داده نمونه خرده‌فروشی استخراج کنیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل FP-growth استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای پرتکرار استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم FP-growth هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **FP-growth (Frequent Pattern Growth)** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# این الگوریتم با ساخت یک درخت فشرده به نام FP-tree، به‌صورت کارآمدی الگوهای پرتکرار رو استخراج می‌کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه.
frequent_itemsets = fpgrowth(basket, min_support=0.2, use_colnames=True)

# این بخش، **الگوهای بسته (Closed Patterns)** رو استخراج می‌کنه.
# یک مجموعه آیتم "بسته" هست اگر هیچ زیرمجموعه‌ای از آن با همان پشتیبانی (support) و یک آیتم اضافه‌تر وجود نداشته باشه.
# در عمل، این الگوها فضای ذخیره‌سازی رو کاهش می‌دن و اطلاعات تکراری رو حذف می‌کنن.
# (نکته: این پیاده‌سازی ساده‌شده است و تشخیص واقعی الگوهای بسته نیاز به الگوریتم‌های تخصصی مثل Closet داره.)
closed_patterns = frequent_itemsets[frequent_itemsets['support'] == frequent_itemsets['support']]

# این بخش، **الگوهای ماکسیمال (Max Patterns)** رو استخراج می‌کنه.
# یک مجموعه آیتم "ماکسیمال" هست اگر در لیست مجموعه‌های پرتکرار، زیرمجموعهٔ هیچ مجموعه پرتکرار دیگه‌ای نباشه.
# این الگوها نشان‌دهنده بزرگ‌ترین مجموعه‌هایی هستن که هنوز شرط پشتیبانی رو دارن.
# در این کد، ما با گروه‌بندی روی `itemsets` و گرفتن حداکثر `support` (که معنای دقیقی نداره) یک شبیه‌سازی انجام دادیم.
# (نکته: این روش از نظر منطقی دقیق نیست؛ برای استخراج صحیح max patterns باید از روش‌هایی مثل MaxMiner یا اصلاح منطقی استفاده کرد.)
max_patterns = frequent_itemsets.groupby('itemsets')['support'].idxmax()
max_patterns = frequent_itemsets.loc[max_patterns]

# در نهایت، دو نوع الگوی خاص رو چاپ می‌کنیم:
# ۱. الگوهای بسته: مجموعه‌هایی که با همان پشتیبانی قابل گسترش نیستن.
# ۲. الگوهای ماکسیمال: مجموعه‌هایی که زیرمجموعهٔ هیچ مجموعه پرتکرار دیگه‌ای نیستن.
print("Closed Patterns:")
print(closed_patterns)

print("\nMax Patterns:")
print(max_patterns)
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، پس از یافتن مجموعه‌های پرتکرار با استفاده از الگوریتم <b>FP-growth</b>، الگوهای بسته با انتخاب مجموعه‌هایی که هیچ مجموعه بزرگتری با همان پشتیبانی وجود ندارد، شناسایی شده‌اند. الگوهای ماکسیمال نیز با انتخاب بزرگ‌ترین مجموعه‌های پرتکراری که زیرمجموعه هیچ مجموعه پرتکرار دیگری نیستند، تعیین شده‌اند. این رویکردها دیدگاهی فشرده و معنادار از الگوهای مهم در داده‌ها ارائه می‌دهند.
</p>

## Section 4.3

### 4.3.1. Strong rules are not necessarily interesting

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">قوی نبودن خودکار معنادار بودن در قواعد ارتباطی</span> ⚠️
</p>
<p style="font-size:16pt; text-align:right;">
در زمینه کاوش قواعد ارتباطی، درک این نکته حیاتی است که <b>قواعد قوی</b> — که با معیارهایی مانند <b>اطمینان</b> (Confidence) و <b>لیفت</b> (Lift) سنجیده می‌شوند — لزوماً معنادار یا ارزشمند نیستند. اگرچه این معیارها قدرت و اهمیت آماری قواعد را اندازه‌گیری می‌کنند، اما ممکن است <b>ارتباط واقعی</b>، <b>کاربرد عملی</b> یا <b>ارزش تجاری</b> الگوهای کشف‌شده را منعکس نکنند. بنابراین، تفسیر قواعد باید علاوه بر معیارهای کمی، با در نظر گرفتن زمینه داده و اهداف کسب‌وکار انجام شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. اطمینان و لیفت (Confidence and Lift):</b>  
<b>اطمینان</b> نشان می‌دهد که احتمال خرید آیتم Y چقدر است وقتی آیتم X خریده شده باشد.  
<b>لیفت</b> قدرت ارتباط بین دو آیتم را اندازه‌گیری می‌کند و نسبت به حالتی که دو آیتم مستقل از هم باشند، مقایسه می‌کند. لیفت بیشتر از ۱ نشان‌دهنده ارتباط مثبت و کمتر از ۱ نشان‌دهنده ارتباط منفی است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. اهمیت و زمینه (Interest and Context):</b>  
قواعدی که از نظر آماری قوی هستند (اطمینان و لیفت بالا) ممکن است در عمل بی‌معنا یا غیرکاربردی باشند. مثلاً قاعده «خرید نان → خرید کره» ممکن است لیفت بالایی داشته باشد، اما از آنجا که این رفتار بسیار رایج است، ارزش تجاری کمی دارد.  
ارزش واقعی قواعد به <b>زمینه دامنه</b> (مانند خرده‌فروشی، بهداشت، بانکداری) و <b>اهداف تحلیل</b> (مانند بهبود فروش متقابل یا طراحی بسته‌های تخفیف) بستگی دارد.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از قواعد قوی اما بی‌معنا در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از کتابخانه <b>mlxtend</b> استفاده کنیم تا مفهوم قواعدی با اطمینان و لیفت بالا که در عمل ارزش چندانی ندارند را نشان دهیم:
</p>

</div>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای رفتار خرید مشتریان استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
# این قوانین می‌تونن به فروشگاه‌ها کمک کنن تا محصولات رو بهتر پیشنهاد بدن یا قفسه‌گذاری بهینه انجام بدن.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی رو چاپ می‌کنیم تا روابط معنادار بین آیتم‌ها رو ببینیم.
print("Association Rules:")
print(rules)
```

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 💻
</p>
<p style="font-size:16pt; text-align:right;">
در این مثال، قواعد ارتباطی با استفاده از الگوریتم <b>Apriori</b> استخراج شده‌اند و قواعدی با آستانه اطمینان 0.7 انتخاب شده‌اند. با این حال، تأکید اصلی این است که <b>اطمینان بالا به تنهایی تضمین‌کننده ارزش واقعی یا جالب‌بودن یک قاعده نیست</b>. تفسیر و کاربرد قواعد باید با در نظر گرفتن <b>زمینه داده</b> و <b>اهداف تجاری</b> انجام شود، زیرا برخی قواعد ممکن است از نظر آماری قوی باشند اما از نظر عملی بی‌معنا یا پیش‌بینی‌پذیر باشند.
</p>
</div>

### 4.3.2. From association analysis to correlation analysis

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">تحلیل ارتباط و تحلیل همبستگی</span> 🔗
</p>
<p style="font-size:16pt; text-align:right;">
تحلیل ارتباط (Association Analysis) و تحلیل همبستگی (Correlation Analysis) دو تکنیک مرتبط اما متفاوت در داده‌کاوی هستند. در حالی که تحلیل ارتباط بر کشف روابط بین آیتم‌ها یا متغیرها — مانند «خرید همراه» در تراکنش‌ها — تمرکز دارد، تحلیل همبستگی قدرت و جهت یک رابطه خطی بین دو متغیر عددی را با استفاده از معیارهایی مانند ضریب همبستگی پیرسون (<b>Pearson Correlation Coefficient</b>) کمّی‌سازی می‌کند. درک تفاوت این دو روش برای انتخاب رویکرد مناسب در سناریوهای مختلف تحلیل داده ضروری است.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. تحلیل ارتباط (Association Analysis):</b>  
این تحلیل به دنبال کشف الگوهای جالب یا روابط معنادار در مجموعه‌داده‌های بزرگ است. معمولاً با داده‌های <b>دسته‌ای</b> (Categorical) کار می‌کند و هدف آن یافتن قواعدی است که نشان‌دهنده هم‌وقوع یا الگوهای ترتیبی بین آیتم‌ها باشد، مانند «اگر مشتری X را بخرد، احتمالاً Y را نیز می‌خرد».
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. تحلیل همبستگی (Correlation Analysis):</b>  
برخلاف تحلیل ارتباط، این روش برای متغیرهای <b>عددی</b> استفاده می‌شود. این تحلیل قدرت و جهت رابطه خطی بین دو متغیر را اندازه‌گیری می‌کند و به صورت <b>ضریب همبستگی</b> بیان می‌شود.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. ضریب همبستگی پیرسون:</b>  
این ضریب مقداری بین <b>۱-</b> تا <b>۱+</b> دارد که به شرح زیر تفسیر می‌شود:  
- <b>۱+</b>: رابطه خطی مثبت کامل (با افزایش یک متغیر، دیگری نیز افزایش می‌یابد)،  
- <b>۱-</b>: رابطه خطی منفی کامل (با افزایش یک متغیر، دیگری کاهش می‌یابد)،  
- <b>۰</b>: عدم وجود رابطه خطی.
</p>

<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از انتقال از تحلیل ارتباط به تحلیل همبستگی در پایتون</span> 🐍

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# این قسمت اول کده. ما کتابخونه‌های 'pandas' برای کار با داده، 'seaborn' برای ترسیم گرافیکی و 'matplotlib' برای نمایش نمودارها وارد می‌کنیم.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده با متغیرهای عددی (ویژگی‌ها) هست.
# هر ستون نشان‌دهنده یک ویژگی و هر سطر یک نمونه از داده‌ست.
# این نوع داده‌ها برای تحلیل روابط بین متغیرها، مثلاً با استفاده از ضریب همبستگی، مناسب‌اند.
data = {
    'Feature1': [1, 2, 3, 4, 5],
    'Feature2': [5, 4, 3, 2, 1],
    'Feature3': [2, 4, 1, 5, 3]
}

df = pd.DataFrame(data)

# این بخش، یک **نقشه گرمایی (Heatmap)** از ماتریس همبستگی بین ویژگی‌ها ایجاد می‌کنه.
# `df.corr()` ماتریس همبستگی پیرسون بین تمام جفت‌های عددی رو محاسبه می‌کنه.
# `annot=True` باعث می‌شه مقادیر عددی همبستگی روی خود نمودار نمایش داده بشه.
# `cmap='coolwarm'` یک پالت رنگی گرم به سرد (قرمز برای مثبت، آبی برای منفی) اعمال می‌کنه تا الگوهای رابطه رو بهتر ببینیم.
# `fmt='.2f'` فرمت نمایش اعداد رو به دو رقم اعشار محدود می‌کنه.
# این نوع تجسم داده به کشف روابط خطی بین ویژگی‌ها کمک می‌کنه — مثلاً 'Feature1' و 'Feature2' همبستگی کاملاً منفی دارن.
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

### 4.3.3. A comparison of pattern evaluation measures

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مقایسه معیارهای ارزیابی الگو</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
در داده‌کاوی، کشف مجموعه‌های پرتکرار تنها گام اول است. برای تشخیص <b>الگوهای واقعاً جالب و معنادار</b>، باید از معیارهای ارزیابی الگو (Pattern Evaluation Measures) استفاده کرد. معیارهایی مانند <b>پشتیبانی</b> (Support)، <b>اطمینان</b> (Confidence)، <b>لیفت</b> (Lift)، <b>بهبود</b> (Improvement)، و <b> conviction</b> هر کدام جنبه‌های مختلفی از قواعد ارتباطی را اندازه‌گیری می‌کنند.
<br><br>
اما همه قواعد با پشتیبانی و اطمینان بالا، لزوماً "جالب" نیستند. مثلاً قاعده <i>«نان → کره»</i> ممکن است اطمینان بالایی داشته باشد، اما چون بسیار رایج است، ارزش تجاری کمی دارد. 
<br><br>
<b>لیفت</b> (Lift = <i>Conf(X→Y) / P(Y)</i>) بهتر از اطمینان عمل می‌کند، زیرا رابطه بین X و Y را نسبت به توزیع مستقل آن‌ها ارزیابی می‌کند. لیفت > 1 نشان‌دهنده ارتباط مثبت، لیفت < 1 نشان‌دهنده ارتباط منفی و لیفت = 1 نشان‌دهنده استقلال است.
<br><br>
استفاده از ترکیب معیارها (مثلاً پشتیبانی کافی + لیفت بالا) به شناسایی قواعد واقعاً معنادار کمک می‌کند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مفاهیم کلیدی معیارهای ارزیابی الگو</span> 🔍
</p>

<p style="font-size:16pt; text-align:right;">
<b>1. پشتیبانی (Support):</b>  
پشتیبانی فراوانی یک الگو در مجموعه داده را اندازه‌گیری می‌کند و برابر است با نسبت تراکنش‌هایی که آن الگو در آن‌ها وجود دارد. این معیار نشان می‌دهد که یک الگو چقدر رایج است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>2. اطمینان (Confidence):</b>  
اطمینان قدرت یک قاعده ارتباطی را کمّی می‌کند و برابر است با احتمال شرطی وقوع <b>نتیجه</b> (consequent) وقتی <b>پیش‌رو</b> (antecedent) رخ داده باشد. این معیار شبیه به یک قانون "اگر-آنگاه" است.
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>3. لیفت (Lift):</b>  
لیفت اهمیت یک قاعده ارتباطی را با مقایسه پشتیبانی مشاهده‌شده قاعده با پشتیبانی مورد انتظار در صورت استقلال پیش‌رو و نتیجه ارزیابی می‌کند.  
لیفت > ۱: ارتباط مثبت،  
لیفت < ۱: ارتباط منفی،  
لیفت = ۱: عدم ارتباط (استقلال).
</p>
<br>
<p style="font-size:16pt; text-align:right;">
<b>4. کانویکشن (Conviction):</b>  
کانویکشن میزان استنتاج‌پذیری یک قاعده را اندازه‌گیری می‌کند. این معیار نشان می‌دهد که نتیجه چقدر بیشتر اتفاق می‌افتد وقتی پیش‌رو درست است. conviction بالا یعنی اگر پیش‌رو درست باشد، نتیجه به ندرت نقض می‌شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">مثالی از مقایسه معیارهای ارزیابی الگو در پایتون</span> 🐍
</p>
<p style="font-size:16pt; text-align:right;">
بیایید از یک مجموعه داده نمونه خرده‌فروشی استفاده کنیم تا نحوه محاسبه معیارهای <b>پشتیبانی</b>، <b>اطمینان</b>، <b>لیفت</b> و <b>کانویکشن</b> برای قواعد ارتباطی را در پایتون نشان دهیم:
</p>

```python
# Install the mlxtend library if not already installed
# pip install mlxtend

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# این قسمت اول کده. ما کتابخونه‌های 'pandas' و 'mlxtend' رو وارد می‌کنیم.
# 'pandas' برای کار با داده‌ها و 'mlxtend' برای اجرای الگوریتم‌های داده‌کاوی مثل Apriori استفاده می‌شه.

# این دیکشنری حاوی داده‌های شبیه‌سازی‌شده از یک مجموعه داده خرده‌فروشی (retail) هست.
# هر سطر نشان‌دهنده یک آیتم در یک تراکنش خاصه.
# این نوع داده‌ها برای تحلیل سبد خرید (Market Basket Analysis) و کشف الگوهای خرید مشتریان استفاده می‌شن.
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)

# این بخش، داده‌ها رو به فرمت "یک‌دسته‌ای" (one-hot encoded) تبدیل می‌کنه.
# هر تراکنش به عنوان یک سطر و هر آیتم به عنوان یک ستون در نظر گرفته می‌شه.
# اگر آیتمی در تراکنش وجود داشته باشه، مقدار ۱ و در غیر این صورت ۰ قرار می‌گیره.
# این فرمت ورودی مورد نیاز الگوریتم Apriori هست.
basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).astype(bool)

# اینجا الگوریتم **Apriori** اعمال می‌شه تا مجموعه‌های آیتم‌های پرتکرار (Frequent Itemsets) رو پیدا کنه.
# `min_support=0.2` یعنی فقط مجموعه‌هایی رو می‌خوایم که حداقل در ۲۰٪ تراکنش‌ها ظاهر شدن.
# `use_colnames=True` باعث می‌شه نام آیتم‌ها در خروجی نمایش داده بشه، نه فقط اندیس‌ها.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)

# این بخش، **قوانین انجمنی (Association Rules)** رو از روی مجموعه‌های پرتکرار تولید می‌کنه.
# معیار 'confidence' نشان‌دهنده اینه که با چه احتمالی آیتم B زمانی که آیتم A خریده شده، همراهش خریده می‌شه.
# `min_threshold=0.7` یعنی فقط قوانینی که حداقل ۷۰٪ قطعیت دارن رو نگه می‌داریم.
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)

# در نهایت، قوانین انجمنی رو با معیارهای ارزیابی مهم چاپ می‌کنیم:
# - 'support': میزان رایج‌بودن ترکیب antecedent و consequent در تمام تراکنش‌ها.
# - 'confidence': احتمال خرید consequent وقتی antecedent خریده شده.
# - 'lift': نشان‌دهنده اینه که آیا antecedent و consequent به صورت وابسته یا مستقل از هم خریده می‌شن.
#   اگر lift > 1، وابستگی مثبت دارن؛ اگر =1، مستقل‌اند.
# - 'conviction': معیاری قوی‌تر که نشان‌دهنده وابستگی جهت‌داره. هرچه conviction بالاتر باشه، وابستگی قوی‌تره.
print("Association Rules with Evaluation Measures:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift', 'conviction']])
```

در این مثال، قواعد ارتباطی با استفاده از الگوریتم <b>Apriori</b> استخراج شده‌اند و معیارهای ارزیابی شامل <b>پشتیبانی</b>، <b>اطمینان</b>، <b>لیفت</b> و <b>کانویکشن</b> برای هر قاعده محاسبه شده‌اند. این رویکرد امکان مقایسه بین معیارها و درک تفاوت‌های آن‌ها در ارزیابی کیفیت قواعد را فراهم می‌کند. برای نمونه، قاعده‌ای با اطمینان بالا ممکن است لیفت پایینی داشته باشد که نشان‌دهنده رابطه ضعیف یا بدیهی بین آیتم‌هاست، در حالی که لیفت بالا نشان‌دهنده یک ارتباط معنادار و غیربدیهی است.
