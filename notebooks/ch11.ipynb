## Section 11.1

### 11.1.2. Types of outliers

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">انواع داده‌های پرت (Types of Outliers)</span> ⚠️
</p>

<p style="font-size: 16pt; text-align: right;">
داده‌های پرت (Outliers) در یک مجموعه داده می‌توانند اشکال مختلفی داشته باشند و درک انواع آن‌ها برای تشخیص مؤثر داده‌های پرت حیاتی است. به طور کلی، داده‌های پرت را می‌توان به انواع زیر دسته‌بندی کرد:
</p>

<ul style="font-size: 16pt;">
<li><b>داده‌های پرت کلی (Global Outliers):</b> داده‌های پرت کلی، نقاط داده‌ای هستند که به طور قابل توجهی از الگوی کلی کل مجموعه داده منحرف می‌شوند. آن‌ها با مقادیر افراطی که از تمایل مرکزی داده‌ها دور هستند، مشخص می‌شوند.</li>
<li><b>داده‌های پرت زمینه‌ای (Contextual Outliers):</b> داده‌های پرت زمینه‌ای، نقاط داده‌ای هستند که در یک زمینه یا زیرمجموعه خاصی از داده‌ها غیرعادی هستند. شناسایی داده‌های پرت زمینه‌ای نیازمند در نظر گرفتن ویژگی‌های محلی داده‌ها است.</li>
<li><b>داده‌های پرت جمعی (Collective Outliers):</b> داده‌های پرت جمعی، که به آن‌ها داده‌های پرت خوشه‌ای (cluster outliers) یا زیرجمعیت‌ها (subpopulations) نیز گفته می‌شود، به گروه‌هایی از نقاط داده اشاره دارند که با هم یک الگوی ناهنجار را تشکیل می‌دهند. تشخیص داده‌های پرت جمعی شامل شناسایی ساختارهای غیرعادی در داخل مجموعه داده است.</li>
<li><b>داده‌های پرت ویژگی (Attribute Outliers):</b> داده‌های پرت ویژگی، مقادیر افراطی را در یک یا چند ویژگی (attributes or features) از خود نشان می‌دهند. این داده‌های پرت می‌توانند با بررسی ویژگی‌های فردی به طور مستقل شناسایی شوند.</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
درک انواع داده‌های پرت برای انتخاب تکنیک‌های تشخیص داده پرت مناسب و تفسیر نتایج در زمینه ویژگی‌های داده‌های خاص ضروری است.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از الگوریتم Isolation Forest ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم Isolation Forest

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# 1️⃣ تولید داده‌های مصنوعی شامل انواع نقاط پرت
np.random.seed(42)
data_normal = np.random.normal(loc=0, scale=1, size=(100, 2))                 # داده‌های نرمال
data_global_outliers = np.random.normal(loc=0, scale=10, size=(5, 2))         # نقاط پرت سراسری
data_contextual_outliers = np.random.normal(loc=20, scale=1, size=(5, 2))     # نقاط پرت زمینه‌ای
data_collector_outliers_1 = np.random.normal(loc=0, scale=1, size=(20, 2))    # خوشه اول پرت جمعی
data_collector_outliers_2 = np.random.normal(loc=30, scale=1, size=(20, 2))   # خوشه دوم پرت جمعی

# 2️⃣ ترکیب همه داده‌ها در یک مجموعه
data = np.vstack([
    data_normal,
    data_global_outliers,
    data_contextual_outliers,
    data_collector_outliers_1,
    data_collector_outliers_2
])

# 3️⃣ اعمال الگوریتم Isolation Forest برای تشخیص نقاط پرت
isolation_forest = IsolationForest(contamination=0.1, random_state=42)
labels = isolation_forest.fit_predict(data)  # برچسب‌های -1 (پرت) و 1 (نرمال)

# 4️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using Isolation Forest')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، الگوریتم <b>Isolation Forest</b> برای تشخیص انواع نقاط پرت در داده‌های دوبعدی مصنوعی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و انواع نقاط پرت (سراسری، زمینه‌ای، جمعی) تولید شده‌اند.</li>
<li style="font-size:16pt;">داده‌ها با استفاده از <code>vstack</code> ترکیب شده‌اند تا یک مجموعه کامل تشکیل شود.</li>
<li style="font-size:16pt;">الگوریتم <b>Isolation Forest</b> با پارامتر <code>contamination=0.1</code> آموزش دیده و برچسب‌های -1 (پرت) و 1 (نرمال) تولید کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از <code>scatter</code> در فضای دوبعدی نمایش داده شده‌اند تا توزیع نقاط پرت قابل مشاهده باشد.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این مثال پایه‌ای مناسب برای درک نحوه عملکرد الگوریتم‌های تشخیص ناهنجاری در داده‌های چندبعدی است. الگوریتم Isolation Forest به‌ویژه برای داده‌های بزرگ و با ابعاد بالا بسیار مؤثر و مقیاس‌پذیر است.
</p>
</p>

### 11.2.1. Parametric methods

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">روش‌های پارامتریک برای تشخیص داده پرت (Parametric methods for outlier detection)</span> 🔔
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های پارامتریک برای تشخیص داده پرت (outlier detection) فرض می‌کنند که توزیع زیربنایی داده‌ها را می‌توان با یک مدل پارامتریک خاص (مانند توزیع گوسی یا Gaussian) مشخص کرد. این روش‌ها برای شناسایی نقاط داده‌ای که به طور قابل توجهی از توزیع مورد انتظار منحرف می‌شوند، به معیارهای آماری و آزمون فرضیه (hypothesis testing) متکی هستند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره روش‌های پارامتریک:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>فرضیه توزیع:</b> روش‌های پارامتریک فرض می‌کنند که داده‌ها از یک توزیع شناخته شده پیروی می‌کنند. توزیع‌های متداول شامل توزیع نرمال (normal) یا گوسی هستند.</li>
<li><b>امتیاز Z:</b> امتیاز Z یک روش پارامتریک رایج است که اندازه‌گیری می‌کند یک نقطه داده چند انحراف معیار (standard deviations) از میانگین فاصله دارد. نقاط داده با امتیاز Z مطلق بالا، داده پرت در نظر گرفته می‌شوند.</li>
<li><b>آزمون فرضیه:</b> روش‌های پارامتریک اغلب شامل آزمون فرضیه برای ارزیابی اینکه آیا یک نقطه داده یا مجموعه‌ای از نقاط داده به طور قابل توجهی از توزیع مورد انتظار منحرف می‌شوند، می‌باشند. آزمون‌های رایج شامل t-test و chi-square test هستند.</li>
<li><b>رویکردهای مبتنی بر مدل:</b> برخی از روش‌های پارامتریک، مدل‌های صریحی از توزیع داده‌ها می‌سازند و داده‌های پرت را بر اساس احتمال داده‌های مشاهده شده تحت مدل شناسایی می‌کنند.</li>
<li><b>حساسیت به فرضیه:</b> اثربخشی روش‌های پارامتریک به دقت فرضیه‌های توزیعی بستگی دارد. اگر داده‌ها به طور قابل توجهی از توزیع فرضی منحرف شوند، روش‌های پارامتریک ممکن است نتایج نادرستی ارائه دهند.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از امتیاز Z ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از Z-Score

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت
np.random.seed(42)
data_normal = np.random.normal(loc=0, scale=1, size=100)       # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=5)       # نقاط پرت شدید

# 2️⃣ ترکیب داده‌های نرمال و پرت در یک آرایه
data = np.concatenate([data_normal, data_outliers])

# 3️⃣ محاسبه Z-Score برای هر نقطه
z_scores = zscore(data)

# 4️⃣ تعیین آستانه برای تشخیص نقاط پرت
threshold = 3.5
outliers = np.abs(z_scores) > threshold  # نقاطی با Z-Score بزرگ‌تر از آستانه

# 5️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(range(len(data)), data, c=outliers, cmap='viridis', alpha=0.7)
plt.axhline(y=threshold, color='r', linestyle='--', label='Outlier Threshold')
plt.title('Outlier Detection using Z-Score')
plt.xlabel('Data Point Index')
plt.ylabel('Data Value')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از روش آماری <b>Z-Score</b> برای تشخیص نقاط پرت در داده‌های عددی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">برای هر نقطه، مقدار Z-Score محاسبه شده تا فاصله آن از میانگین مشخص شود.</li>
<li style="font-size:16pt;">نقاطی که مقدار Z-Score آن‌ها از آستانه تعیین‌شده (۳.۵) بیشتر باشد، به‌عنوان پرت شناسایی شده‌اند.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
روش Z-Score برای داده‌های تک‌بعدی و توزیع‌های نرمال بسیار مؤثر است، اما در داده‌های چندبعدی یا غیرنرمال بهتر است از روش‌هایی مانند <b>Isolation Forest</b> یا <b>LOF</b> استفاده شود.
</p>
</p>

### 11.2.2. Nonparametric methods

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">روش‌های ناپارامتریک برای تشخیص داده پرت (Nonparametric methods for outlier detection)</span> 📈
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های ناپارامتریک برای تشخیص داده پرت (outlier detection) فرضیات قوی در مورد توزیع زیربنایی داده‌ها ندارند. این روش‌ها برای شناسایی نقاط داده‌ای که از الگوی مورد انتظار منحرف می‌شوند، به رتبه‌بندی یا سایر تکنیک‌های مستقل از توزیع متکی هستند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره روش‌های ناپارامتریک:</b>
</p>

<ol style="font-size: 16pt; text-align: right;">
<li><b>تکنیک‌های مبتنی بر رتبه:</b> روش‌های ناپارامتریک اغلب از آمار مبتنی بر رتبه، مانند میانه و دامنه بین چارکی (interquartile range - IQR) استفاده می‌کنند که نسبت به میانگین و انحراف معیار، حساسیت کمتری به مقادیر افراطی دارند.</li>
<li><b>رویکردهای مبتنی بر صدک:</b> داده‌های پرت می‌توانند بر اساس صدک‌ها (percentiles) یا چارک‌های (quantiles) توزیع داده‌ها شناسایی شوند. نقاط داده‌ای که خارج از یک محدوده صدکی خاص قرار می‌گیرند، داده پرت در نظر گرفته می‌شوند.</li>
<li><b>تخمین چگالی هسته (KDE):</b> KDE یک روش ناپارامتریک است که تابع چگالی احتمال داده‌ها را تخمین می‌زند. داده‌های پرت می‌توانند بر اساس نواحی با چگالی کم در توزیع تخمین زده شده، شناسایی شوند.</li>
<li><b>فاکتور داده پرت محلی (LOF):</b> LOF یک روش ناپارامتریک است که انحراف چگالی محلی یک نقطه داده را نسبت به همسایگانش اندازه‌گیری می‌کند. نقاط با چگالی محلی به طور قابل توجهی کمتر، داده پرت در نظر گرفته می‌شوند.</li>
<li><b>آزمون‌های مستقل از توزیع:</b> روش‌های ناپارامتریک اغلب از آزمون‌های آماری استفاده می‌کنند که به فرضیات توزیعی خاصی متکی نیستند. به عنوان مثال، آزمون کولموگوروف-اسمیرنوف (Kolmogorov-Smirnov) ارزیابی می‌کند که آیا یک نمونه از یک توزیع خاص می‌آید یا خیر.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از الگوریتم فاکتور داده پرت محلی (LOF) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم Local Outlier Factor (LOF)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت
np.random.seed(42)
data_normal = np.random.normal(loc=0, scale=1, size=100)       # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=5)       # نقاط پرت شدید

# 2️⃣ ترکیب داده‌های نرمال و پرت در یک آرایه و تبدیل به شکل مناسب
data = np.concatenate([data_normal, data_outliers]).reshape(-1, 1)

# 3️⃣ اعمال الگوریتم LOF برای تشخیص نقاط پرت
lof = LocalOutlierFactor(contamination=0.1)
labels = lof.fit_predict(data)  # برچسب‌های -1 (پرت) و 1 (نرمال)

# 4️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(range(len(data)), data, c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using Local Outlier Factor (LOF)')
plt.xlabel('Data Point Index')
plt.ylabel('Data Value')
plt.show()
```

<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، الگوریتم <b>Local Outlier Factor (LOF)</b> برای تشخیص نقاط پرت در داده‌های عددی تک‌بعدی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">داده‌ها به شکل مناسب برای ورودی الگوریتم LOF تبدیل شده‌اند (آرایه دو‌بعدی).</li>
<li style="font-size:16pt;">الگوریتم LOF با پارامتر <code>contamination=0.1</code> آموزش دیده و برچسب‌های -1 (پرت) و 1 (نرمال) تولید کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
الگوریتم LOF با مقایسه چگالی محلی هر نقطه نسبت به همسایگانش، نقاط پرت را شناسایی می‌کند. این روش برای داده‌های غیرخطی و توزیع‌های پیچیده بسیار مؤثر است.
</p>
</p>

### 11.3.1. Distance-based outlier detection

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">تشخیص داده پرت مبتنی بر فاصله (Distance-based outlier detection)</span> 📏
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های تشخیص داده پرت مبتنی بر فاصله، داده‌های پرت را با اندازه‌گیری عدم تشابه یا فاصله بین نقاط داده شناسایی می‌کنند. این روش‌ها فرض می‌کنند که داده‌های پرت اغلب ایزوله هستند یا از اکثریت نقاط داده فاصله زیادی دارند، که منجر به فواصل بزرگ‌تر می‌شود.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره تشخیص داده پرت مبتنی بر فاصله:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>فاصله اقلیدسی:</b> فاصله اقلیدسی بین نقاط داده یک معیار رایج است که در تشخیص داده پرت مبتنی بر فاصله استفاده می‌شود. این معیار فاصله خط مستقیم بین دو نقطه در فضای ویژگی (feature space) را محاسبه می‌کند.</li>
<li><b>فاصله ماهالانوبیس:</b> فاصله ماهالانوبیس همبستگی بین ویژگی‌ها را در نظر می‌گیرد و یک معیار دقیق‌تر از فاصله در حضور ویژگی‌های همبسته ارائه می‌دهد.</li>
<li><b>رویکردهای نزدیک‌ترین همسایه:</b> تشخیص داده پرت مبتنی بر فاصله اغلب شامل مقایسه فواصل نقاط داده با k نزدیک‌ترین همسایه‌شان است. داده‌های پرت به عنوان نقاطی با فواصل غیرعادی بزرگ به همسایگان خود شناسایی می‌شوند.</li>
<li><b>امتیازهای داده پرت:</b> امتیازهای داده پرت بر اساس فواصل محاسبه می‌شوند و یک آستانه برای تمایز داده‌های پرت از نقاط داده عادی تعیین می‌گردد.</li>
<li><b>مقاوم در برابر توزیع:</b> روش‌های مبتنی بر فاصله اغلب در برابر توزیع زیربنایی داده‌ها مقاوم هستند که آن‌ها را برای انواع مختلفی از مجموعه‌های داده مناسب می‌سازد.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از الگوریتم Isolation Forest که یک روش مبتنی بر فاصله است، ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم Isolation Forest (مبتنی بر فاصله)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت
np.random.seed(42)
data_normal = np.random.normal(loc=0, scale=1, size=(100, 2))     # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=(5, 2))     # نقاط پرت شدید

# 2️⃣ ترکیب داده‌های نرمال و پرت در یک مجموعه
data = np.vstack([data_normal, data_outliers])

# 3️⃣ اعمال الگوریتم Isolation Forest برای تشخیص نقاط پرت
isolation_forest = IsolationForest(contamination=0.1, random_state=42)
labels = isolation_forest.fit_predict(data)  # برچسب‌های -1 (پرت) و 1 (نرمال)

# 4️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using Isolation Forest (Distance-Based)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از الگوریتم <b>Isolation Forest</b> برای تشخیص نقاط پرت در داده‌های دوبعدی استفاده شده است. این الگوریتم بر اساس فاصله و جداسازی تصادفی نمونه‌ها عمل می‌کند:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">الگوریتم Isolation Forest با پارامتر <code>contamination=0.1</code> آموزش دیده و نقاط پرت را با برچسب -1 مشخص کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
الگوریتم Isolation Forest برای داده‌های بزرگ و چندبعدی بسیار مؤثر است و برخلاف روش‌های آماری، نیازی به فرض توزیع خاص ندارد. این مثال پایه‌ای مناسب برای درک نحوه عملکرد آن در فضای دوبعدی است.
</p>
</p>

### 11.3.2. Density-based outlier detection

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">تشخیص داده پرت مبتنی بر چگالی (Density-based outlier detection)</span> 🔍
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های تشخیص داده پرت مبتنی بر چگالی، داده‌های پرت را بر اساس چگالی محلی نقاط داده شناسایی می‌کنند. این روش‌ها فرض می‌کنند که داده‌های پرت با چگالی محلی به طور قابل توجهی پایین‌تر در مقایسه با همسایگان خود مشخص می‌شوند. رویکردهای مبتنی بر چگالی به ویژه در شناسایی داده‌های پرت در مناطقی با چگالی داده متغیر مؤثر هستند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره تشخیص داده پرت مبتنی بر چگالی:</b>
</p>

<ol style="font-size: 16pt; text-align: right;">
<li><b>تخمین چگالی محلی:</b> روش‌های مبتنی بر چگالی، چگالی محلی را در اطراف هر نقطه داده تخمین می‌زنند، که معمولاً با در نظر گرفتن تعداد نقاط همسایه در یک فاصله مشخص انجام می‌شود.</li>
<li><b>رویکردهای مبتنی بر خوشه‌بندی:</b> رویکردهای مبتنی بر چگالی اغلب شامل خوشه‌بندی (clustering) هستند، جایی که مناطق متراکم، خوشه در نظر گرفته می‌شوند و داده‌های پرت، نقاطی هستند که به هیچ خوشه‌ای تعلق ندارند یا به یک خوشه با چگالی پایین تعلق دارند.</li>
<li><b>امتیازهای داده پرت:</b> امتیازهای داده پرت بر اساس چگالی محلی محاسبه می‌شوند و نقاط با چگالی کمتر به عنوان داده‌های پرت بالقوه در نظر گرفته می‌شوند.</li>
<li><b>روش مثال: DBSCAN:</b> DBSCAN یک روش محبوب مبتنی بر چگالی است که خوشه‌ها را بر اساس چگالی داده شناسایی می‌کند و نقاط خارج از این خوشه‌ها را به عنوان داده پرت علامت‌گذاری می‌نماید.</li>
<li><b>تطبیق با الگوهای محلی:</b> روش‌های مبتنی بر چگالی با الگوهای محلی تطبیق‌پذیر هستند و می‌توانند داده‌های پرت را در مناطق با چگالی داده متغیر شناسایی کنند، که آن‌ها را برای مجموعه‌های داده پیچیده مناسب می‌سازد.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از DBSCAN ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم DBSCAN (مبتنی بر چگالی)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

# 1️⃣ تولید داده‌های مصنوعی با ساختار غیرخطی و نویز
data, _ = make_moons(n_samples=200, noise=0.05, random_state=42)

# 2️⃣ اعمال الگوریتم DBSCAN برای تشخیص نقاط پرت و خوشه‌بندی
dbscan = DBSCAN(eps=0.3, min_samples=5)
labels = dbscan.fit_predict(data)  # برچسب‌های -1 (پرت) و عددهای خوشه

# 3️⃣ نمایش تصویری نتایج تشخیص و خوشه‌بندی
plt.figure(figsize=(10, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using DBSCAN (Density-Based)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، الگوریتم <b>DBSCAN</b> برای تشخیص نقاط پرت و خوشه‌بندی داده‌های غیرخطی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ساختار <code>make_moons</code> تولید شده‌اند که دارای نویز و شکل غیرخطی هستند.</li>
<li style="font-size:16pt;">الگوریتم DBSCAN با پارامترهای <code>eps=0.3</code> و <code>min_samples=5</code> اجرا شده و نقاط پرت را با برچسب -1 مشخص کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا خوشه‌ها و نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
DBSCAN یکی از الگوریتم‌های قدرتمند برای تشخیص ناهنجاری در داده‌هایی با ساختار پیچیده و غیرخطی است. این روش بدون نیاز به تعیین تعداد خوشه‌ها، بر اساس چگالی محلی عمل می‌کند و نقاط پرت را به‌صورت طبیعی جدا می‌سازد.
</p>

### 11.4.1. Matrix factorization–based methods for numerical data

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">روش‌های مبتنی بر تجزیه ماتریس (Matrix factorization–based methods) برای تشخیص داده پرت</span> 🧩
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های مبتنی بر تجزیه ماتریس (Matrix factorization) برای تشخیص داده پرت (outlier detection) شامل تجزیه ماتریس داده اصلی به ماتریس‌های با ابعاد پایین‌تر هستند، با هدف ثبت ساختار زیربنایی داده‌ها. سپس داده‌های پرت بر اساس تفاوت بین داده‌های مشاهده شده و بازسازی شده شناسایی می‌شوند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره روش‌های مبتنی بر تجزیه ماتریس:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>تجزیه ماتریس:</b> تکنیک‌های تجزیه ماتریس، ماتریس داده اصلی را به دو یا چند ماتریس با ابعاد پایین‌تر تجزیه می‌کنند. روش‌های رایج شامل تجزیه مقدار منفرد (Singular Value Decomposition - SVD) و تجزیه ماتریس غیرمنفی (Non-Negative Matrix Factorization - NMF) هستند.</li>
<li><b>خطای بازسازی:</b> تفاوت بین ماتریس داده اصلی و بازسازی آن به عنوان خطای بازسازی محاسبه می‌شود. داده‌های پرت بر اساس خطاهای بازسازی غیرعادی بالا شناسایی می‌گردند.</li>
<li><b>تقریب با رتبه پایین:</b> تجزیه ماتریس با هدف تقریب ماتریس داده اصلی با یک تقریب با رتبه پایین (low-rank approximation)، ثبت ویژگی‌های ضروری و در عین حال کاهش تأثیر نویز و داده‌های پرت انجام می‌شود.</li>
<li><b>مقاوم در برابر ناهنجاری‌ها:</b> روش‌های تجزیه ماتریس اغلب در برابر داده‌های پرت مقاوم هستند، زیرا آن‌ها بر ثبت الگوهای غالب در داده‌ها تمرکز می‌کنند.</li>
<li><b>کاربردها:</b> این روش‌ها معمولاً برای تشخیص داده پرت در داده‌های عددی، مانند قرائت‌های حسگر، داده‌های تصویری، و سایر مجموعه‌های داده چندبعدی استفاده می‌شوند.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از تجزیه ماتریس با تجزیه مقدار منفرد (SVD) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از تجزیه ماتریسی SVD و خطای بازسازی

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت
np.random.seed(42)
data_normal = np.random.normal(loc=0, scale=1, size=(100, 2))     # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=(5, 2))     # نقاط پرت شدید

# 2️⃣ ترکیب داده‌ها و استانداردسازی ویژگی‌ها
data = np.vstack([data_normal, data_outliers])
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data)

# 3️⃣ اعمال SVD برای کاهش بعد و بازسازی داده‌ها
svd = TruncatedSVD(n_components=2)
svd.fit(data_standardized)
reconstructed_data = svd.inverse_transform(svd.transform(data_standardized))

# 4️⃣ محاسبه خطای بازسازی برای هر نمونه
reconstruction_error = np.mean(np.square(data_standardized - reconstructed_data), axis=1)

# 5️⃣ تعیین آستانه برای تشخیص نقاط پرت
threshold = 2.5
outliers = reconstruction_error > threshold

# 6️⃣ نمایش تصویری خطای بازسازی و نقاط پرت
plt.figure(figsize=(10, 6))
plt.scatter(range(len(data)), reconstruction_error, c=outliers, cmap='viridis', alpha=0.7)
plt.axhline(y=threshold, color='r', linestyle='--', label='Outlier Threshold')
plt.title('Outlier Detection using Matrix Factorization (SVD)')
plt.xlabel('Data Point Index')
plt.ylabel('Reconstruction Error')
plt.legend()
plt.show()
```

در این مثال، از روش <b>تجزیه مقدار منفرد (SVD)</b> برای تشخیص نقاط پرت در داده‌های دوبعدی استفاده شده است. مراحل انجام‌شده به شرح زیر است:

<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">داده‌ها با استفاده از <code>StandardScaler</code> استانداردسازی شده‌اند تا میانگین صفر و واریانس یک داشته باشند.</li>
<li style="font-size:16pt;">با استفاده از <code>TruncatedSVD</code> داده‌ها به فضای فشرده‌تر نگاشته شده و سپس بازسازی شده‌اند.</li>
<li style="font-size:16pt;">برای هر نمونه، خطای بازسازی محاسبه شده و نمونه‌هایی با خطای بالا به‌عنوان نقاط پرت شناسایی شده‌اند.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
روش‌های مبتنی بر بازسازی ماتریسی مانند SVD برای تشخیص ناهنجاری در داده‌های عددی و کاهش بعد بسیار مؤثر هستند. این روش‌ها به‌ویژه در مسائل تشخیص ناهنجاری در داده‌های سنسوری، مالی و تصویری کاربرد دارند.

### 11.4.2. Pattern-based compression methods for categorical data

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">روش‌های فشرده‌سازی مبتنی بر الگو (Pattern-based compression methods) برای تشخیص داده پرت</span> 🗜️
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های فشرده‌سازی مبتنی بر الگو برای داده‌های دسته‌بندی‌شده (categorical data) شامل نمایش داده‌های دسته‌بندی‌شده اصلی با استفاده از الگوها یا کدهای فشرده هستند. سپس داده‌های پرت بر اساس عدم تشابه بین الگوهای دسته‌بندی‌شده مشاهده شده و بازسازی شده شناسایی می‌شوند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره روش‌های فشرده‌سازی مبتنی بر الگو:</b>
</p>

<ol style="font-size: 16pt; text-align: right;">
<li><b>نمایش داده دسته‌بندی‌شده:</b> داده‌های دسته‌بندی‌شده با استفاده از الگوها یا کدهای فشرده نمایش داده می‌شوند، که معمولاً از طریق تکنیک‌هایی مانند کاوش الگوهای مکرر (frequent pattern mining) یا طرح‌های کدگذاری به دست می‌آیند.</li>
<li><b>خطای بازسازی:</b> تفاوت بین داده‌های دسته‌بندی‌شده اصلی و شکل بازسازی‌شده آن به عنوان خطای بازسازی محاسبه می‌شود. داده‌های پرت بر اساس خطاهای بازسازی غیرعادی بالا شناسایی می‌گردند.</li>
<li><b>تکنیک‌های کاوش الگو:</b> الگوریتم‌های کاوش الگوهای مکرر، مانند Apriori یا FP-growth، می‌توانند برای کشف الگوهای مهم در داده‌های دسته‌بندی‌شده به کار گرفته شوند.</li>
<li><b>کدگذاری یک‌-داغ:</b> کدگذاری یک‌-داغ (One-Hot Encoding) یک طرح کدگذاری رایج است که در آن هر دسته به عنوان یک بردار دودویی نمایش داده می‌شود و فشرده‌سازی با ثبت حضور یا عدم حضور دسته‌ها حاصل می‌شود.</li>
<li><b>کاربردها:</b> این روش‌ها به ویژه برای تشخیص داده پرت در مجموعه‌های داده‌ای که در آن‌ها متغیرهای دسته‌بندی‌شده غالب هستند، مانند داده‌های تراکنش مشتری یا تاریخچه خرید اقلام، مفید می‌باشند.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از روش‌های فشرده‌سازی مبتنی بر الگو با کدگذاری یک‌-داغ ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت در داده‌های دسته‌ای با استفاده از SVD و خطای بازسازی

import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.decomposition import TruncatedSVD

# 1️⃣ تولید داده‌های دسته‌ای مصنوعی شامل مقادیر پرت
data = pd.DataFrame({
    'Category_A': ['A', 'B', 'A', 'C', 'B', 'A', 'A'],
    'Category_B': ['X', 'Y', 'X', 'Z', 'Y', 'Z', 'X'],
    'Category_C': ['High', 'Low', 'Medium', 'High', 'Medium', 'Low', 'Low']
})

# 2️⃣ تبدیل ویژگی‌های دسته‌ای به بردارهای دودویی با One-Hot Encoding
encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(data).toarray()

# 3️⃣ اعمال SVD برای کاهش بعد و بازسازی داده‌ها
svd = TruncatedSVD(n_components=2)
svd.fit(encoded_data)
reconstructed_data = svd.inverse_transform(svd.transform(encoded_data))

# 4️⃣ محاسبه خطای بازسازی برای هر نمونه
reconstruction_error = ((encoded_data - reconstructed_data) ** 2).sum(axis=1)

# 5️⃣ تعیین آستانه برای تشخیص نقاط پرت
threshold = 0.1
outliers = reconstruction_error > threshold

# 6️⃣ افزودن نتایج به دیتافریم اصلی و نمایش
data['Reconstruction Error'] = reconstruction_error
data['Outlier'] = outliers.astype(int)
print(data)
```

<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از روش <b>تجزیه مقدار منفرد (SVD)</b> برای تشخیص نقاط پرت در داده‌های دسته‌ای استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های دسته‌ای مصنوعی شامل سه ویژگی متنی تولید شده‌اند.</li>
<li style="font-size:16pt;">با استفاده از <code>OneHotEncoder</code> داده‌ها به بردارهای دودویی تبدیل شده‌اند تا قابل پردازش عددی باشند.</li>
<li style="font-size:16pt;">با استفاده از <code>TruncatedSVD</code> داده‌ها به فضای فشرده‌تر نگاشته شده و سپس بازسازی شده‌اند.</li>
<li style="font-size:16pt;">برای هر نمونه، خطای بازسازی محاسبه شده و نمونه‌هایی با خطای بالا به‌عنوان نقاط پرت شناسایی شده‌اند.</li>
<li style="font-size:16pt;">نتایج نهایی شامل خطای بازسازی و برچسب پرت بودن به دیتافریم اصلی اضافه شده‌اند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این روش برای تشخیص ناهنجاری در داده‌های دسته‌ای بسیار مفید است، به‌ویژه زمانی که داده‌ها به‌صورت بردارهای sparse نمایش داده می‌شوند. استفاده از SVD به کاهش بعد و استخراج ساختارهای پنهان کمک می‌کند.
</p>

### 11.5.1. Clustering-based approaches

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">رویکردهای مبتنی بر خوشه‌بندی (Clustering-based approaches) برای تشخیص داده پرت</span> 🤝
</p>

<p style="font-size: 16pt; text-align: right;">
رویکردهای مبتنی بر خوشه‌بندی برای تشخیص داده پرت شامل گروه‌بندی نقاط داده به خوشه‌ها بر اساس شباهت و سپس شناسایی داده‌های پرت به عنوان نقاطی است که به خوبی با هیچ خوشه‌ای مطابقت ندارند. داده‌های پرت اغلب به عنوان نقاط داده‌ای در نظر گرفته می‌شوند که در خوشه‌های تشکیل شده قرار نمی‌گیرند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره رویکردهای مبتنی بر خوشه‌بندی:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>تشکیل خوشه:</b> نقاط داده بر اساس شباهت یا نزدیکی به خوشه‌ها گروه‌بندی می‌شوند. الگوریتم‌های خوشه‌بندی رایج شامل K-Means، خوشه‌بندی سلسله‌مراتبی (hierarchical clustering) و DBSCAN هستند.</li>
<li><b>شناسایی داده پرت:</b> داده‌های پرت، نقاطی هستند که به هیچ خوشه مشخصی تعلق ندارند یا به خوشه‌های کوچک یا نویز (noise) اختصاص داده می‌شوند. این نقاط از مرکزهای (centroids) خوشه‌های تشکیل شده دور هستند.</li>
<li><b>معیارهای فاصله:</b> معیارهای فاصله، مانند فاصله اقلیدسی (Euclidean distance) یا فاصله ماهالانوبیس (Mahalanobis distance)، اغلب برای ارزیابی شباهت بین نقاط داده و مراکز خوشه استفاده می‌شوند.</li>
<li><b>اندازه و چگالی خوشه:</b> داده‌های پرت اغلب در خوشه‌هایی با اندازه‌های کوچک یا چگالی کم شناسایی می‌شوند، زیرا این خوشه‌ها ممکن است نشان‌دهنده ناهنجاری‌هایی (anomalies) در داده‌ها باشند.</li>
<li><b>کاربردها:</b> تشخیص داده پرت مبتنی بر خوشه‌بندی در طیف وسیعی از مجموعه‌های داده، از جمله آن‌هایی که دارای ویژگی‌های عددی و دسته‌بندی‌شده هستند، قابل اجرا است.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از الگوریتم خوشه‌بندی DBSCAN ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم DBSCAN (مبتنی بر خوشه‌بندی)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs

# 1️⃣ تولید داده‌های مصنوعی با سه خوشه و دو نقطه پرت
data, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)
outliers = np.array([[0, 10], [5, 5]])  # نقاط پرت خارج از خوشه‌ها

# 2️⃣ ترکیب داده‌های نرمال با نقاط پرت
data = np.vstack([data, outliers])

# 3️⃣ اعمال الگوریتم DBSCAN برای خوشه‌بندی و تشخیص نقاط پرت
dbscan = DBSCAN(eps=1.5, min_samples=5)
labels = dbscan.fit_predict(data)  # برچسب‌های -1 (پرت) و عددهای خوشه

# 4️⃣ نمایش تصویری نتایج خوشه‌بندی و نقاط پرت
plt.figure(figsize=(10, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using DBSCAN (Clustering-Based)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از الگوریتم <b>DBSCAN</b> برای تشخیص نقاط پرت و خوشه‌بندی داده‌های دوبعدی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با سه خوشه مشخص تولید شده‌اند و دو نقطه پرت به‌صورت دستی اضافه شده‌اند.</li>
<li style="font-size:16pt;">الگوریتم DBSCAN با پارامترهای <code>eps=1.5</code> و <code>min_samples=5</code> اجرا شده و نقاط پرت را با برچسب -1 مشخص کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا خوشه‌ها و نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
DBSCAN یکی از الگوریتم‌های قدرتمند برای خوشه‌بندی داده‌هایی با ساختار پیچیده و تشخیص ناهنجاری است. این روش بدون نیاز به تعیین تعداد خوشه‌ها، بر اساس چگالی محلی عمل می‌کند و نقاط پرت را به‌صورت طبیعی جدا می‌سازد.
</p>

### 11.5.2. Classification-based approaches

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">رویکردهای مبتنی بر طبقه‌بندی (Classification-based approaches) برای تشخیص داده پرت</span> ✅
</p>

<p style="font-size: 16pt; text-align: right;">
رویکردهای مبتنی بر طبقه‌بندی برای تشخیص داده پرت شامل آموزش یک مدل یادگیری ماشین برای تمایز بین نمونه‌های عادی و ناهنجار (anomalous instances) است. مدل بر روی یک مجموعه داده دارای برچسب (labeled dataset) آموزش داده می‌شود، جایی که نمونه‌های عادی کلاس اکثریت و داده‌های پرت کلاس اقلیت هستند. پس از آموزش، مدل می‌تواند پیش‌بینی کند که آیا یک نمونه جدید عادی است یا یک داده پرت.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره رویکردهای مبتنی بر طبقه‌بندی:</b>
</p>

<ol style="font-size: 16pt; text-align: right;">
<li><b>یادگیری با نظارت:</b> رویکردهای مبتنی بر طبقه‌بندی روش‌های یادگیری با نظارت هستند که برای آموزش به داده‌های دارای برچسب نیاز دارند. مجموعه داده باید شامل نمونه‌هایی باشد که به عنوان عادی یا داده پرت برچسب‌گذاری شده‌اند.</li>
<li><b>انتخاب مدل:</b> الگوریتم‌های طبقه‌بندی رایج مانند ماشین‌های بردار پشتیبان (Support Vector Machines - SVM)، جنگل‌های تصادفی (Random Forests) یا شبکه‌های عصبی (Neural Networks) می‌توانند برای ساخت مدل‌های تشخیص داده پرت به کار گرفته شوند.</li>
<li><b>مدیریت داده نامتعادل:</b> مجموعه‌های داده تشخیص داده پرت اغلب نامتعادل هستند، به طوری که اکثریت نمونه‌ها عادی می‌باشند. مدیریت عدم تعادل کلاس برای عملکرد مدل حیاتی است و تکنیک‌هایی مانند بیش‌نمونه‌گیری (oversampling)، کم‌نمونه‌گیری (undersampling) یا استفاده از الگوریتم‌های تخصصی به کار می‌روند.</li>
<li><b>آستانه تصمیم:</b> تصمیم مدل بر اساس یک آستانه احتمال یا امتیاز (score) است. نمونه‌هایی با احتمال زیر آستانه، به عنوان داده پرت طبقه‌بندی می‌شوند.</li>
<li><b>کاربردها:</b> تشخیص داده پرت مبتنی بر طبقه‌بندی متنوع بوده و برای حوزه‌های مختلفی از جمله تشخیص تقلب، امنیت شبکه و مراقبت‌های بهداشتی قابل استفاده است.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تشخیص داده پرت با استفاده از یک طبقه‌بند ماشین بردار پشتیبان (SVM) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با استفاده از الگوریتم One-Class SVM (مبتنی بر طبقه‌بندی)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import OneClassSVM
from sklearn.datasets import make_blobs

# 1️⃣ تولید داده‌های مصنوعی با سه خوشه و دو نقطه پرت
data, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)
outliers = np.array([[0, 10], [5, 5]])  # نقاط پرت خارج از خوشه‌ها

# 2️⃣ ترکیب داده‌های نرمال با نقاط پرت
data = np.vstack([data, outliers])

# 3️⃣ اعمال الگوریتم One-Class SVM برای تشخیص نقاط پرت
svm = OneClassSVM(nu=0.05)  # نسبت تقریبی نقاط پرت
labels = svm.fit_predict(data)  # برچسب‌های -1 (پرت) و 1 (نرمال)

# 4️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Outlier Detection using One-Class SVM (Classification-Based)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠

<p style="font-size:18pt; text-align:right;">
در این مثال، از الگوریتم <b>One-Class SVM</b> برای تشخیص نقاط پرت در داده‌های دوبعدی استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با سه خوشه مشخص تولید شده‌اند و دو نقطه پرت به‌صورت دستی اضافه شده‌اند.</li>
<li style="font-size:16pt;">الگوریتم One-Class SVM با پارامتر <code>nu=0.05</code> اجرا شده که بیانگر نسبت تقریبی نقاط پرت در داده‌هاست.</li>
<li style="font-size:16pt;">مدل با استفاده از داده‌های نرمال آموزش دیده و نقاط پرت را با برچسب -1 شناسایی کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت و نرمال به‌صورت رنگی قابل تشخیص باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
الگوریتم One-Class SVM برای تشخیص ناهنجاری در داده‌هایی که فقط نمونه‌های نرمال در دسترس هستند بسیار مناسب است. این روش بر اساس مرز تصمیم‌گیری در فضای ویژگی‌ها عمل می‌کند و نقاط خارج از آن را پرت در نظر می‌گیرد.
</p>

### 11.6.1. Transforming contextual outlier detection to conventional outlier detection

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">تبدیل تشخیص داده پرت زمینه‌ای به تشخیص داده پرت مرسوم</span> ⚙️
</p>

<p style="font-size: 16pt; text-align: right;">
تشخیص داده پرت زمینه‌ای (Contextual outlier detection) شامل در نظر گرفتن زمینه یا محیطی است که نقاط داده در آن وجود دارند. تبدیل تشخیص داده پرت زمینه‌ای به تشخیص داده پرت مرسوم اغلب شامل استخراج ویژگی‌ها (features) یا نمایش‌های مرتبط از اطلاعات زمینه‌ای است که امکان استفاده از تکنیک‌های استاندارد تشخیص داده پرت را فراهم می‌سازد.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره تبدیل تشخیص داده پرت زمینه‌ای:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>اطلاعات زمینه‌ای:</b> تشخیص داده پرت زمینه‌ای، اطلاعات یا ویژگی‌های اضافی را که زمینه یا محیط نقاط داده را توصیف می‌کنند، در نظر می‌گیرد. این می‌تواند شامل اطلاعات زمانی، اطلاعات مکانی، یا هر زمینه مرتبط دیگری باشد.</li>
<li><b>استخراج ویژگی:</b> برای تبدیل تشخیص داده پرت زمینه‌ای به تشخیص داده پرت مرسوم، اغلب تکنیک‌های استخراج ویژگی برای تقطیر جنبه‌های مهم اطلاعات زمینه‌ای به ویژگی‌های عددی اعمال می‌شوند.</li>
<li><b>تکنیک‌های مرسوم:</b> هنگامی که ویژگی‌های مرتبط استخراج شدند، تکنیک‌های مرسوم تشخیص داده پرت، مانند روش‌های مبتنی بر خوشه‌بندی (clustering) یا طبقه‌بندی (classification)، می‌توانند برای شناسایی داده‌های پرت به کار روند.</li>
<li><b>کاربردها:</b> این رویکرد زمانی مفید است که با داده‌هایی سروکار داریم که اطلاعات زمینه‌ای در آن‌ها موجود است، اما باید به قالبی مناسب برای روش‌های استاندارد تشخیص داده پرت ترجمه شود.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای تبدیل تشخیص داده پرت زمینه‌ای با استفاده از استخراج ویژگی و اعمال یک رویکرد مبتنی بر خوشه‌بندی ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت زمینه‌ای با استفاده از PCA و DBSCAN

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت
np.random.seed(42)
data_contextual = np.random.normal(loc=0, scale=1, size=(100, 2))     # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=(5, 2))        # نقاط پرت زمینه‌ای

# 2️⃣ ترکیب داده‌ها و افزودن اطلاعات زمینه‌ای (زمانی)
data = np.vstack([data_contextual, data_outliers])
temporal_context = np.arange(0, 105).reshape(-1, 1)                   # اطلاعات زمانی به‌عنوان زمینه
data_with_context = np.hstack([data, temporal_context])              # ترکیب ویژگی‌ها با زمینه

# 3️⃣ استانداردسازی داده‌ها برای حذف اثر مقیاس
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data_with_context)

# 4️⃣ استخراج ویژگی‌های اصلی با استفاده از PCA
pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_standardized)

# 5️⃣ اعمال DBSCAN برای تشخیص نقاط پرت بر اساس ویژگی‌های استخراج‌شده
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(data_pca)  # برچسب‌های -1 (پرت) و عددهای خوشه

# 6️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(data_pca[:, 0], data_pca[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Transformed Contextual Outlier Detection using PCA and DBSCAN')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از ترکیب <b>تحلیل مؤلفه‌های اصلی (PCA)</b> و <b>DBSCAN</b> برای تشخیص نقاط پرت زمینه‌ای استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">اطلاعات زمینه‌ای (مانند زمان یا ترتیب وقوع) به‌عنوان ویژگی سوم به داده‌ها اضافه شده است.</li>
<li style="font-size:16pt;">داده‌ها با استفاده از <code>StandardScaler</code> استانداردسازی شده‌اند تا اثر مقیاس حذف شود.</li>
<li style="font-size:16pt;">با استفاده از <code>PCA</code> داده‌ها به فضای دوبعدی فشرده شده‌اند تا ویژگی‌های اصلی استخراج شوند.</li>
<li style="font-size:16pt;">الگوریتم <b>DBSCAN</b> روی فضای PCA اجرا شده و نقاط پرت را با برچسب -1 شناسایی کرده است.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا خوشه‌ها و نقاط پرت قابل مشاهده باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این روش برای تشخیص ناهنجاری‌های زمینه‌ای (contextual outliers) بسیار مؤثر است، به‌ویژه زمانی که اطلاعات اضافی مانند زمان، مکان یا شرایط محیطی در داده‌ها وجود دارد. ترکیب کاهش بعد و خوشه‌بندی می‌تواند ساختارهای پنهان را آشکار کند.
</p>

### 11.6.2. Modeling normal behavior with respect to contexts

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">مدل‌سازی رفتار عادی با توجه به زمینه‌ها (Modeling normal behavior with respect to contexts)</span> 🧠
</p>

<p style="font-size: 16pt; text-align: right;">
مدل‌سازی رفتار عادی با توجه به زمینه‌ها شامل ساخت مدل‌هایی است که الگوها یا رفتارهای معمول را در زمینه‌های خاص ثبت می‌کنند. این رویکرد با هدف تمایز بین نمونه‌های عادی و ناهنجار (anomalous instances) بر اساس مدل‌های رفتار عادی که مختص زمینه یاد گرفته شده‌اند، صورت می‌گیرد.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره مدل‌سازی رفتار عادی با توجه به زمینه‌ها:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>اطلاعات زمینه‌ای:</b> در نظر گرفتن اطلاعات زمینه‌ای ضروری است. این می‌تواند شامل عوامل محیطی، الگوهای مبتنی بر زمان، یا هر ویژگی (attribute) دیگری که مختص زمینه است، باشد.</li>
<li><b>مدل‌سازی رفتار عادی:</b> رفتار عادی در هر زمینه مدل‌سازی می‌شود تا الگوها یا توزیع‌های مورد انتظار درک شوند. این مدل‌سازی اغلب با استفاده از روش‌های آماری، مدل‌های یادگیری ماشین، یا دانش خاص حوزه انجام می‌شود.</li>
<li><b>تشخیص ناهنجاری:</b> ناهنجاری‌ها با مقایسه نمونه‌های مشاهده‌شده با رفتار عادی مدل‌سازی‌شده شناسایی می‌شوند. نمونه‌هایی که به طور قابل توجهی از الگوهای مورد انتظار منحرف می‌شوند، داده پرت (outliers) در نظر گرفته می‌شوند.</li>
<li><b>زمینه‌های پویا:</b> در زمینه‌های پویا، ممکن است مدل‌ها نیاز به انطباق در طول زمان داشته باشند؛ زیرا رفتار عادی تکامل می‌یابد. نظارت مداوم و آموزش مجدد مدل‌ها ممکن است ضروری باشد.</li>
<li><b>کاربردها:</b> این رویکرد برای سناریوهایی مناسب است که رفتار عادی در زمینه‌های مختلف متفاوت است و برای تشخیص مؤثر داده پرت، درک آگاهانه از زمینه مورد نیاز است.</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای مدل‌سازی رفتار عادی با توجه به زمینه‌ها با استفاده از یک مدل ترکیبی گوسی (Gaussian Mixture Model - GMM) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: مدل‌سازی رفتار نرمال با استفاده از GMM و تشخیص نقاط پرت زمینه‌ای

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler

# 1️⃣ تولید داده‌های مصنوعی شامل داده‌های نرمال و نقاط پرت زمینه‌ای
np.random.seed(42)
data_contextual = np.random.normal(loc=0, scale=1, size=(100, 2))     # داده‌های نرمال
data_outliers = np.random.normal(loc=10, scale=1, size=(5, 2))        # نقاط پرت زمینه‌ای

# 2️⃣ ترکیب داده‌ها و افزودن اطلاعات زمینه‌ای (مانند زمان یا ترتیب وقوع)
data = np.vstack([data_contextual, data_outliers])
temporal_context = np.arange(0, 105).reshape(-1, 1)                   # ویژگی زمینه‌ای
data_with_context = np.hstack([data, temporal_context])              # ترکیب ویژگی‌ها با زمینه

# 3️⃣ استانداردسازی داده‌ها برای حذف اثر مقیاس
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data_with_context)

# 4️⃣ اعمال مدل GMM برای مدل‌سازی رفتار نرمال
gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)
gmm.fit(data_standardized)

# 5️⃣ محاسبه log-likelihood برای هر نمونه
probabilities = gmm.score_samples(data_standardized)

# 6️⃣ تعیین آستانه برای تشخیص نقاط پرت
threshold = -8
outliers = probabilities < threshold  # نمونه‌هایی با احتمال پایین‌تر از آستانه

# 7️⃣ نمایش تصویری نتایج تشخیص
plt.figure(figsize=(10, 6))
plt.scatter(data_with_context[:, 0], data_with_context[:, 1], c=outliers, cmap='viridis', alpha=0.7)
plt.title('Modeling Normal Behavior with GMM and Contextual Information')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از مدل <b>Gaussian Mixture Model (GMM)</b> برای مدل‌سازی رفتار نرمال و تشخیص نقاط پرت زمینه‌ای استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های نرمال و نقاط پرت مصنوعی تولید شده‌اند و با هم ترکیب شده‌اند.</li>
<li style="font-size:16pt;">اطلاعات زمینه‌ای (مانند زمان یا ترتیب وقوع) به‌عنوان ویژگی سوم به داده‌ها اضافه شده است.</li>
<li style="font-size:16pt;">داده‌ها با استفاده از <code>StandardScaler</code> استانداردسازی شده‌اند تا اثر مقیاس حذف شود.</li>
<li style="font-size:16pt;">مدل <b>GMM</b> با دو مؤلفه اجرا شده و برای هر نمونه مقدار log-likelihood محاسبه شده است.</li>
<li style="font-size:16pt;">نمونه‌هایی با احتمال پایین‌تر از آستانه تعیین‌شده به‌عنوان نقاط پرت شناسایی شده‌اند.</li>
<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت قابل مشاهده باشند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
مدل‌های آماری مانند GMM برای تشخیص ناهنجاری در داده‌هایی با ساختار پیچیده و زمینه‌ای بسیار مؤثر هستند. این روش‌ها می‌توانند رفتار نرمال را مدل‌سازی کرده و انحراف از آن را به‌صورت کمی شناسایی کنند.
</p>

### 11.6.3. Mining collective outliers

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">کاوش داده‌های پرت جمعی (Mining collective outliers)</span> 🔗
</p>

<p style="font-size: 16pt; text-align: right;">
کاوش داده‌های پرت جمعی شامل شناسایی گروه‌ها یا زیرمجموعه‌هایی از نمونه‌ها است که به طور جمعی رفتار ناهنجار از خود نشان می‌دهند. به جای تمرکز بر روی نقاط داده فردی، این رویکرد به دنبال الگوهای ناهنجاری می‌گردد که با در نظر گرفتن تعاملات یا روابط بین چندین نمونه ظهور می‌کنند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره کاوش داده‌های پرت جمعی:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>ناهنجاری‌های جمعی:</b> داده‌های پرت جمعی ناهنجاری‌هایی هستند که هنگام در نظر گرفتن روابط، تعاملات یا وابستگی‌ها در میان چندین نمونه به جای نمونه‌های فردی، پدیدار می‌شوند.</li>
<li><b>مدل‌سازی رابطه:</b> کاوش داده‌های پرت جمعی اغلب شامل مدل‌سازی روابط یا وابستگی‌ها بین نقاط داده است، مانند قوانین انجمنی (association rules)، رویکردهای مبتنی بر گراف یا سایر اشکال مدل‌سازی رابطه‌ای.</li>
<li><b>الگوهای نوظهور:</b> داده‌های پرت جمعی ممکن است الگوهای نوظهوری از خود نشان دهند که هنگام تحلیل نمونه‌های فردی به صورت ایزوله آشکار نیستند. رفتار جمعی به دلیل اثرات ترکیبی گروه، غیرعادی می‌شود.</li>
<li><b>کاربردها:</b> این رویکرد برای سناریوهایی مفید است که در آن‌ها ناهنجاری‌ها در بستر روابط یا تعاملات بهتر درک می‌شوند، مانند تشخیص تقلب در شبکه‌ها، تحلیل رفتار گروهی یا فیلترگذاری مشارکتی (collaborative filtering).</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای کاوش داده‌های پرت جمعی با استفاده از یک رویکرد مبتنی بر گراف ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت جمعی با استفاده از گراف و نمایش درجه گره‌ها به‌صورت رنگ و اندازه

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from sklearn.datasets import make_blobs
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

# 1️⃣ تولید داده‌های مصنوعی شامل سه خوشه و دو نقطه پرت
data, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)
outliers = np.array([[0, 10], [5, 5]])  # نقاط پرت خارج از خوشه‌ها
data = np.vstack([data, outliers])

# 2️⃣ ساخت گراف بدون وزن بر اساس نزدیکی نقاط
G = nx.Graph()
for i in range(data.shape[0]):
    G.add_node(i)

threshold = 1  # آستانه فاصله برای اتصال یال
for i in range(data.shape[0]):
    for j in range(i + 1, data.shape[0]):
        distance = np.linalg.norm(data[i] - data[j])
        if distance < threshold:
            G.add_edge(i, j)

# 3️⃣ محاسبه درجه هر گره (تعداد همسایه‌ها)
degrees = np.array([G.degree[i] for i in range(data.shape[0])])

# 4️⃣ تنظیم اندازه و رنگ نقاط بر اساس درجه گره
sizes = 50 + 50 * degrees
norm = Normalize(vmin=min(degrees), vmax=max(degrees))
cmap = plt.cm.coolwarm_r
colors = cmap(norm(degrees))

# 5️⃣ رسم نقاط با رنگ و اندازه متناسب با درجه
fig, ax = plt.subplots(figsize=(12, 8))
scatter = ax.scatter(data[:, 0], data[:, 1], c=colors, s=sizes, edgecolor='k', alpha=0.8)

# 6️⃣ افزودن colorbar برای نمایش درجه گره‌ها
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Degree (Number of Neighbors)', fontsize=12)

# 7️⃣ برجسته‌سازی نقاط پرت جمعی (گره‌هایی با درجه کم)
collective_outliers = [i for i in G.nodes if G.degree[i] <= 1]
ax.scatter(data[collective_outliers, 0], data[collective_outliers, 1],
           facecolors='none', edgecolors='red', s=250, linewidths=2, marker='o', label='Collective Outliers')

# 8️⃣ تنظیم عنوان، برچسب‌ها و legend
ax.set_title('Graph-Based Collective Outliers with Degree Gradient 🌈', fontsize=16, fontweight='bold')
ax.set_xlabel('Feature 1', fontsize=14)
ax.set_ylabel('Feature 2', fontsize=14)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=12)
plt.show()
```

<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🌐
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از ساختار گراف برای تشخیص <b>نقاط پرت جمعی</b> استفاده شده و درجه گره‌ها به‌صورت رنگ و اندازه نمایش داده شده‌اند. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با سه خوشه تولید شده‌اند و دو نقطه پرت به‌صورت دستی اضافه شده‌اند.</li>
<li style="font-size:16pt;">بین هر دو نقطه‌ای که فاصله‌شان کمتر از آستانه تعیین‌شده باشد، یال ایجاد شده است.</li>
<li style="font-size:16pt;">درجه هر گره (تعداد همسایه‌ها) محاسبه شده و به‌صورت رنگی و اندازه‌ای در نمودار نمایش داده شده است.</li>
<li style="font-size:16pt;">گره‌هایی با درجه کمتر یا مساوی ۱ به‌عنوان نقاط پرت جمعی شناسایی و با حاشیه قرمز برجسته شده‌اند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این روش برای تحلیل ساختارهای گرافی و شناسایی ناهنجاری‌های گروهی بسیار مؤثر است. ترکیب گراف و نمایش بصری درجه گره‌ها می‌تواند در تحلیل شبکه‌های اجتماعی، داده‌های مکانی و سیستم‌های پیچیده کاربرد فراوانی داشته باشد.

### 11.7.1. Extending conventional outlier detection

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">گسترش روش‌های مرسوم تشخیص داده پرت به داده‌های با ابعاد بالا (High-dimensional data)</span> 📊
</p>

<p style="font-size: 16pt; text-align: right;">
گسترش روش‌های مرسوم تشخیص داده پرت به داده‌های با ابعاد بالا (High-dimensional data) شامل پرداختن به چالش‌هایی است که هنگام کار با مجموعه‌های داده با تعداد زیادی ویژگی به وجود می‌آیند. رویکردهای سنتی ممکن است به دلیل پدیده نفرین ابعاد (Curse of Dimensionality) با مشکل مواجه شوند و برای شناسایی مؤثر داده‌های پرت در فضاهای با ابعاد بالا، به اصلاحات یا تکنیک‌های تخصصی نیاز است.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره گسترش تشخیص داده پرت مرسوم برای داده‌های با ابعاد بالا:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>نفرین ابعاد (Curse of Dimensionality):</b> در فضاهای با ابعاد بالا، نقاط داده پراکنده‌تر می‌شوند و فواصل بین نقاط تمایل به افزایش دارند، که باعث می‌شود روش‌های مرسوم مبتنی بر فاصله کمتر مؤثر باشند.</li>
<li><b>کاهش ابعاد (Dimensionality Reduction):</b> تکنیک‌هایی مانند تحلیل مؤلفه اصلی (Principal Component Analysis - PCA) یا تعبیه‌سازی همسایگی تصادفی توزیع‌شده t (t-Distributed Stochastic Neighbor Embedding - t-SNE) می‌توانند برای کاهش ابعاد داده‌ها در عین حفظ اطلاعات مهم به کار گرفته شوند.</li>
<li><b>مدل‌های پراکنده (Sparse Models):</b> داده‌های با ابعاد بالا اغلب پراکندگی از خود نشان می‌دهند، به طوری که تنها زیرمجموعه کوچکی از ویژگی‌ها ممکن است به داده‌های پرت کمک کنند. مدل‌های پراکنده، مانند LASSO، می‌توانند به شناسایی ویژگی‌های مرتبط کمک کنند.</li>
<li><b>روش‌های ترکیبی (Ensemble Methods):</b> ترکیب چندین مدل تشخیص داده پرت یا استفاده از روش‌های ترکیبی (Ensemble) که برای داده‌های با ابعاد بالا طراحی شده‌اند، می‌تواند عملکرد کلی را بهبود بخشد.</li>
<li><b>کاربردها (Applications):</b> این رویکرد برای حوزه‌هایی با مجموعه‌های داده حاوی تعداد زیادی ویژگی، مانند ژنومیک، تحلیل تصویر، یا داده‌های مالی با متغیرهای متعدد، حیاتی است.
</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای گسترش تشخیص داده پرت مرسوم به داده‌های با ابعاد بالا با استفاده از روش‌های ترکیبی ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت در داده‌های با ابعاد بالا با استفاده از PCA، Isolation Forest و Sparse Models (LASSO)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.datasets import make_classification, load_iris
from sklearn.decomposition import PCA
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler

# 1️⃣ تولید داده‌های مصنوعی با ۲۰ ویژگی (۱۰ اطلاع‌رسان، ۵ افزونه)
data, _ = make_classification(n_samples=500, n_features=20, n_informative=10,
                              n_redundant=5, n_clusters_per_class=1, random_state=42)

# 2️⃣ تولید ۱۰ نقطه پرت با پراکندگی بالا
outliers = np.random.normal(loc=0, scale=5, size=(10, 20))

# 3️⃣ ترکیب داده‌های نرمال و پرت
data = np.vstack([data, outliers])

# ⚡ استفاده از مدل پراکنده (LASSO) برای انتخاب ویژگی‌های مهم
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)
lasso = LassoCV(cv=5, random_state=42).fit(data_scaled, np.random.randn(data_scaled.shape[0]))
important_features = np.sum(lasso.coef_ != 0)
print(f"ویژگی‌های انتخاب‌شده توسط LASSO: {important_features} از {data.shape[1]}")

# 4️⃣ کاهش ابعاد با PCA برای تصویرسازی
pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_scaled)

# 5️⃣ تشخیص نقاط پرت با Isolation Forest
isolation_forest = IsolationForest(contamination=0.02, random_state=42)
labels = isolation_forest.fit_predict(data_scaled)

# 6️⃣ نمایش تصویری داده‌ها در فضای PCA
plt.figure(figsize=(10, 6))
plt.scatter(data_pca[:, 0], data_pca[:, 1], c=labels, cmap='coolwarm', alpha=0.7, s=80, edgecolor='k')
plt.title('High-Dimensional Outlier Detection using PCA + Isolation Forest + LASSO', fontsize=14)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 🔎 7️⃣ مثال کاربردی روی داده واقعی (Iris با ویژگی‌های نویزی اضافه‌شده)
iris = load_iris()
iris_data = iris.data
random_noise = np.random.normal(0, 1, size=(iris_data.shape[0], 10))
iris_high_dim = np.hstack([iris_data, random_noise])

# کاهش بعد با PCA
pca_iris = PCA(n_components=2)
iris_pca = pca_iris.fit_transform(iris_high_dim)

# تشخیص نقاط پرت با Isolation Forest
iso_iris = IsolationForest(contamination=0.05, random_state=42)
labels_iris = iso_iris.fit_predict(iris_high_dim)

# نمایش تصویری
plt.figure(figsize=(10, 6))
plt.scatter(iris_pca[:, 0], iris_pca[:, 1], c=labels_iris, cmap='coolwarm', alpha=0.7, s=80, edgecolor='k')
plt.title('Real Dataset (Iris) with High-Dimensional Noise + Outlier Detection', fontsize=14)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از ترکیب سه روش قدرتمند برای تشخیص نقاط پرت در داده‌های با ابعاد بالا استفاده شده است:
<ul>
<li style="font-size:16pt;">مدل <b>LASSO</b> برای انتخاب ویژگی‌های مهم و حذف ویژگی‌های غیرضروری استفاده شده است.</li>
<li style="font-size:16pt;">با استفاده از <b>PCA</b> داده‌ها به فضای دوبعدی نگاشته شده‌اند تا امکان تصویرسازی فراهم شود.</li>
<li style="font-size:16pt;">الگوریتم <b>Isolation Forest</b> برای تشخیص نقاط پرت اجرا شده و برچسب‌های -1 و 1 تولید کرده است.</li>
<li style="font-size:16pt;">در بخش دوم، همین رویکرد روی داده واقعی <b>Iris</b> با ویژگی‌های نویزی اضافه‌شده اعمال شده است تا کاربرد عملی آن نشان داده شود.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این ترکیب برای مسائل واقعی مانند تشخیص تقلب، تحلیل رفتار کاربران، یا پایش سلامت سیستم‌ها بسیار مؤثر است. استفاده از مدل‌های پراکنده به کاهش پیچیدگی و افزایش تفسیرپذیری کمک می‌کند.

### 11.7.2. Finding outliers in subspaces

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">یافتن داده‌های پرت در زیرفضاها (Finding outliers in subspaces)</span> 🗺️
</p>

<p style="font-size: 16pt; text-align: right;">
یافتن داده‌های پرت (outliers) در زیرفضاها (subspaces) شامل شناسایی ناهنجاری‌ها در زیرمجموعه‌های خاصی از ویژگی‌ها به جای در نظر گرفتن کل فضای با ابعاد بالا است. این رویکرد تشخیص می‌دهد که داده‌های پرت ممکن است فقط در ترکیبات خاصی از ویژگی‌ها ظاهر شوند و تکنیک‌های سنتی تشخیص داده پرت ممکن است چنین ناهنجاری‌های محلی را ثبت نکنند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره یافتن داده‌های پرت در زیرفضاها:</b>
</p>

<ol style="font-size: 16pt;">
<li><b>ناهنجاری‌های محلی (Localized Anomalies):</b> داده‌های پرت ممکن است در زیرفضاها یا ترکیبات خاصی از ویژگی‌ها وجود داشته باشند و این ناهنجاری‌ها ممکن است هنگام تحلیل کل مجموعه ویژگی‌ها به طور همزمان آشکار نباشند.</li>
<li><b>کاوش زیرفضا (Subspace Exploration):</b> تکنیک‌ها شامل کاوش زیرمجموعه‌های ویژگی یا زیرفضاهای مختلف برای شناسایی اینکه کدام ترکیبات از ویژگی‌ها به بروز داده‌های پرت کمک می‌کنند، می‌باشند.</li>
<li><b>رویکردهای ترکیبی (Combinatorial Approaches):</b> روش‌های ترکیبی، مانند الگوریتم‌های انتخاب ویژگی یا کاوش بَزورانه (brute-force) ترکیبات ویژگی، می‌توانند به شناسایی زیرفضاهای مرتبط برای تشخیص داده پرت کمک کنند.</li>
<li><b>نفرین ابعاد (Curse of Dimensionality):</b> در فضاهای با ابعاد بالا، داده‌ها ممکن است به‌طور یکنواخت پخش شوند و تشخیص ناهنجاری‌ها دشوار شود. تمرکز بر زیرفضاها می‌تواند این مشکل را کاهش دهد.</li>
<li><b>کاربردها (Applications):</b> این رویکرد برای مجموعه‌های داده‌ای که داده‌های پرت در زمینه‌ها یا زیرمجموعه‌های خاصی از ویژگی‌ها برجسته هستند، مفید است، مانند شبکه‌های حسگر، تشخیص‌های پزشکی، یا هر حوزه‌ای که در آن ترکیبات مختلفی از ویژگی‌ها ممکن است رفتار ناهنجار را نشان دهند.</li>
</ol>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای یافتن داده‌های پرت در زیرفضاها با استفاده از یک رویکرد ترکیبی ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت در زیر‌فضاهای تصادفی با استفاده از Isolation Forest

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.ensemble import IsolationForest
from itertools import combinations
import random

# 1️⃣ تولید داده‌های مصنوعی با ۲۰ ویژگی (۱۰ اطلاع‌رسان، ۵ افزونه)
data, _ = make_classification(
    n_samples=500,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    n_clusters_per_class=1,
    random_state=42
)

# 2️⃣ تولید ۱۰ نقطه پرت با پراکندگی بالا
outliers = np.random.normal(loc=0, scale=5, size=(10, 20))

# 3️⃣ ترکیب داده‌های نرمال و پرت
data = np.vstack([data, outliers])

# 4️⃣ تعریف تعداد ویژگی‌های انتخاب‌شده در هر زیر‌فضا
num_features_per_subspace = 3

# 5️⃣ تولید تمام ترکیب‌های ممکن از ویژگی‌ها و نمونه‌گیری تصادفی از آن‌ها
all_feature_combinations = list(combinations(range(data.shape[1]), num_features_per_subspace))
feature_combinations = random.sample(all_feature_combinations, 300)

# 6️⃣ اجرای Isolation Forest در هر زیر‌فضا و شمارش دفعات تشخیص پرت بودن
outliers_subspaces = np.zeros(data.shape[0])
for combination in feature_combinations:
    data_subspace = data[:, combination]
    isolation_forest = IsolationForest(
        contamination=0.02,
        n_estimators=50,
        max_samples=100,
        random_state=42
    )
    labels = isolation_forest.fit_predict(data_subspace)
    outliers_subspaces[labels == -1] += 1

# 7️⃣ تعیین آستانه برای تشخیص نهایی نقاط پرت
threshold = 3
outliers_final = outliers_subspaces >= threshold

# 8️⃣ نمایش تصویری: رنگ و اندازه متناسب با قدرت تشخیص در زیر‌فضاها
plt.figure(figsize=(10, 6))
sc = plt.scatter(
    data[:, 0], data[:, 1],
    c=outliers_subspaces,
    s=20 + outliers_subspaces*20,
    cmap='plasma', alpha=0.7, edgecolor="k"
)

plt.title('Outlier Detection in Subspaces using Isolation Forest')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

# 9️⃣ افزودن legend برای نمایش شدت پرت بودن
for count in [1, 3, 5, 10]:
    plt.scatter([], [], c='k', alpha=0.5, s=20 + count*20,
                label=f'Detected {count}x')
plt.legend(scatterpoints=1, frameon=True, labelspacing=1, title="Outlier Strength")

# 🔟 افزودن colorbar برای نمایش تعداد دفعات تشخیص
cbar = plt.colorbar(sc)
cbar.set_label('Number of Subspaces Detected as Outlier')

plt.show()
```

در این مثال، از رویکرد <b>تشخیص نقاط پرت در زیر‌فضاهای تصادفی</b> استفاده شده است. مراحل انجام‌شده به شرح زیر است:

### 11.7.3. Outlier detection ensemble

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">روش‌های ترکیبی (Ensemble methods) در تشخیص داده پرت</span> 🤝
</p>

<p style="font-size: 16pt; text-align: right;">
روش‌های ترکیبی در تشخیص داده پرت (Outlier Detection Ensemble Methods) شامل ترکیب خروجی‌های چندین مدل تشخیص داده پرت برای افزایش دقت و استحکام کلی تشخیص داده پرت در داده‌های با ابعاد بالا (High-dimensional Data) است. این رویکرد با هدف کاهش محدودیت‌های مدل‌های فردی و ارائه شناسایی قابل‌اعتمادتر از ناهنجاری‌ها صورت می‌گیرد.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره روش‌های ترکیبی در تشخیص داده پرت:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>مدل‌های متنوع (Diverse Models):</b> روش‌های ترکیبی از نقاط قوت الگوریتم‌های متنوع تشخیص داده پرت بهره می‌برند. هر مدل ممکن است در جنبه‌های مختلفی عالی باشد یا در زیرفضاهای خاصی مؤثرتر عمل کند.</li>
<li><b>استراتژی‌های تجمیع (Aggregation Strategies):</b> خروجی‌های مدل‌های فردی تجمیع می‌شوند تا یک تصمیم جمعی درباره اینکه آیا یک نقطه داده، یک داده پرت است یا خیر، گرفته شود. استراتژی‌های تجمیع رایج شامل رأی‌گیری (Voting)، میانگین‌گیری (Averaging) یا استفاده از فرامدل‌ها (Meta-models) است.</li>
<li><b>استحکام (Robustness):</b> روش‌های ترکیبی با کاهش تأثیر ضعف‌ها یا سوگیری‌های مدل فردی، استحکام تشخیص داده پرت را افزایش می‌دهند. این امر به ویژه در داده‌های با ابعاد بالا (High-dimensional Data) که اثربخشی یک مدل واحد ممکن است محدود باشد، حیاتی است.</li>
<li><b>کاربردها (Applications):</b> در سناریوهایی که داده‌های با ابعاد بالا برای مدل‌های فردی چالش‌برانگیز هستند، مفید است. کاربردها شامل تشخیص تقلب (Fraud Detection)، امنیت شبکه (Network Security) و هر حوزه‌ای است که در آن داده‌های پرت ممکن است الگوهای متنوعی از خود نشان دهند.</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای روش‌های ترکیبی در تشخیص داده پرت با استفاده از ترکیبی از Isolation Forest و Local Outlier Factor (LOF) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت با ترکیب Isolation Forest و LOF در داده‌های با ابعاد بالا

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.datasets import make_classification
from sklearn.decomposition import PCA

# 1️⃣ تولید داده‌های مصنوعی با ۲۰ ویژگی (۱۰ اطلاع‌رسان، ۵ افزونه)
data, _ = make_classification(n_samples=500, n_features=20, n_informative=10,
                              n_redundant=5, n_clusters_per_class=1, random_state=42)

# 2️⃣ تولید ۱۰ نقطه پرت با پراکندگی بالا
outliers = np.random.normal(loc=0, scale=6, size=(10, 20))
data = np.vstack([data, outliers])

# 3️⃣ کاهش ابعاد با PCA برای تصویرسازی
pca = PCA(n_components=2, random_state=42)
data_pca = pca.fit_transform(data)

# 4️⃣ اجرای Isolation Forest و استخراج نمره پرت بودن
iso = IsolationForest(contamination=0.02, random_state=42)
iso_labels = iso.fit_predict(data)
iso_scores = -iso.decision_function(data)  # نمره پرت بودن (بزرگ‌تر → پرت‌تر)

# 5️⃣ اجرای Local Outlier Factor و استخراج نمره پرت بودن
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.02)
lof_labels = lof.fit_predict(data)
lof_scores = -lof.negative_outlier_factor_

# 6️⃣ نرمال‌سازی نمرات برای ترکیب
iso_scores = (iso_scores - iso_scores.min()) / (iso_scores.max() - iso_scores.min())
lof_scores = (lof_scores - lof_scores.min()) / (lof_scores.max() - lof_scores.min())

# 7️⃣ ترکیب نمرات با میانگین‌گیری
ensemble_scores = (iso_scores + lof_scores) / 2

# 8️⃣ تعیین برچسب نهایی بر اساس آستانه (۲٪ بالاترین نمرات)
threshold = np.percentile(ensemble_scores, 98)
labels = np.where(ensemble_scores >= threshold, -1, 1)

# 9️⃣ نمایش تصویری داده‌ها در فضای PCA با رنگ و اندازه متناسب با شدت پرت بودن
plt.figure(figsize=(10, 6))
plt.scatter(data_pca[:, 0], data_pca[:, 1],
            c=ensemble_scores, cmap="coolwarm",
            s=60 + 200 * ensemble_scores,
            edgecolor='k', alpha=0.7)

plt.title("Ensemble Outlier Detection (Isolation Forest + LOF) with PCA", fontsize=14)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label="Anomaly Score (Ensemble)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

<p style="font-size:18pt; text-align:right;">
در این مثال، از ترکیب دو الگوریتم قدرتمند برای تشخیص نقاط پرت در داده‌های با ابعاد بالا استفاده شده است:
<ul>
<li style="font-size:16pt;">الگوریتم <b>Isolation Forest</b> بر اساس جداسازی تصادفی نمونه‌ها عمل می‌کند و برای داده‌های بزرگ و پیچیده بسیار مؤثر است.</li>
<li style="font-size:16pt;">الگوریتم <b>Local Outlier Factor (LOF)</b> با مقایسه چگالی محلی هر نقطه نسبت به همسایگانش، نقاط پرت را شناسایی می‌کند.</li>
<li style="font-size:16pt;">نمرات هر دو الگوریتم نرمال‌سازی شده و با میانگین‌گیری ترکیب شده‌اند تا یک نمره نهایی برای هر نقطه به‌دست آید.</li>
<li style="font-size:16pt;">با استفاده از <b>PCA</b> داده‌ها به فضای دوبعدی نگاشته شده‌اند تا امکان تصویرسازی فراهم شود.</li>
<li style="font-size:16pt;">در نمودار نهایی، رنگ و اندازه هر نقطه نشان‌دهنده شدت پرت بودن آن است.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این رویکرد ترکیبی برای تشخیص ناهنجاری در داده‌های واقعی مانند تراکنش‌های مالی، داده‌های حسگر، یا تحلیل رفتار کاربران بسیار مؤثر است. ترکیب چند الگوریتم می‌تواند دقت و پایداری تشخیص را افزایش دهد.

### 11.7.4. Taming high dimensionality by deep learning

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">مهار ابعاد بالا (High dimensionality) با یادگیری عمیق (Deep learning)</span> 🧠
</p>

<p style="font-size: 16pt; text-align: right;">
مهار ابعاد بالا با یادگیری عمیق شامل بهره‌گیری از معماری‌های شبکه عصبی (neural network architectures) برای یادگیری خودکار نمایش‌ها و الگوها در داده‌های با ابعاد بالا است. مدل‌های یادگیری عمیق، به ویژه خودرمزگذارها (autoencoders)، برای کاهش ابعاد، یادگیری ویژگی (feature learning) و ثبت ساختارهای پیچیده در داده‌ها به کار گرفته می‌شوند که به شناسایی داده‌های پرت (outliers) کمک می‌کند.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره مهار ابعاد بالا با یادگیری عمیق:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>خودرمزگذارها (Autoencoders):</b> خودرمزگذارها معماری‌های شبکه عصبی هستند که برای وظایف یادگیری بدون نظارت (unsupervised learning) از جمله کاهش ابعاد طراحی شده‌اند. آن‌ها از یک رمزگذار (encoder) و یک رمزگشا (decoder) تشکیل شده‌اند، که رمزگذار داده ورودی را به یک نمایش با ابعاد پایین‌تر نگاشت می‌کند و رمزگشا ورودی را از این نمایش بازسازی می‌نماید.</li>
<li><b>یادگیری ویژگی (Feature Learning):</b> مدل‌های یادگیری عمیق، با یادگیری نمایش‌های سلسله‌مراتبی، می‌توانند ویژگی‌ها و روابط پیچیده را در داده‌های با ابعاد بالا ثبت کنند. این امر با تمرکز بر الگوهای مرتبط، تشخیص مؤثرتر داده پرت را ممکن می‌سازد.</li>
<li><b>تبدیلات غیرخطی (Nonlinear Transformations):</b> مدل‌های یادگیری عمیق قادر به ثبت روابط غیرخطی در داده‌ها هستند، که به ویژه در فضاهای با ابعاد بالا که روش‌های خطی ممکن است کارایی نداشته باشند، ارزشمند است.</li>
<li><b>نفرین ابعاد (Curse of Dimensionality):</b> یادگیری عمیق با کاهش مؤثر ابعاد و تمرکز بر ویژگی‌های مهم، می‌تواند اثرات منفی نفرین ابعاد را کاهش دهد و تحلیل داده‌های پرت را بهبود بخشد.</li>
<li><b>کاربردها (Applications):</b> مهار ابعاد بالا با یادگیری عمیق در حوزه‌های مختلفی از جمله پردازش تصویر، ژنومیک و هر رشته‌ای که با داده‌های پیچیده و با ابعاد بالا سروکار دارد و روش‌های سنتی ممکن است در آن دچار مشکل شوند، مفید است.</li>
</ul>

<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای مهار ابعاد بالا با یادگیری عمیق با استفاده از یک خودرمزگذار ارائه دهیم:</b>
</p>

```python
# آموزش: تشخیص نقاط پرت در داده‌های با ابعاد بالا با استفاده از Autoencoder

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1️⃣ تولید داده‌های مصنوعی با ۲۰ ویژگی (۱۰ اطلاع‌رسان، ۵ افزونه)
data, _ = make_classification(
    n_samples=500,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    n_clusters_per_class=1,
    random_state=42
)

# 2️⃣ تولید ۱۰ نقطه پرت با پراکندگی بالا
outliers = np.random.normal(loc=0, scale=5, size=(10, 20))
data = np.vstack([data, outliers])

# 3️⃣ استانداردسازی داده‌ها برای آموزش پایدار شبکه
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data)

# 4️⃣ تعریف و آموزش مدل Autoencoder
input_dim = data_standardized.shape[1]
model = Sequential([
    Dense(15, activation='relu', input_dim=input_dim),
    Dense(8, activation='relu'),
    Dense(15, activation='relu'),
    Dense(input_dim, activation='linear')
])
model.compile(optimizer='adam', loss='mean_squared_error')
history = model.fit(
    data_standardized, data_standardized,
    epochs=50, batch_size=32,
    shuffle=True, validation_split=0.2,
    verbose=0
)

# 5️⃣ بازسازی داده‌ها و محاسبه خطای بازسازی برای هر نمونه
decoded_data = model.predict(data_standardized)
mse = np.mean(np.square(data_standardized - decoded_data), axis=1)

# 6️⃣ تعیین آستانه برای تشخیص نقاط پرت (میانگین + ۲ انحراف معیار)
threshold = np.mean(mse) + 2*np.std(mse)
outliers = mse > threshold

# 7️⃣ نمایش تصویری داده‌ها با رنگ‌بندی بر اساس پرت بودن
plt.figure(figsize=(10, 6))
plt.scatter(data_standardized[:, 0], data_standardized[:, 1],
            c=outliers, cmap='coolwarm', alpha=0.7, s=60, edgecolor='k')
plt.title('Outlier Detection using Autoencoder in High-Dimensional Data')
plt.xlabel('Feature 1 (Standardized)')
plt.ylabel('Feature 2 (Standardized)')
plt.colorbar(label="Outlier (1=True, 0=False)")
plt.show()
```

</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از مدل <b>Autoencoder</b> برای تشخیص نقاط پرت در داده‌های با ابعاد بالا استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ویژگی‌های اطلاع‌رسان و افزونه تولید شده‌اند و نقاط پرت به‌صورت دستی اضافه شده‌اند.</li>
<li style="font-size:16pt;">داده‌ها با استفاده از <code>StandardScaler</code> استانداردسازی شده‌اند تا آموزش شبکه پایدارتر باشد.</li>
<li style="font-size:16pt;">مدل Autoencoder با ساختار ساده آموزش دیده تا داده‌ها را بازسازی کند.</li>

<li style="font-size:16pt;">نتایج با استفاده از نمودار پراکندگی نمایش داده شده‌اند تا نقاط پرت به‌صورت رنگی قابل تشخیص باشند.</li>

<p style="font-size:18pt; text-align:right;">
مدل‌های Autoencoder برای تشخیص ناهنجاری در داده‌های پیچیده و بدون برچسب بسیار مؤثر هستند. این روش به‌ویژه در مسائل واقعی مانند تشخیص تقلب، پایش سلامت تجهیزات، و تحلیل رفتار کاربران کاربرد دارد.

### 11.7.5. Modeling high-dimensional outliers

<div dir="rtl" style="font-family: Vazirmatn; line-height: 300%; background: #fafafa; border: 1px solid #ccc; border-radius: 12px; padding: 25px; text-align: right;">

<p style="font-size: 18pt; text-align: center;">
<span style="font-size: 24pt; font-weight: bold; color: #1B263B;">مدل‌سازی داده‌های پرت با ابعاد بالا (Modeling High-Dimensional Outliers)</span> 📊
</p>

<p style="font-size: 16pt; text-align: right;">
مدل‌سازی داده‌های پرت با ابعاد بالا (High-Dimensional Outliers) شامل استفاده از مدل‌های آماری و یادگیری ماشین است که به طور خاص برای ثبت الگوهای پیچیده موجود در داده‌های با ابعاد بالا طراحی شده‌اند. این مدل‌ها با هدف تمایز بین نمونه‌های عادی و ناهنجار (Anomalous Instances) به شیوه‌ای مؤثر برای مجموعه‌های داده با تعداد زیادی ویژگی، صورت می‌گیرد.
</p>

<p style="font-size: 16pt; text-align: right;">
<b>نکات کلیدی درباره مدل‌سازی داده‌های پرت با ابعاد بالا:</b>
</p>

<ul style="font-size: 16pt;">
<li><b>فرضیات توزیعی (Distributional Assumptions):</b> برخی مدل‌ها یک توزیع خاص از داده‌های عادی را فرض می‌کنند و داده‌های پرت را به عنوان نمونه‌هایی شناسایی می‌کنند که به طور قابل توجهی از این توزیع فرضی منحرف می‌شوند. با این حال، در فضاهای با ابعاد بالا، انتخاب یک توزیع مناسب چالش‌برانگیز می‌شود.</li>
<li><b>معیارهای فاصله (Distance Metrics):</b> روش‌های مبتنی بر فاصله، مانند فاصله ماهالانوبیس (Mahalanobis Distance)، برای در نظر گرفتن همبستگی‌ها در میان ویژگی‌ها در مجموعه‌های داده با ابعاد بالا تطبیق داده شده‌اند. این روش‌ها فاصله یک نقطه داده از مرکز توزیع را با در نظر گرفتن ساختار کوواریانس (Covariance) اندازه‌گیری می‌کنند.</li>
<li><b>تخمین‌گرهای مقاوم (Robust Estimators):</b> تخمین‌گرهای آماری مقاوم، مانند تعیین‌کننده کوواریانس حداقل (Minimum Covariance Determinant - MCD)، طوری طراحی شده‌اند که حساسیت کمتری نسبت به داده‌های پرت داشته باشند. آن‌ها می‌توانند به ویژه در سناریوهای با ابعاد بالا که حضور داده‌های پرت می‌تواند تخمین‌گرهای سنتی را مخدوش کند، مفید باشند.</li>
<li><b>مدل‌های یادگیری ماشین (Machine Learning Models):</b> روش‌های ترکیبی (Ensemble Methods)، ماشین‌های بردار پشتیبان (Support Vector Machines - SVM) و سایر مدل‌های یادگیری ماشین می‌توانند برای تشخیص داده پرت با ابعاد بالا، با در نظر گرفتن چالش‌های ناشی از پدیده نفرین ابعاد (Curse of Dimensionality)، تطبیق داده یا تخصصی شوند.</li>
</ul>

<p style="font-size: 16pt; text-align: right;">
<b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای مدل‌سازی داده‌های پرت با ابعاد بالا با استفاده از تخمین‌گر تعیین‌کننده کوواریانس حداقل (Minimum Covariance Determinant - MCD) ارائه دهیم:</b>
</p>

</div>


```python
# آموزش: تشخیص نقاط پرت در داده‌های با ابعاد بالا با استفاده از MCD و فاصله Mahalanobis

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.preprocessing import StandardScaler
from sklearn.covariance import EllipticEnvelope
from scipy.spatial.distance import mahalanobis
from sklearn.decomposition import PCA

# 1️⃣ تولید داده‌های مصنوعی با ۲۰ ویژگی (۱۰ اطلاع‌رسان، ۵ افزونه)
data, _ = make_classification(
    n_samples=500,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    n_clusters_per_class=1,
    random_state=42
)

# 2️⃣ تولید ۱۰ نقطه پرت با پراکندگی بالا
outliers = np.random.normal(loc=0, scale=5, size=(10, 20))
data = np.vstack([data, outliers])

# 3️⃣ استانداردسازی داده‌ها برای پایداری آماری
scaler = StandardScaler()
data_std = scaler.fit_transform(data)

# 4️⃣ کاهش ابعاد با PCA برای تثبیت تخمین کوواریانس
pca = PCA(n_components=10)
data_pca = pca.fit_transform(data_std)

# 5️⃣ اعمال الگوریتم MCD برای تخمین کوواریانس مقاوم
mcd = EllipticEnvelope(contamination=0.02, support_fraction=0.95, random_state=42)
mcd.fit(data_pca)
labels = mcd.predict(data_pca)  # برچسب‌های -1 (پرت) و 1 (نرمال)

# 6️⃣ محاسبه فاصله Mahalanobis برای هر نمونه
mean = mcd.location_
cov = mcd.covariance_
inv_cov = np.linalg.pinv(cov)
mahal_dist = np.array([mahalanobis(x, mean, inv_cov) for x in data_pca])

# 7️⃣ نرمال‌سازی فاصله‌ها برای نمایش تصویری
mahal_dist_norm = (mahal_dist - mahal_dist.min()) / (mahal_dist.max() - mahal_dist.min())

# 8️⃣ نمایش تصویری داده‌ها با رنگ و اندازه متناسب با شدت پرت بودن
plt.figure(figsize=(10, 6))
sc = plt.scatter(
    data_std[:, 0], data_std[:, 1],
    c=mahal_dist_norm,
    s=20 + mahal_dist_norm*80,
    cmap='plasma', alpha=0.7, edgecolor='k'
)
plt.title('High-Dimensional Outlier Detection using MCD & Mahalanobis Distance')
plt.xlabel('Feature 1 (Standardized)')
plt.ylabel('Feature 2 (Standardized)')

# 9️⃣ افزودن colorbar و برجسته‌سازی نقاط پرت
cbar = plt.colorbar(sc)
cbar.set_label('Normalized Mahalanobis Distance (Outlier Strength)')
plt.scatter(
    data_std[labels==-1, 0],
    data_std[labels==-1, 1],
    facecolors='none', edgecolors='red', s=100,
    label='Detected Outliers (MCD)'
)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

<p style="font-size:18pt; text-align:right;">
در این مثال، از روش آماری <b>Minimum Covariance Determinant (MCD)</b> برای تشخیص نقاط پرت در داده‌های با ابعاد بالا استفاده شده است. مراحل انجام‌شده به شرح زیر است:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ویژگی‌های اطلاع‌رسان و افزونه تولید شده‌اند و نقاط پرت به‌صورت دستی اضافه شده‌اند.</li>

<li style="font-size:16pt;">برای هر نمونه، فاصله Mahalanobis نسبت به مرکز داده محاسبه شده و نرمال‌سازی شده است.</li>
<li style="font-size:16pt;">در نمودار نهایی، رنگ و اندازه هر نقطه نشان‌دهنده شدت پرت بودن آن است و نقاط پرت با حاشیه قرمز برجسته شده‌اند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
روش MCD برای تشخیص ناهنجاری‌های آماری در داده‌های چندبعدی بسیار مناسب است، به‌ویژه زمانی که داده‌ها شامل نویز یا نقاط پرت شدید باشند. ترکیب آن با فاصله Mahalanobis و کاهش بعد می‌تواند در مسائل واقعی مانند تشخیص تقلب، تحلیل داده‌های مالی، یا پایش کیفیت داده‌ها کاربرد داشته باشد.
