### 9.1.1. Fuzzy clusters

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی فازی (Fuzzy Clustering)</span> 🧩
</p>

<p style="font-size:22pt; text-align:right;">
<br>
خوشه‌بندی فازی (Fuzzy clustering) یک گسترش از روش‌های خوشه‌بندی سنتی است که به نقاط داده‌ای اجازه می‌دهد به طور همزمان به چندین خوشه، هر کدام با یک درجه عضویت (degree of membership)، تعلق داشته باشند. برخلاف خوشه‌بندی سخت (hard clustering) که در آن یک نقطه به طور انحصاری به یک خوشه واحد تعلق دارد، خوشه‌بندی فازی یک درجه عدم قطعیت را معرفی می‌کند. مدل اصلی مورد استفاده برای خوشه‌بندی فازی، الگوریتم <b>فازی سی-مینز (Fuzzy C-Means - FCM)</b> است.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
در FCM، هر نقطه داده دارای مقادیر عضویت (membership values) در تمام خوشه‌ها است که به صورت یک ماتریس عضویت فازی (fuzzy membership matrix) نمایش داده می‌شود. این مقادیر عضویت بین ۰ و ۱ متغیر هستند که نشان‌دهنده احتمال تعلق یک نقطه داده‌ای به یک خوشه خاص است. خوشه‌بندی فازی به ویژه زمانی مفید است که نقاط داده‌ای عضویت جزئی (partial membership) به چندین خوشه را نشان می‌دهند.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی فازی با استفاده از الگوریتم Fuzzy C-Means:</b>
</p>

</div>

```python
pip install scikit-learn-extra
```

```python
# وارد کردن کتابخانه‌های مورد نیاز
import numpy as np
from sklearn.datasets import make_blobs
from sklearn_extra.cluster import KMedoids
import matplotlib.pyplot as plt

# تعریف تابع خوشه‌بندی Fuzzy C-Means با مقداردهی اولیه از طریق K-Medoids
def fuzzy_c_means(X, n_clusters, m=2, max_iter=100, tol=1e-4):
    # 1. مقداردهی اولیه مراکز خوشه با استفاده از الگوریتم K-Medoids
    kmedoids = KMedoids(n_clusters=n_clusters, random_state=42)
    kmedoids.fit(X)
    centers = kmedoids.cluster_centers_

    for _ in range(max_iter):
        # 2. محاسبه فاصله بین هر نقطه و مراکز خوشه (ماتریس N×K)
        distances = np.linalg.norm(X[:, None] - centers[None, :], axis=2) + 1e-10  # جلوگیری از تقسیم بر صفر

        # 3. محاسبه ماتریس عضویت فازی (Fuzzy Partition Matrix)
        membership_matrix = 1.0 / distances ** (2 / (m - 1))
        membership_matrix = membership_matrix / np.sum(membership_matrix, axis=1, keepdims=True)

        # 4. به‌روزرسانی مراکز خوشه بر اساس وزن عضویت‌ها
        new_centers = (membership_matrix.T @ X) / np.sum(membership_matrix, axis=0)[:, None]

        # 5. بررسی شرط همگرایی
        if np.linalg.norm(new_centers - centers) < tol:
            break
        centers = new_centers

    # 6. تخصیص برچسب نهایی بر اساس بیشترین عضویت
    labels = np.argmax(membership_matrix, axis=1)
    return labels, centers, membership_matrix

# تولید داده‌های مصنوعی با سه خوشه
X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0)

# اعمال الگوریتم Fuzzy C-Means روی داده‌ها
n_clusters = 3
fuzzy_labels, fuzzy_centers, U = fuzzy_c_means(X, n_clusters)

# رسم نتایج خوشه‌بندی
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=fuzzy_labels, cmap='viridis', alpha=0.7)
plt.scatter(fuzzy_centers[:, 0], fuzzy_centers[:, 1], c='red', marker='X', s=200, label='Cluster Centers')
plt.title('Fuzzy C-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧮✨ </p> <p style="font-size:18pt; text-align:right;"> در این مثال، الگوریتم <b>خوشه‌بندی فازی C-Means</b> برای گروه‌بندی داده‌های مصنوعی به کار گرفته شده است. برخلاف خوشه‌بندی سخت (مانند K-Means)، در این روش هر نقطه می‌تواند به چند خوشه با درجات مختلف تعلق داشته باشد: <ul> <li style="font-size:16pt;">ابتدا داده‌های مصنوعی با سه خوشه مشخص تولید شده‌اند تا سناریوی خوشه‌بندی شبیه‌سازی شود.</li> <li style="font-size:16pt;">برای مقداردهی اولیه مراکز خوشه‌ها، از الگوریتم <b>K-Medoids</b> استفاده شده که نسبت به K-Means مقاوم‌تر به داده‌های پرت است.</li> <li style="font-size:16pt;">در هر تکرار، فاصله بین نقاط و مراکز خوشه محاسبه شده و سپس <b>ماتریس عضویت فازی</b> ساخته می‌شود که نشان می‌دهد هر نقطه با چه درجه‌ای به هر خوشه تعلق دارد.</li> <li style="font-size:16pt;">مراکز خوشه‌ها بر اساس وزن عضویت‌ها به‌روزرسانی می‌شوند تا زمانی که تغییرات مراکز کمتر از آستانه همگرایی شود.</li> <li style="font-size:16pt;">در پایان، برچسب هر نقطه بر اساس بیشترین عضویت تعیین شده و نتایج خوشه‌بندی به صورت تصویری نمایش داده شده‌اند.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>خوشه‌بندی فازی</b>، ساختارهای پیچیده‌تری از داده‌ها را مدل‌سازی کرد. این روش به‌ویژه در کاربردهایی که مرز بین خوشه‌ها واضح نیست یا داده‌ها دارای هم‌پوشانی هستند، بسیار مؤثر است. </p> </p> </div>

### 9.1.2. Probabilistic model-based clusters

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی مبتنی بر مدل احتمالی (Probabilistic Model-Based Clustering)</span> 📈
</p>

<p style="font-size:22pt; text-align:right;">
<br>
خوشه‌بندی مبتنی بر مدل احتمالی شامل استفاده از مدل‌های آماری برای توصیف توزیع اصلی داده‌ها و شناسایی خوشه‌ها بر اساس پارامترهای این مدل‌ها است. یک رویکرد رایج، مدل ترکیبی گوسی (Gaussian Mixture Model - GMM) است، که در آن فرض می‌شود نقاط داده‌ای از ترکیبی از چندین توزیع گوسی تولید شده‌اند.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
در خوشه‌بندی مبتنی بر GMM، هر خوشه با یک توزیع گوسی مرتبط است که با میانگین (mean)، ماتریس کوواریانس (covariance matrix) و وزن آن مشخص می‌شود. این مدل امکان انعطاف‌پذیری بیشتری در درک شکل‌های پیچیده خوشه را فراهم می‌کند و یک توزیع احتمال بر روی انتساب خوشه‌ها ارائه می‌دهد. الگوریتم Expectation-Maximization - EM اغلب برای تخمین پارامترها در GMMها به کار گرفته می‌شود.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی مبتنی بر مدل احتمالی با استفاده از مدل ترکیبی گوسی:</b>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt
import numpy as np

# --- تولید داده‌های مصنوعی با ۳ خوشه ---
X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0)

# --- اعمال مدل آماری GMM با ۳ مؤلفه ---
n_components = 3
gmm = GaussianMixture(n_components=n_components, random_state=42)
gmm.fit(X)

# --- محاسبه احتمال تعلق هر نمونه به هر خوشه ---
probabilities = gmm.predict_proba(X)
labels = np.argmax(probabilities, axis=1)  # انتخاب خوشه با بیشترین احتمال

# --- رسم نمودار پراکندگی داده‌ها و مراکز خوشه‌ها ---
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c='red', marker='X', s=200, label='Cluster Centers')
plt.title('Gaussian Mixture Model Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🎲
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از مدل <b>Gaussian Mixture Model (GMM)</b> برای خوشه‌بندی داده‌های مصنوعی استفاده شده است. برخلاف K-Means که هر نقطه را به یک خوشه خاص اختصاص می‌دهد، GMM به هر نمونه یک <b>احتمال تعلق</b به هر خوشه اختصاص می‌دهد:
<ul>
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با سه خوشه تولید شده‌اند.</li>
<li style="font-size:16pt;">مدل GMM با سه مؤلفه آموزش دیده و پارامترهای آماری هر خوشه (میانگین و کوواریانس) برآورد شده‌اند.</li>
<li style="font-size:16pt;">برای هر نمونه، احتمال تعلق به هر خوشه محاسبه شده و بر اساس بیشترین احتمال، برچسب خوشه تعیین شده است.</li>
<li style="font-size:16pt;">در نمودار، نقاط داده با رنگ خوشه‌ها و مراکز خوشه‌ها با علامت <b>X قرمز</b> نمایش داده شده‌اند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>مدل‌های آماری احتمالاتی</b>، خوشه‌بندی دقیق‌تری انجام داد. GMM به‌ویژه در داده‌هایی که خوشه‌ها هم‌پوشانی دارند یا شکل آن‌ها غیر کروی است، عملکرد بهتری نسبت به K-Means دارد.
</p>
</p>

### 9.1.3. Expectation-maximization algorithm

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">الگوریتم Expectation-Maximization (EM)</span> 🔄
</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم Expectation-Maximization (EM) یک چارچوب کلی برای یافتن تخمین‌های حداکثر درست‌نمایی (maximum likelihood estimates) از پارامترها در مدل‌های احتمالی (probabilistic models) با متغیرهای نهان (latent variables) است. در زمینه خوشه‌بندی مبتنی بر مدل احتمالی (probabilistic model-based clustering)، الگوریتم EM اغلب برای تخمین پارامتر در مدل‌های ترکیبی گوسی (Gaussian Mixture Models - GMMs) استفاده می‌شود.
<br><br>
در اینجا یک نمای کلی از الگوریتم EM در زمینه خوشه‌بندی مبتنی بر GMM آورده شده است:
<ul style="font-size:16pt;">
<li><b>گام امید (Expectation Step - E-step):</b> مقدار مورد انتظار (expected value) متغیرهای نهان (انتساب خوشه‌ها) را با توجه به داده‌های مشاهده شده و تخمین‌های فعلی پارامتر محاسبه کنید.</li>
<li><b>گام حداکثرسازی (Maximization Step - M-step):</b> پارامترهای توزیع‌های گوسی (میانگین، کوواریانس و وزن‌ها) را بر اساس داده‌های مشاهده شده و مقادیر مورد انتظار به دست آمده در گام E-step به‌روزرسانی کنید.</li>
<li><b>تکرار (Iteration):</b> گام‌های E و M را تا رسیدن به همگرایی (convergence)، جایی که تغییر در درست‌نمایی لگاریتمی (log-likelihood) یا پارامترها کمتر از یک آستانه از پیش تعریف شده باشد، تکرار کنید.</li>
</ul>
<br>
<b> الگوریتم EM به صورت تکراری تخمین‌های پارامتر را اصلاح می‌کند و درست‌نمایی داده‌های مشاهده شده را با توجه به مدل حداکثر می‌سازد.</b>
<br><br>
<b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای الگوریتم Expectation-Maximization اعمال شده بر روی مدل‌های ترکیبی گوسی:</b>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt
import numpy as np

# --- تولید داده‌های مصنوعی با ۳ خوشه ---
X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0)

# --- اعمال مدل آماری GMM با الگوریتم EM ---
n_components = 3
gmm = GaussianMixture(n_components=n_components, random_state=42)
gmm.fit(X)
labels = gmm.predict(X)  # برچسب خوشه‌ها بر اساس بیشترین احتمال تعلق

# --- رسم نمودار پراکندگی داده‌ها و مراکز خوشه‌ها ---
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c='red', marker='X', s=200, label='Cluster Centers')
plt.title('Gaussian Mixture Model Clustering with EM Algorithm')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

در این مثال، از مدل <b>Gaussian Mixture Model (GMM)</b> برای خوشه‌بندی داده‌های مصنوعی استفاده شده است. این مدل با استفاده از الگوریتم <b>Expectation-Maximization (EM)</b> پارامترهای آماری هر خوشه را تخمین می‌زند:
<ul>
<li style="font-size:16pt;">ابتدا داده‌هایی با سه خوشه تولید شده‌اند که به‌صورت تصادفی در فضای دوبعدی پخش شده‌اند.</li>
<li style="font-size:16pt;">مدل GMM با سه مؤلفه آموزش دیده و برای هر نمونه، احتمال تعلق به هر خوشه محاسبه شده است.</li>
<li style="font-size:16pt;">برچسب نهایی هر نمونه بر اساس بیشترین احتمال تعلق تعیین شده است.</li>
<li style="font-size:16pt;">در نمودار، نقاط داده با رنگ خوشه‌ها و مراکز خوشه‌ها با علامت <b>X قرمز</b> نمایش داده شده‌اند.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این روش خوشه‌بندی مبتنی بر مدل‌های آماری است و نسبت به K-Means انعطاف‌پذیری بیشتری دارد، به‌ویژه زمانی که خوشه‌ها هم‌پوشانی دارند یا شکل آن‌ها غیر کروی است.
</p>

### 9.2.2. Axis-parallel subspace approaches

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی زیرفضای موازی با محور (Axis-parallel subspace clustering)</span> 📐
</p>

<p style="font-size:22pt; text-align:right;">
<br>
خوشه‌بندی داده‌های با ابعاد بالا چالش‌های منحصربه‌فردی را به همراه دارد، مانند نفرین ابعاد (curse of dimensionality) و وجود ویژگی‌های نامرتبط یا اضافی. رویکردهای زیرفضا (subspace) موازی با محور (Axis-parallel) با هدف رسیدگی به این چالش‌ها از طریق شناسایی زیرفضاها در فضای ویژگی با ابعاد بالا که در آن خوشه‌ها واضح‌تر هستند، عمل می‌کنند.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>ویژگی‌های کلیدی رویکردهای زیرفضای موازی با محور عبارتند از:</b>
</p>
<ul style="font-size:16pt; text-align:right;">
<li><b>انتخاب زیرمجموعه ویژگی (Feature Subset Selection):</b> این روش‌ها شامل انتخاب یک زیرمجموعه از ویژگی‌ها یا ابعاد است که برای خوشه‌بندی آموزنده‌ترین هستند، و ابعاد نامرتبط یا اضافی را کنار می‌گذارند.</li>
<li><b>خوشه‌بندی موازی با محور (Axis-Parallel Clustering):</b> خوشه‌بندی در زیرفضای انتخاب شده با استفاده از روش‌هایی انجام می‌شود که به هندسه داده‌ها در امتداد هر محور حساس هستند.</li>
<li><b>نمایش و بصری‌سازی (Projection and Visualization):</b> خوشه‌بندی زیرفضا اغلب شامل نمایش داده‌ها بر روی زیرفضای انتخاب شده برای بصری‌سازی و تفسیر بهتر است.</li>
</ul>
<br>
<p style="font-size:22pt; text-align:right;">
تکنیک‌های رایج برای خوشه‌بندی زیرفضای موازی با محور شامل خوشه‌بندی زیرفضا بر اساس k-means یا خوشه‌بندی مبتنی بر چگالی (density-based clustering) در داخل زیرفضاها است.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی زیرفضای موازی با محور با استفاده از الگوریتم k-means را ارائه دهیم:
</p>

</div>

```python
# وارد کردن کتابخانه‌های مورد نیاز
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# تولید داده‌های مصنوعی با ابعاد بالا (۱۰ ویژگی) و ۳ خوشه
X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0, n_features=10)

# اعمال تحلیل مؤلفه‌های اصلی (PCA) برای کاهش ابعاد به ۲ مؤلفه اصلی
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# اعمال خوشه‌بندی KMeans روی فضای کاهش‌یافته
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_pca)

# رسم نتایج خوشه‌بندی در فضای دو مؤلفه اصلی
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='X', s=200, label='Cluster Centers')
plt.title('KMeans Clustering in Reduced Subspace using PCA')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠📉 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، ترکیب دو تکنیک قدرتمند یعنی <b>تحلیل مؤلفه‌های اصلی (PCA)</b> و <b>خوشه‌بندی KMeans</b> برای تحلیل داده‌های مصنوعی با ابعاد بالا به کار گرفته شده است: <ul> <li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ۱۰ ویژگی و ۳ خوشه تولید شده‌اند تا سناریوی خوشه‌بندی در فضای چندبعدی شبیه‌سازی شود.</li> <li style="font-size:16pt;">برای کاهش پیچیدگی و امکان تصویرسازی، از <b>PCA</b> برای کاهش ابعاد به ۲ مؤلفه اصلی استفاده شده است. این مؤلفه‌ها بیشترین واریانس داده را حفظ می‌کنند.</li> <li style="font-size:16pt;">سپس الگوریتم <b>KMeans</b> روی داده‌های کاهش‌یافته اعمال شده تا خوشه‌ها در فضای جدید شناسایی شوند.</li> <li style="font-size:16pt;">در پایان، نقاط داده با رنگ‌بندی بر اساس خوشه‌ها و مراکز خوشه با علامت قرمز رسم شده‌اند تا ساختار خوشه‌بندی به‌وضوح قابل مشاهده باشد.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>کاهش ابعاد</b>، داده‌های پیچیده را ساده‌سازی کرد و سپس با <b>خوشه‌بندی</b>، الگوهای پنهان در آن‌ها را کشف نمود. این ترکیب در بسیاری از کاربردهای عملی مانند تحلیل رفتار مشتری، فشرده‌سازی داده و تصویرسازی داده‌های چندبعدی بسیار مؤثر است. </p> </p> </div>

### 9.2.3. Arbitrarily oriented subspace approaches

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">رویکردهای زیرفضای با جهت‌گیری دلخواه (Arbitrarily oriented subspace approaches)</span> 🌀
</p>

<p style="font-size:22pt; text-align:right;">
<br>
رویکردهای زیرفضای با جهت‌گیری دلخواه (Arbitrarily oriented subspace approaches) به چالش‌های خوشه‌بندی داده‌های با ابعاد بالا با شناسایی زیرفضاهایی که لزوماً با محورهای فضای ویژگی اصلی هم‌راستا نیستند، می‌پردازند. برخلاف رویکردهای موازی با محور، این روش‌ها زیرفضاها را با جهت‌گیری‌های دلخواه در نظر می‌گیرند که انعطاف‌پذیری بیشتری را در ثبت روابط پیچیده در داده‌ها فراهم می‌سازد.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>ویژگی‌های کلیدی رویکردهای زیرفضای با جهت‌گیری دلخواه عبارتند از:</b>
</p>
<ul style="font-size:16pt; text-align:right;">
<li><b>تعریف زیرفضا (Subspace Definition):</b> این روش‌ها امکان تعریف زیرفضاها را بر اساس ترکیب‌های خطی (linear combinations) از ویژگی‌های اصلی فراهم می‌سازند و شناسایی خوشه‌ها را در زیرفضاهای غیرموازی با محور ممکن می‌سازند.</li>
<li><b>تکنیک‌های خوشه‌بندی زیرفضا (Subspace Clustering Techniques):</b> تکنیک‌هایی مانند خوشه‌بندی زیرفضا بر اساس انتشار وابستگی (affinity propagation) یا خوشه‌بندی طیفی (spectral clustering) اغلب برای شناسایی خوشه‌ها در داخل زیرفضاهای تعریف شده به کار گرفته می‌شوند.</li>
<li><b>مقاومت در برابر افزونگی ویژگی (Robustness to Feature Redundancy):</b> رویکردهای زیرفضای با جهت‌گیری دلخواه عموماً در برابر افزونگی ویژگی مقاوم‌تر هستند، که آن‌ها را برای مجموعه‌های داده با ابعاد بالا و ویژگی‌های همبسته (correlated features) مناسب می‌سازد.</li>
</ul>
<br>
<p style="font-size:22pt; text-align:right;">
اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی زیرفضای با جهت‌گیری دلخواه با استفاده از الگوریتم خوشه‌بندی طیفی (Spectral Clustering) ارائه دهیم:
</p>

</div>

```python
# وارد کردن کتابخانه‌های مورد نیاز
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
from sklearn.cluster import SpectralClustering
import matplotlib.pyplot as plt
import numpy as np

# مرحله ۱: تولید داده‌های مصنوعی با ابعاد بالا
X, _ = make_blobs(
    n_samples=300, 
    centers=3, 
    n_features=10, 
    random_state=42, 
    cluster_std=1.0
)

# مرحله ۲: کاهش ابعاد با PCA برای تصویرسازی
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# مرحله ۳a: خوشه‌بندی طیفی با affinity از نوع RBF (هسته گاوسی)
spectral_rbf = SpectralClustering(
    n_clusters=3, 
    affinity="rbf", 
    random_state=42
)
labels_rbf = spectral_rbf.fit_predict(X)

# مرحله ۳b: خوشه‌بندی طیفی با affinity از نوع Nearest Neighbors

# affinity="nearest_neighbors" یعنی:
# - یک graph روی k نزدیک‌ترین همسایه‌ها ساخته میشه
# - eigenvectors روی این graph محاسبه میشه
# نکته علمی/technical:
# - Spectral Clustering روی یک گراف connected درست کار می‌کنه
# - اگر graph disconnected باشه، eigenvectors درست محاسبه نمیشن و warning میده
# انتخاب n_neighbors:
# - باید به اندازه‌ای باشه که graph connected بشه
# - rule of thumb: بین log(n_samples) تا sqrt(n_samples)
# - در practice: کمی بزرگ‌تر می‌گیریم تا connectivity تضمین بشه
# trade-off:
# - connectivity بالا ولی ممکنه اطلاعات محلی کمی غرق بشه

spectral_nn = SpectralClustering(
    n_clusters=3, 
    affinity="nearest_neighbors", 
    n_neighbors=120,  # تضمین اتصال گراف برای ۳۰۰ نمونه
    random_state=42
)
labels_nn = spectral_nn.fit_predict(X)

# مرحله ۴: رسم نتایج خوشه‌بندی در فضای کاهش‌یافته
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# رسم نتایج RBF
axes[0].scatter(
    X_pca[:, 0], 
    X_pca[:, 1], 
    c=labels_rbf, 
    cmap="viridis", 
    alpha=0.7
)
axes[0].set_title("Spectral Clustering (RBF affinity)")
axes[0].set_xlabel("PC 1")
axes[0].set_ylabel("PC 2")

# رسم نتایج Nearest Neighbors
axes[1].scatter(
    X_pca[:, 0], 
    X_pca[:, 1], 
    c=labels_nn, 
    cmap="viridis", 
    alpha=0.7
)
axes[1].set_title("Spectral Clustering (Nearest Neighbors, k=120)")
axes[1].set_xlabel("PC 1")
axes[1].set_ylabel("PC 2")

# بهبود چیدمان نمودارها
plt.tight_layout()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> 
<p style="font-size:20pt; text-align:center;"> 
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🌐🔍 
</p> 

<p style="font-size:18pt; text-align:right;"> 
در این مثال، الگوریتم <b>خوشه‌بندی طیفی (Spectral Clustering)</b> روی داده‌های مصنوعی با ابعاد بالا اعمال شده و نتایج آن با استفاده از <b>PCA</b> در فضای دوبعدی تصویرسازی شده‌اند. این روش برای شناسایی ساختارهای پیچیده و غیرخطی در داده‌ها بسیار مؤثر است:
<ul> 
<li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ۱۰ ویژگی و ۳ خوشه تولید شده‌اند تا سناریوی خوشه‌بندی در فضای چندبعدی شبیه‌سازی شود. این مرحله مشابه بسیاری از مطالعات علمی و پروژه‌های صنعتی است که نیاز به شبیه‌سازی داده‌های high-dimensional دارند.</li> 

<li style="font-size:16pt;">برای کاهش ابعاد و امکان تصویرسازی، از <b>PCA</b> استفاده شده است. PCA مؤلفه‌های اصلی داده را استخراج می‌کند و واریانس بیشینه را حفظ می‌نماید تا ساختار اصلی داده‌ها از بین نرود. این مرحله هم در تحقیقات علمی و هم در کاربردهای صنعتی برای فهم بهتر داده‌ها ضروری است.</li> 

<li style="font-size:16pt;">دو نوع خوشه‌بندی طیفی اعمال شده‌اند:
<ul> 
<li><b>RBF affinity</b>: با هسته گاوسی، مناسب برای داده‌های غیرخطی و خوشه‌های پیچیده است و در بسیاری از مقالات و پروژه‌های عملی استاندارد محسوب می‌شود.</li> 
<li><b>Nearest Neighbors affinity</b>: ساخت گراف روی k نزدیک‌ترین همسایه‌ها و حفظ ساختار محلی داده‌ها. برای این روش، تعداد همسایه‌ها به گونه‌ای انتخاب شده که گراف متصل باشد تا eigenvectorها به‌درستی محاسبه شوند.</li> 
</ul> 
</li> 

<li style="font-size:16pt;">انتخاب <b>n_neighbors</b> بر اساس تعداد نمونه‌ها انجام شده است. قاعده تقریبی: بین log(n_samples) تا sqrt(n_samples) و در عمل کمی بزرگ‌تر برای تضمین connectivity انتخاب می‌شود. این نکته بسیار مهم است چرا که Spectral Clustering روی گراف disconnected نتایج نادرست و هشدار می‌دهد.</li> 

<li style="font-size:16pt;">نتایج هر دو روش در فضای دوبعدی رسم شده‌اند تا تفاوت ساختار خوشه‌ها به‌وضوح قابل مشاهده باشد و تحلیل نتایج راحت‌تر شود.</li> 
</ul> 

<p style="font-size:18pt; text-align:right;"> 
این مثال نشان می‌دهد که چگونه می‌توان با ترکیب <b>کاهش ابعاد</b> و <b>خوشه‌بندی طیفی</b>، ساختارهای پنهان در داده‌های پیچیده را کشف و تحلیل کرد. این روش در کاربردهای واقعی مانند تحلیل تصاویر، داده‌های زیستی، شبکه‌های اجتماعی و تحلیل رفتار کاربران بسیار پرکاربرد است و هم در مقالات پژوهشی و هم در پروژه‌های صنعتی اهمیت بالایی دارد. 
</p>

### 9.3.2. Types of biclusters

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی دوقطبی (Biclustering)</span> 📊
</p>

<p style="font-size:22pt; text-align:right;">
<br>
خوشه‌بندی دوقطبی (Biclustering) تکنیکی است که به صورت همزمان سطرها و ستون‌های یک مجموعه داده را خوشه‌بندی می‌کند و زیرماتریس‌هایی را شناسایی می‌کند که الگوهای منسجمی از خود نشان می‌دهند. چندین نوع خوشه‌ دوقطبی بر اساس ماهیت الگوهایی که ثبت می‌کنند وجود دارد:
</p>
<br>
<ul style="font-size:16pt; text-align:right;">
<li><b>خوشه‌های دوقطبی با مقدار ثابت (Constant Value Biclusters):</b> خوشه‌های دوقطبی که در آن‌ها تمام عناصر دارای یک مقدار ثابت هستند و نشان‌دهنده یکنواختی در داخل زیرماتریس شناسایی شده است.</li>
<li><b>خوشه‌های دوقطبی با جابجایی ثابت (Shift Biclusters):</b> خوشه‌های دوقطبی که در آن‌ها مقادیر درون زیرماتریس یک جابجایی ثابت را نشان می‌دهند و الگوی کلی را حفظ می‌کنند، اما امکان تغییر را فراهم می‌سازند.</li>
<li><b>خوشه‌های دوقطبی افزایشی (Additive Biclusters):</b> خوشه‌های دوقطبی که در آن‌ها مقادیر می‌توانند به عنوان مجموع یک ثابت و یک جابجایی مدل‌سازی شوند، و الگوهای ثابت و جابجایی را به طور همزمان ثبت می‌کنند.</li>
<li><b>خوشه‌های دوقطبی ضربی (Multiplicative Biclusters):</b> خوشه‌های دوقطبی که در آن‌ها مقادیر می‌توانند به عنوان حاصل‌ضرب یک ثابت و یک جابجایی مدل‌سازی شوند، و الگوهای ثابت و مقیاس‌گذاری را ثبت می‌کنند.</li>
</ul>
<br>
<p style="font-size:22pt; text-align:right;">
درک نوع خوشه دوقطبی برای تفسیر معنای بیولوژیکی یا خاص دامنه در پس الگوهای شناسایی شده در داده‌ها ضروری است.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی دوقطبی با استفاده از کلاس Bicluster از کتابخانه sklearn ارائه دهیم:
</p>

</div>


```python
import numpy as np
from sklearn.cluster import SpectralBiclustering
import matplotlib.pyplot as plt

# --- تعریف تعداد خوشه‌های سطری و ستونی ---
n_row_clusters, n_col_clusters = 3, 3
block_size = 100  # اندازه هر بلوک

# --- ساخت الگوی شطرنجی پایه (ماتریس ۳×۳ از ۰ و ۱) ---
pattern = np.array([[0, 1, 0],
                    [1, 0, 1],
                    [0, 1, 0]])

# --- گسترش الگو با ضرب کرونکر برای ایجاد ماتریس بزرگ‌تر ---
data = np.kron(pattern, np.ones((block_size, block_size)))

# --- افزودن نویز نرمال به داده‌ها ---
noise = np.random.normal(0, 0.1, size=data.shape)
data_noisy = data + noise

# --- اعمال الگوریتم Spectral Biclustering ---
model = SpectralBiclustering(n_clusters=(n_row_clusters, n_col_clusters), random_state=42)
model.fit(data_noisy)

# --- دریافت برچسب‌های سطری و ستونی ---
rows, cols = model.row_labels_, model.column_labels_

# --- رسم داده‌ی اصلی و داده‌ی خوشه‌بندی‌شده ---
plt.figure(figsize=(12, 5))

# --- نمایش داده‌ی اصلی ---
plt.subplot(1, 2, 1)
plt.title("Original Data")
plt.imshow(data_noisy, cmap="viridis", aspect="auto")
plt.colorbar()

# --- نمایش داده‌ی خوشه‌بندی‌شده با مرتب‌سازی سطرها و ستون‌ها ---
plt.subplot(1, 2, 2)
plt.title("Biclustered Data")
plt.imshow(data_noisy[np.argsort(rows)][:, np.argsort(cols)], cmap="viridis", aspect="auto")
plt.colorbar()

plt.tight_layout()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:20pt; text-align:center;">
<span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧩
</p>
<p style="font-size:18pt; text-align:right;">
در این مثال، از الگوریتم <b>Spectral Biclustering</b> برای شناسایی خوشه‌های هم‌زمان در سطرها و ستون‌های یک ماتریس داده استفاده شده است. این الگوریتم به‌ویژه برای داده‌هایی با ساختار شطرنجی یا بلوکی مناسب است:
<ul>
<li style="font-size:16pt;">ابتدا یک الگوی شطرنجی ۳×۳ با استفاده از ضرب کرونکر گسترش داده شده تا ماتریسی بزرگ‌تر با ساختار بلوکی ایجاد شود.</li>
<li style="font-size:16pt;">برای واقع‌گراتر شدن داده‌ها، نویز نرمال به ماتریس افزوده شده است.</li>
<li style="font-size:16pt;">الگوریتم Spectral Biclustering اجرا شده و خوشه‌های سطری و ستونی استخراج شده‌اند.</li>
<li style="font-size:16pt;">در نمودار اول، داده‌ی اصلی نمایش داده شده و در نمودار دوم، داده‌ها بر اساس خوشه‌های شناسایی‌شده مرتب شده‌اند تا ساختار بلوکی بهتر دیده شود.</li>
</ul>
<p style="font-size:18pt; text-align:right;">
این روش در تحلیل داده‌های ژنومیکس، توصیه‌گرها، و ماتریس‌های تعامل کاربرد دارد و می‌تواند الگوهای پنهان بین زیرمجموعه‌های سطرها و ستون‌ها را آشکار کند.
</p>

### 9.3.3. Biclustering methods

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">روش‌های خوشه‌بندی دوقطبی (Biclustering methods)</span> 📊
</p>
<p style="font-size:16pt; text-align:right;">
روش‌های خوشه‌بندی دوقطبی (Biclustering methods) برای کشف الگوها در مجموعه‌های داده‌ای طراحی شده‌اند که در آن‌ها زیرمجموعه‌هایی از سطرها و ستون‌ها رفتار مشابهی از خود نشان می‌دهند. این روش‌ها ابزاری قدرتمند برای کشف ساختارهای پیچیده در انواع مختلف داده‌ها فراهم می‌کنند. برخی از الگوریتم‌های خوشه‌بندی دوقطبی رایج عبارتند از:
<ul style="font-size:16pt;">
<li><b>خوشه‌بندی دوقطبی طیفی (Spectral Biclustering):</b> این روش از تجزیه مقادیر ویژه (eigenvalue decomposition) برای شناسایی خوشه‌های دوقطبی در داده‌ها استفاده می‌کند. این روش هم روابط سطر و هم ستون را در نظر می‌گیرد و به ویژه برای کشف خوشه‌های دوقطبی افزایشی (additive biclusters) مؤثر است.</li>
<li><b>خوشه‌بندی دوقطبی با مدل پِلِید (Plaid Model Biclustering):</b> مدل پلید (Plaid Model) داده‌ها را به عنوان یک سوپرپوزیشن از خوشه‌های دوقطبی افزایشی نمایش می‌دهد. خوشه‌های دوقطبی با تطبیق الگوهای سطر و ستون به صورت تکراری یافت می‌شوند.</li>
<li><b>خوشه‌بندی دوقطبی بیمکس (Bimax Biclustering):</b> بیمکس (Bimax) یک روش تجزیه ماتریس دودویی (binary matrix factorization) است که با هدف یافتن زیرماتریس‌هایی که تنها شامل ۰ و ۱ هستند، عمل می‌کند. این روش برای داده‌های دودویی مناسب است.</li>
<li><b>خوشه‌بندی دوقطبی حفظ‌کننده ترتیب (Order-Preserving Biclustering):</b> این روش با حفظ ترتیب نقاط داده در سطرها و ستون‌ها، خوشه‌های دوقطبی را شناسایی می‌کند و الگوها را بر اساس روابط یکنواخت (monotonic relationships) ثبت می‌کند.</li>
</ul>
<br>
هر روش خوشه‌بندی دوقطبی برای رسیدگی به ویژگی‌های خاص داده‌ها و انواعی از الگوهایی که انتظار می‌رود یافت شوند، طراحی شده است.
<br><br>
<b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی دوقطبی با استفاده از کلاس SpectralBiclustering از کتابخانه sklearn:</b>
</p>

</div>

```python
# وارد کردن کتابخانه‌های مورد نیاز
from sklearn.datasets import make_checkerboard
from sklearn.cluster import SpectralBiclustering
import matplotlib.pyplot as plt
import numpy as np

# تولید داده‌های مصنوعی به صورت ماتریس شطرنجی با ۳ خوشه در ردیف و ۳ خوشه در ستون
data = make_checkerboard(shape=(300, 300), n_clusters=(3, 3), noise=10, random_state=42)[0]

# بررسی نوع و شکل داده‌ها
print(type(data), data.shape)  # خروجی باید numpy.ndarray با ابعاد (300, 300) باشد

# اعمال الگوریتم خوشه‌بندی طیفی دوطرفه (Spectral Biclustering)
model = SpectralBiclustering(n_clusters=(3, 3), random_state=42)
model.fit(data)

# استخراج برچسب‌های خوشه برای ردیف‌ها و ستون‌ها
rows, cols = model.row_labels_, model.column_labels_

# مرتب‌سازی داده‌ها بر اساس برچسب‌های خوشه برای نمایش واضح‌تر بلوک‌ها
sorted_data = data[np.argsort(rows)][:, np.argsort(cols)]

# رسم داده‌ها قبل و بعد از خوشه‌بندی
plt.figure(figsize=(12, 5))

# نمایش داده اولیه
plt.subplot(1, 2, 1)
plt.title('Original Data')
plt.imshow(data, cmap='viridis', aspect='auto')
plt.colorbar()

# نمایش داده مرتب‌شده بر اساس خوشه‌ها
plt.subplot(1, 2, 2)
plt.title('Biclustered Data')
plt.imshow(sorted_data, cmap='viridis', aspect='auto')
plt.colorbar()

plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧩📊 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، از الگوریتم <b>خوشه‌بندی طیفی دوطرفه (Spectral Biclustering)</b> برای شناسایی ساختار بلوکی در داده‌های مصنوعی استفاده شده است. این الگوریتم به‌طور همزمان خوشه‌هایی در ردیف‌ها و ستون‌ها پیدا می‌کند: <ul> <li style="font-size:16pt;">ابتدا داده‌ای به شکل ماتریس شطرنجی با ۳ خوشه در ردیف و ۳ خوشه در ستون تولید شده است. این نوع داده‌ها برای تست الگوریتم‌های biclustering بسیار مناسب هستند.</li> <li style="font-size:16pt;">الگوریتم <b>Spectral Biclustering</b> روی داده اعمال شده تا ساختارهای پنهان در هر دو محور شناسایی شوند. این الگوریتم از تجزیه طیفی برای یافتن الگوهای هم‌زمان در ردیف‌ها و ستون‌ها استفاده می‌کند.</li> <li style="font-size:16pt;">برچسب‌های خوشه برای ردیف‌ها و ستون‌ها استخراج شده و داده‌ها بر اساس آن‌ها مرتب شده‌اند تا بلوک‌های خوشه‌ای به‌وضوح قابل مشاهده باشند.</li> <li style="font-size:16pt;">در نهایت، داده اولیه و داده خوشه‌بندی‌شده به صورت تصویری نمایش داده شده‌اند تا تفاوت ساختار قبل و بعد از biclustering مشخص شود.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>خوشه‌بندی دوطرفه</b>، الگوهای پیچیده و هم‌زمان در ردیف‌ها و ستون‌های داده را کشف کرد. این روش در تحلیل داده‌های ژنومیک، توصیه‌گرها و ماتریس‌های تعامل بسیار کاربردی و مؤثر است. </p> </p> </div>

### 9.4.1. Linear dimensionality reduction methods for clustering

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">کاهش ابعاد خطی (Linear dimensionality reduction)</span> 📉
</p>
<p style="font-size:16pt; text-align:right;">
روش‌های کاهش ابعاد خطی (Linear dimensionality reduction methods) با هدف کاهش تعداد ویژگی‌ها در یک مجموعه داده در عین حفظ اطلاعات ضروری، عمل می‌کنند. این تکنیک‌ها به ویژه برای خوشه‌بندی داده‌های با ابعاد بالا مفید هستند، زیرا می‌توانند کارایی و تفسیرپذیری الگوریتم‌های خوشه‌بندی را افزایش دهند. برخی از روش‌های رایج کاهش ابعاد خطی برای خوشه‌بندی عبارتند از:
<ol style="font-size:16pt;">
<li><b>تحلیل مؤلفه‌های اصلی (Principal Component Analysis - PCA):</b> PCA مجموعه‌ای از محورهای متعامد (orthogonal axes) (مؤلفه‌های اصلی) را شناسایی می‌کند که حداکثر واریانس را در داده‌ها ثبت می‌کنند. این روش برای کاهش ابعاد در عین حفظ ساختار کلی، مؤثر است.</li>
<li><b>تحلیل تفکیک خطی (Linear Discriminant Analysis - LDA):</b> LDA با هدف حداکثر کردن جداسازی بین کلاس‌های مختلف در داده‌ها عمل می‌کند. این روش به ویژه زمانی برای خوشه‌بندی مفید است که اطلاعات کلاس در دسترس باشد.</li>
<li><b>تجزیه ماتریس غیرمنفی (Non-Negative Matrix Factorization - NMF):</b> NMF ماتریس داده را به ماتریس‌های غیرمنفی تجزیه می‌کند و ویژگی‌های مبتنی بر بخش (parts-based) و قابل تفسیر را نمایش می‌دهد. این روش برای داده‌هایی با مقادیر غیرمنفی مناسب است.</li>
<li><b>تحلیل مؤلفه‌های مستقل (Independent Component Analysis - ICA):</b> ICA به دنبال یافتن مؤلفه‌های مستقل (independent components) در داده‌ها است. این روش برای جداسازی سیگنال‌های مختلط (mixed signals) و شناسایی منابع زیربنایی مفید است.</li>
</ol>
<br>
<p style="font-size:18pt; text-align:right;"> این روش‌های کاهش ابعاد خطی می‌توانند قبل از خوشه‌بندی اعمال شوند تا نویز را کاهش داده، کارایی محاسباتی را بهبود بخشند و کیفیت انتساب خوشه‌ها را افزایش دهند.</p>
<p style="font-size:18pt; text-align:right;"><b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای کاهش ابعاد خطی با استفاده از PCA برای خوشه‌بندی:</b></p>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# تولید داده‌های مصنوعی با ابعاد بالا (۱۰ ویژگی) و ۳ خوشه
X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0, n_features=10)

# اعمال تحلیل مؤلفه‌های اصلی (PCA) برای کاهش ابعاد به ۲ مؤلفه اصلی
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# اعمال خوشه‌بندی KMeans روی فضای کاهش‌یافته
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_pca)

# رسم نتایج خوشه‌بندی در فضای دو مؤلفه اصلی
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='X', s=200, label='Cluster Centers')
plt.title('KMeans Clustering after PCA')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠📉 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، ترکیب دو تکنیک قدرتمند یعنی <b>تحلیل مؤلفه‌های اصلی (PCA)</b> و <b>خوشه‌بندی KMeans</b> برای تحلیل داده‌های مصنوعی با ابعاد بالا به کار گرفته شده است: <ul> <li style="font-size:16pt;">ابتدا داده‌های مصنوعی با ۱۰ ویژگی و ۳ خوشه تولید شده‌اند تا سناریوی خوشه‌بندی در فضای چندبعدی شبیه‌سازی شود.</li> <li style="font-size:16pt;">برای کاهش پیچیدگی و امکان تصویرسازی، از <b>PCA</b> برای کاهش ابعاد به ۲ مؤلفه اصلی استفاده شده است. این مؤلفه‌ها بیشترین واریانس داده را حفظ می‌کنند.</li> <li style="font-size:16pt;">سپس الگوریتم <b>KMeans</b> روی داده‌های کاهش‌یافته اعمال شده تا خوشه‌ها در فضای جدید شناسایی شوند.</li> <li style="font-size:16pt;">در پایان، نقاط داده با رنگ‌بندی بر اساس خوشه‌ها و مراکز خوشه با علامت قرمز رسم شده‌اند تا ساختار خوشه‌بندی به‌وضوح قابل مشاهده باشد.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>کاهش ابعاد</b>، داده‌های پیچیده را ساده‌سازی کرد و سپس با <b>خوشه‌بندی</b>، الگوهای پنهان در آن‌ها را کشف نمود. این ترکیب در بسیاری از کاربردهای عملی مانند تحلیل رفتار مشتری، فشرده‌سازی داده و تصویرسازی داده‌های چندبعدی بسیار مؤثر است. </p> </p> </div>

### 9.4.2. Nonnegative matrix factorization (NMF)

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">تجزیه ماتریس غیرمنفی (Nonnegative Matrix Factorization - NMF)</span> ➕
</p>

<p style="font-size:22pt; text-align:right;">
<br>
تجزیه ماتریس غیرمنفی (Nonnegative Matrix Factorization - NMF) یک تکنیک کاهش ابعاد (dimensionality reduction) است که یک ماتریس داده غیرمنفی را به دو ماتریس با ابعاد پایین‌تر تجزیه می‌کند، که در آن تمام عناصر غیرمنفی هستند. یک ماتریس نشان‌دهنده بردارهای پایه (ویژگی‌ها) و دیگری نشان‌دهنده ضرایب برای هر نقطه داده بر اساس این بردارهای پایه است. NMF به ویژه زمانی مفید است که داده‌ها دارای مقادیر غیرمنفی باشند و زمانی که ویژگی‌های مبتنی بر بخش (parts-based) و قابل تفسیر مورد نظر هستند.
</p>
<br>
<p style="font-size:22pt; text-align:right;">
<b>نکات کلیدی درباره NMF:</b>
</p>
<ul style="font-size:16pt; text-align:right;">
<li><b>محدودیت غیرمنفی بودن (Nonnegativity Constraint):</b> هم بردارهای پایه و هم ضرایب در NMF محدود به غیرمنفی بودن هستند، که باعث می‌شود فاکتورهای حاصل جمعی و قابل تفسیر باشند.</li>
<li><b>کاربردها در خوشه‌بندی (Applications in Clustering):</b> NMF اغلب برای وظایف خوشه‌بندی (clustering tasks) به کار گرفته می‌شود، زیرا می‌تواند نمایش‌های مبتنی بر بخش از داده‌ها را آشکار کند و تفسیر ساختار زیربنایی را آسان‌تر سازد.</li>
<li><b>مناسب برای داده‌های متنی و تصویری (Suitability for Text and Image Data):</b> NMF معمولاً در پردازش زبان طبیعی (natural language processing) برای مدل‌سازی موضوع (topic modeling) و در تحلیل تصویر (image analysis) برای استخراج مؤلفه‌های معنادار استفاده می‌شود.</li>
<li><b>پارامترها (Parameters):</b> تعداد مؤلفه‌ها در تجزیه، یک پارامتر حیاتی است که ابعاد فضای کاهش یافته را تعیین می‌کند.</li>
</ul>
<br>
<p style="font-size:22pt; text-align:right;">
اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای NMF اعمال شده بر خوشه‌بندی ارائه دهیم:
</p>

</div>

```python
# وارد کردن کتابخانه‌های مورد نیاز
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF
from sklearn.cluster import KMeans
from sklearn.pipeline import make_pipeline
import numpy as np
import re

# مرحله ۱: بارگذاری مجموعه داده متنی 20 Newsgroups
# این مجموعه شامل هزاران سند متنی از ۲۰ گروه خبری مختلف است
newsgroups = fetch_20newsgroups(
    subset='all',
    remove=('headers', 'footers', 'quotes')  # حذف اطلاعات متا برای تمرکز روی محتوای اصلی
)

# مرحله ۲: پیش‌پردازش متون
# تبدیل به حروف کوچک + حذف کاراکترهای غیرالفبایی
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    return text

cleaned_data = [preprocess(doc) for doc in newsgroups.data]

# مرحله ۳: تعریف پایپ‌لاین پردازش و خوشه‌بندی
# شامل سه مرحله: بردار TF-IDF → استخراج موضوع با NMF → خوشه‌بندی با KMeans
vectorizer = TfidfVectorizer(
    stop_words='english',  # حذف کلمات رایج انگلیسی
    max_df=0.95,            # حذف کلمات بسیار پرتکرار
    min_df=5                # حذف کلمات بسیار نادر
)

nmf = NMF(
    n_components=20,        # تعداد مؤلفه‌ها (موضوعات) برابر با تعداد خوشه‌ها
    random_state=42
)

kmeans = KMeans(
    n_clusters=20,          # تعداد خوشه‌ها برابر با تعداد گروه‌های خبری
    random_state=42
)

pipeline = make_pipeline(vectorizer, nmf, kmeans)

# مرحله ۴: آموزش پایپ‌لاین روی داده‌های متنی پیش‌پردازش‌شده
pipeline.fit(cleaned_data)

# مرحله ۵: نمایش ۱۰ کلمه برتر برای هر موضوع استخراج‌شده توسط NMF
feature_names = vectorizer.get_feature_names_out()
top_words = np.argsort(nmf.components_, axis=1)[:, :-11:-1]  # انتخاب ۱۰ کلمه با بیشترین وزن

print("Top words per topic:\n")
for i, topic in enumerate(top_words):
    words = [feature_names[j] for j in topic]
    print(f"Topic {i + 1}: {', '.join(words)}")
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🗞️🧠 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، یک پایپ‌لاین کامل برای <b>تحلیل موضوعی و خوشه‌بندی متون خبری</b> با استفاده از مجموعه داده معروف <b>20 Newsgroups</b> پیاده‌سازی شده است. این مجموعه شامل هزاران سند متنی از ۲۰ گروه خبری مختلف است: <ul> <li style="font-size:16pt;">ابتدا داده‌های متنی بارگذاری شده و با حذف کاراکترهای غیرالفبایی و تبدیل به حروف کوچک، پیش‌پردازش شده‌اند.</li> <li style="font-size:16pt;">سپس با استفاده از <b>TF-IDF</b>، متون به بردارهای عددی تبدیل شده‌اند که وزن هر واژه را نسبت به اهمیت آن در سند مشخص می‌کند.</li> <li style="font-size:16pt;">با الگوریتم <b>NMF</b>، مؤلفه‌های اصلی (موضوعات) استخراج شده‌اند. این روش بر پایه تجزیه ماتریس و ساختار افزایشی عمل می‌کند و برای تحلیل موضوعی بسیار مناسب است.</li> <li style="font-size:16pt;">در مرحله آخر، با استفاده از <b>KMeans</b>، اسناد بر اساس نمایش موضوعی‌شان خوشه‌بندی شده‌اند.</li> <li style="font-size:16pt;">برای هر موضوع، ۱۰ واژه کلیدی با بیشترین وزن نمایش داده شده‌اند که نمایانگر محتوای غالب آن موضوع هستند.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با ترکیب <b>بردارسازی متنی، تحلیل مؤلفه‌ها و خوشه‌بندی</b>، ساختار پنهان در مجموعه‌های بزرگ متنی را کشف کرد. این روش در کاربردهایی مانند دسته‌بندی خودکار اسناد، تحلیل احساسات، و توصیه‌گرهای متنی بسیار مؤثر است. </p> </p> </div>

### 9.4.3. Spectral clustering

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی طیفی (Spectral Clustering)</span> 📈
</p>
<p style="font-size:16pt; text-align:right;">
خوشه‌بندی طیفی (Spectral Clustering) تکنیکی است که برای انجام خوشه‌بندی (clustering) از ویژگی‌های طیفی داده‌ها بهره می‌برد. این روش به ویژه هنگام کار با خوشه‌های غیرمحدب (non-convex) و ساختارهای هندسی پیچیده مؤثر است. خوشه‌بندی طیفی شامل گام‌های زیر است:
<ol style="font-size:16pt;">
<li><b>ساخت گراف (Graph Construction):</b> ساخت یک ماتریس وابستگی (affinity matrix) که شباهت دو به دوی بین نقاط داده را نمایش می‌دهد. معیارهای رایج وابستگی شامل شباهت گوسی (Gaussian similarity) یا نزدیک‌ترین همسایه‌ها (nearest neighbors) هستند.</li>
<li><b>تبدیل گراف (Graph Transformation):</b> تبدیل ماتریس وابستگی به یک ماتریس لاپلاسی گراف (graph Laplacian matrix). ماتریس لاپلاسی اتصالات و روابط بین نقاط داده را ثبت می‌کند.</li>
<li><b>تجزیه مقادیر ویژه (Eigenvalue Decomposition):</b> محاسبه مقادیر ویژه (eigenvalues) و بردارهای ویژه (eigenvectors) ماتریس لاپلاسی. بردارهای ویژه مربوط به کوچک‌ترین مقادیر ویژه، ساختار خوشه را در داده‌ها رمزگذاری می‌کنند.</li>
<li><b>کاهش ابعاد (Dimensionality Reduction):</b> استفاده از بردارهای ویژه مربوط به کوچک‌ترین مقادیر ویژه به عنوان یک نمایش کاهش‌یافته از داده‌ها.</li>
<li><b>خوشه‌بندی (Clustering):</b> اعمال یک الگوریتم خوشه‌بندی استاندارد (مانند KMeans) بر روی نمایش کاهش‌یافته برای به دست آوردن خوشه‌های نهایی.</li>
</ol>
<br>
<p style="font-size:18pt; text-align:right;"> این روش برای شناسایی خوشه‌هایی با شکل‌های پیچیده قدرتمند است و در برابر نویز و داده‌های پرت (outliers) مقاوم می‌باشد.</p>
<p style="font-size:18pt; text-align:right;"><b>مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی طیفی:</b></p>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import SpectralClustering
import matplotlib.pyplot as plt

# تولید داده‌های مصنوعی با ۴ خوشه و پراکندگی مشخص
X, _ = make_blobs(n_samples=300, centers=4, random_state=42, cluster_std=1.0)

# اعمال خوشه‌بندی طیفی با affinity از نوع Nearest Neighbors
spectral = SpectralClustering(
    n_clusters=4,
    affinity='nearest_neighbors',
    n_neighbors=120,  # چرا ۱۲۰؟ چرا کمتر نه؟ چرا بیشتر نه؟ / اندکی بیاندیشید و بحث کنید
    random_state=42
)
labels = spectral.fit_predict(X)

# رسم نتایج خوشه‌بندی
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', alpha=0.7)
plt.title('Spectral Clustering (fixed)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🔗📈 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، الگوریتم <b>خوشه‌بندی طیفی (Spectral Clustering)</b> با استفاده از ساختار گراف همسایگی روی داده‌های مصنوعی اعمال شده است. این روش برای شناسایی خوشه‌های پیچیده و غیرخطی بسیار مؤثر است: <ul> <li style="font-size:16pt;">ابتدا داده‌هایی با ۴ خوشه و ۳۰۰ نمونه تولید شده‌اند تا ساختار خوشه‌ای قابل تحلیل باشد.</li> <li style="font-size:16pt;">برای ساخت گراف همسایگی، از پارامتر <code>n_neighbors=120</code> استفاده شده است. این انتخاب نقش کلیدی در کیفیت خوشه‌بندی دارد.</li> <li style="font-size:16pt;">در الگوریتم Spectral Clustering، گراف همسایگی باید <b>connected</b> باشد تا محاسبه eigenvectors برای embedding داده‌ها به‌درستی انجام شود.</li> <li style="font-size:16pt;">اگر مقدار <code>n_neighbors</code> خیلی کم باشد (مثلاً زیر 30)، احتمال دارد گراف disconnect شود و الگوریتم با هشدار یا نتایج ضعیف مواجه شود.</li> <li style="font-size:16pt;">اگر مقدار خیلی زیاد باشد (مثلاً نزدیک به تعداد کل نمونه‌ها)، گراف بیش از حد متصل شده و ساختار محلی داده‌ها از بین می‌رود.</li> <li style="font-size:16pt;">انتخاب ۱۲۰ برای ۳۰۰ نمونه، یک تعادل مناسب بین <b>اتصال گراف</b> و <b>حفظ ساختار محلی</b> ایجاد می‌کند. این مقدار تقریباً بین <code>log(n)</code> و <code>sqrt(n)</code> ضربدر یک ضریب ایمنی است.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که در خوشه‌بندی طیفی، انتخاب مناسب <code>n_neighbors</code> نه‌تنها از نظر عددی بلکه از نظر ساختار گرافی اهمیت بالایی دارد. در پروژه‌های واقعی، این پارامتر باید با آزمون و خطا یا تحلیل ساختار داده تنظیم شود. </p> </p> </div>

### 9.5.2. Similarity measures

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">معیارهای شباهت (Similarity Measures)</span> 🔗
</p>
<p style="font-size:16pt; text-align:right;">
در زمینه خوشه‌بندی داده‌های گراف و شبکه، معیارهای شباهت (similarity measures) نقش حیاتی در کمی‌سازی میزان شباهت یا ارتباط بین گره‌ها ایفا می‌کنند. این معیارها به ساخت ماتریس‌های وابستگی (affinity matrices) کمک می‌کنند که برای الگوریتم‌های خوشه‌بندی مختلف ضروری هستند. برخی از معیارهای رایج شباهت برای داده‌های گراف و شبکه عبارتند از:
<ul style="font-size:16pt;">
<li><b>Jaccard Similarity (شباهت ژاکارد):</b> شباهت بین دو مجموعه را با تقسیم اندازه اشتراک آن‌ها بر اندازه اجتماع آن‌ها اندازه‌گیری می‌کند.</li>
<li><b>Cosine Similarity (شباهت کسینوسی):</b> کسینوس زاویه بین دو بردار را اندازه‌گیری می‌کند. در زمینه گراف‌ها، بردارها نشان‌دهنده همسایگی گره (node neighborhoods) یا بردارهای ویژگی (attribute vectors) هستند.</li>
<li><b>Graph Edit Distance (فاصله ویرایش گراف):</b> با اندازه‌گیری حداقل هزینه تبدیل یک گراف به گراف دیگر از طریق یک سری عملیات ویرایشی (مانند حذف، درج گره یا اصلاح یال)، عدم شباهت بین دو گراف را کمی‌سازی می‌کند.</li>
<li><b>Node and Edge Overlap (همپوشانی گره و یال):</b> همپوشانی گره‌ها یا یال‌ها بین دو گراف را اندازه‌گیری می‌کند و یک معیار ساده از شباهت بر اساس عناصر مشترک را فراهم می‌سازد.</li>
<li><b>Pearson Correlation Coefficient (ضریب همبستگی پیرسون):</b> همبستگی خطی (linear correlation) بین دو متغیر را اندازه‌گیری می‌کند. در زمینه گراف‌ها، می‌توان از آن برای اندازه‌گیری همبستگی بین ویژگی‌های گره (node attributes) استفاده کرد.</li>
</ul>
<br>
<p style="font-size:18pt; text-align:right;">انتخاب یک معیار شباهت مناسب به ویژگی‌های داده‌های گراف و وظیفه خوشه‌بندی خاص بستگی دارد.</p>
<p style="font-size:18pt; text-align:right;"><b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای محاسبه Jaccard Similarity بر روی داده‌های گراف ارائه دهیم:</b></p>
</p>

</div>

```python
import networkx as nx
from sklearn.metrics import jaccard_score
import numpy as np

# ایجاد دو گراف نمونه برای مقایسه
G1 = nx.Graph([(1, 2), (2, 3), (3, 4)])
G2 = nx.Graph([(1, 2), (2, 3), (3, 4), (4, 5)])

# تبدیل گراف‌ها به ماتریس مجاورت (Adjacency Matrix)
adj_matrix1 = nx.to_numpy_array(G1)
adj_matrix2 = nx.to_numpy_array(G2)

# یکسان‌سازی ابعاد ماتریس‌ها با padding صفر
max_size = max(adj_matrix1.shape[0], adj_matrix2.shape[0])
if adj_matrix1.shape[0] < max_size:
    adj_matrix1 = np.pad(adj_matrix1, ((0, max_size - adj_matrix1.shape[0]), (0, max_size - adj_matrix1.shape[1])))
if adj_matrix2.shape[0] < max_size:
    adj_matrix2 = np.pad(adj_matrix2, ((0, max_size - adj_matrix2.shape[0]), (0, max_size - adj_matrix2.shape[1])))

# تبدیل ماتریس‌ها به بردارهای یک‌بعدی برای محاسبه شباهت
flat_matrix1 = adj_matrix1.flatten()
flat_matrix2 = adj_matrix2.flatten()

# محاسبه شباهت جاکارد بین دو گراف
jaccard_similarity = jaccard_score(flat_matrix1, flat_matrix2)

# نمایش نتیجه
print(f"Jaccard Similarity between G1 and G2: {jaccard_similarity}")
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🔗📐 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، هدف مقایسه ساختار دو گراف با استفاده از <b>شباهت جاکارد (Jaccard Similarity)</b> است. این معیار نشان می‌دهد که تا چه حد یال‌های دو گراف با یکدیگر همپوشانی دارند: <ul> <li style="font-size:16pt;">ابتدا دو گراف ساده با تعداد متفاوتی از یال‌ها ایجاد شده‌اند. گراف دوم یک یال بیشتر دارد.</li> <li style="font-size:16pt;">هر گراف به <b>ماتریس مجاورت</b> تبدیل شده که نشان‌دهنده ارتباط بین گره‌هاست.</li> <li style="font-size:16pt;">برای مقایسه، ابعاد ماتریس‌ها با padding صفر یکسان‌سازی شده‌اند تا بتوان آن‌ها را به بردارهای یک‌بعدی تبدیل کرد.</li> <li style="font-size:16pt;">سپس با استفاده از <b>jaccard_score</b>، میزان شباهت بین دو بردار محاسبه شده است. این مقدار بین ۰ تا ۱ است و هرچه بیشتر باشد، شباهت ساختاری بین گراف‌ها بیشتر است.</li> </ul> <p style="font-size:18pt; text-align:right;"> این روش برای مقایسه گراف‌ها در کاربردهایی مانند تحلیل شبکه‌های اجتماعی، تشخیص تغییرات ساختاری در گراف‌ها، و ارزیابی الگوریتم‌های تولید گراف بسیار مفید است. البته برای گراف‌های بزرگ‌تر و پیچیده‌تر، معیارهای پیشرفته‌تری مانند Graph Edit Distance یا Graph Kernels نیز استفاده می‌شود. </p> </p> </div>

### 9.5.3. Graph clustering methods

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">روش‌های خوشه‌بندی گراف (Graph Clustering Methods)</span> 🔗
</p>
<p style="font-size:16pt; text-align:right;">
روش‌های خوشه‌بندی گراف (Graph clustering methods) با هدف شناسایی گروه‌هایی از گره‌ها در داخل یک گراف که به طور چگال‌تری به یکدیگر متصل هستند تا به گره‌های خارج از گروه، عمل می‌کنند. خوشه‌بندی در گراف‌ها به ویژه برای کشف ساختارهای جامعه (community structures)، شناسایی ماژول‌های کارکردی (functional modules) یا تشخیص گروه‌های منسجم از موجودیت‌های مرتبط مفید است. برخی از روش‌های رایج خوشه‌بندی گراف عبارتند از:
<ol style="font-size:16pt;">
<li><b>Louvain Modularity Optimization:</b> مدیولاریتی (modularity) یک پارتیشن را حداکثر می‌کند، که چگالی یال‌ها را در داخل خوشه‌ها در مقایسه با چگالی مورد انتظار در یک شبکه تصادفی اندازه‌گیری می‌کند.</li>
<li><b>Spectral Clustering (خوشه‌بندی طیفی):</b> تجزیه طیفی (spectral decomposition) را بر روی ماتریس لاپلاسی گراف (Laplacian matrix) اعمال می‌کند تا خوشه‌ها را بر اساس بردارهای ویژه (eigenvectors) مربوط به کوچک‌ترین مقادیر ویژه شناسایی کند.</li>
<li><b>Walktrap:</b> شباهت بین گره‌ها را بر اساس پیمایش تصادفی (random walks) در گراف اندازه‌گیری می‌کند، به طوری که گره‌هایی که اغلب با هم پیمایش می‌شوند، مشابه‌تر در نظر گرفته می‌شوند.</li>
<li><b>Infomap:</b> شبکه را به عنوان یک جریان اطلاعات مدل‌سازی کرده و به دنبال یافتن پارتیشن‌هایی است که طول توصیف (description length) جریان اطلاعات را فشرده می‌کند.</li>
<li><b>Kernighan-Lin Bisection:</b> یک روش اکتشافی (heuristic method) است که به صورت بازگشتی گراف را به دو نیمه تقسیم می‌کند و یک تابع هزینه مربوط به برش یال (edge cuts) را بهینه می‌سازد.</li>
</ol>
<br>
<p style="font-size:18pt; text-align:right;">این روش‌ها می‌توانند بر روی انواع مختلفی از گراف‌ها، از جمله شبکه‌های اجتماعی، شبکه‌های بیولوژیکی، شبکه‌های استنادی (citation networks) و غیره اعمال شوند.</p>

```python
import networkx as nx
import community.community_louvain as community_louvain  # کتابخانه Louvain برای خوشه‌بندی گراف
import matplotlib.pyplot as plt

# ایجاد گراف معروف باشگاه کاراته (Karate Club Graph)
# این گراف شامل ۳۴ گره و روابط اجتماعی بین اعضای باشگاه است
G = nx.karate_club_graph()

# اعمال الگوریتم Louvain برای خوشه‌بندی گراف بر اساس ماژولاریتی
partition = community_louvain.best_partition(G)

# تعیین موقعیت گره‌ها برای رسم گراف با layout فنری
pos = nx.spring_layout(G, seed=42)  # برای تکرارپذیری نتایج

# تعیین رنگ هر گره بر اساس خوشه‌ای که به آن تعلق دارد
colors = [partition[node] for node in G.nodes]

# رسم گراف با رنگ‌بندی خوشه‌ها
plt.figure(figsize=(10, 6))
nx.draw(G, pos, node_color=colors, with_labels=True, cmap=plt.cm.viridis, node_size=500)
plt.title('Graph Clustering using Louvain Modularity Optimization')
plt.show()
```

### 9.6.1. Semisupervised clustering on partially labeled data

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی نیمه‌نظارتی (Semi-supervised clustering)</span> 🧩
</p>
<p style="font-size:16pt; text-align:right;">
خوشه‌بندی نیمه‌نظارتی (Semi-supervised clustering) به سناریوهایی می‌پردازد که در آن‌ها تنها زیرمجموعه‌ای از نقاط داده در یک مجموعه داده دارای برچسب هستند. این رویکرد داده‌های دارای برچسب و بدون برچسب را ترکیب می‌کند تا فرآیند خوشه‌بندی را بهبود بخشد. در داده‌های با برچسب جزئی (partially labeled data)، تنها بخشی از نقاط داده دارای برچسب‌های واقعیت زمینی (ground truth labels) هستند، در حالی که بقیه بدون برچسب می‌باشند. روش‌های خوشه‌بندی نیمه‌نظارتی با هدف بهره‌برداری از اطلاعات دارای برچسب موجود برای هدایت خوشه‌بندی کل مجموعه داده عمل می‌کنند.
<br><br>
<b>نکات کلیدی درباره خوشه‌بندی نیمه‌نظارتی بر روی داده‌های با برچسب جزئی:</b>
<ul style="font-size:16pt;">
<li><b>ترکیب اطلاعات با برچسب (Incorporating Labeled Information):</b> خوشه‌بندی نیمه‌نظارتی نمونه‌های دارای برچسب را در فرآیند خوشه‌بندی ادغام می‌کند و دقت انتساب خوشه‌ها را افزایش می‌دهد.</li>
<li><b>محدودیت‌های نرم (Soft Constraints):</b> الگوریتم‌ها اغلب از محدودیت‌های نرمی استفاده می‌کنند که رفتار مشابهی را بین نقاط داده با برچسب و بدون برچسب تشویق می‌کند و به مدل اجازه می‌دهد از هر دو منبع یاد بگیرد.</li>
<li><b>رویکردهای ترکیبی (Hybrid Approaches):</b> روش‌های خوشه‌بندی نیمه‌نظارتی می‌توانند تکنیک‌های خوشه‌بندی سنتی را با الگوریتم‌های یادگیری نظارتی (supervised learning algorithms) ترکیب کنند تا یک مدل ترکیبی ایجاد شود.</li>
<li><b>رسیدگی به برچسب‌های نویزی (Handling Noisy Labels):</b> این روش‌ها برای رسیدگی به نویز و خطاها در داده‌های دارای برچسب طراحی شده‌اند و خوشه‌بندی قوی را در حضور برچسب‌های ناقص فراهم می‌کنند.</li>
</ul>
<br>
<p style="font-size:18pt; text-align:right;"><b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی نیمه‌نظارتی بر روی داده‌های با برچسب جزئی ارائه دهیم:</b></p>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.semi_supervised import LabelPropagation
import numpy as np
import matplotlib.pyplot as plt

# تولید داده‌های مصنوعی با ۳ خوشه مشخص
X, y = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.0)

# شبیه‌سازی داده‌های نیمه‌برچسب‌خورده با حذف تصادفی ۹۰٪ از برچسب‌ها
rng = np.random.RandomState(42)
random_unlabeled_points = rng.rand(len(y)) < 0.9
y[random_unlabeled_points] = -1  # مقدار -1 نشان‌دهنده داده بدون برچسب است

# اعمال خوشه‌بندی KMeans روی کل داده‌ها (بدون توجه به برچسب‌ها)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X)

# اعمال الگوریتم Label Propagation با استفاده از داده‌های نیمه‌برچسب‌خورده
label_propagation = LabelPropagation(kernel='knn', n_neighbors=10)
label_propagation.fit(X, y)
lp_labels = label_propagation.predict(X)

# رسم نتایج خوشه‌بندی برای هر الگوریتم
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', alpha=0.7)
plt.title('KMeans Clustering')

plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=lp_labels, cmap='viridis', alpha=0.7)
plt.title('LabelPropagation Clustering')

plt.show()
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:320%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;"> <p style="font-size:20pt; text-align:center;"> <span style="font-size:26pt; font-weight:bold; color:#1B263B;">توضیحات مثال کد پایتون</span> 🧠🔍 </p> <p style="font-size:18pt; text-align:right;"> در این مثال، دو الگوریتم متفاوت برای خوشه‌بندی داده‌های مصنوعی با سه خوشه به کار گرفته شده‌اند: <b>KMeans</b> و <b>Label Propagation</b>. هدف، مقایسه عملکرد خوشه‌بندی سنتی با یادگیری نیمه‌نظارتی است: <ul> <li style="font-size:16pt;">ابتدا داده‌هایی با سه خوشه تولید شده‌اند که ساختار مشخصی دارند.</li> <li style="font-size:16pt;">برای شبیه‌سازی یادگیری نیمه‌نظارتی، فقط ۱۰٪ از داده‌ها دارای برچسب نگه داشته شده‌اند و باقی با مقدار -1 به عنوان داده‌های بدون برچسب علامت‌گذاری شده‌اند.</li> <li style="font-size:16pt;">الگوریتم <b>KMeans</b> بدون توجه به برچسب‌ها، داده‌ها را خوشه‌بندی کرده و ساختار کلی را بر اساس فاصله‌ها شناسایی می‌کند.</li> <li style="font-size:16pt;">در مقابل، الگوریتم <b>Label Propagation</b> با استفاده از داده‌های برچسب‌خورده، برچسب‌ها را به داده‌های بدون برچسب منتقل می‌کند. این روش از ساختار گرافی داده‌ها و همسایگی برای انتشار اطلاعات استفاده می‌کند.</li> <li style="font-size:16pt;">نتایج هر دو الگوریتم به صورت تصویری نمایش داده شده‌اند تا تفاوت در نحوه خوشه‌بندی قابل مشاهده باشد.</li> </ul> <p style="font-size:18pt; text-align:right;"> این مثال نشان می‌دهد که چگونه می‌توان با استفاده از <b>یادگیری نیمه‌نظارتی</b>، از داده‌های ناقص نیز برای ساخت مدل‌های طبقه‌بندی یا خوشه‌بندی استفاده کرد. همچنین مقایسه با KMeans کمک می‌کند تا مزایا و محدودیت‌های هر روش بهتر درک شوند. </p> </p> </div>

### 9.6.2. Semisupervised clustering on pairwise constraints

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">خوشه‌بندی نیمه‌نظارتی با استفاده از محدودیت‌های زوجی (Pairwise constraints)</span> ⛓️
</p>
<p style="font-size:16pt; text-align:right;">
خوشه‌بندی نیمه‌نظارتی با استفاده از محدودیت‌های زوجی (pairwise constraints) شامل ارائه روابط یا محدودیت‌های زوجی بین نقاط داده به الگوریتم است. این محدودیت‌ها می‌توانند به شکل محدودیت‌های must-link (نشان‌دهنده اینکه دو نقطه باید به یک خوشه تعلق داشته باشند) یا محدودیت‌های cannot-link (نشان‌دهنده اینکه دو نقطه نمی‌توانند به یک خوشه تعلق داشته باشند) باشند. با ترکیب این اطلاعات زوجی، الگوریتم خوشه‌بندی گروه‌بندی نقاط داده را با در نظر گرفتن روابط مشخص شده، هدایت می‌کند.
<br><br>
<b>نکات کلیدی درباره خوشه‌بندی نیمه‌نظارتی بر روی محدودیت‌های زوجی:</b>
<ul style="font-size:16pt;">
<li><b>محدودیت‌های Must-Link و Cannot-Link:</b> محدودیت‌های must-link مشخص می‌کنند که دو نقطه داده باید به یک خوشه تعلق داشته باشند، در حالی که محدودیت‌های cannot-link مشخص می‌کنند که آن‌ها باید در خوشه‌های متفاوتی باشند.</li>
<li><b>ترکیب محدودیت:</b> الگوریتم‌ها از این محدودیت‌های زوجی برای هدایت فرآیند خوشه‌بندی بهره می‌برند و انتساب‌های خوشه را بر اساس روابط مشخص شده تنظیم می‌کنند.</li>
<li><b>رسیدگی به محدودیت‌های نویزی:</b> روش‌های خوشه‌بندی نیمه‌نظارتی بر روی محدودیت‌های زوجی برای رسیدگی به محدودیت‌های نویزی یا متناقض طراحی شده‌اند و مقاومت در برابر اطلاعات ناقص را فراهم می‌کنند.</li>
<li><b>بهبود دقت خوشه‌بندی:</b> با ترکیب محدودیت‌های زوجی، الگوریتم با هدف بهبود دقت انتساب‌های خوشه، به ویژه هنگام کار با ساختارهای پیچیده یا همپوشان، عمل می‌کند.</li>
</ul>

```python
from sklearn.datasets import make_blobs
from metric_learn import MMC_Supervised
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np

# مرحله ۱: تولید داده‌های مصنوعی با ۳ خوشه مشخص
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)

# مرحله ۲: یادگیری متریک با استفاده از الگوریتم نیمه‌نظارتی MMC
# این الگوریتم با استفاده از برچسب‌ها، محدودیت‌های must-link و cannot-link را به‌صورت ضمنی ایجاد می‌کند
mmc = MMC_Supervised()
mmc.fit(X, y_true)
X_transformed = mmc.transform(X)  # تبدیل داده‌ها به فضای جدید با متریک آموخته‌شده

# مرحله ۳: بررسی داده‌های تبدیل‌شده با کاهش بعد PCA برای تصویرسازی
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_transformed)

plt.figure(figsize=(6, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_true, cmap='viridis', alpha=0.7)
plt.title('PCA of Transformed Data (after MMC)')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.show()

# مرحله ۴: اعمال خوشه‌بندی KMeans روی داده‌های تبدیل‌شده
kmeans = KMeans(n_clusters=3, random_state=42)
labels_kmeans = kmeans.fit_predict(X_transformed)

# مرحله ۵: رسم نتایج خوشه‌بندی قبل و بعد از یادگیری متریک
plt.figure(figsize=(12, 5))

# نمایش خوشه‌های واقعی
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis', alpha=0.7)
plt.title('True Clusters')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

# نمایش خوشه‌بندی پس از یادگیری متریک با MMC و اعمال KMeans
plt.subplot(1, 2, 2)
plt.scatter(X_transformed[:, 0], X_transformed[:, 1], c=labels_kmeans, cmap='viridis', alpha=0.7)
plt.title('Semi-supervised Clustering with MMC + KMeans')
plt.xlabel('Transformed Feature 1')
plt.ylabel('Transformed Feature 2')

plt.show()
```

### 9.6.3. Other types of background knowledge for semisupervised clustering

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">استفاده از دانش پس‌زمینه در خوشه‌بندی نیمه‌نظارتی (Semi-supervised clustering with background knowledge)</span> 🧠
</p>
<p style="font-size:16pt; text-align:right;">
در خوشه‌بندی نیمه‌نظارتی (semi-supervised clustering)، بهره‌برداری از دانش پس‌زمینه اضافی فراتر از محدودیت‌های زوجی (pairwise constraints) یا نمونه‌های دارای برچسب، می‌تواند فرآیند خوشه‌بندی را بیشتر بهبود بخشد. انواع دیگر دانش پس‌زمینه ممکن است شامل موارد زیر باشد:
<ul style="font-size:16pt;">
<li><b>محدودیت‌های ویژگی (Attribute Constraints):</b> محدودیت‌ها بر روی ویژگی‌ها یا صفات نقاط داده می‌توانند الگوریتم خوشه‌بندی را هدایت کنند. به عنوان مثال، مشخص کردن اینکه ویژگی‌های خاصی باید در میان خوشه‌ها مشابه یا نامشابه باشند.</li>
<li><b>اطلاعات سلسله‌مراتبی (Hierarchy Information):</b> اطلاعات مربوط به روابط سلسله‌مراتبی بین خوشه‌ها یا نقاط داده می‌تواند ترکیب شود. این مورد به ویژه زمانی مفید است که داده‌ها ساختار تودرتو یا سلسله‌مراتبی از خود نشان می‌دهند.</li>
<li><b>احتمالات پیشین (Prior Probabilities):</b> ارائه احتمالات پیشین برای تعلق نقاط داده به خوشه‌های خاص می‌تواند الگوریتم را در شرایطی که انتظار می‌رود خوشه‌های خاصی غالب‌تر باشند، هدایت کند.</li>
<li><b>محدودیت‌های فضایی (Spatial Constraints):</b> ترکیب محدودیت‌های مبتنی بر آرایش فضایی (spatial arrangement) نقاط داده می‌تواند مفید باشد، به ویژه در کاربردهایی که در آن‌ها نزدیکی یا آرایش حاوی اطلاعات معنادار است.</li>
<li><b>اطلاعات زمانی (Temporal Information):</b> برای داده‌های سری زمانی، محدودیت‌های زمانی یا اطلاعات مربوط به ترتیب زمانی نقاط داده می‌تواند در هدایت فرآیند خوشه‌بندی ارزشمند باشد.</li>
</ul>
<br>
<p style="font-size:18pt; text-align:right;">ترکیب انواع متنوع دانش پس‌زمینه به الگوریتم‌های خوشه‌بندی نیمه‌نظارتی اجازه می‌دهد تا با ویژگی‌های داده‌های مختلف و الزامات خاص کاربردها سازگار شوند.</p>
<p style="font-size:18pt; text-align:right;"><b>اکنون، بیایید یک مثال دنیای واقعی از کاربرد عملی در پایتون (Python) برای خوشه‌بندی نیمه‌نظارتی با محدودیت‌های ویژگی ارائه دهیم:</b></p>
</p>

</div>

```python
from sklearn.datasets import make_blobs
from sklearn.semi_supervised import LabelSpreading
from sklearn.cluster import SpectralClustering
import numpy as np
import matplotlib.pyplot as plt

# مرحله ۱: تولید داده‌های مصنوعی چالش‌برانگیز با همپوشانی بالا
X, y = make_blobs(n_samples=300, centers=3, cluster_std=2.5, random_state=42)

# اطلاعات ویژگی (برای آموزش یا نمایش، در الگوریتم استفاده نمی‌شود)
attribute_information = np.random.rand(len(y), 5)

# شبیه‌سازی محدودیت‌های ویژگی بین خوشه‌ها (برای اهداف آموزشی)
attribute_constraints = []
for i in range(5):
    cluster1, cluster2 = np.random.choice(range(3), size=2, replace=False)
    attribute_constraints.append((cluster1, cluster2, i))

# مرحله ۲: شبیه‌سازی یادگیری نیمه‌نظارتی با حذف ۹۰٪ از برچسب‌ها
y_partial = np.copy(y)
rng = np.random.RandomState(42)
for cls in np.unique(y):
    cls_indices = np.where(y == cls)[0]
    unlabeled_count = int(0.9 * len(cls_indices))
    unlabeled_indices = rng.choice(cls_indices, size=unlabeled_count, replace=False)
    y_partial[unlabeled_indices] = -1  # مقدار -1 نشان‌دهنده داده بدون برچسب است

# مرحله ۳: خوشه‌بندی بدون استفاده از برچسب‌ها با Spectral Clustering
spectral = SpectralClustering(n_clusters=3, random_state=42)
spectral_labels = spectral.fit_predict(X)

# مرحله ۴: خوشه‌بندی نیمه‌نظارتی با استفاده از LabelSpreading
label_spreading = LabelSpreading(kernel='knn', n_neighbors=10, alpha=0.8)
label_spreading.fit(X, y_partial)
lp_labels = label_spreading.transduction_

# مرحله ۵: رسم نتایج خوشه‌بندی
plt.figure(figsize=(12, 5))

# نمودار اول: خوشه‌بندی طیفی بدون دانش نیمه‌نظارتی
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=spectral_labels, cmap='viridis', alpha=0.7)
plt.title('Spectral Clustering (No Semi-supervised Knowledge)')

# نمودار دوم: خوشه‌بندی نیمه‌نظارتی با LabelSpreading
plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=lp_labels, cmap='viridis', alpha=0.7)
plt.title('LabelSpreading (Semi-supervised)')

plt.show()
```
